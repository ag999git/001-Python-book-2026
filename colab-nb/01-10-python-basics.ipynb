## Background: how Python runs your code (Sections 1.2 and 1.3)

There is no code in these sections, but one idea is needed for everything that follows.

A computer cannot run Python text directly. Some program has to turn your code into actions. There are two ways to do this.

- **Compiler:** reads the whole program first, translates all of it into a new file (object code or an executable), and only then is that file run. C++ works this way.
- **Interpreter:** reads your program a little at a time and runs each part straight away. There is no separate translated file for you to manage. Python works this way.

![Interpreted language compared with a compiled language](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-01-nb-01-interpreted-vs-compiled.png)

Path A is an interpreted language like Python. Path B is a compiled language like C++.

Because Python is interpreted, you can use it in two ways:

| | Interactive mode (command line) | Script mode (.py file) |
|---|---|---|
| How you use it | Type one line, press Enter, see the result at once | Write many lines in a file, save it, then run the whole file |
| Good for | Quick tests, trying out an idea, using Python as a calculator | Real programs you want to keep, share and run again |
| Is your code saved? | No | Yes |

A notebook like this one sits in between. Each cell can be run on its own, like interactive mode. But all the cells are saved in one file, like a script.

## Section 1.11: Comments

Python has two ways to write notes inside code:

| | Hash `#` | Triple quotes `"""..."""` or `'''...'''` |
|---|---|---|
| What it really is | A true comment | A string (text value) |

## Section 1.11.1: Statements over many lines, and many statements on one line

A **statement** is one instruction that Python can run. For example, `x = 10` is an assignment statement. `if`, `for` and `while` statements are other kinds.

Normally, a statement ends where the line ends. The next four examples show how to bend this rule, and when you should not.

| Way | How it looks | Needs a special character? | Recommended? |
|---|---|---|---|
| Explicit line continuation | End the line with a backslash `\` | Yes, `\` | Only when there is no better way |
| Implicit line continuation | Keep the statement inside `( )`, `[ ]` or `{ }` | No | Yes, the preferred way |
| Many statements on one line | Separate them with semicolons `;` | Yes, `;` | No, avoid it |
| One statement per line | The normal way | No | Yes |

### Example 9: Explicit line continuation with a backslash

**What it does:** works out `2 + 3 + 4` and shows `9`, even though the sum is split over two lines.

**How it does it:**

- The first three lines are comments. Python skips them.
- Line 4 ends with a backslash `\`. This tells Python: "the statement is not finished, carry on to the next line."
- Python joins line 4 and line 5 and reads them as `2 + 3 + 4`.

**Why:** it lets you break a very long statement into shorter lines that fit on the screen.

**Learning points:**

- The backslash must be the **very last character** on the line. Nothing can come after it: not a comment, and not even a space. Otherwise Python raises `SyntaxError: unexpected character after line continuation character`.
- This is why the comment is written on separate lines above the code, and not after the `\`.

````python
# Backslash (\) indicates explicit line continuation
# But there should be nothing after \ else SyntaxError:
# unexpected character after line continuation character
2 + 3 \
+ 4
````

### Example 10: Implicit line continuation inside brackets

**What it does:** adds `10 + 20 + 30 + 40` and prints `100`.

**How it does it:**

1. Line 1 opens a round bracket `(`.
2. As long as a bracket is open, Python knows the statement is not finished. It keeps reading the next lines.
3. At the closing bracket `)` on line 4, the statement is complete. Python adds the numbers and stores `100` in `result`.
4. Line 5 prints `result`.

**Why:** this is the clean, preferred way to split a long line. You do not need a backslash, and a stray space at the end of a line cannot break it.

**Learning points:**

- An open `(`, `[` or `{` lets a statement continue over many lines.
- Python waits for the matching closing bracket before it treats the statement as complete.
- The indentation of lines 2 and 3 is only for looks. Inside brackets, Python does not care about it.

````python
result = (
    10 + 20 +
    30 + 40
)
print(result)
````

### Example 11: Many statements on one line with semicolons

**What it does:** stores 5 in `a`, stores 6 in `b`, and prints their sum `11`. All three steps are on one line.

**How it does it:** a semicolon `;` separates statements. Python runs them one after the other, from left to right:

1. `a = 5` puts the value 5 in the variable `a`.
2. `b = 6` puts the value 6 in the variable `b`.
3. `print(a + b)` adds them and prints `11`.

**Why it is shown:** so that you can recognise it when you see it. It is allowed, but it is **not** good practice.

**Learning point:** a semicolon lets you put several statements on one line. Avoid it, because:

- the code is harder to read,
- errors are harder to find, since one line now holds several steps,
- the logic is hidden, and mistakes are easier to miss.

````python
a = 5; b = 6; print(a + b)
````

### Example 12: The better version, one statement per line

**What it does:** the same work as Example 11, and prints `11`.

**How:** each statement is on its own line. No semicolons are needed.

**Why:** the Python style guide (PEP 8) recommends one statement per line. Each line does one job, so the code is easy to read and to fix.

**Learning point:** Examples 11 and 12 give the same result. The difference is only readability, and readability matters a lot when you or someone else comes back to the code later.

````python
a = 5
b = 6
print(a + b)
````

## Sections 1.14 and 1.15: Identifiers, naming conventions and PEP 8

An **identifier** is a name you give to something in your code: a variable, a function, a class, a module and so on.

There are two kinds of naming rules:

1. **Legal rules.** Python checks these. Break one and your code will not run.
2. **Conventions.** Python does not check these. They are good habits that most Python programmers follow, mostly taken from PEP 8. Your code will run if you ignore them, but it will be harder for others to read.

### The legal rules

| Rule | What it means | Valid examples | Invalid examples |
|---|---|---|---|
| 1. Allowed characters | Only letters (a-z, A-Z), digits (0-9) and the underscore `_` | `total_marks`, `value2`, `Student_Name` | `hello-world`, `age@school` |
| 2. First character | Must be a letter or an underscore, never a digit | `value1`, `_count`, `name2` | `1value`, `2ndNumber` |
| 3. No keywords | A keyword cannot be used as a name | (none) | `for`, `while`, `class`, `break` |
| 4. No special symbols | Symbols like `@`, `$`, `!`, `#`, `%` and `-` are not allowed | (none) | `first-name`, `price$`, `value#1` |
| 5. Case matters | Names that differ only in capital letters are different names | `student`, `Student`, `STUDENT` are three separate names | (none) |

A word of care about `print`: it is not a keyword in Python 3, so `print = 10` is legal. But it is a very bad idea. After that line, `print` means 10, and `print("Hi")` stops working until you restart the notebook. The same goes for other built-in names such as `list`, `str`, `type` and `id`.

![Checking whether a name is a valid identifier](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-01-nb-07-valid-identifier.png)

### PEP and PEP 8

- **PEP** stands for **Python Enhancement Proposal**. A PEP is a document that proposes or describes something about Python.
- **PEP 8** is the official **style guide** for Python code. It covers naming, spacing, line length and more.

The main naming conventions from PEP 8:

| What you are naming | Style | Example | Avoid |
|---|---|---|---|
| Variables | snake_case: lowercase words joined by underscores | `student_age` | `StudentAge`, `studentage` |
| Functions | snake_case | `calculate_average` | `CalculateAverage` |
| Constants (values that should not change) | UPPER_CASE with underscores | `MAX_LIMIT` | `maxlimit` |
| Classes | PascalCase: each word starts with a capital, no underscores | `StudentRecord` | `studentRecord`, `Attendance_Manager` |
| Names meant for use only inside a module | A single leading underscore | `_internal_utility_function` | (no underscore, if it is meant to be internal) |

About camelCase: in camelCase the first word is lowercase and each later word starts with a capital, as in `myAge`. The capital letter looks like the hump of a camel. camelCase is common in Java and JavaScript. PEP 8 prefers snake_case (`my_age`) for variables and functions in Python.

Also choose names that tell you what they hold. `n = 10` is legal, but `number_of_students = 10` explains itself.

### Example 14: PEP 8 style versus non-PEP 8 style

This script is not printed in the book. It supports Sections 1.14 and 1.15, where the book points to the online resource for more on PEP 8.

**What it does:** shows, side by side, the right way (PEP 8) and the way to avoid (non-PEP 8) for five kinds of code. It prints one line for each section.

**How it does it:** the script has five parts. Each part has a "PEP 8 STYLE (CORRECT)" block and a "NON-PEP 8 STYLE (AVOID)" block.

| Part | Topic | The PEP 8 way | The way to avoid |
|---|---|---|---|
| 1 | Variables and constants | `student_age`, `MAX_LIMIT` | `StudentAge`, `totalScore`, `maxlimit` |
| 2 | Functions | `def calculate_average(...)` with the body on its own indented lines | `CalculateAverage`, with the whole body squeezed onto the `def` line with semicolons |
| 3 | Classes | `StudentRecord`, `AttendanceManager` | `studentRecord`, `Attendance_Manager` |
| 4 | `if` statements | Spaces around `>` and `==`, body on its own indented line | `if x>5: print("x>5")` on one line |
| 5 | Internal names | `_internal_utility_function` with a leading underscore | (not shown) |

**Why:** code is read far more often than it is written. When everyone follows the same style, anyone can pick up anyone's code and read it quickly.

**Learning points:**

- **Both styles run.** Python does not check style. The "AVOID" code runs without any error. You can even see its output: the line `x>5` is printed by the squeezed `if` statement.

````python
# ===============================================================
# PYTHON STYLE GUIDE: PEP 8 VS. NON-PEP 8
# This script demonstrates the "Pythonic" way to write code.
# ===============================================================

# ---------------------------------------------------------------
# 1. VARIABLES: The "Identity" of your data
# Rule: Use snake_case (all lowercase with underscores)
# ---------------------------------------------------------------

# PEP 8 STYLE (CORRECT)
# This is easy to read because words are separated by underscores.
student_age = 20
total_score = 95

# CONSTANTS: Use all UPPERCASE to show this value should not change.
MAX_LIMIT = 100

# NON-PEP 8 STYLE (AVOID)
StudentAge = 20  # Avoid starting variables with Capital Letters
totalScore = 95  # This is "camelCase" - common in Java, but not Python
maxlimit = 100    # Hard to read; always use underscores to separate words

print("Section 1: Variable naming conventions demonstrated.")


# ---------------------------------------------------------------
# 2. FUNCTIONS: Actions your code performs
# Rule: Use snake_case and keep it readable
# ---------------------------------------------------------------

# PEP 8 STYLE (CORRECT)
# Use descriptive names and give the code space to "breathe."
def calculate_average(score1, score2, score3):
    total = score1 + score2 + score3
    return total / 3

# NON-PEP 8 STYLE (AVOID)
# Never put the logic on the same line as the 'def' statement.
def CalculateAverage(score1,score2,score3): total=score1+score2+score3;return total/3

print(f"Section 2: Average is {calculate_average(80, 90, 100)}")


# ---------------------------------------------------------------
# 3. CLASSES: Blueprints for objects
# Rule: Use PascalCase (Capitalize every word, no underscores)
# ---------------------------------------------------------------

# PEP 8 STYLE (CORRECT)
class StudentRecord:
    """Classes represent 'Things' so we capitalize them."""
    pass

class AttendanceManager:
    pass

# NON-PEP 8 STYLE (AVOID)
class studentRecord: # Should be Capitalized
    pass

class Attendance_Manager: # Avoid underscores in class names
    pass

print("Section 3: Class naming (PascalCase) demonstrated.")


# ---------------------------------------------------------------
# 4. CONDITIONALS: Decision making
# Rule: Use spaces around operators and proper indentation
# ---------------------------------------------------------------

x = 10

# PEP 8 STYLE (CORRECT)
# Space around the '>' and the '==' makes it much cleaner.
if x > 5:
    print("Section 4: x is greater than 5")
elif x == 5:
    print("Section 4: x is exactly 5")

# NON-PEP 8 STYLE (AVOID)
# Don't squash code together; it makes debugging very difficult.
if x>5: print("x>5")


# ---------------------------------------------------------------
# 5. INTERNAL NAMES (Advanced)
# Rule: Start with a single underscore for internal/private use
# ---------------------------------------------------------------

# PEP 8 STYLE (CORRECT)
# This tells other programmers: "Use this only inside this script."
def _internal_utility_function():
    return "This is a hidden helper function."

print(f"Section 5: {_internal_utility_function()}")

# ===============================================================
# END OF SCRIPT: All code is PEP 8 compliant and runs perfectly!
# ===============================================================

````

## Section 1.16: Variables, literals, references and built-in types

Some ideas in this part may feel new if you have never programmed before. Do not worry. Run each example, look at the output, and they will become clear.

### A variable is a tag, not a box

A **variable** is a name that refers to a value (an **object**) stored in memory. The way this works is different in C++ and Python.

- **In C++,** a variable is like a **box**. Each box is made for one kind of thing, just as a shoe box is made for shoes and a match box for matches. You must say what kind of box it is (`int`, `float` and so on) before you use it, and it keeps that kind for life.
- **In Python,** a variable is like a **price tag**. The tag has no type of its own. You can take it off a shoe and tie it to a book. The **object** the tag is tied to has the type.

| Feature | C++ (statically typed) | Python (dynamically typed) |
|---|---|---|
| What a variable is | A box that holds a value | A tag (label) tied to a value |
| Must you declare the type? | Yes (`int`, `float`, `string` ...) | No |
| Who has the type? | The box (the variable) | The object (the value) |
| Can the same name later refer to a different type? | No. An `int` variable cannot later hold a string | Yes. The same name can point to any type |
| Everyday picture | Boxes for shoes, boxes for matches | A price tag that can be tied to any item |

**Static and dynamic typing.** C++ is called **statically typed**, because the type of each variable is fixed when you write the code. Python is **dynamically typed**, because Python works out the type from the object at the moment the code runs.

**Python is also strongly typed.** Dynamic does not mean careless. Python will not quietly mix types that do not fit together. For example, `"5" + 5` gives an error instead of guessing what you meant. (See Beyond the Book, Extra 4.)

### Example 15: Two tags on one object, then one tag moves

**What it does:** prints `2` and then `3`.

**How it does it, step by step:**

1. `x = 2`: Python creates the object `2` and ties the tag `x` to it.
2. `y = x`: Python does **not** copy the number. It ties a second tag, `y`, to the **same** object `2`.
3. `y = 3`: Python creates the object `3` and moves the tag `y` to it. Tag `x` is not touched. It is still on `2`.
4. `print(x)` prints `2`, and `print(y)` prints `3`.

**Why:** many beginners expect `x` to change to 3 as well. It does not, because `y = 3` only moves the tag `y`.

**Learning points:**

- `=` means "tie this name to that object". It does not mean "equal" as in maths.
- `y = x` makes both names refer to the same object. It does not create a copy.
- Giving a new value to one name never changes another name.

![Two names on one object, then one name moves](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-01-nb-09-two-tags-one-object.png)

````python
x = 2  # x "points" to 2
y = x  # Now y also "points" to 2
y = 3  # Now y "points" to 3
print(x)  # x still "points" to 2, so it prints 2
print(y)  # y now "points" to 3, so it prints 3
````

**Learning points:**

### Example 17: Casting a float to an int

**Casting** (or **type conversion**) means making a new object of one type from an object of another type.

**What it does:** turns the float `78.11` into the whole number `78`, and prints the **id** (identity number) of each object.

**How it does it:**

1. `my_float = 78.11` creates a float object and ties the tag `my_float` to it.
2. `id(my_float)` gives the identity number of that float object.
3. `int(my_float)` builds a **new** integer object from it. The decimal part `.11` is dropped. The tag `my_int` is tied to this new object `78`.
4. `print(my_int)` prints `78`.
5. `id(my_int)` gives the identity of the integer object. It is a different number from step 2, because it is a different object.
6. `id(78)` gives the same number as `id(my_int)`. This shows that `my_int` points to the object `78`.

**Why:** in Python it is the **object** that is converted, not the name. `int()` does not change the float. It makes a new integer, and the float stays as it was. Numbers cannot be changed in place (they are **immutable**), so converting always creates a new object.

**Learning points:**

- `int()` drops the decimal part. It does not round: `int(78.99)` is also `78`.
- The original object (`78.11`) stays unchanged. Check it by printing `my_float` after the conversion.
- The comments in the code mention `myInt`. They mean the variable `my_int`.

**Signature of `int()`:**

```python
int(x=0)
int(x, base=10)
```

| Parameter | Meaning |
|---|---|
| `x` | The value to convert, for example a float such as `78.11` or a string such as `"78"` |
| `base` | Only used when `x` is a string. The number base, for example 2 for binary. Default is 10 |
| Returns | A new `int` object. `int()` with nothing inside gives `0` |

![Casting a float to an int and checking the ids](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-01-nb-10-float-to-int-cast.png)

````python
my_float = 78.11
print('id of my_float->',id(my_float)) # 1945253350608
my_int = int(my_float)    # int() is a casting function
print(my_int) # myInt will now point to value 78
print('id of my_int->', id(my_int)) # 1945171874576
print(id(78)) # Note id(78) is 1945171874576, same as id(myInt)
````

| | Identifier | Variable |

So every variable has an identifier, but not every identifier is a variable.

### Extra 4: Dynamic, but also strong (the last line gives an error on purpose)
