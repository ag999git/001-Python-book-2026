### <a name="2-1-1-numbers-int-float-complex-and-bool"></a>2.1.1 Numbers (int, float, complex and bool)

Some facts that hold for all number types:

- They store numeric values.
- They are **immutable**: a number object can never be changed. A calculation always creates a new number object.
- A number written directly in the code, like `23`, is a **literal**. A name that points to a number, like `my_numb = 24`, is a **variable**. Both are objects in Python.

| Type | What it holds | Examples | Notes |
|---|---|---|---|
| `int` | Whole numbers, positive or negative | `7`, `-120`, `10**30` | No size limit. Python 3 has only one integer type |
| `float` | Numbers with a decimal point | `3.14`, `-0.5`, `3.33e-2` | `e` or `E` means "times 10 to the power". It is not Euler's number |
| `complex` | A real part and an imaginary part | `3 + 4j`, `5j` | Python uses `j` for the square root of -1 |
| `bool` | `True` or `False` | `True`, `False` | The first letter must be a capital |

**Division always gives a float.** The `/` operator gives a float even when both numbers are whole numbers: `6 / 3` is `2.0`, not `2`. Python converts the result for you. This is an example of **implicit type conversion**. (If you want a whole-number answer, use `//`. You will meet it in the chapter on operators.)

**Mixed arithmetic.** When an operator gets two numbers of different types, Python first converts the "narrower" one to the "wider" one, and then does the sum:

| Narrower | Wider | Example | Result |
|---|---|---|---|
| `int` | `float` | `4 + 3.5` | `7.5` (a float) |
| `float` | `complex` | `2.5 + 1j` | `(2.5+1j)` (a complex) |
| `int` | `complex` | `3 + 2j` | `(3+2j)` (a complex) |

#### <a name="example-1-a-float-that-is-not-exact"></a>Example 1: A float that is not exact

**What it does:** adds 0.1 and 0.2 and prints the result. You might expect `0.3`, but you get `0.30000000000000004`.

**How:** a computer stores a float in **binary** (base 2). Some simple decimal numbers, like 0.1 and 0.2, cannot be written exactly in binary, just as 1/3 cannot be written exactly in decimal (0.3333...). Python stores the closest binary value it can. When the two tiny errors are added, the result is a very small amount above 0.3.

**Why:** this is not a bug in Python. Every language that uses standard floats (C, C++, Java, JavaScript) behaves the same way.

**Learning points:**

- Floats are very close to the true value, but not always exact.
- Do not use floats where every last digit matters, such as money in a bank account. Python has other tools for that (for example the `decimal` module).

````python
print(0.1 + 0.2)  # Output 0.30000000000000004
````

#### <a name="example-2-do-not-compare-floats-with"></a>Example 2: Do not compare floats with `==`

**What it does:** checks whether `0.1 + 0.2` is equal to `0.3`. The answer is `False`.

**How:** `==` checks whether two values are exactly equal. Because of the tiny error seen in Example 1, `0.30000000000000004` and `0.3` are not exactly equal.

**Why:** this can cause hidden bugs. An `if` test that "should" be true quietly fails.

**Learning point:** never test floats for exact equality with `==`. Use `math.isclose()` instead (Example 3).

````python
0.1 + 0.2 == 0.3    # False
````

#### <a name="example-3-comparing-floats-safely-with-math-isclose"></a>Example 3: Comparing floats safely with `math.isclose()`

**What it does:** checks whether `0.1 + 0.2` and `0.3` are close enough to be treated as equal. The answer is `True`.

**How it does it:**

1. `import math` loads the `math` module, which holds many maths functions.
2. `math.isclose(a, b)` returns `True` when `a` and `b` differ by only a very tiny amount.

**Why:** this is the safe way to compare two floats.

**Signature of `math.isclose()`:**

```python
math.isclose(a, b, *, rel_tol=1e-09, abs_tol=0.0)
```

| Parameter | Meaning | Default |
|---|---|---|
| `a`, `b` | The two numbers to compare | (required) |
| `rel_tol` | How close they must be, as a fraction of the larger value. `1e-09` means "equal to about 9 decimal places" | `1e-09` |
| `abs_tol` | A fixed smallest difference to accept. Useful when comparing with values near zero | `0.0` |

The `*` in the signature means that `rel_tol` and `abs_tol` must be given by name, for example `math.isclose(a, b, rel_tol=0.01)`.

````python
import math
math.isclose(0.1 + 0.2, 0.3)   # True
````

### <a name="2-1-2-complex-numbers"></a>2.1.2 Complex numbers

A complex number has two parts: a **real** part and an **imaginary** part. It is written `x + yj`:

- `x` is the real part.
- `y` is the imaginary part. The `j` stands for the square root of -1. Maths books use `i`, but Python uses `j`, as engineers do.
- Both parts can be ints or floats, but Python always stores them as floats.

The order matters: `3 + 2j` is not the same as `2 + 3j`. This is why a complex number is called an **ordered pair**.

Three ways to make a complex number:

| Way | Example | Result |
|---|---|---|
| Write it directly | `z = 3 + 4j` | `(3+4j)` |
| Use the `complex()` function | `z = complex(3, 4)` | `(3+4j)` |
| Imaginary part only | `z = 5j` | `5j` (the real part is 0) |

Tools for working with a complex number `z`:

| Tool | Kind | What it gives |
|---|---|---|
| `z.real` | attribute | The real part |
| `z.imag` | attribute | The imaginary part |
| `z.conjugate()` | method | The same number with the sign of the imaginary part flipped |
| `abs(z)` | built-in function | The size (magnitude) of the number |

Notice that `real` and `imag` have no brackets. They are **attributes**: stored values. `conjugate()` has brackets because it is a **method**: an action that works something out.

Complex numbers are used in science and engineering, for example with signals and sine waves, AC circuits, quantum physics, control systems and Fourier transforms. Python handles them without any extra library.

#### <a name="example-4-the-parts-of-a-complex-number"></a>Example 4: The parts of a complex number

**What it does:** creates `z = 3 + 4j` and prints its real part, imaginary part and conjugate.

**How it does it:**

1. `z = 3 + 4j` creates the complex number.
2. `z.real` gives `3.0`. It is a float, even though we wrote `3`.
3. `z.imag` gives `4.0`, again a float.
4. `z.conjugate()` flips the sign of the imaginary part and gives `(3-4j)`.

**Learning points:**

- The parts of a complex number are always stored as floats.
- Attributes are used without brackets; methods need brackets.

````python
z = 3 + 4j
print(z.real)         # 3.0
print(z.imag)         # 4.0
print(z.conjugate())  # (3-4j)
````

#### <a name="example-5-arithmetic-with-complex-numbers"></a>Example 5: Arithmetic with complex numbers

**What it does:** multiplies `(3+4j)` by `(2-1j)` and prints `(10+5j)`.

**How:** Python multiplies out the brackets, just as in algebra, and uses the rule that `j * j = -1`:

| Step | Working |
|---|---|
| 1. Multiply each part | `3*2 + 3*(-1j) + 4j*2 + 4j*(-1j)` |
| 2. Simplify | `6 - 3j + 8j - 4j*j` |
| 3. Use `j*j = -1` | `6 - 3j + 8j + 4` |
| 4. Collect the parts | `10 + 5j` |

**Learning point:** complex numbers support `+`, `-`, `*` and `/` just like other numbers. Python does all the working for you.

````python
print((3+4j) * (2-1j))   # (10+5j)
````

#### <a name="example-6-the-size-magnitude-of-a-complex-number"></a>Example 6: The size (magnitude) of a complex number

**What it does:** works out the size of `3 + 4j`, which is `5.0`.

**How:** `abs()` of a complex number `x + yj` gives the square root of `x*x + y*y`. Here that is the square root of 9 + 16 = 25, which is 5.0. It is the length of the line from the origin to the point (3, 4), the same as the long side of a 3-4-5 right-angled triangle.

**Learning point:** for ordinary numbers, `abs()` removes the minus sign (`abs(-7)` is `7`). For complex numbers, it gives the magnitude, and the answer is always a float.

````python
abs(3 + 4j)   # 5.0
````

**What it does:** tries to check whether `3+4j` is greater than `1+2j`. Python stops with a `TypeError`.

#### <a name="example-9-complex-numbers-all-together"></a>Example 9: Complex numbers, all together

**What it does:** creates complex numbers in three different ways and then prints the parts, the conjugate and the size of one of them.

**How it does it, line by line:**

| Line | Code | What happens | Output |
|---|---|---|---|
| 1 | `print(2j)` | A complex number with only an imaginary part | `2j` |
| 2 | `print(1 + 2j)` | A complex number with both parts | `(1+2j)` |

| 6 | `print(z.imag)` | The imaginary part | `3.0` |
| 7 | `print(z.conjugate())` | The sign of the imaginary part is flipped | `(2-3j)` |
| 8 | `print(abs(z))` | The square root of 2*2 + 3*3 = 13 | `3.605551275463989` |

**Learning point:** Python prints a complex number with brackets, like `(1+2j)`, when it has a real part. When the real part is zero, it prints just the imaginary part, like `2j`.

```python
complex(real=0, imag=0)

```

### <a name="2-1-3-bool"></a>2.1.3 Bool

A **Boolean** value is either `True` or `False`. The type is called `bool`.

- The first letter must be a capital. `true` and `false` are not Boolean values in Python. Python would treat them as variable names.
- PEP 8 (the Python style guide) says Boolean variables should be in lowercase with underscores, like any other variable.

| Prefix | Use it for | Examples |
|---|---|---|
| `is_` | A state or identity. The most common prefix | `is_valid`, `is_active` |
| `has_` | Whether something is owned or present | `has_permission`, `has_data` |
| `can_` | Whether something is possible or allowed | `can_delete`, `can_upload` |
| `should_` | Whether an action is recommended, or a flag for later | `should_update`, `should_log` |

A name like `is_active` reads almost like English in an `if` statement: `if is_active:`.

### <a name="2-1-4-sequences-and-other-containers"></a>2.1.4 Sequences and other containers

A **container** is an object that holds other objects. Python has two kinds:

- **Sequences** keep their items in order. Each item has a position number called an **index**. Think of beads on a string: each bead has a place.
- **Non-sequence containers** do not use positions. Think of a bag: the items are inside, but in no particular order. You find an item by what it is, or by its key, not by its place.

| Type | Kind | Written with | Can you change it? | Duplicates allowed? | Get an item by |
|---|---|---|---|---|---|
| `str` | Sequence | `'...'` or `"..."` | No | Yes | Index, `s[0]` |
| `list` | Sequence | `[ ]` | Yes | Yes | Index, `a[0]` |
| `tuple` | Sequence | `( )` | No | Yes | Index, `t[0]` |
| `dict` | Mapping (not a sequence) | `{key: value}` | Yes | Keys: no. Values: yes | Key, `d["name"]` |
| `set` | Set (not a sequence) | `{ }` with items, or `set()` | Yes | No | Not possible. You can only test membership with `in` |

The next sections look at each of these in turn.

### <a name="2-1-5-strings"></a>2.1.5 Strings

A **string** (type `str`) is a piece of text: a sequence of characters inside quotes.

- You can use single quotes `'Hello'` or double quotes `"Hello"`. Both give the same kind of string.
- Each character has an **index** (position number). Counting starts at 0, so for a string of length `n` the last character is at index `n-1`.
- A **slice** `s[m:n]` gives a part of the string. It starts at index `m` (included) and stops just before index `n` (excluded).
- `+` joins two strings (this is called **concatenation**).
- `*` with a whole number repeats a string.

[Back to the Table of Contents](#toc)

#### <a name="example-10-working-with-a-string"></a>Example 10: Working with a string

**What it does:** creates the string `'Hello World!'` and tries eleven common string operations on it: indexing, slicing, joining, repeating, length, upper and lower case, searching and replacing.

**How:** each `print()` line does one operation. The expected output is in the comment on the same line, and a short explanation is on the line below it.

**Why:** strings are used in almost every program: names, messages, file contents, web pages. These few operations cover a large part of everyday string work.

**Learning points:**

- Indexing `s[i]` gives one character. Slicing `s[m:n]` gives a piece of the string.
- Negative indexes count from the end: `-1` is the last character.
- `+` joins strings and `*` repeats them.
- Methods such as `upper()`, `lower()` and `replace()` do **not** change `my_str`. Strings cannot be changed (they are immutable). These methods return a **new** string. You can check this: `my_str` still prints `Hello World!` at the end.

The index numbers of `'Hello World!'`:

````python
my_str = 'Hello World!' # A string assigned to variable my_str
print(str(my_str))  # Expected output: Hello World!
# str() converts to a string (although my_str is already a string)

print(my_str[0])  # Expected output: H
# Indexing starts from 0, so my_str[0] gives the first character.

print(my_str[1:5])  # Expected output: ello
# Characters from index 1 to index 4 (5 is excluded).
# Slice length = 5 - 1 = 4 characters.

print(my_str + ' From me')  # Expected output: Hello World! From me
# + joins two strings together.

print(my_str * 3)  # Expected output: Hello World!Hello World!Hello World!
# * repeats the string 3 times.

print(len(my_str))  # Expected output: 12
# len() returns the number of characters in the string.

print(my_str[-1])  # Expected output: !
# Negative indexing: -1 refers to the last character.

print(my_str.upper())  # Expected output: HELLO WORLD!
# .upper() converts all letters to uppercase.

print(my_str.lower())  # Expected output: hello world!
# .lower() converts all letters to lowercase.

print('World' in my_str)  # Expected output: True
# 'in' checks whether the substring exists inside the string.

print(my_str.replace('World', 'Python'))  # Expected output: Hello Python!
# replace(old, new) substitutes part of the string.
````

### <a name="2-1-6-lists"></a>2.1.6 Lists

A **list** is a container that holds other objects in order.

- It is written with square brackets, with items separated by commas: `['dog', 'cat', 'parrot']`.
- It can hold any objects: numbers, strings, other lists, or a mix: `[10, 'apple', 3.5, True]`.
- It is a **sequence**: every item has an index, starting at 0.

A list is similar to an array in C++. The main difference is that a C++ array holds items of one type only, while a Python list can mix types.

**How:** each block does one operation and prints the list, so you can watch it change.

**Why:** lists are the most used container in Python. You will use these operations all the time.

**Learning points:**

- `append()` adds to the end.
- `remove()` deletes the first item that matches the value.
- `pets[2] = 'turtle'` **replaces** the item at index 2. It does not insert a new one.
- `+` joins two lists into a new list. The original lists are not changed.

The starting list and its indexes:

| Index | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|---|

### <a name="2-1-7-tuples"></a>2.1.7 Tuples

A **tuple** is like a list that cannot be changed.

- It is a sequence, so its items have indexes starting at 0.
- It is written with round brackets: `('cat', 'dog')`. (A list uses square brackets.)
- It is **immutable**: once made, you cannot change, add or remove items. Think of it as a "read-only list".

#### <a name="example-12-working-with-tuples"></a>Example 12: Working with tuples

**How:** each numbered block shows one idea. Block 7 is commented out, because it would stop the program with an error.

**Why:** tuples are used for data that should stay fixed, such as the coordinates of a point or a date. Python also uses them behind the scenes, for example when a function returns several values.

**Learning points:**

- Changing an item of a tuple gives a `TypeError` (block 7).
- A one-item tuple needs a comma: `('apple',)`. Without the comma, `('apple')` is just a string in brackets (block 8). It is the comma, not the bracket, that makes a tuple.
- **Packing** (block 9): writing values separated by commas makes a tuple, even without brackets.

### <a name="2-1-8-dictionaries"></a>2.1.8 Dictionaries

A **dictionary** (type `dict`) stores data as **key-value pairs**, like `{"name": "John", "age": 20}`.

- Think of a real dictionary: you look up a word (the key) to find its meaning (the value).
- It is written with curly braces. A colon joins each key to its value, and commas separate the pairs.
- It is not a sequence. You cannot use an index like `d[0]`. You use the key: `d["name"]`.
- Keys must be unique. If you use the same key twice, the second value replaces the first. Different keys may have the same value.
- Keys must be **immutable** (strings, numbers or tuples). A list cannot be a key. Values can be anything.
- A dictionary is **mutable**: you can add, change and remove pairs.
- Since Python 3.7, a dictionary remembers the order in which pairs were added. But you still cannot get an item by position.

A dictionary works a little like a maths function `f(x) = y`: the key is like `x`, and the value is like `f(x)`.

The dictionary used in Example 13, drawn as keys and values:

| Key | `"name"` | `"age"` | `"course"` |
|---|---|---|---|
| Value | `"Alice"` | `20` | `"Maths"` |

#### <a name="example-13-working-with-a-dictionary"></a>Example 13: Working with a dictionary

**What it does:** creates a student record and shows ten common dictionary operations.

**How it does it:**

| Block | Code | What happens |

| 2 | `student["name"]` | Looks up the value for the key `"name"` |

| 4 | `student["name"] = "Bob"` | The key `"name"` already exists, so its value is replaced |

````python
# 1. Create dictionary student with key-value pairs inside {}.
student = {"name": "Alice", "age": 20, "course": "Maths"}
print(student) # {'name': 'Alice', 'age': 20, 'course': 'Maths'}

# 2. Accessing a value using its key inside square brackets
print(student["name"])  # Alice

# 3. Adding a new key-value pair. Dictionaries are mutable
student["year"] = 2026 # You can add new key/ value pairs anytime.
print(student) # {'name': 'Alice', 'age': 20, 'course': 'Maths', 'year': 2026}

# 4. Updating the value of an existing key.
student["name"] = "Bob"  # Assign a new value to update dictionary.
print(student)  # {'name': 'Bob', 'age': 20, 'course': 'Maths', 'year': 2026}

# 5. Removing a key-value pair
del student["course"]  # del removes a key/ value from dictionary.
print(student)  # {'name': 'Bob', 'age': 20, 'year': 2026}

# 6. Checking if a key exists using 'in' operator
print("age" in student)  # True

# 7. Getting the list of all keys using .keys()
print(student.keys()) # dict_keys(['name', 'age', 'year'])

# 8. Getting the list of all values using .values()
print(student.values()) # dict_values(['Bob', 20, 2026])

# 9. Getting key-value pairs together using .items()
print(student.items()) # dict_items([('name', 'Bob'), ('age', 20), ('year', 2026)])
# .items() returns each key and its value as a tuple.

# 10. Using .get() to safely access values
print("student get course:->", student.get("course")) # None
# .get() returns 'None' instead of causing an error if the key does not exist.
````

### <a name="2-1-9-sets"></a>2.1.9 Sets

A **set** is an unordered collection of **unique** items.

- Items have no fixed position, so you cannot use indexing (`s[0]`) or slicing (`s[1:3]`).
- Duplicates are removed automatically.
- The set itself is **mutable**: you can add and remove items.
- But each **item** must be **immutable**: numbers, strings and tuples are allowed; lists and dictionaries are not. This is because a set uses a technique called **hashing** to find items quickly, and hashing only works for values that cannot change.
- You can turn a list or a tuple into a set with `set()`.

A set is like a dictionary that has keys but no values. Both use curly braces `{ }`. But `{}` on its own makes an empty **dictionary**, not an empty set. For an empty set you must write `set()`.

[Back to the Table of Contents](#toc)

**What it does:** creates sets in three ways (from items, from a string, from a list), shows that a list cannot be an item, and then counts, adds and removes items.

**How it does it:**

1. `{"apple", "banana", "mango", "banana"}`: the second `"banana"` is dropped, so the set has 3 items.
2. `set("hello world")`: a string is a sequence of characters, so `set()` keeps one copy of each character, including the space.
3. `set([1, 2, 2, 3, 4, 4, 5])`: the duplicates 2 and 4 are dropped.
4. `{1, 2, [3, 4]}` would fail, because a list is mutable. That line is commented out.
5. `len(fruits)` counts the items: 3.
6. `add("orange")` adds one item.
7. `remove("banana")` deletes an item. It gives a `KeyError` if the item is not there.
8. `discard("papaya")` deletes an item if it is there, and quietly does nothing if it is not.

**Why:** sets are the quickest way to remove duplicates and to check whether an item is present.

**Learning points:**

| Method | If the item is present | If the item is missing |
|---|---|---|
| `s.remove(x)` | Removes it | Stops with `KeyError` |
| `s.discard(x)` | Removes it | Does nothing, no error |

````python
# 1. Creating a simple set of fruits
fruits = {"apple", "banana", "mango", "banana"}
print(fruits)  # {'apple', 'banana', 'mango'}
# Explanation: Duplicate 'banana' is removed because sets only keep unique elements.

# 2. Creating a set using a string
charSet = set("hello world")
print(charSet)  # {'d', 'e', 'h', 'l', 'o', 'r', 'w', ' '}
# Explanation: set() creates a set of unique characters from the string.

# 3. Creating a set from a list
numSet = set([1, 2, 2, 3, 4, 4, 5])
print(numSet)  # {1, 2, 3, 4, 5}
# Explanation: set() removes duplicate items automatically.

# 4. Error creating a set with mutable items
# The following line is commented out to avoid runtime error
# invalidSet = {1, 2, [3, 4]}  # Raises a TypeError
# Explanation: Lists are mutable and cannot be added to a set.

# 5. Checking the number of items in a set
print(len(fruits))  # 3
# Explanation: The set has 3 unique items.

# 6. Adding an item to a set
fruits.add("orange")
print(fruits)  # {'apple', 'banana', 'mango', 'orange'}
# Explanation: add() inserts new element (order not guaranteed).

# 7. Removing an item from a set using remove()
fruits.remove("banana")
print(fruits)  # {'apple', 'mango', 'orange'}
# Explanation: remove() deletes the specified item. Error if item not present.

# 8. Safely removing an item from a set using discard()
fruits.discard("papaya")  # papaya not in the set. But no error
print(fruits)  # {'apple', 'mango', 'orange'}
# Explanation: discard() does nothing if the item is missing.
````

#### <a name="example-15-sets-part-2"></a>Example 15: Sets, part 2

**What it does:** checks for an item, then combines two sets in three ways (union, intersection, difference), and finally creates an empty set.

**How:** with `A = {1, 2, 3}` and `B = {3, 4, 5}`:

| Operation | Method | Same thing with an operator | Result | Meaning |
|---|---|---|---|---|
| Union | `A.union(B)` | `A \| B` | `{1, 2, 3, 4, 5}` | Everything in A or B (3 appears once) |
| Intersection | `A.intersection(B)` | `A & B` | `{3}` | Only what is in both |
| Difference | `A.difference(B)` | `A - B` | `{1, 2}` | What is in A but not in B |

**Why:** these are the same set operations you may know from school maths (Venn diagrams). They are very handy, for example to find students who are in both the cricket team and the chess club.

**Learning points:**

- These methods return a **new** set. `A` and `B` are not changed.
- `B.difference(A)` is not the same as `A.difference(B)`. Try it: it gives `{4, 5}`.
- An empty set must be written `set()`. Writing `{}` makes an empty dictionary. Check with `type({})`.

![Union, intersection and difference of two sets](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-02-nb-05-set-operations.png)

````python
# 1. Checking whether an item exists in a set
fruits = {"apple", "banana", "mango", "banana"}
print("mango" in fruits)  # True
# Explanation: 'in' checks membership in the set.

# 2. Union of two sets. Removes duplicates if present
A = {1, 2, 3}  # 3 is common in both A and B
B = {3, 4, 5}
print(A.union(B))  # {1, 2, 3, 4, 5}. 3 appears only once in result.
# Explanation: union() joins items from both sets.

# 3. Intersection of two sets
print(A.intersection(B))  # {3}
# Explanation: intersection() returns only common elements.

# 4. Difference of two sets
print(A.difference(B))  # {1, 2}
# Explanation: Items in A that are not in B.

# 5. Creating an empty set (IMPORTANT)
empty_set = set()
print(empty_set)  # set()
# Explanation: use set() to make an empty set.
````

### <a name="2-1-10-the-value-none"></a>2.1.10 The value None

`None` is a special value that means "no value here" or "nothing yet".

- It has its own type: `NoneType`. `None` is the only value of that type.
- You can assign it like any other value: `x = None`.
- Other languages, such as Java and C++, use `null` for a similar idea.

`None` is **not** the same as:

| Value | What it means | Is it equal to `None`? |
|---|---|---|

## <a name="2-2-mutable-and-immutable-objects"></a>2.2 Mutable and immutable objects

In Python, every piece of data is an **object**: numbers, strings, lists, functions, everything. Every object has three things:

| Property | What it is | How to see it |
|---|---|---|
| Identity | A number unique to the object while it exists (in standard Python, its memory address) | `id(obj)` |
| Type | What kind of object it is | `type(obj)` |
| Value | The data it holds | `print(obj)` |

- A **mutable** object can be changed after it is created. Its value changes, but its identity stays the same. Examples: `list`, `dict`, `set`, `bytearray`.
- An **immutable** object cannot be changed. If you "change" it, Python quietly creates a **new** object with a new identity. Examples: `int`, `float`, `str`, `tuple`, `bool`, `frozenset`.

How to find out which one you have: compare the id before and after the change.

#### <a name="example-18-lists-and-dictionaries-are-mutable"></a>Example 18: Lists and dictionaries are mutable

**What it does:** changes a list and a dictionary, and prints the id before and after each change. The ids stay the same.

**How it does it:**

1. `numbers = [1, 2, 3]` creates a list. Its id is printed.
2. `numbers.append(4)` adds an item **to the same list**. The printed id does not change.
3. `d = {"a": 1, "b": 2}` creates a dictionary. Its id is printed.
4. `d["c"] = 3` adds a pair **to the same dictionary**. The printed id does not change.

**Why:** changing an object in place is fast, and every name that points to the object sees the change.

**Learning points:**

- A mutable object keeps its identity when its contents change.
- The long id numbers in the comments are the ones the author got. Yours will be different, and they change each time you restart the notebook. What matters is that the two ids in each pair are the same.

````python
# 1. Lists are mutable. Can be changed after they are created.
# Modifying list, does not create a new list; instead, it changes
# the existing list in place.
numbers = [1, 2, 3]
print("Before appending:", numbers, "id:", id(numbers))
# Output: Before appending: [1, 2, 3] id: 140353303123456
numbers.append(4)
# The list changed, but its id() stays the same.
print("After appending:", numbers, "id:", id(numbers))
# Output: After appending: [1, 2, 3, 4] id: 140353303123456

# 2. Dictionaries are mutable. They can be changed after they are created.
d = {"a": 1, "b": 2}
print(id(d))  # 140733544647664
d["c"] = 3  # Modifying dictionary by adding new key-value pair
print(id(d))  # 140733544647664 (same as previous id)
# Explanation: Dictionary was modified in-place (same object).
````

#### <a name="example-19-integers-and-strings-are-immutable"></a>Example 19: Integers and strings are immutable

**What it does:** "changes" an integer and a string, and prints the id before and after. This time the ids are different.

**How it does it:**

1. `x = 10` ties `x` to the integer object `10`. Its id is printed.
2. `x = x + 5` works out `15`. The object `10` cannot be changed, so Python creates a **new** object `15` and moves the name `x` to it. The id is different.
3. `s = "hello"` ties `s` to a string. Its id is printed.
4. `s = s + " world"` creates a **new** string `"hello world"` and moves `s` to it. The id is different.

**Why:** immutable objects are safe. Nothing can change them behind your back. Because they never change, they can also be used as dictionary keys and set items.

**Learning point:** for an immutable object, "changing the value" really means "making a new object and moving the name to it".

````python
# 1. Immutable example using integers
x = 10
print(id(x))  # 140733544647384
x = x + 5
print(id(x))  # 140733544647456 (different from previous id)
# Explanation: NEW int object created because ints are immutable

# 2. Immutable example using strings
s = "hello"
print(id(s))  # 140733544647520
s = s + " world"  # Modifying string creates a NEW string object
print(id(s))  # 140733544647584 (different from previous id)
# Explanation: Strings cannot change; Python creates NEW string object.

````

##### Notes on Example 19

**Why have both kinds of object?**

| Mutable objects are useful because | Immutable objects are useful because |
|---|---|
| Data that needs updating (lists, dictionaries) can be changed directly | They cannot be changed by accident, so they are safer |
| Changing in place is efficient when you make many changes | They can be used as dictionary keys and set items (they are **hashable**) |
| | Programs that use them are easier to reason about |

**Two points that often confuse beginners:**

1. Strings **look** changeable (you can write `s = s + " world"`), but they are not. A new string is made each time.
2. A tuple is immutable, but it **can hold** mutable items. For example, in `t = ([1, 2], 3)`, you cannot replace `t[0]`, but you can still `append` to the list inside it.

**When to use which:** use mutable types when your data needs to change. Use immutable types for fixed values and for dictionary keys.

##### Notes on Example 28: mixed arithmetic, narrow and wide types

Python allows mixed arithmetic only when the types can be converted safely.

- `'Hello' + 1` is not allowed in Python. Java would allow it: it would turn the `1` into a string and give `'Hello1'`.
- Python's rule comes from the "Zen of Python": **explicit is better than implicit**. The programmer should make such conversions on purpose, not by accident.
- So to add the text `'234'` and the number `5`, you must choose: `int('234') + 5` gives `239`, while `'234' + str(5)` gives `'2345'`.

A conversion can be **narrow** or **wide**:

### <a name="2-3-5-and-2-3-6-implicit-conversion-in-a-boolean-context"></a>2.3.5 and 2.3.6 Implicit conversion in a Boolean context

Some places in Python expect a `True` or `False` answer. These are called **Boolean contexts**:

1. `if` statements
2. `while` loops
3. the logical operators `and`, `or` and `not`
4. the function `bool()`

You can put a value that is **not** a Boolean in such a place. Python then turns it into `True` or `False` by itself. This is called **implicit type conversion to bool**. For example:

```python
if my_list:
    print("List is not empty")
```

Here the list is not compared with `True`. Python converts the list to a Boolean: an empty list counts as `False`, and a list with items counts as `True`. (A full, runnable version of this pattern is in Example 31.)

### <a name="2-3-7-and-2-3-8-values-that-count-as-false-and-values-that-c"></a>2.3.7 and 2.3.8 Values that count as False and values that count as True

A value that counts as `False` is called **falsy**. A value that counts as `True` is called **truthy**.

| Falsy (count as `False`) | Truthy (count as `True`) |
|---|---|
| `None` | Everything else, including: |
| `False` | `True` |
| Zero of any number type: `0`, `0.0`, `0j` | Any non-zero number, positive or negative: `5`, `-6`, `3.14` |
| Empty string `""` | Any non-empty string, even `" "` (a single space) |
| Empty list `[]`, empty tuple `()` | Any list or tuple with at least one item |
| Empty dictionary `{}`, empty set `set()` | Any dictionary or set with at least one item |
| Empty range `range(0)` | Any object that does not define its own "false" behaviour |

**Why this matters:** you can write short, readable tests like these:

```python
if message:
    print("Message is not empty")

if items:
    print("List contains elements")

if num:
    print("Number is not zero")
```

#### <a name="example-29-converting-values-to-bool"></a>Example 29: Converting values to bool

**What it does:** converts thirteen different values to `True` or `False` with `bool()` and prints the answers.

**How:** `bool(value)` applies the falsy and truthy rules from the table above.

**Why:** you will rarely call `bool()` yourself, but Python does exactly this every time you write `if value:`. Seeing the answers helps you predict what an `if` will do.

**Learning points:**

- Zero is falsy; every other number, even a negative one, is truthy.
- An empty string is falsy, but a string with just a space is truthy. A space is a character.
- Empty containers are falsy; containers with anything in them are truthy.
- `None` is always falsy.

````python
print(bool(5))        # True  -> non-zero integer is True
print(bool(-6))       # True  -> negative non-zero integer is True
print(bool(0))        # False -> zero is False

print(bool("hello"))  # True  -> non-empty string is True
print(bool(""))       # False -> empty string is False
print(bool(" "))      # True  -> string with space is NOT empty

print(bool(0.0))      # False -> float zero is False
print(bool(3.14))     # True  -> any non-zero float is True

print(bool([]))       # False -> empty list is False
print(bool([1, 2, 3]))# True  -> non-empty list is True

print(bool({}))       # False -> empty dictionary is False
print(bool({"a": 1})) # True  -> non-empty dictionary is True

print(bool(None))     # False -> None is always False
````

## <a name="2-4-suggestions-for-writing-better-code"></a>2.4 Suggestions for writing better code

#### <a name="example-32-do-s-and-don-ts-for-simple-tests"></a>Example 32: Do's and don'ts for simple tests

**What it does:** shows seven pairs of `if` statements. In each pair, both give the same result. The first is marked "Dont do this", and the second is the better, "Pythonic" form. (Pythonic means "written the way experienced Python programmers write it".)

| Value tested | Don't write | Write instead | Reason |
|---|---|---|---|
| A number `x` | `if x > 0:` | `if x:` | See the note below |
| A string | `if len(given_name) > 0:` | `if given_name:` | A non-empty string is truthy |
| A list | `if len(items_list) > 0:` | `if items_list:` | A non-empty list is truthy |
| A Boolean | `if my_flag == True:` | `if my_flag:` | It is already `True` or `False` |
| A string that may be empty | `if len(my_input) > 0:` | `if my_input:` | An empty string is falsy |
| A number | `if my_number != 0:` | `if my_number:` | `0` is falsy, any other number is truthy |
| A Boolean | `if is_ok == False:` | `if not is_ok:` | `not` reads naturally |

**Reading the output:** each message is printed twice, once by each form. The only pair that prints nothing is the `my_input` pair, because `my_input` is an empty string.

**An important note on the first pair:** `if x > 0:` and `if x:` are **not** the same test. `if x:` means "x is not zero", so it is also true for negative numbers. With `x = -5`, `if x > 0:` is false, but `if x:` is true. Here `x` is `100`, so both print. The Pythonic form replaces a check for "not zero", not a check for "positive". If you really need "positive", keep `if x > 0:`.

````python
x = 100
if x > 0:  #  Dont do this. It is redundant to compare with 0
    print("x is positive")
if x:  # Do this instead. It is more concise and Pythonic
    print("x is positive")

given_name = "Alice"
if len(given_name) > 0:  #  Dont do this .It is verbose to check length against 0
    print("Name is not empty")
if given_name:  # Do this instead. It is more concise and Pythonic
    print("Name is not empty")

items_list = [10, 20, 30]
if len(items_list) > 0:  #  Dont do this because it is verbose to check length against 0
    print("Items available")
if items_list:  # Do this instead because it is more concise and Pythonic
    print("Items available")
# Explanation: Non-empty lists are truthy, empty lists are falsy.

my_flag = True
if my_flag == True:  #  Dont do this. It is redundant to compare with True
    print("Flag is True")
if my_flag:  # Do this instead. It is more concise and Pythonic
    print("Flag is True")

my_input = ""
if len(my_input) > 0:  #  Dont do this because it is verbose to check length against 0
    print("Value is not empty")
if my_input:  # Do this instead because it is more concise and Pythonic
    print("Value is not empty")
# Explanation: Non-empty strings are truthy, empty strings are falsy.

my_number = 42
if my_number != 0:  #  Dont do this because it is redundant to compare with 0
    print("Number is non-zero")
if my_number:  # Do this instead because it is more concise and Pythonic
    print("Number is non-zero")
# Explanation: 0 is falsy, any other number is truthy.

is_ok = False
if is_ok == False:  #  Dont do this. It is redundant to compare with False
    print("Inactive")
if not is_ok:  # Do this instead because it is more concise and Pythonic
    print("Inactive")
# Explanation: Using 'not' is more Pythonic for boolean checks.
````

#### <a name="example-33-use-is-none-when-you-mean-none"></a>Example 33: Use `is None` when you mean None

**What it does:** shows why `if data:` and `if data is not None:` can give different answers.

**How it does it:**

1. `data = 0`. The value is a real number, zero, so data **is** available.
2. `if data:` asks "is `data` truthy?". `0` is falsy, so the first message is skipped, even though there is data.
3. `if data is not None:` asks "is `data` something other than `None`?". `0` is not `None`, so the second message is printed.

**Why:** when you want to know whether a value has been **set** at all, test for `None` directly. There are two reasons:

1. **Clarity:** `is not None` says exactly what you mean.
2. **Safety:** `if data:` wrongly treats real values such as `0`, `""` and `[]` as "no data".

**Learning point:** use `if x:` to ask "is x empty or zero?". Use `if x is not None:` to ask "has x been given a value?".

![if data compared with if data is not None](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-02-nb-10-is-not-none.png)

````python
data = 0  # 0 is Falsy
if data:
    print("1. Data is available")  # This is skipped!
# But if you only wanted to check for None:
if data is not None:  # This condition is True because 0 is not None
    print("2. Data is available")  # This runs! # 2. Data is available
````

## <a name="2-6-modular-programming-and-python-modules"></a>2.6 Modular programming and Python modules

**Modular programming** means breaking a large program into smaller parts, where each part does one job. In Python, each part is called a **module**. Splitting a program this way makes it:

1. easier to read,
2. easier to test,
3. easier to fix,
4. easier to reuse in other programs, and
5. better organised.

### <a name="2-6-1-what-is-a-python-module"></a>2.6.1 What is a Python module?

A module is simply a Python file whose name ends in `.py`. It can contain:

- variables,
- functions,
- classes,
- constants, and
- code that runs straight away (runnable code).

[Back to the Table of Contents](#toc)

### <a name="2-6-2-runnable-and-non-runnable-code"></a>2.6.2 Runnable and non-runnable code

When a module is loaded, either by running it or by importing it, Python reads it from top to bottom.

| Kind of code | Example | What happens when the module is loaded |
|---|---|---|
| Runnable code | `print("Loading...")`, `x = 10` | Runs at once |
| A function definition | `def add(a, b): ...` | The function is **created**, but its body does **not** run. The body runs only when the function is called |
| A class definition | `class Student: ...` | The class is **created**, but its methods do not run. They run only when they are called on an object |

So function and class definitions are "non-runnable" in the sense that nothing in them happens until you use them. "Beyond the Book", Extra 1 lets you see this for yourself.

```python
# Assuming we are writing in main.py which is located in the project directory

project/

    mypackage/
        __init__.py
        helper.py
        math_functions.py
        utils/
            __init__.py
            file_utils.py

# So we must specify the full mypackage.utils.file_utils

# 4. Import a function from file_utils.py which is in the utils subpackage

from mypackage.utils.file_utils import read_file
```

##### Notes on Example 36: what happens during an import

When Python meets an `import` statement, it follows the steps below. The step numbers match the numbers in the book's flowchart (Figure 2.5).

![What happens during an import, in 13 steps](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-02-nb-13-import-flow.png)

| Step | What happens, in short |
|---|---|
| 1 | Python meets an `import` statement, such as `import math` |
| 2 | It looks in `sys.modules`, a dictionary of every module already imported while the program runs |
| 3 | If the module is there, Python simply hands it back. Nothing is loaded twice |
| 4 and 5 | If not, Python checks whether it is a **built-in** module, one that is part of the Python program itself |
| 6 | A built-in module is written in C. Python sets it up straight from memory, stores it in `sys.modules`, and is done |
| 7 | Otherwise Python looks for a `.py` file in the folders listed in `sys.path`: your current folder, the standard library folders, and the folder where `pip` installs packages |
| 8 and 9 | If no file is found anywhere, the program stops with `ModuleNotFoundError` |
| 10 | If the file is found, Python translates it into **bytecode**, a simpler form that the Python interpreter runs. It often saves this in a `__pycache__` folder so the next import is faster |
| 11 | Python creates an empty module object and stores it in `sys.modules` **before** running any code in it. This prevents endless loops when two modules import each other |
| 12 | Python runs the module's code from top to bottom. This creates its variables, functions and classes, which become the module's attributes |
| 13 | The module is ready to use |

To write the name of a module you made yourself, the rule is the same: no `.py`. For a file named `myModule.py`:

```python
import myModule     #Note no extension .py in import statement parameter
```

(PEP 8 prefers lowercase module names, such as `my_module.py`. See Section 2.7.)

A consistent naming style makes your code easy to read, for you and for others. Python mainly uses **snake_case** and **PascalCase**.

### <a name="2-7-1-common-naming-conventions"></a>2.7.1 Common naming conventions

| What you are naming | Convention | How it looks | Examples |
|---|---|---|---|
| Variables and functions | snake_case | All lowercase, words joined by underscores | `user_age`, `calculate_total()` |
| Classes | PascalCase (also called UpperCamelCase) | Every word starts with a capital, including the first; no separators | `UserProfile`, `DataAnalyzer`, `Animal` |
| Constants (set once, never changed) | UPPER_SNAKE_CASE | All capitals, words joined by underscores | `MAX_RETRY_LIMIT`, `PI` |
| Modules | lowercase | Short, all lowercase. Underscores are allowed if they help readability | `math_utils`, `requests` |
| Packages | lowercase | Short, all lowercase. Underscores are best avoided | `mypackage`, `numpy` |

| Topic | Guideline |
|---|---|

### <a name="2-7-3-styles-to-avoid-in-python"></a>2.7.3 Styles to avoid in Python

#### <a name="example-39-more-ways-to-read-documentation-script-2"></a>Example 39: More ways to read documentation (Script 2)

**What it does:** reads the documentation of a third-party module (NumPy) safely, gets the documentation of `math` as a string, and finally starts a small documentation website.

**How it does it:**

- **Part A:** the `try` and `except` lines protect the code. If NumPy is installed, the code inside `try` runs and lists NumPy's names, then shows the help and docstring of `numpy.array`. If NumPy is **not** installed, Python jumps to the `except ModuleNotFoundError` part and prints a message instead of crashing. (You will learn `try` and `except` in the chapter on exceptions.)
- **Part B:** `pydoc.render_doc(math)` builds the same text that `help(math)` shows, but returns it as a string instead of printing it. `math_doc[:500]` is a slice: it prints just the first 500 characters.
- **Part C:** `pydoc.browse()` starts a small web server that shows documentation for every installed module.

**What happens in Colab when the last line runs:**

1. Python prints `Server ready at http://localhost:...` and a line of server commands.
2. A small input box appears, and the cell keeps running while it waits.

On your own computer, `pydoc.browse()` opens the documentation in your web browser. Type `q` and press Enter in the terminal when you have finished.

**Learning points:**

- `try` and `except` let a program continue when something goes wrong.
- `render_doc()` is useful when you want the documentation as text, for example to save it to a file.

![Steps of documentation Script 2](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-02-nb-14-read-docs-script.png)

````python
# Script 2: Reading Documentation of Third-Party Python Modules

# This script demonstrates how to:
# 1. Check whether a third-party module is installed
# 2. Read its documentation safely
# 3. Use help(), dir(), and docstrings
# 4. Use pydoc programmatically
# We use try / except so that the program does NOT crash
# if a third-party module is not installed.

# A. Reading documentation for NumPy (third-party module)

try:
    import numpy as np

    print("\nNumPy module found!")  # Will print if NumPy is imported successfully
    print("Exploring NumPy documentation...\n")

    # dir() lists all top-level names in the module
    print("Listing some NumPy attributes and functions:")
    print(dir(np))     # Shows array, zeros, ones, random, etc.

    print("\nHelp on numpy.array:")
    help(np.array)     # Detailed documentation for array()

    print("\nDocstring of numpy.array:")
    print(np.array.__doc__)   # Short description of array constructor

except ModuleNotFoundError:
    print("NumPy is NOT installed.")
    print("Install it using: pip install numpy")

# B. Using pydoc programmatically (built-in module)
import math
import pydoc

# pydoc.render_doc() generates formatted documentation as text
math_doc = pydoc.render_doc(math)  # Get documentation for math module as a string

# Print only the first 500 characters to keep output readable
print(math_doc[:500])
# pydoc is useful when:
# - You want offline documentation
# - You want to save docs to a file
# - You want clean text output

# C. Opening documentation in a web browser (Optional)
# pydoc can also start a local documentation web server.
# This is useful for browsing documentation in a browser.

# Uncomment the line below if you want to open documentation
# in your default web browser.
pydoc.browse()
# Message like Server ready at http://localhost:59635/
# This will open a local web page with documentation for all installed modules.
# You can search for any module, including third-party ones, in that web interface.
````

### <a name="extra-4-binary-octal-and-hexadecimal-numbers"></a>Extra 4: Binary, octal and hexadecimal numbers

You can write whole numbers in other bases by starting them with a zero and a letter:

| Base | Prefix | Digits allowed | Example | Value in decimal |
|---|---|---|---|---|
| Binary (2) | `0b` | 0 and 1 | `0b1010` | 10 |
| Octal (8) | `0o` | 0 to 7 | `0o10` | 8 |
| Hexadecimal (16) | `0x` | 0 to 9 and A to F | `0x1F` | 31 |

The prefix letter can be lowercase or capital (`0o10` or `0O10`), but lowercase is easier to read. Whatever base you write, Python stores an ordinary `int`. The functions `bin()`, `oct()` and `hex()` go the other way: they turn a number into text in that base.

````python
print(0b1010)   # binary literal: 10
print(0o10)     # octal literal: 8
print(0x1F)     # hexadecimal literal: 31
print(bin(10), oct(8), hex(31))   # from a number to text in each base
````
