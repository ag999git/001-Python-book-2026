## <a name="3-2-assignment"></a>3.2 Assignment

In Python, `=` is the **assignment operator**. `x = 5` means "take the value on the right and tie the name on the left to it". It helps to read `=` as an arrow pointing left:

| You write | Read it as |
|---|---|
| `a = 5` | a gets 5 (a &larr; 5) |
| `a = a + 5` | a gets (the old a) + 5 |

As you saw in Chapter 1, the name `a` is a tag, and the value `5` is the object the tag is tied to.

**`=` is not the "equals" of school maths.**

- `x = 12` is fine: the name `x` now refers to 12.
- `12 = x` is an error: you cannot assign anything to the number 12, because 12 is a literal, not a name.
- To **test** whether two values are equal, Python uses a different operator, `==` (see Section 3.9).

**Re-assignment.** A statement such as `a = a + 5` may look strange at first. Python runs it in two steps:

1. It works out the right-hand side, using the current value of `a`.
2. It ties the name on the left to the result.

Because the right-hand side is worked out first, `a` must already have a value. If `a` has never been given a value, `a = a + 5` stops with a `NameError`.

![How a = a + 5 is carried out](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-01-reassignment-steps.png)

**The rules of assignment in Python:**

| No. | Rule |
|---|---|
| 1 | The value on the right side is worked out first |
| 2 | The result is then stored in the name on the left side |
| 3 | `=` assigns. It does not compare |
| 4 | A variable can refer to any type of value |
| 5 | A variable is created automatically the first time it is assigned |

## <a name="3-4-compound-augmented-assignment-operators"></a>3.4 Compound (augmented) assignment operators

A **compound assignment operator** does a calculation and an assignment in one go. For example, `a += 15` means "add 15 to `a` and store the result back in `a`". It is a short form of `a = a + 15`.

| Operator | Example | Same as | What it does |
|---|---|---|---|
| `=` | `a = 10` | (plain assignment) | a gets 10 |
| `+=` | `a += 15` | `a = a + 15` | Increase a by 15 |
| `-=` | `a -= 5` | `a = a - 5` | Decrease a by 5 |
| `*=` | `a *= 15` | `a = a * 15` | Multiply a by 15 |
| `/=` | `a /= 15` | `a = a / 15` | Divide a by 15. The result is always a float |
| `%=` | `a %= 5` | `a = a % 5` | a gets the remainder of a divided by 5 |
| `//=` | `a //= 5` | `a = a // 5` | a gets the result of floor division of a by 5 |
| `**=` | `a **= 5` | `a = a ** 5` | a gets a raised to the power 5 |

As with `a = a + 5`, the variable must already have a value before you use a compound operator on it. "Beyond the Book", Extra 1 runs all of these in one cell.

## <a name="3-6-arithmetic-operators"></a>3.6 Arithmetic operators

Arithmetic operators work on numbers. Most take two operands (**binary** form, as in `x + y`). `+` and `-` can also take one operand (**unary** form, as in `-20`).

| No. | Operator | Meaning | Example | Result |
|---|---|---|---|---|
| 1 | `+` | Add two operands | `7 + 2` | `9` |
| | | Unary plus | `+20` | `20` |
| 2 | `-` | Subtract the right operand from the left | `7 - 2` | `5` |
| | | Unary minus | `-20` | `-20` |
| 3 | `*` | Multiply | `7 * 2` | `14` |
| 4 | `/` | Divide. The result is always a float | `6 / 2` | `3.0` (not `3`) |
| 5 | `%` | Modulus: the remainder after division | `13 % 8` | `5` |
| 6 | `//` | Floor division: divide, then round **down** to the next whole number on the number line | `7 // 3` | `2` (not `2.5`) |
| | | Rounding down also applies to negative results | `-7 // 3` | `-3` (not `-2`) |
| 7 | `**` | Power: the left operand raised to the right operand | `2 ** 5` | `32` |

`%` works with floats too: `7.5 % 2` gives `1.5`.

**Spaces around an operator do not matter.** `2+3`, `2 + 3` and `2   +   3` all give `5`. But an operator made of two characters must be written without a space inside it. `2 ** 3` is 8, while `2 * * 3` is a `SyntaxError`. The same goes for `//`, `==`, `!=`, `<=`, `>=`, `+=` and the other two-character operators.

#### <a name="example-1-the-sign-of-the-remainder"></a>Example 1: The sign of the remainder

**What it does:** works out three remainders where one or both numbers are negative: `-7 % 3`, `7 % -3` and `-7 % -3`.

**How it does it:** Python finds the remainder in two steps:

1. It does floor division first: `q = a // b`. This divides and then rounds **down** (towards minus infinity).
2. The remainder is whatever is left: `r = a - b * q`.

| Expression | a // b (rounded down) | a - b * q | Remainder |
|---|---|---|---|
| `-7 % 3` | `-7 // 3` = `-3` | `-7 - (3 * -3)` = `-7 + 9` | `2` |
| `7 % -3` | `7 // -3` = `-3` | `7 - (-3 * -3)` = `7 - 9` | `-2` |
| `-7 % -3` | `-7 // -3` = `2` | `-7 - (-3 * 2)` = `-7 + 6` | `-1` |

**Why:** because `//` always rounds down, the remainder in Python always has the **same sign as the divisor** (the number on the right), or is zero. C and Java work differently: there, the remainder takes the sign of the number on the left, so `-7 % 3` gives `-1` in C.

**Learning points:**

- `a % b` has the same sign as `b`.
- `(a // b) * b + (a % b)` always gives back `a`. This is true for every pair of numbers (as long as `b` is not zero). Check it for each line of the table.
- This rule is handy. For example, `n % 7` is always between 0 and 6 for any whole number `n`, even a negative one, which makes it easy to work out days of the week.

![Working out -7 % 3 step by step](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-02-remainder-sign.png)

````python
# In Python, sign of remainder (%) always follows sign of denominator

# Example 1 — Negative numerator, positive denominator
print(-7 % 3)  # Output: 2
# Explanation: -7 divided by 3 is -3 with remainder 2, since -3 * 3 + 2 = -7

# Example 2 — Positive numerator, negative denominator
print(7 % -3)  # Output: -2
# Explanation: 7 divided by -3 is -3 with remainder -2, since -3 * -3 + -2 = 7

# Example 3 — Negative numerator, negative denominator
print(-7 % -3)  # Output: -1
# Explanation: -7 divided by -3 is 2 with remainder -1, since 2 * -3 + -1 = -7
````

## <a name="3-7-order-of-precedence-of-operators"></a>3.7 Order of precedence of operators

When an expression has more than one operator, Python must decide which one to work out first. This is decided by **precedence**. The rules are the same ones you learned in school maths, extended to Python's other operators.

- **Parentheses come first.** Anything inside `( )` is worked out before the rest. `2 * (3 - 1)` is `4`. Brackets are also useful just to make an expression easier to read, even when they do not change the result, as in `(minute * 100) / 60`.
- **Power comes next.** `2 ** 1 + 1` is `3`, not `4`, and `3 * 1 ** 3` is `3`, not `27`, because the power is worked out before the `+` or `*`.
- **Multiplication and division** (`*`, `/`, `//`, `%`) come before **addition and subtraction**. `2 * 3 - 1` is `5`, not `4`, and `5 - 2 * 2` is `1`, not `6`.
- **Operators with the same precedence are worked out from left to right.** They are called **left-associative**. So `6 - 3 + 2` is `(6 - 3) + 2`, which is `5`. Working from right to left would give `6 - (3 + 2)`, which is `1`.
- **There is one exception: the power operator `**` is worked out from right to left.** So `2 ** 3 ** 2` means `2 ** (3 ** 2)`, which is `2 ** 9` = `512`, not `(2 ** 3) ** 2` = `64`.

The full order, from highest (done first) to lowest (done last):

![Order of precedence of Python operators, highest first](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-03-precedence-ladder.png)

| Level | Operators | Meaning |
|---|---|---|
| 1 (highest) | `( )` | Parentheses |
| 2 | `**` | Power (right to left) |
| 3 | `+x`, `-x`, `~x` | Unary plus, unary minus, bit-wise NOT |
| 4 | `*`, `/`, `//`, `%`, `@` | Multiplication, division, floor division, remainder, matrix multiplication |
| 5 | `+`, `-` | Addition, subtraction |
| 6 | `<<`, `>>` | Bit shifts |
| 7 | `&` | Bit-wise AND |
| 8 | `^` | Bit-wise XOR |
| 9 | `\|` | Bit-wise OR |
| 10 | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `is not`, `in`, `not in` | Comparisons, identity, membership |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 (lowest) | `or` | Logical OR |

**Watch out:** `-2 ** 2` is `-4`, not `4`. The power is worked out before the minus sign on its left. Write `(-2) ** 2` if you want `4`.

When in doubt, add brackets. They cost nothing and make your meaning clear. "Beyond the Book", Extra 2 lets you test all of these rules.

## <a name="3-8-types-of-operands-and-the-type-of-the-result"></a>3.8 Types of operands and the type of the result

When an operator works on two operands, three things can happen:

| Case | Example | Result |
|---|---|---|
| Both operands have the same type, and so does the result | `2 + 3` (int + int) | `5` (int) |
| Both operands have the same type, but the result has a different type | `6 / 3` (int / int) | `2.0` (float) |
| The operands have different types; the result has the type of one of them | `2.5 * 4` (float * int) | `10.0` (float) |

So always be aware of what type of result an operation gives. (This links to type casting, which you met in Chapter 2.)

You can check the type of any value or variable with the built-in function `type()`, for example `type(6 / 3)` gives `float`.

**The same operator can do different jobs.** `+` adds numbers, but it **joins** (concatenates) strings: `5 + 3` is `8`, while `"5" + "3"` is `"53"`. `*` multiplies numbers, but **repeats** strings, lists and tuples. Example 2 shows this.

#### <a name="example-2-and-on-different-types"></a>Example 2: `+` and `*` on different types

**What it does:** uses `+` and `*` on numbers, strings, lists and tuples, and prints the seven results.

**How it does it:** Python looks at the types of the operands and picks the matching job:

| No. | Code | Types | Job done | Result |
|---|---|---|---|---|
| 1 | `5 + 3` | int + int | Addition | `8` |
| 2 | `"Hello, " + "world!"` | str + str | Joining | `Hello, world!` |
| 3 | `[1, 2] + [3, 4]` | list + list | Joining | `[1, 2, 3, 4]` |
| 4 | `4 * 2` | int * int | Multiplication | `8` |
| 5 | `"abc " * 3` | str * int | Repetition | `abc abc abc ` |
| 6 | `[1, 2, 3] * 3` | list * int | Repetition | the list three times |
| 7 | `(4, 5) * 2` | tuple * int | Repetition | `(4, 5, 4, 5)` |

**Why:** the meaning of an operator depends on the types of its operands. This is called **operator overloading**.

**Learning points:**

- `+` joins two sequences of the **same** type. `[1, 2] + (3, 4)` (a list and a tuple) is an error, and so is `"5" + 3`.
- `*` repeats a sequence only when the other operand is a whole number.
- In block 5 the string is `"abc "`, with a space at the end, so the output ends with a space too.

````python
# 1. Using + with numbers → Performs addition
print(5 + 3)  # Output: 8
# Explanation: The + operator adds two numeric values together.

# 2. Using + with strings → Performs concatenation
print("Hello, " + "world!")  # Output: Hello, world!
# Explanation: The + operator concatenates two strings together.

# 3. Using + with lists → Merges two lists
print([1, 2] + [3, 4])  # Output: [1, 2, 3, 4]
# Explanation: The + operator merges two lists into one.

# 4. Using * with numbers → Performs multiplication
print(4 * 2)  # Output: 8
# Explanation: The * operator multiplies two numeric values together.

# 5. Using * with strings → Repeats the string
print("abc " * 3)  # Output: abc abc abc
# Explanation: The * operator repeats the string a specified number of times.

# 6. Using * with lists → Repeats the list
print([1, 2, 3] * 3)  # Output: [1, 2, 3, 1, 2, 3, 1, 2, 3]
# Explanation: The * operator repeats the list a specified number of times.

# 7. Using * with tuples → Repeats the tuple
print((4, 5) * 2)  # Output: (4, 5, 4, 5)
# Explanation: The * operator repeats the tuple a specified number of times.
````

## <a name="3-9-comparison-operators"></a>3.9 Comparison operators

A **comparison operator** compares two values and always gives back a Boolean: `True` if the comparison holds, `False` if it does not. Just as `+` takes two numbers and gives back their sum, a comparison operator takes two values and gives back `True` or `False`.

| No. | Operator | Meaning | True when | Example | Result |
|---|---|---|---|---|---|
| 1 | `<` | Less than | the left operand is strictly less than the right | `3 < 5` | `True` |
| 2 | `<=` | Less than or equal to | the left operand is less than or equal to the right | `5 <= 5` | `True` |
| 3 | `>` | Greater than | the left operand is strictly greater than the right | `10 > 2` | `True` |
| 4 | `>=` | Greater than or equal to | the left operand is greater than or equal to the right | `7 >= 9` | `False` |
| 5 | `==` | Equal to | both operands have the same value | `4 == 4` | `True` |
| 6 | `!=` | Not equal to | the operands have different values | `8 != 3` | `True` |

[Back to the Table of Contents](#toc)

#### <a name="example-3-the-six-comparison-operators"></a>Example 3: The six comparison operators

**What it does:** runs one example of each comparison operator and prints `True` or `False`.

**How:** each `print()` line compares two numbers. The comment on the same line gives the expected answer and the reason.

**Why:** comparisons are the heart of every decision a program makes. Every `if` and every `while` uses one.

**Learning points:**

- A comparison always gives back `True` or `False`, never a number.
- `==` (two equals signs) compares. `=` (one equals sign) assigns. Mixing them up is one of the most common beginner mistakes.
- `<=` and `>=` include equality: `5 <= 5` is `True`, but `5 < 5` is `False`.
- Comparisons also work on other types. Strings are compared letter by letter in dictionary order: `"apple" < "banana"` is `True`.

````python
# 1. <  (Less than)
print(3 < 5)   # True  -> 3 is strictly less than 5
# Explanation: The < operator checks if the value on the left is less than the value on the right.

# 2. <= (Less than or equal to)
print(5 <= 5)   # True  -> 5 is equal to 5, condition satisfied
# Explanation: The <= operator checks if the value on the left is less than or equal to the value on the right.

# 3. >  (Greater than)
print(10 > 2)   # True  -> 10 is strictly greater than 2
# Explanation: The > operator checks if the value on the left is greater than the value on the right.

# 4. >= (Greater than or equal to)
print(7 >= 9)   # False -> 7 is NOT greater than or equal to 9
# Explanation: The >= operator checks if the value on the left is greater than or equal to the value on the right.

# 5. == (Equal to)
print(4 == 4)   # True  -> both values are equal
# Explanation: The == operator checks if the value on the left is equal to the value on the right.

# 6. != (Not equal to)
print(8 != 3)   # True  -> 8 and 3 are different values
# Explanation: The != operator checks if the value on the left is not equal to the value on the right.
````

## <a name="3-10-identity-operators"></a>3.10 Identity operators

**Identity operators** check whether two names refer to the **very same object** in memory. They do not compare values. Python has two:

| No. | Operator | Meaning | True when | Example | Result |
|---|---|---|---|---|---|
| 1 | `is` | Same object | both names refer to the same object | `x = [1, 2]`, `y = x`, then `x is y` | `True` |
| 2 | `is not` | Not the same object | the names refer to different objects | `a = [1, 2]`, `b = [1, 2]`, then `a is not b` | `True` (two separate lists) |

**When should you use `is`?** Only when you really care about identity, not value. The two common cases are:

1. checking for `None`: `if x is None:`
2. checking whether two names refer to the same list or other object.

**`is` compared with `==`:**

| | `==` (equality) | `is` (identity) |
|---|---|---|
| Question it asks | "Are the values equal?" | "Are both names tied to the same object?" |
| Kind of test | Value comparison | Reference comparison |
| `[1, 2] == [1, 2]` | `True` | |
| `[1, 2] is [1, 2]` | | `False` (two different list objects) |
| `a = 5; b = 5` | `a == b` is `True` | `a is b` is `True` here because of interning (see Example 4), but this is not guaranteed in every case |

Two different objects can have the same value and still not be the same object. That is why:

- use `==` to check **equality**, and
- use `is` to check **identity**.

![Identity (is) compared with equality (==)](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-04-identity-vs-equality.png)

Small integers and short strings can look "identical" because Python reuses (**interns**) some objects to save memory. So `a is b` may happen to be `True` when you did not expect it. Never rely on that. Example 4 explores this.

#### <a name="example-4-interning-and-why-is-can-surprise-you"></a>Example 4: Interning, and why `is` can surprise you

This example goes a little deeper than the book. It supports the note on interning in Section 3.10.

**What it does:** compares six pairs of values with both `is` and `==`: small integers, large integers, short strings, long strings and lists.

**How it works:** **interning** is a memory-saving trick. For some immutable values, Python keeps one shared object and hands out that same object every time, instead of making a new copy.

| Block | Values | What standard Python (CPython) does | `is` result in this notebook | `==` result |
|---|---|---|---|---|
| 1 | `256` and `256` | Whole numbers from -5 to 256 are always shared | `True` | `True` |
| 2 | `10000` and `10000` | Not always shared | `False` | `True` |
| 3 | `1000000` and `1000000` | Not always shared | `False` | `True` |
| 4 | `"hello"` and `"hello"` | Short strings that look like names are usually shared | `True` | `True` |
| 5 | `"a" * 1000` twice | Long strings are usually not shared | `False` | `True` |
| 6 | `[1, 2, 3]` twice | Lists are never shared: each list is a new object | `False` | `True` |

**Why:** this shows why `is` must not be used to compare values. The `is` results depend on hidden details of Python. The `==` results are always `True`, because the values really are equal.

**Learning points:**

- `==` compares values. `is` compares identity.
- The comments say "False in some implementations". The results of blocks 2, 3 and 5 can change: if you put the same code in a `.py` file and run it as a script, Python may reuse the same object and print `True`. Your code must never depend on this.
- The only safe everyday uses of `is` are `x is None` and checking whether two names point to the same object.

````python
# Example 1: Small Integer Interning
a = 256
b = 256
print(a is b)  # True -> small integers are interned
print(a == b)  # True -> values are equal

# Example 2: Large integers (not interned by default)
num1 = 10000
num2 = 10000
print(num1 is num2)  # False in some implementations
print(num1 == num2)  # True -> values are equal
# Explanation: 'is' checks for object identity, while '==' checks for value equality.

# Example 3: Interning does NOT affect ==, because the values
# are compared, not the object locations.
x = 1000000
y = 1000000
print(x is y)  # False in some implementations
print(x == y)  # True -> values are equal

# Example 4: String Interning
str1 = "hello"
str2 = "hello"
print(str1 is str2)  # True -> due to string interning
print(str1 == str2)  # True -> values are equal

# Example 5: Large strings (not interned by default)
large_str1 = "a" * 1000
large_str2 = "a" * 1000
print(large_str1 is large_str2)  # False in some implementations
print(large_str1 == large_str2)  # True -> values are equal

# Example 6: Lists are not interned
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print(list1 is list2)  # False -> different objects in memory
print(list1 == list2)  # True -> values are equal
````

## <a name="3-13-bit-wise-operators"></a>3.13 Bit-wise operators

**Bit-wise operators** work on whole numbers as if they were strings of binary digits (bits: 0s and 1s), one bit at a time. They are used in low-level work, such as handling on/off flags, controlling devices, network code and fast calculations.

[Back to the Table of Contents](#toc)

### <a name="3-13-1-bit-wise-operators-a-simple-explanation"></a>3.13.1 Bit-wise operators, a simple explanation

| Operator | Name | Rule for each pair of bits |
|---|---|---|
| `&` | AND | The result bit is 1 only if **both** bits are 1 |
| `\|` | OR | The result bit is 1 if **at least one** bit is 1 |
| `^` | XOR (exclusive OR) | The result bit is 1 only if the two bits are **different** |
| `~` | NOT | Flips every bit. For any whole number A, `~A` equals `-(A + 1)` |
| `<<` | Left shift | Moves the bits to the left and fills with 0s on the right. Each shift by one place multiplies by 2 |
| `>>` | Right shift | Moves the bits to the right and drops the rightmost bits. Each shift by one place divides by 2 (floor division) |

The easiest way to follow a bit-wise operation is to write both numbers in binary, line up the bits, and work column by column. Here the numbers are written in 8 bits (1 byte):

| | Bits | Decimal |
|---|---|---|
| 6 | `0 0 0 0 0 1 1 0` | 6 |
| 5 | `0 0 0 0 0 1 0 1` | 5 |
| `6 & 5` (both 1) | `0 0 0 0 0 1 0 0` | 4 |
| `6 \| 5` (at least one 1) | `0 0 0 0 0 1 1 1` | 7 |
| `6 ^ 5` (different) | `0 0 0 0 0 0 1 1` | 3 |

More results:

| No. | Example | How it works |
|---|---|---|
| 1 | `6 << 2` gives `24` | `00000110` shifted left by 2 places becomes `00011000`, which is 16 + 8 = 24. It is the same as 6 * 2 * 2 |
| 2 | `6 >> 2` gives `1` | `00000110` shifted right by 2 places becomes `00000001`, which is 1. It is the same as 6 // 4 |
| 3 | `~6` gives `-7` | Flipping every bit of `00000110` gives `11111001`. In the "two's complement" system used for negative numbers, the leftmost bit counts as -128, so this is -128 + 64 + 32 + 16 + 8 + 1 = -7. The quick rule is `~A = -(A + 1)` |

![Steps in a bit-wise operation on decimal numbers](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-05-bitwise-steps.png)

[Back to the Table of Contents](#toc)

#### <a name="example-5-the-six-bit-wise-operators"></a>Example 5: The six bit-wise operators

**What it does:** applies each bit-wise operator to the numbers 6 and 5 (or to 6 alone) and prints the result.

**How:** the comments above each `print()` show the binary working. Compare them with the tables above.

| Block | Code | Binary working | Result |
|---|---|---|---|
| 1 | `6 & 5` | `00000110 & 00000101 = 00000100` | `4` |
| 2 | `6 \| 5` | `00000110 \| 00000101 = 00000111` | `7` |
| 3 | `6 ^ 5` | `00000110 ^ 00000101 = 00000011` | `3` |
| 4 | `~6` | flip `00000110` to get `11111001` | `-7` |
| 5 | `6 << 1` | `00000110` becomes `00001100` | `12` |
| 6 | `6 >> 1` | `00000110` becomes `00000011` | `3` |

**Why:** these operators are fast and compact. For example, a single number can store eight yes/no settings, one in each bit.

**Learning points:**

- Bit-wise operators work only on whole numbers (and Booleans), not on floats.
- `x << n` is the same as `x * 2**n`, and `x >> n` is the same as `x // 2**n`.
- You can see the binary form of a number with `bin()`: `bin(6)` gives `'0b110'`.

````python
# Example 1: Bitwise AND Operator(&) compares each
# bit of two integers. # 6 → binary: 00000110
# 5 → binary: 00000101. AND means 1 only if both bits are 1.
# So 6 & 5 is → 00000100 (in binary) → 4 (In decimal)
print(6 & 5)  # Output: 4

# Example 2: Bitwise OR Operator(|) compares each bit of two
# integers. 6 → binary: 00000110. 5 → binary: 00000101, OR
# OR (|) means 1 if at least one of the two bits is 1. So 6 | 5 is→ 00000111 → 7
print(6 | 5)  # Output: 7

# Example 3: Bitwise XOR Operator (^) compares each bit of
# two integers 6 → binary: 00000110 5 → binary: 00000101.
# XOR means 1 only if the bits are different.
# So 6 ^ 5 → 00000011 → 3 (Bitwise XOR)
print(6 ^ 5)  # Output: 3

# Example 4: Bitwise NOT Operator (~) inverts all the bits of
# an integer. 6 in binary:-> 00000110. In two's complement
# representation, this equals -7. So ~6 → 11111001 → -7
print(~6)  # Output: -7

# Example 5: Left Shift Operator (<<) shifts the bits of an
# integer to the left by a specified number of positions.
# 6 in binary: -> 00000110. So 6 << 1 → 00001100 → 12
print(6 << 1)  # Output: 12

# Example 6: Right Shift Operator (>>) shifts the bits of an
# integer to the right by a specified number of positions.
# 6 in binary: -> 00000110. So 6 >> 1 → 00000011 → 3
print(6 >> 1)  # Output: 3
````

## <a name="3-15-special-operators-identity-and-membership"></a>3.15 Special operators: identity and membership

Python's identity operators (`is`, `is not`) and membership operators (`in`, `not in`) are sometimes called **special operators**, because they behave differently from ordinary comparisons.

**Membership operators** (`in`, `not in`):

- `in` is `True` if the value exists in the collection.
- `not in` is `True` if the value does not exist.
- They work with strings, lists, tuples, sets and dictionaries. With dictionaries, they check the keys only, not the values.

**Identity operators** (`is`, `is not`):

1. They check whether two names refer to the same object in memory, not whether their contents are equal.
2. `is` is `True` if both names refer to the same object; `is not` is `True` if they refer to different objects.
3. Identity is about the object itself, not its value. You can see an object's identity number with the built-in function `id()`.
4. Two objects can have equal values (`==` gives `True`) but different identities (`is` gives `False`).
5. An object's identity never changes while it exists.

[Back to the Table of Contents](#toc)

#### <a name="example-6-identity-and-membership-together"></a>Example 6: Identity and membership together

**What it does:** shows three things: the difference between `is` and `==`, membership in a string and a list, and membership in a dictionary.

**How it does it:**

| Block | Code | Result | Reason |
|---|---|---|---|
| 1 | `a = b = [1, 2, 3]` | | One list is created, and both `a` and `b` are tied to it |
| | `c = [1, 2, 3]` | | A second, separate list with the same values |
| | `a is b` | `True` | Same object |
| | `a is c` | `False` | Different objects |
| | `a == c` | `True` | Same values |
| 2 | `'a' in 'abcd'` | `True` | The letter is found in the string |
| | `5 not in [1, 2, 3]` | `True` | 5 is not in the list |
| 3 | `'x' in my_dict` | `True` | `'x'` is a key |
| | `10 in my_dict` | `False` | 10 is a **value**, not a key |

**Learning points:**

- `a = b = [1, 2, 3]` is a **chained assignment**: one object, two names. If you now do `b.append(4)`, then `a` changes too, because it is the same list.
- To check whether a dictionary contains a value, use `.values()`: `10 in my_dict.values()` gives `True`.

````python
# Example 1: Identity vs Equality
a = b = [1, 2, 3]      # a and b refer to the SAME list object
c = [1, 2, 3]          # c has the same value but is a DIFFERENT object
print(a is b)  # True -> a and b point to the same object
print(a is c)  # False -> a and c are different objects
print(a == c)  # True -> a and c have the same value

# Example 2: Membership Operators
print('a' in 'abcd')   # True -> 'a' is found in the string 'abcd'
print(5 not in [1, 2, 3])  # True -> 5 is not in the list [1, 2, 3]

# Example 3: Dictionary Membership tests keys only
my_dict = {'x': 10, 'y': 20}
print('x' in my_dict)  # True -> 'x' is a key in the dictionary
print(10 in my_dict)   # False -> 10 is a value, not a key
````

### <a name="3-16-1-the-walrus-operator-python-3-8-and-later"></a>3.16.1 The walrus operator `:=` (Python 3.8 and later)

The **walrus operator** `:=` assigns a value to a variable **as part of an expression**. (It is called "walrus" because `:=` looks like the eyes and tusks of a walrus.) It helps you avoid working out the same value twice, and can make code shorter.

It is meant for use inside a larger expression, such as the condition of an `if` statement or a `while` loop. Written on its own as a plain statement, as in `n := 5`, it is a `SyntaxError`. Use the ordinary `n = 5` there.

[Back to the Table of Contents](#toc)

#### <a name="example-7-the-walrus-operator-in-an-if-statement"></a>Example 7: The walrus operator in an `if` statement

**What it does:** works out the length of `"hello"`, stores it in `n`, checks whether it is greater than 3, and prints `5`.

**How it does it:**

1. `len("hello")` gives `5`.
2. `n := ...` stores `5` in `n`, and at the same time passes `5` on to the rest of the expression.
3. `5 > 3` is `True`, so the `if` block runs.
4. `print(n)` prints `5`.

The brackets around `(n := len("hello"))` are needed. Without them, Python would work out `len("hello") > 3` first and store `True` in `n`.

**Why:** without the walrus operator, you would need two lines: `n = len("hello")` and then `if n > 3:`. With it, you assign and test in one step.

**Learning points:**

- `=` is a statement; it cannot be used inside an expression. `:=` is an operator; it can.
- Keep walrus expressions short and simple. If they make the line hard to read, use two lines instead.

![How the walrus operator works inside an if statement](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-06-walrus.png)

````python
# In this example, the walrus operator is used to assign the length of
# the string "hello" to the variable n while also checking if n is
# greater than 3. If the condition is true, it prints the value
# of n, which is 5 in this case.

if (n := len("hello")) > 3:
    print(n)  # Output: 5
````

### <a name="3-16-2-chained-comparisons"></a>3.16.2 Chained comparisons

A **chained comparison** compares several values in one expression, just as you would write it in maths: `1 < x < 10` means "x is between 1 and 10".

[Back to the Table of Contents](#toc)

#### <a name="example-8-chained-comparisons"></a>Example 8: Chained comparisons

**What it does:** checks whether `x` is between 1 and 10, and whether `y` is between 10 and 20 (inclusive), and prints a message for each.

**How it does it:**

- `1 < x < 10` is worked out as `(1 < x) and (x < 10)`. Both parts must be `True` for the whole expression to be `True`.
- `10 <= y <= 20` works the same way, with `<=` so that 10 and 20 themselves count as "between".
- The messages use f-strings: the `{x}` inside `f"..."` is replaced by the value of `x`.

**Why:** a chained comparison is shorter and easier to read than joining two comparisons with `and`.

**Learning points:**

- Any comparison operators can be chained, including `==` and `!=`. For example, `a == b == c` is `True` when all three are equal.
- In a chain, the middle value is worked out only once.
- If the first comparison is `False`, Python does not check the rest. The whole chain is already `False`.

![How 1 < x < 10 is checked](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-03-nb-07-chained-comparison.png)

````python
# Chained comparisons: to compare multiple values in a single expression.
# Example, check if a number is between two other numbers like this:
x = 5
if 1 < x < 10:
    print(f"{x} is between 1 and 10")  # Output: 5 is between 1 and 10
# Here, the expression 1 < x < 10 is evaluated as (1 < x) and (x < 10).
# Both comparisons must be true for the entire expression to be true.

# Chained comparisons can also be used with other types of comparisons, such
# as greater than or equal to (>=) and less than or equal to (<=). For example:
y = 15
if 10 <= y <= 20:
    print(f"{y} is between 10 and 20")  # Output: 15 is between 10 and 20
````

### <a name="3-16-3-the-ternary-conditional-operator"></a>3.16.3 The ternary conditional operator

The **ternary conditional operator** lets you write a simple if-else choice on one line. It is called "ternary" because it has three parts: a value if true, a condition, and a value if false.

**Syntax:**

```python
value_if_true if condition else value_if_false
```

[Back to the Table of Contents](#toc)

#### <a name="example-9-choosing-between-two-values-in-one-line"></a>Example 9: Choosing between two values in one line

**What it does:** stores `"Adult"` in `status` if `age` is 18 or more, and `"Minor"` otherwise. Since `age` is 20, it prints `Adult`.

**How it does it:**

1. Python checks the condition in the middle: `age >= 18`.
2. If it is `True`, the whole expression gives the value on the left, `"Adult"`.
3. If it is `False`, it gives the value on the right, `"Minor"`.
4. The chosen value is stored in `status`.

**Why:** it is a compact way to pick one of two values. The same thing written as a full `if` statement takes four lines:

```python
if age >= 18:

else:

```

**Learning point:** use the ternary operator only for simple choices. For anything longer, a full `if` statement is easier to read.

````python
# Ternary conditional operator in Python
# Syntax: value_if_true if condition else value_if_false
# Example 1: Basic usage
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)  # Output: Adult
````

### <a name="the-matrix-multiplication-operator"></a>The matrix multiplication operator `@`

The `@` operator was added in Python 3.5. It multiplies matrices and vectors. It is mostly used with the NumPy library, which is already installed in Colab.

To multiply matrix `A` by matrix `B`, the number of **columns** of `A` must equal the number of **rows** of `B`. Each entry of the result is made by taking a row of `A` and a column of `B`, multiplying the matching numbers, and adding them up:

For `A = [[1, 2], [3, 4]]` and `B = [[5, 6], [7, 8]]`:

|---|---|---|---|---|

| Row 1, column 2 | `[1, 2]` | `[6, 8]` | 1*6 + 2*8 | 22 |
| Row 2, column 1 | `[3, 4]` | `[5, 7]` | 3*5 + 4*7 | 43 |
| Row 2, column 2 | `[3, 4]` | `[6, 8]` | 3*6 + 4*8 | 50 |

[Back to the Table of Contents](#toc)

#### <a name="example-10-matrix-multiplication-with"></a>Example 10: Matrix multiplication with `@`

**What it does:** multiplies two 2 by 2 matrices, and then a matrix by a vector, and prints everything.

**How it does it:**

1. `import numpy as np` loads NumPy under the short name `np`.
2. `np.array(...)` turns nested lists into NumPy arrays (matrices).
3. `A @ B` multiplies the two matrices and gives `[[19, 22], [43, 50]]`, as worked out in the table above.
4. `M @ v` multiplies a matrix by a vector: 1*5 + 2*6 = 17 and 3*5 + 4*6 = 39, so the result is `[17, 39]`.

**Why:** matrix multiplication is at the heart of graphics, engineering, statistics and machine learning. `A @ B` reads almost like the maths.

**Learning points:**

- `A @ B` is matrix multiplication. `A * B` with NumPy arrays is something different: it multiplies matching entries one by one (`[[5, 12], [21, 32]]` here). Try it.
- The order matters: `A @ B` is usually not the same as `B @ A`.
- `@` does not work on ordinary Python lists. `[[1, 2]] @ [[3], [4]]` gives a `TypeError`. That is why NumPy arrays are used.

````python
# Note: The @ operator is only available in Python 3.5 and later versions.
# For matrix operations, it's often more efficient to use NumPy arrays instead of nested lists.
# In this example, we used NumPy to create matrices and perform multiplication, which is optimized for performance.
# The @ operator is a convenient way to express matrix multiplication in a clear and concise manner.
# Always ensure that the dimensions of the matrices or vectors are compatible for multiplication when using the @ operator.
# For matrix multiplication, the number of columns in the first matrix must equal the number of rows in the second matrix.
# For matrix-vector multiplication, the number of columns in the matrix must equal the number of elements in the vector.


# This script demonstrates the use of the @ operator for matrix multiplication
# in Python. The @ operator is used to perform matrix multiplication between two
# matrices (2D lists) or between a matrix and a vector (1D list).
# =============================================================
import numpy as np
# Example 1: Matrix-Matrix Multiplication
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = A @ B  # Matrix multiplication of A and B
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("Result of A @ B:\n", C)  # Should output the product of A and B
# Example 2: Matrix-Vector Multiplication
M = np.array([[1, 2], [3, 4]])
v = np.array([5, 6])
result = M @ v  # Matrix multiplication of M and v
print("Matrix M:\n", M)
print("Vector v:\n", v)
print("Result of M @ v:\n", result)  # Should output the product of M and v
````

### <a name="set-operators"></a>Set operators

#### <a name="example-11-the-four-set-operators"></a>Example 11: The four set operators

**What it does:** builds two small sets and prints their union, intersection, difference and symmetric difference.

**How:** the triple-quoted text at the top is just a note listing the operators. It is a string that is not stored anywhere, so Python ignores it. Each example then uses one operator and prints the result.

**Learning points:**

- Each operator gives back a **new** set. `set_a` and `set_b` are not changed.
- `set_a - set_b` is not the same as `set_b - set_a`. Try it: the second gives `{4, 5}`.
- The same results are available as methods: `set_a.union(set_b)`, `set_a.intersection(set_b)`, `set_a.difference(set_b)` and `set_a.symmetric_difference(set_b)`.

````python
"""
Set Operators
Sets support:

|   union
&   intersection
-   difference
^   symmetric difference

"""

# Example 1: Union
set_a = {1, 2, 3}
set_b = {3, 4, 5}
union_set = set_a | set_b
print("Union:", union_set)  # Output: {1, 2, 3, 4, 5}

# Example 2: Intersection
intersection_set = set_a & set_b
print("Intersection:", intersection_set)  # Output: {3}

# Example 3: Difference
difference_set = set_a - set_b
print("Difference (A - B):", difference_set)  # Output: {1, 2}

# Example 4: Symmetric Difference
symmetric_diff_set = set_a ^ set_b
print("Symmetric Difference:", symmetric_diff_set)  # Output: {1, 2, 4, 5}
````

### <a name="extra-3-logical-operators-and-logical-xor"></a>Extra 3: Logical operators and logical XOR

### <a name="extra-4-checking-the-floor-division-and-remainder-rule"></a>Extra 4: Checking the floor division and remainder rule

Example 1 said that `(a // b) * b + (a % b)` always gives back `a`. This cell checks it for all four sign combinations.

- The `for` line takes each pair of numbers from the list in turn (you will learn loops in the next chapter).
- `q, r = a // b, a % b` works out the quotient and the remainder in one line.
- The last column prints `True` if the rule holds.

Python also has a built-in function that gives both at once: `divmod(-7, 3)` returns `(-3, 2)`.

````python
for a, b in [(7, 3), (-7, 3), (7, -3), (-7, -3)]:
    q, r = a // b, a % b
    print(a, b, "->", "a // b =", q, " a % b =", r, " check:", q * b + r == a)
````
