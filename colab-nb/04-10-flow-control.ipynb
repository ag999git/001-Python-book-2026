## <a name="4-1-introduction"></a>4.1 Introduction

**Flow control** is about the order in which the statements of a program run. By default, Python runs statements one after another, from top to bottom. Flow control lets a program choose a path, repeat work, or jump out early.

[Back to the Table of Contents](#toc)

### <a name="the-three-building-blocks-of-flow"></a>The three building blocks of flow

| No. | Building block | What it does | Python keywords |
|---|---|---|---|
| 1 | **Selection** (conditionals) | Chooses between paths, based on whether a condition is `True` or `False` | `if`, `elif`, `else`, `match` |
| 2 | **Iteration** (loops) | Repeats a block of code until a job is finished | `while`, `for` |
| 3 | **Transfer** statements | Change the flow in the middle, for example to leave a loop early | `break`, `continue`, `return` |

(`pass` is also covered in this chapter. It does not change the flow at all. It is a placeholder that does nothing.)

### <a name="a-real-world-picture-taking-cash-from-an-atm"></a>A real-world picture: taking cash from an ATM

A visit to an ATM uses all three kinds of flow:

1. **Sequential flow:** insert your card, enter your PIN, choose "Withdraw Cash". Each step must finish before the next one starts.
2. **Conditional flow:** is there enough money in your account? If yes, the machine pays out. If no, it shows "Insufficient Funds".
3. **Iterative flow:** the machine counts out one Rs 100 note at a time, and keeps going **while** the total counted is less than the amount you asked for.

## <a name="4-2-the-if-elif-else-decision-structure"></a>4.2 The if...elif...else decision structure

The `if...elif...else` statement lets a program choose **exactly one** path out of several.

The general form is:

```python

if condition_1:# condition_1 must be a Bool expression and eval to True or False
    statement_block_1
elif condition_2:	# condition_2 also must be a bool expression
    statement_block_2
else:
    statement_block_3

```

(This is a pattern, not a program you can run, so it is shown as text.)

**Key rules:**

| No. | Rule | What it means |
|---|---|---|
| 1 | Conditions are Boolean | Every condition after `if` or `elif` must work out to `True` or `False` |
| 2 | Only one block runs | Python checks the conditions from the top. It runs the block of the **first** condition that is `True`, and then jumps past the whole structure |
| 3 | `else` is the catch-all | `else` has no condition. It runs only if every condition above it was `False` |
| 4 | Indentation marks a block | All the statements in a block must be indented by the same amount, usually 4 spaces. The first line that is not indented ends the block |
| 5 | Any number of `elif`, at most one `else` | You may have many `elif` blocks, but only one `else`, and it must come last |

`elif` is short for "else if".

**Truthiness.** A condition does not have to be a real Boolean. Python treats any non-zero number (including negative numbers) as `True`, and treats `0`, `0.0`, `None`, `False` and empty containers such as `""` and `[]` as `False`. (You met this in Chapter 2.)

#### <a name="example-1-is-a-number-positive-negative-or-zero"></a>Example 1: Is a number positive, negative or zero?

**What it does:** asks you for a number and tells you whether it is positive, negative or zero.

**How it does it:**

1. `input()` reads what you type, as a string.
2. `int()` turns that string into a whole number.
3. `if num > 0:` checks for a positive number. If this is `True`, Python prints the "positive" message and skips the rest of the structure.
4. Otherwise `elif num < 0:` checks for a negative number.
5. If both tests are `False`, the number can only be zero, so the `else` block runs.
6. The last `print()` is not indented, so it runs in every case.

**Why:** this is the classic three-way choice. Exactly one of the three messages is printed.

**Learning points:**

- The `else` block does not need a test. By the time Python reaches it, only one possibility is left.
- Try `0`, `7` and `-3` to see all three paths.
- If you type something like `abc` or `2.5`, `int()` raises a `ValueError`. The comment in the code warns about this.

![Flow of the positive, negative or zero script](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-03-positive-negative-zero.png)

````python
# --- STEP 1: Get User Input ---
user_entry = input("Enter a number: ")

# --- STEP 2: Convert to Integer ---
# We convert the input string directly to a number using int().
# Note: Will raise an error if input not a valid integer,
# but we are skipping error handling for simplicity.
num = int(user_entry)

# --- STEP 3: The Decision Structure (Selection Flow) ---
if num > 0:
    # Runs if num is greater than zero
    print(f"Success: {num} is a positive number.")

elif num < 0:
    # Runs if num is less than zero
    print(f"Success: {num} is a negative number.")

else:
    # Runs only if the number is exactly zero
    print("Success: The number is zero.")

# --- STEP 4: End of Script ---
print("Program execution finished.")
````

## <a name="4-3-if-else-statement-without-elif"></a>4.3 if...else statement (without elif)

When there are only two possible paths, you need just `if` and `else`:

```python

if bool_expression:   #bool_expression must eval to True or False
    #Code block for bool_expression True
    pass
else:
    # Code block for bool_expression False
    pass


```

(`pass` is a placeholder that means "do nothing". It is used here only because a block cannot be empty. See Section 4.8.)

**Key rules:**

- **The colon:** both the `if` line and the `else` line must end with a colon `:`. It tells Python that a block is starting.
- **Indentation:** all statements in each block must be indented by the same amount.
- **One or the other:** Python runs either the `if` block or the `else` block, never both.
- **No condition on `else`:** it simply handles every case that the `if` did not.

[Back to the Table of Contents](#toc)

#### <a name="example-2-even-or-odd"></a>Example 2: Even or odd?

**What it does:** asks for a whole number and says whether it is even or odd.

**How it does it:**

- `num % 2` is the remainder when `num` is divided by 2. For an even number it is `0` (for example `10 % 2` is `0`). For an odd number it is `1` (for example `11 % 2` is `1`).
- `if num % 2 == 0:` is `True` for an even number, so the "Even" message is printed. Otherwise the `else` block prints "Odd".

**Why:** a number is either even or odd, never both. With exactly two paths, `if...else` is the natural choice.

**Learning points:**

- `==` compares. `=` assigns. Writing `if num % 2 = 0:` is a `SyntaxError`.
- `%` (the remainder operator) is the standard way to test whether one number divides another.

![Flow of the even or odd script](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-04-even-odd.png)

````python
# --- STEP 1: Get User Input ---
# We use input() to get a string from the user and int() to convert it
user_entry = input("Enter a whole number to check: ")
num = int(user_entry)

# --- STEP 2: The Two-Way Decision (Selection Flow) ---
# We use the Modulo Operator (%) to find the remainder.
# If a number divided by 2 has a remainder of 0, it is EVEN.

if num % 2 == 0:
    # This block executes if the boolean expression evaluates to True
    print(f"The number {num} is Even.")
else:
    # This block executes if the boolean expression evaluates to False
    print(f"The number {num} is Odd.")

# --- STEP 3: Sequential Flow Resumes ---
print("Flow Control test complete.")

````

## <a name="4-4-omitting-the-else-clause-the-one-way-decision"></a>4.4 Omitting the else clause (the "one-way" decision)

Sometimes you want to act only when a condition is `True`, and do nothing at all when it is `False`. Then you leave out the `else` completely.

- **True case:** the indented block runs, and then the program carries on with the rest of the script.
- **False case:** Python skips the indented block and goes straight to the first line after it that is not indented.

[Back to the Table of Contents](#toc)

#### <a name="example-3-switching-on-the-air-conditioner"></a>Example 3: Switching on the air conditioner

**What it does:** switches the air conditioner on only if the temperature is above 25 degrees. Either way, it then prints "Monitoring system is active...".

**How it does it:**

1. `temp = 30` sets the temperature.
2. `if temp > 25:` is `True` (30 is more than 25), so the two indented `print()` lines run.
3. The last `print()` is not indented. It is outside the `if`, so it always runs.

**Why:** like a thermostat, you only need to act when it is hot. When it is cool, there is nothing special to do.

**Learning points:**

- An `if` does not need an `else`.
- Indentation decides which lines belong to the `if`.
- Change `temp` to `20` and run the cell again. Only the last line is printed.

![A one-way if: the block runs only when the condition is True](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-05-one-way-if.png)

````python
# --- STEP 1: Set the Temperature (Data) ---
temp = 30  # Try changing this to 20 to see the skip in action

# --- STEP 2: The One-Way if Statement ---
if temp > 25:
    # This block ONLY runs if temp is greater than 25
    print(f"Current temperature is {temp}°C.")
    print("Action: Switching the Air Conditioner ON.")

# --- STEP 3: Rest of the Script ---
# This line is NOT indented.
# It runs regardless of whether the 'if' was True or False.
print("Monitoring system is active...")
````

## <a name="4-5-nested-if-else-statements"></a>4.5 Nested if...else statements

**Nesting** means putting an `if...elif...else` statement inside a block of another one. Python lets you nest as deeply as you like. Each level of nesting is shown by one more level of indentation.

**The problem with nesting:** deep nesting pushes the code further and further to the right, and it becomes hard to read and to change. A good rule of thumb: if your code starts drifting far to the right, look for a simpler way to write it, for example with `elif` or with logical operators (Section 4.6).

[Back to the Table of Contents](#toc)

#### <a name="example-4-a-staircase-of-nested-checks-on-age"></a>Example 4: A staircase of nested checks on age

**What it does:** asks for your age and prints a comment, such as "You can vote" or "You are a teenager".

**How it does it:** each `else` block contains a new `if`. So each check happens only when all the checks above it were `False`:

| Age typed | First test that is True | Comment |
|---|---|---|
| 18 or more | `my_age >= 18` | You can vote |
| 13 to 17 | `my_age >= 13` | You are a teenager |
| 10 to 12 | `my_age >= 10` | You are in middle school |
| 6 to 9 | `my_age >= 6` | You are in primary school |
| Below 6 | (none) | You are too small to learn Python |

The comment is stored in `my_comment`. The last line, which is not indented, prints the age and the comment.

**Why:** it shows how nesting works, and also why deep nesting gets hard to read. Each level moves the code four spaces to the right.

**Learning points:**

- The order of the checks matters. Checking from the largest age downwards means each test only has to check a lower limit.
- The same logic can be written flat, with one `if`, several `elif` and one `else`. Try rewriting it that way. It is much easier to read.

![Flow of the nested if statements that check age](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-06-nested-age.png)

````python
# --- STEP 1: Get User Input ---
# We convert the input to an integer immediately
my_age = int(input("Enter Age: "))

# --- STEP 2: The Nested Logic ---
# Each 'else' block contains a new 'if' statement,
# pushing the logic deeper into the nest.
if my_age >= 18:
    my_comment = 'You can vote'
else:
    # First level of nesting: Happens if age is < 18
    if my_age >= 13:
        my_comment = 'You are a teenager'
    else:
        # Second level of nesting: Happens if age is < 13
        if my_age >= 10:
            my_comment = 'You are in middle school'
        else:
            # Third level of nesting: Happens if age is < 10
            if my_age >= 6:
                my_comment = 'You are in primary school'
            else:
                # The deepest part of the nest
                my_comment = 'You are too small to learn Python'

# --- STEP 3: Sequential Flow Resumes ---
# This line is unindented, so it always executes at the end.
print(f"At age: {my_age} -> {my_comment}")
````

#### <a name="example-5-two-numbers-each-even-or-odd-nested-version"></a>Example 5: Two numbers, each even or odd (nested version)

**What it does:** asks for two numbers and says, for each one, whether it is even or odd.

**How it does it:**

1. The outer `if` checks the first number.
2. Inside **both** branches of the outer `if`, an inner `if...else` checks the second number.
3. So there are four possible paths, and each one prints two lines.

**Why:** it shows nesting two levels deep. Notice that the inner check on `num2` is written twice, once in each branch. This repetition is a sign that the code could be simpler, as Example 7 shows.

**Learning points:**

- Here `input()` gives a string, and `int(num1)` converts it each time it is used. Converting once, straight after `input()`, is tidier.
- `(int(num1) % 2) == 0` has extra brackets. They are not needed, but they make the order of working clear.

![Flow of the nested script that checks two numbers](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-07-two-numbers-nested.png)

````python
num1 = input('First number ')
num2 = input('Second number ')
if (int(num1) % 2) == 0:    # Modulo operator % gives remainder.
    print('First number even')
    if (int(num2) % 2) ==0:
        print("Second number even")
    else:
        print('Second number odd')
else:   # This else block is entered if first number is odd
    print('First number odd')
    if (int(num2) % 2) ==0:
        print("Second number even")
    else:
        print('Second number odd')
````

## <a name="4-6-how-to-avoid-nested-if-statements-good-programming-techn"></a>4.6 How to avoid nested if statements (good programming technique)

As programs grow, nesting (`if` statements inside other `if` statements) makes code hard to read and maintain. A good way to simplify is to combine conditions with the **logical operators** `and`, `or` and `not`. This reduces the indentation and makes the logic clear at a glance. `and` is especially useful, because it needs both conditions to be `True` at the same time.

[Back to the Table of Contents](#toc)

#### <a name="example-6-the-teenager-check-in-one-line"></a>Example 6: The "teenager" check in one line

**What it does:** checks whether `age` (set to 15) is between 13 and 19, first with `and` and then with a chained comparison. Both print a message.

**How it does it:**

- **Option 1:** `age > 12 and age < 20` combines a lower limit and an upper limit into one condition. It is `True` only when both parts are `True`.
- **Option 2:** `12 < age < 20` is a chained comparison (Chapter 3). Python reads it as `(12 < age) and (age < 20)`.

**Why:** instead of a "staircase" of two `if` statements, one inside the other, you get a single, readable line.

| Style | Code | Levels of indentation |
|---|---|---|
| Nested | `if age > 12:` then, inside it, `if age < 20:` | 2 |
| With `and` | `if age > 12 and age < 20:` | 1 |
| Chained | `if 12 < age < 20:` | 1 |

**Learning point:** change `age` to `25` or `10` and run the cell again. Neither message is printed.

````python
# --- THE REFACTORED APPROACH ---
# We assume 'age' is already defined as an integer.
age = 15

# Option 1: Using 'and' to Combine Conditions combine lower bound (>12)
# and upper bound (<20).Much cleaner than nesting one 'if' inside another.
if age > 12 and age < 20:
    # This block runs only if BOTH conditions are met.
    print(f"Age {age}: Status is 'Teenager'.")

# Option 2: Using Python's Comparison Chaining. Even cleaner
if 12 < age < 20:
    print("This is also valid and very readable!")
````

#### <a name="example-7-two-numbers-even-or-odd-without-nesting"></a>Example 7: Two numbers, even or odd, without nesting

**What it does:** the same job as Example 5, but with one flat `if...elif...else` structure instead of nesting.

**How it does it:** each branch tests one of the four possible combinations:

| Branch | Condition | num1 | num2 |
|---|---|---|---|
| `if` | `(num1 % 2 == 0) and (num2 % 2 == 0)` | even | even |
| first `elif` | `(num1 % 2 != 0) and (num2 % 2 == 0)` | odd | even |
| second `elif` | `(num1 % 2 == 0) and (num2 % 2 != 0)` | even | odd |
| `else` | (no condition needed) | odd | odd |

**Why:** the code is flatter, each branch says exactly which case it handles, and there is no repeated inner check.

**Learning points:**

- With four combinations of two yes-or-no questions, three conditions are enough. The fourth case is whatever is left, so `else` handles it.
- The inputs are converted with `int()` straight away, so the conditions stay short.

![Flow of the script that uses and instead of nesting](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-08-two-numbers-and.png)

````python
# --- STEP 1: Get User Input ---
# We convert the inputs to integers to perform the modulo calculation.
num1 = int(input('Enter the first number: '))
num2 = int(input('Enter the second number: '))

# --- STEP 2: Multi-way Decision using 'and' ---
# Each 'elif' represents a specific combination of conditions.

if (num1 % 2 == 0) and (num2 % 2 == 0):
    # Runs only if BOTH numbers have a remainder of 0
    print('Both numbers are Even.')

elif (num1 % 2 != 0) and (num2 % 2 == 0):
    # Runs if num1 is Odd AND num2 is Even
    print('First number is Odd, second number is Even.')

elif (num1 % 2 == 0) and (num2 % 2 != 0):
    # Runs if num1 is Even AND num2 is Odd
    print('First number is Even, second number is Odd.')

else:
    # If none of the above are True, both must be Odd.
    # Note: No boolean expression is needed for 'else'.
    print('Both numbers are Odd.')

# --- STEP 3: Sequential Flow Resumes ---
print("Comparison complete.")
````

## <a name="4-7-the-while-loop-iterative-flow"></a>4.7 The while loop (iterative flow)

A `while` loop repeats a block of code **as long as** a condition stays `True`. An `if` statement checks its condition once. A `while` loop checks the condition, runs the body, and then goes back to check the condition again.

**How it works:**

| Stage | What happens |
|---|---|
| The pre-check | The condition is checked **before** each round, including the first. If it is `False` at the very start, the body never runs |
| The iteration | If the condition is `True`, the indented body runs. Then Python goes back to the `while` line and checks the condition again |
| The exit | As soon as the condition is `False`, Python skips the body and carries on with the first line after the loop |
| Indentation | As with `if`, the body of the loop is marked by indentation |

The syntax, with the optional `else` part:

```python

while test_condition:   # Must evaluate to True or False
    # Loop body: Runs repeatedly while condition is True
    statements
else:
    # Optional: Runs once when the condition becomes False
    statements

```

The `else` part is explained in Section 4.11.

#### <a name="example-8-numbers-and-their-squares"></a>Example 8: Numbers and their squares

**What it does:** prints the numbers 1 to 5 and their squares in two columns, then prints "Done!".

**How it does it:**

1. `x = 1` sets up the **counter**, the variable that controls the loop.
2. Two `print()` lines print the column headings. `\t` is an **escape character** that works like the Tab key, so the columns line up.
3. `while x <= 5:` checks the condition before every round.
4. Inside the loop, `x ** 2` works out the square, and the f-string prints `x`, a tab, and the square.
5. `x = x + 1` moves the counter on. After the round with `x = 5`, `x` becomes 6, the condition is `False`, and the loop ends.
6. `print('Done!')` is not indented, so it runs once, after the loop.

**Why:** this is the standard pattern for a counting loop: set up a counter, test it, do the work, and update the counter.

**Learning points:**

- **Always update the counter.** If you delete `x = x + 1`, `x` stays 1 forever and the loop never ends. This is called an **infinite loop**. (In Colab, stop a runaway cell with the Stop button on the left of the cell, or **Runtime > Interrupt execution**.)
- f-strings turn numbers into text for you, so you do not need `str()`.

| Round | x at the start | x <= 5? | Printed | x at the end |
|---|---|---|---|---|
| 1 | 1 | True | 1 and 1 | 2 |
| 2 | 2 | True | 2 and 4 | 3 |
| 3 | 3 | True | 3 and 9 | 4 |
| 4 | 4 | True | 4 and 16 | 5 |
| 5 | 5 | True | 5 and 25 | 6 |
| (check) | 6 | False | the loop ends | |

![Flow of the while loop that prints squares](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-09-while-squares.png)

````python
# --- STEP 1: Initialize the Counter ---
# We start x at 1. This variable controls how many times the loop repeats.
x = 1

# --- STEP 2: Print the Table Header ---
# \t is an 'escape character' that acts like the Tab key on your keyboard.
print('Number\tSquare')
print('------\t------')

# --- STEP 3: The while Loop ---
# The condition (x <= 5) is checked BEFORE every iteration.
while x <= 5:

    # Calculate the square (x to the power of 2)
    square = x ** 2

    # Print the result using an f-string (formatted string).
    # This automatically converts numbers to strings for us!
    print(f"{x}\t{square}")

    # --- STEP 4: Update the Counter ---
    # We add 1 to x so the loop eventually reaches 6 and stops.
    # If we forgot this line, the loop would run forever (Infinite Loop)!
    x = x + 1

# --- STEP 5: Loop Exit ---
# This line is unindented, so it only runs once the loop is finished.
print('Done!')
````

#### <a name="example-9-chopping-a-string-two-characters-at-a-time"></a>Example 9: Chopping a string, two characters at a time

**What it does:** prints `'112233'`, then removes the first two characters and prints what is left, again and again until the string is empty.

**How it does it:**

1. `while s1:` uses **truthiness**. A string with characters counts as `True`; the empty string `''` counts as `False`. So the loop runs while `s1` still has characters.
2. `print(s1)` shows the string.
3. `s1 = s1[2:]` is a slice from index 2 to the end. It makes a **new**, shorter string and ties `s1` to it.
4. After three rounds, `s1` is `''`, the condition is `False`, and the loop ends.

| Round | s1 at the start | Printed | s1 at the end |
|---|---|---|---|
| 1 | `'112233'` | 112233 | `'2233'` |
| 2 | `'2233'` | 2233 | `'33'` |
| 3 | `'33'` | 33 | `''` |
| (check) | `''` | the loop ends | |

**Why:** it shows how a loop can process data step by step, and how truthiness keeps the condition short.

**Learning points:**

- **Implicit Boolean check:** you do not need to write `while s1 != "":`. `while s1:` does the same job and is the "Pythonic" way.
- **Slicing does not delete anything.** Strings cannot be changed (they are immutable), so `s1[2:]` builds a new string, and the name `s1` is moved to it.
- **The exit condition:** the loop stops because the slicing eventually produces an empty string.

![Flow of the while loop that chops a string](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-10-while-slicing.png)

````python
# --- STEP 1: Define the initial string ---
s1 = '112233'

# --- STEP 2: The Loop Condition ---
# In Python, an empty string '' is considered False.
# A string with characters is considered True.
# Therefore, 'while s1:' means 'Keep going until the string is empty'.
while s1:
    # Print the current state of the string
    print(s1)

    # --- STEP 3: String Slicing (The Update) ---
    # s1[2:] creates a NEW string starting from index 2 to the end.
    # This effectively "chops off" the first two characters.
    s1 = s1[2:]

# --- STEP 4: Loop Exit ---
print("The string is now empty!")
# Output:
# 112233
# 2233
# 33
# The string is now empty!
````

## <a name="4-8-control-flow-tools-break-continue-and-pass"></a>4.8 Control flow tools: break, continue and pass

Sometimes you need finer control over a loop. Python has three keywords for this:

| Keyword | Nickname | Effect on this round | Effect on later rounds | Typical use |
|---|---|---|---|---|
| `break` | The emergency exit | Stops at once | None: the loop ends completely | Stop as soon as you find what you are looking for |
| `continue` | The skip button | Stops at once | The loop goes on with the next round | Skip bad data or particular values |
| `pass` | The placeholder | Does nothing | None | Keep a place for code you have not written yet |

After a `break`, Python carries on with the first line after the loop.

[Back to the Table of Contents](#toc)

#### <a name="example-10-break-continue-and-pass-in-one-loop"></a>Example 10: break, continue and pass in one loop

**What it does:** counts `x` up from 1. It behaves differently at 3 (`pass`), 5 (`continue`) and 8 (`break`).

**How it does it:** at the top of each round, `x = x + 1`. Then three `if` tests are checked, in this order:

| x | What happens | Is "Current number is: x" printed? |
|---|---|---|
| 1, 2, 4, 6, 7 | None of the tests match | Yes |
| 3 | `pass` does nothing. The note is printed, and the round carries on as normal | Yes |
| 5 | The note is printed, then `continue` jumps straight back to the `while` line | No |
| 8 | The note is printed, then `break` ends the loop at once | No |
| 9, 10 | Never reached: the loop already ended at 8 | |

**Why:** it puts the three keywords side by side, so you can see exactly how they differ.

**Learning points:**

- `pass` changes nothing about the flow. Here the `print()` after it still runs.
- `continue` skips only the rest of **this** round.
- `break` ends the **whole** loop, even though `x < 10` is still `True`.

![What break, continue and pass do inside one round of the loop](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-11-break-continue-pass.png)

````python
x = 0
print("Starting the loop...")

while x < 10:
    x = x + 1

    # 1. BREAK STATEMENT Exits loop immediately even if condition is still True.
    if x == 8:
        print(f"--- Found {x}: Breaking (Exiting loop entirely) ---")
        break

    # 2. CONTINUE STATEMENT Skips rest of current iteration. Moves to next
    if x == 5:
        print(f"--- Found {x}: Continuing (Skipping rest of this iteration) ---")
        continue

    # 3. THE PASS STATEMENT Does nothing, acts as a placeholder.
    if x == 3:
        # We know 3 is special, but we haven't decided what to do yet.
        # 'pass' keeps the code from crashing.
        pass
        print(f"--- Found {x}: Pass was used, but execution continues below ---")

    # This print statement is skipped when 'continue' or 'break' is triggered
    print(f"Current number is: {x}")

print("Loop is finished!")
````

## <a name="4-9-the-continue-statement"></a>4.9 The continue statement

`break` ends the loop. `continue` is gentler: it says "stop this round here, ignore the code below me, and go on with the next item".

- **The jump:** when Python reaches `continue`, it goes straight to the next round.
- **Use it** to skip bad data or particular values, without stopping the whole process.

#### <a name="example-11-skipping-the-unlucky-7"></a>Example 11: Skipping the unlucky 7

**What it does:** goes through the numbers 1 to 10 and prints each one, except 7, which it skips with a message.

**How it does it:**

1. `for num in range(1, 11):` gives `num` the values 1, 2, ..., 10 in turn. (`for` and `range()` are explained in Sections 4.12 and 4.16.)
2. When `num == 7`, the skip message is printed and `continue` sends Python to the next number.
3. For every other number, the "Processing number" line runs.

**Learning points:**

- `continue` works the same way in `for` loops and `while` loops.
- Replace `continue` with `break` and run the cell again. Now the loop stops at 7, and 8, 9 and 10 are never processed.

![Flow of the for loop that skips 7 with continue](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-12-continue-skip-7.png)

````python
# --- THE CONTINUE DEMONSTRATION ---

print("Starting the countdown (skipping unlucky 7)...")

for num in range(1, 11):
    if num == 7:
        print("--- SKIPPING 7: Jumping to the next number! ---")
        continue  # Everything below this line is ignored for '7'

    # This line only runs if 'continue' was NOT triggered
    print(f"Processing number: {num}")

print("Done!")
````

## <a name="4-10-the-match-statement-python-3-10-and-later"></a>4.10 The match statement (Python 3.10 and later)

`match` is Python's version of the "switch" statement found in C++ and Java. It was added in Python 3.10. Its full name is **structural pattern matching**. It is a cleaner alternative to a long chain of `if...elif...else`, and is handy for menu-driven programs.

**Key ideas:**

| Part | Meaning |
|---|---|
| The subject | The value after `match`, which is compared with each pattern |
| The cases | Each `case` line gives a pattern to look for. Python tries them from the top and runs the **first** one that matches |
| The wildcard `_` | `case _:` matches anything. It works like `else`, and should come last |
| A guard | An `if` after a pattern, as in `case char if ...`. The case matches only if the guard is also `True` |

**The general shape:**

```python
match subject:
    case pattern_1:
        statements
    case pattern_2 if condition:   # a pattern with a guard
        statements
    case _:                        # the wildcard: anything else
        statements
```

[Back to the Table of Contents](#toc)

#### <a name="example-12-what-kind-of-key-did-you-press"></a>Example 12: What kind of key did you press?

**What it does:** asks you to press one key (then **Enter**), and tells you whether it was a letter, a digit, or something else.

**How it does it:**

1. `case char if char.isalpha() and len(char) == 1:` The pattern `char` on its own matches **any** value and gives it the name `char`. The guard then checks that it is a single letter.
2. `case num if num.isdigit() and len(num) == 1:` works the same way for a single digit.
3. `case _:` catches everything else: a symbol, a space, several keys, or nothing at all.

| Method | Returns `True` when |
|---|---|
| `s.isalpha()` | every character of `s` is a letter, and `s` is not empty |
| `s.isdigit()` | every character of `s` is a digit, and `s` is not empty |

**Why:** the three cases read like a list of rules, which is easier to follow than a chain of `elif`s.

**Learning points:**

- A bare name in a `case` (like `char`) captures the value. It does not compare with a variable of that name.
- Only the first matching case runs. There is no "fall through" to the next case, as there is in C.
- Try `a`, `7`, `#` and `ab`.

![How the match statement picks a case for the key pressed](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-13-match-key.png)

````python
# --- ADVANCED MATCH-CASE: PATTERN GUARDS ---

key = input("Press a single key: ")

match key:
    # 1. Case for an Alphabet (a-z or A-Z)
    # We use 'if' inside the case to add a condition (a "Guard")
    case char if char.isalpha() and len(char) == 1:
        print(f"Success: '{char}' is an Alphabet.")

    # 2. Case for a Number (0-9)
    case num if num.isdigit() and len(num) == 1:
        print(f"Success: '{num}' is a Number.")

    # 3. Wildcard Case for any other key (Symbols, Space, etc.)
    case _:
        print("Note: You pressed a special character or multiple keys.")

print("Key analysis complete.")
````

## <a name="4-11-the-while-else-statement"></a>4.11 The while...else statement

`while...else` is a feature found in Python but not in most languages. It looks odd at first, but it becomes useful once you know the "exit rule".

**Key ideas:**

| Rule | Meaning |
|---|---|
| The "condition False" trigger | The `else` block runs when the `while` condition becomes `False` |
| The "break" rule | If the loop is ended by `break`, the `else` block is skipped |
| Think "completed" | `else` means: "if the loop finished normally, without being stopped early, run this" |
| The exception rule | If an error stops the program inside the loop, the `else` block does not run either |

**if...else compared with while...else:**

| Feature | `if...else` | `while...else` |
|---|---|---|
| Logic | Runs the `if` block **or** the `else` block, once | Runs the body again and again, then runs `else` once at the end |
| When `else` runs | When the condition is `False` | When the condition becomes `False` in the normal way |
| Effect of `break` | Not used | The `else` block is skipped |

[Back to the Table of Contents](#toc)

#### <a name="example-13-using-while-to-act-like-if"></a>Example 13: Using while to act like if

**What it does:** asks for a number and says whether it is even or odd, using `while...else` instead of `if...else`.

**How it does it:**

- **Even number:** the condition `num % 2 == 0` is `True`, so the body prints "even". Then `break` ends the loop, and because of `break`, the `else` block is skipped.
- **Odd number:** the condition is `False` from the start, so the body never runs. The loop ends normally, and the `else` block prints "odd".

**Why:** this is an unusual way to write an if-else, and you would not normally do it. It is here because it shows the `break` rule very clearly.

**Learning points:**

- The `break` is essential. Without it, `num` would stay even for ever, and the loop would print "even" without stopping.
- `else` after a loop means "no break happened".

![Flow of the script that uses while to act like if](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-14-while-mimics-if.png)

````python
# --- Script: Check Even/Odd using while...else ---
user_input = input('Give a number -> ')
num = int(user_input)

while num % 2 == 0:
    # This block runs if the condition is True (Even)
    print(f"{num} is even")

    # IMPORTANT: We must 'break' here.
    # If we don't, 'num' stays even forever, and we get an Infinite Loop!
    break

else:
    # This block runs if the condition was False (Odd)
    # OR if the loop finished without hitting a 'break'
    print(f"{num} is odd")
````

#### <a name="example-14-a-practical-use-searching"></a>Example 14: A practical use: searching

**What it does:** looks for the letter `'z'` in the word `"Python"`. It is not there, so the `else` block prints "No 'z' was found".

**How it does it:**

1. `index` starts at 0 and the loop runs while `index < len(word)`, so it visits every position of the word.
2. If `word[index] == 'z'`, it prints "Found a 'z'!" and `break` ends the loop. The `else` is skipped.
3. Otherwise `index += 1` moves on to the next letter.
4. If the loop reaches the end of the word without a `break`, the condition becomes `False` and the `else` block runs.

**Why:** this is where `while...else` really helps. It lets you say: "I looked at everything, and since I never found it (never used `break`), now I will run the else part". Without `else`, you would need an extra variable to remember whether the letter was found.

**Learning points:**

- Change `word` to `"pizza"` and run the cell again. The loop now finds a `'z'`, and the `else` message is not printed.
- The comment suggests `"Zebra"`. Try it too. The search compares exactly, and `'Z'` is not the same as `'z'`, so no `'z'` is found.

![Flow of the while...else search for the letter z](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-15-while-else-search.png)

````python
# --- Script: Search for the letter 'z' ---
word = "Python"  # Change this to test with different words,"Zebra"
index = 0

while index < len(word):
    if word[index] == 'z':
        print("Found a 'z'!")  # This will print if 'z' is found
        break # Exit the loop and SKIP the else block
    index += 1
else:
    # This only runs because the loop finished without finding 'z'
    print("Search complete: No 'z' was found in this word.")

print("End of program.")
````

| The body | Runs at once, at least one time |
| The exit condition | An `if` checks whether it is time to stop |
| `break` | Leaves the loop when the condition is met |

#### <a name="example-15-the-input-trap"></a>Example 15: The input trap

**What it does:** keeps asking for your age until you type only digits. Then it prints a success message and moves on.

**How it does it:**

1. `while True:` starts a loop whose condition is always `True`.
2. `input()` asks for your age. This always happens at least once.
3. `my_age.isdigit()` checks whether the text is made only of the digits 0 to 9.
4. If it is, the success message is printed, and `break` leaves the loop.
5. If not, the error message is printed, and the loop goes round again to ask once more.

**Why:** this is a very common real use: **input validation**. The user is "trapped" in the loop until they give the right kind of data.

**Learning points:**

- **The "at least once" rule:** the prompt is always shown once, because the check comes after `input()`.
- `\n` at the end of the error message adds a blank line, which separates the failed attempts.
- `isdigit()` is `False` for `-5` and for `3.5`, because `-` and `.` are not digits. Try them, and then try `abc`, and finally a proper age.

![Flow of the do...until style input check](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-16-do-until.png)

````python
# --- SIMULATING DO...UNTIL ---
# We want to force the user to enter a valid number.

while True:
    # 1. Ask for input (This always runs at least once)
    my_age = input('Give your age in years -> ')

    # 2. Check the condition (Is the input composed only of digits?)
    if my_age.isdigit():
        # Success! Print the result and exit the loop.
        print(f'Success: You are {my_age} years old.')
        break

    # 3. This runs only if the 'if' condition was False (break didn't happen)
    print('Error: You did not enter a valid integer. Please try again.\n')

print('Program continues...')
````

## <a name="4-12-the-for-loop-iterating-over-a-sequence"></a>4.12 The for loop: iterating over a sequence

A `for` loop steps through the items of a sequence or collection, such as a string, a list, a tuple or a range, one item at a time.

- A `while` loop is **indefinite**: it runs as long as a condition is `True`, and you may not know in advance how many rounds that will take.
- A `for` loop is **definite**: the number of rounds is fixed by the size of the collection.

**Key ideas:**

| Idea | Meaning |
|---|---|
| Sequence-based | It "steps through" the items of a string, list, tuple, range and so on |
| Automatic progress | You do not need a counter like `x = x + 1`. Python moves to the next item for you |
| The loop variable | In each round, the current item is put into the loop variable (also called the target variable) |
| Optional `else` | As with `while`, an `else` block runs once after the loop ends normally, without a `break` |

The general syntax:

```python

for item in collection:
    # This code runs for every single item in the collection
    statements
else:
    # Optional: Runs once after the loop finishes naturally
    statements
```

[Back to the Table of Contents](#toc)

#### <a name="example-16-spelling-a-word-one-letter-at-a-time"></a>Example 16: Spelling a word, one letter at a time

**What it does:** prints each letter of `"Python"` on its own line, then a closing message from the `else` block, then "Program continues...".

**How it does it:**

1. `for letter in word:` puts `'P'` into `letter` in round 1, `'y'` in round 2, and so on, up to `'n'`.
2. After the last letter, the loop ends normally, so the `else` block runs.
3. The last `print()` is outside the loop and runs once.

**Why:** walking through a string letter by letter is one of the most common uses of `for`.

**Learning points:**

- The same kind of loop works over lists, tuples and dictionaries. (Over a dictionary, the loop variable gets the **keys**.)
- There is no counter to forget, so there is no risk of an infinite loop.

![Flow of the for loop over the letters of Python](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-17-for-letters.png)

````python
# --- STEP 1: Define a collection ---
word = "Python"

print(f"Let's spell '{word}' one letter at a time:")
print("--------------------------------")

# --- STEP 2: The for Loop ---
# 'letter' is our loop variable.
# It will take the value of each character in 'word' one by one.
for letter in word:
    # In the 1st round, letter = 'P'
    # In the 2nd round, letter = 'y', and so on.
    print(f"Current character: {letter}")

else:
    # This runs only after the loop has visited EVERY letter.
    print("--------------------------------")
    print("Loop complete: All letters processed!")

# --- STEP 3: Outside the loop ---
print("Program continues...")
````

## <a name="4-13-nested-for-loops"></a>4.13 Nested for loops

A **nested loop** is a loop inside another loop. It is useful when you need to compare every item of one collection with every item of another.

Think of a clock: the minute hand (the inner loop) must go all the way round before the hour hand (the outer loop) moves on by one step.

**Key ideas:**

- **Each-to-each comparison:** for every single item of the first collection, the whole second collection is scanned from start to finish.
- **Execution order:** the inner loop runs much more often than the outer loop.
- **Cost:** the total number of comparisons is the length of the first collection multiplied by the length of the second. For large collections, nested loops can be slow.

[Back to the Table of Contents](#toc)

#### <a name="example-17-finding-matching-characters"></a>Example 17: Finding matching characters

**What it does:** compares every letter of `"hello"` with every letter of `"world"`, and prints each match.

**How it does it:**

1. The outer loop takes one letter `ch1` from `"hello"`.
2. For that `ch1`, the inner loop takes each letter `ch2` of `"world"` in turn.
3. If `ch1 == ch2`, a match is printed.
4. When the inner loop has used every letter of `"world"`, the outer loop moves to the next letter of `"hello"`.

| ch1 from "hello" | Matches found in "world" |
|---|---|
| h | none |
| e | none |
| l | l |
| l | l |
| o | o |

So there are 5 x 5 = 25 comparisons and 3 matches.

**Why:** it shows the "every item against every item" pattern that nested loops are used for.

**Learning point:** `'l'` is printed twice because it appears twice in `"hello"`. The loops report **every** match, not just unique ones. (With sets from Chapter 2, `set(str1) & set(str2)` gives just the unique common letters: `{'l', 'o'}`.)

![Flow of the nested for loops that find matching characters](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-18-nested-for.png)

````python
# --- STEP 1: Define the two strings to compare ---
str1 = "hello"
str2 = "world"

print(f"Comparing: '{str1}' & '{str2}'")
print("----------------------------")

# --- STEP 2: The Outer Loop ---
# This picks one character (ch1) from the first string at a time.
for ch1 in str1:

    # --- STEP 3: The Inner Loop ---
    # For the current ch1, we look at every character (ch2) in str2.
    for ch2 in str2:

        # --- STEP 4: The Comparison ---
        # We check if the character from the first word matches the second.
        if ch1 == ch2:
            print(f"Match found: '{ch1}' is in both words.")

# --- NOTE TO STUDENTS ---
# Notice that 'l' is printed multiple times.
# This happens because 'l' appears twice in 'hello'
# and once in 'world'. The loop finds EVERY possible match!
````

## <a name="4-14-choosing-your-tool-for-or-while"></a>4.14 Choosing your tool: for or while

Both loops can often do the same job, but each has a "natural home" where it is shorter and clearer.

| Feature | `for` loop | `while` loop |
|---|---|---|
| Kind of repetition | **Definite:** the number of rounds is known or fixed | **Indefinite:** the number of rounds depends on a condition |
| Driven by | A sequence (list, string, range and so on) | A Boolean condition |
| Moving on | **Automatic:** Python hands you the next item | **Manual:** you must update the variable yourself, for example `x += 1` |
| Main purpose | Process every item in a collection | Repeat until something changes |
| Risk of an infinite loop | Very low: it ends when the items run out | Higher: it never ends if the condition stays `True` |
| Reads best for | Counting and walking through sequences | Open-ended logic, waiting for something to happen |
| Does the body always run at least once? | No, not if the collection is empty | No, not if the condition is `False` at the start |
| Typical example | "Print every name in this list" | "Keep asking for a password until it is correct" |

## <a name="4-15-dead-code-unreachable-code"></a>4.15 Dead code (unreachable code)

**Dead code** (also called unreachable code) is code that can never run, whatever happens. It does not break the program, but it makes the script messy and confusing, because a reader may think it does something.

Dead code usually falls into one of three groups:

| No. | Kind | Example | Why it never runs |
|---|---|---|---|
| 1 | Code after an exit statement | a line after `break`, `continue`, `return` or `raise` in the same block | Python has already left the block |
| 2 | An impossible condition | `if 1 > 2:` or `if x < 5 and x > 15:` | The condition can never be `True` |
| 3 | A shadowed (redundant) branch | `if x > 0:` followed by `elif x > 10:` | Any value that passes the second test has already passed the first, so the `elif` is never reached |

[Back to the Table of Contents](#toc)

#### <a name="example-18-three-kinds-of-dead-code"></a>Example 18: Three kinds of dead code

**What it does:** contains one example of each kind of dead code. Only three lines are printed: "This line runs!", "You passed!" and "Program finished.".

**How it does it:**

| Type | Code | What happens |
|---|---|---|
| 1. After an exit | a `print()` after `break` | `break` leaves the loop first, so that `print()` is never reached |
| 2. Impossible condition | `if x < 5 and x > 15:` | No number is both below 5 and above 15, so the block never runs |
| 3. Redundant logic | `if score > 50:` then `elif score > 90:` | 95 passes the first test, so Python never looks at the `elif`. In fact **every** score above 90 is also above 50, so the `elif` can never run for any score |

**Why:** knowing these patterns helps you spot mistakes in your own code.

**Learning point:** the fix for type 3 is to put the stricter test first:

```python
if score > 90:
    print("Excellent! You got an A!")
elif score > 50:
    print("You passed!")
else:
    print("Better luck next time.")
```

````python
# Example: Identifying Dead Code

# --- Type 1: Post-Termination Code ---
while True:
    print("This line runs!")
    break
    print("DEAD CODE: This will never print because of the break above.")

# --- Type 2: Impossible Conditions ---
x = 10
if x < 5 and x > 15:
    print("DEAD CODE: A number cannot be less than 5 AND greater than 15.")

# --- Type 3: Redundant Logic ---
# Imagine a student's score is 95
score = 95
# 1. This condition is checked first.
# Since 95 is greater than 50, this block executes.
if score > 50:
    print("You passed!")

# 2. DEAD CODE / REDUNDANT BLOCK:
# Even though 95 is also greater than 90, Python will NEVER look at this elif.
# Once the first 'if' is True, Python skips all related 'elif' and 'else' blocks.
elif score > 90:
    print("Excellent! You got an A!")

# 3. This else is also skipped because the first 'if' was True.
else:
    print("Better luck next time.")

print("Program finished.")
````

## <a name="4-16-the-range-function"></a>4.16 The range() function

`range()` creates a sequence of whole numbers. It is very memory-efficient: it does not store all the numbers at once, but produces each one when it is needed.

**Signature:**

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

| Parameter | Meaning | Default |
|---|---|---|
| `start` | The first number | `0` |
| `stop` | The sequence stops **before** this number. `stop` itself is never included | (required) |
| `step` | How much to add each time. It can be negative, to count down, but it cannot be `0` | `1` |

All three must be whole numbers. A `step` of `0` raises a `ValueError`.

**The "up to but not including" rule:** if you want the numbers to reach 10, `stop` must be 11.

**How the values are worked out:** the number at position `i` is `start + step * i`, for `i = 0, 1, 2, ...`.

- With a **positive** step, the sequence continues while the value is **less than** `stop`.
- With a **negative** step, it continues while the value is **greater than** `stop`.
- If the step points the wrong way (start below stop with a negative step, or start above stop with a positive step), the range is **empty**.

| Call | Numbers produced | Why |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | start defaults to 0, step to 1 |
| `range(2, 7)` | 2, 3, 4, 5, 6 | starts at 2 and stops before 7 |
| `range(1, 10, 2)` | 1, 3, 5, 7, 9 | starts at 1 and goes up in steps of 2 |
| `range(10, 0, -2)` | 10, 8, 6, 4, 2 | a negative step counts down; 0 is not included |
| `range(5, 1, 1)` | (empty) | you cannot go down from 5 to 1 with a positive step |
| `range(1, 5, -1)` | (empty) | you cannot go up from 1 to 5 with a negative step |

![How range(start, stop, step) produces its numbers](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-19-range-values.png)

[Back to the Table of Contents](#toc)

#### <a name="example-19-the-different-ways-to-use-range"></a>Example 19: The different ways to use range()

**What it does:** shows seven ranges, from simple to tricky, and prints each one as a list.

**How it does it:** `list(range(...))` turns the range into a list, so that you can see all its numbers at once. (Printing `range(5)` directly just shows `range(0, 5)`.)

| No. | Call | Result |
|---|---|---|
| 1 | `range(5)` | `[0, 1, 2, 3, 4]` |
| 2 | `range(2, 6)` | `[2, 3, 4, 5]` |
| 3 | `range(0, 12, 3)` | `[0, 3, 6, 9]` |
| 4 | `range(0, -5, -1)` | `[0, -1, -2, -3, -4]` |
| 5 | `range(-2, -12, -2)` | `[-2, -4, -6, -8, -10]` |
| 6 | `range(-2, -12, 2)` | `[]`: you cannot count up from -2 to -12 |
| 7 | `range(2, 8, -1)` | `[]`: you cannot count down from 2 to 8 |

**Learning points:**

- An empty range is not an error. A loop over it simply does not run.
- Check the direction: a negative `step` needs `start` greater than `stop`.

````python
# ==========================================================
# Demonstrating various usages of the range() function
# ==========================================================

# 1. Using range() with a SINGLE argument (stop)
# Starts at 0 by default. Ends at (stop - 1).
print("Single argument (0 to 4):")
print(list(range(5)))         # Output: [0, 1, 2, 3, 4]

# 2. Using range() with START and STOP arguments
# Starts at 2. Ends at (6 - 1 = 5).
print("\nStart and Stop (2 to 5):")
print(list(range(2, 6)))      # Output: [2, 3, 4, 5]

# 3. Using range() with START, STOP, and STEP arguments
# Starts at 0. Increments by 3. Stops before 12.
print("\nWith positive step (0 to 9):")
print(list(range(0, 12, 3)))  # Output: [0, 3, 6, 9]

# 4. Using range() with NEGATIVE STEP (Counting Backwards)
# Starts at 0. Decrements by 1. Stops before -5.
print("\nNegative step (0 to -4):")
print(list(range(0, -5, -1))) # Output: [0, -1, -2, -3, -4]

# 5. Negative START and STOP with NEGATIVE STEP
# Starts at -2. Decrements by 2. Stops before -12.
print("\nNegative start/stop/step (-2 to -10):")
print(list(range(-2, -12, -2))) # Output: [-2, -4, -6, -8, -10]

# ----------------------------------------------------------
# LOGIC TRAPS: Why some ranges return an empty list []
# ----------------------------------------------------------

# 6. Impossible Range: Negative start/stop with POSITIVE step
# Logic: You cannot count UP from -2 to reach -12.
print("\nImpossible (Start > Stop with positive step):")
print(list(range(-2, -12, 2)))  # Output: []

# 7. Impossible Range: Positive start/stop with NEGATIVE step
# Logic: You cannot count DOWN from 2 to reach 8.
print("\nImpossible (Start < Stop with negative step):")
print(list(range(2, 8, -1)))    # Output: []

print("\n--- End of Demonstration ---")
````

## <a name="4-17-using-the-range-function-in-a-loop"></a>4.17 Using the range() function in a loop

#### <a name="example-20-range-in-for-loops"></a>Example 20: range() in for loops

**What it does:** uses `range()` in nine small `for` loops: six that print something and three that print nothing.

**How it does it:**

| Part | Loop | What it shows |
|---|---|---|
| 1 | `for i in range(len(s)):` | Loops over the **positions** of a string. `len(s)` is 6, so `i` goes 0 to 5, and `s[i]` gives the letter at each position |
| 2 | `for i in range(len(my_list)):` | The same idea for a list |
| 3 | `range(0, 10, 2)` | Counts in twos: 0 2 4 6 8 |
| 4 | `range(10, 0, -1)` | Counts down from 10 to 1 (0 is not included) |
| 5 | `range(-1, -6, -1)` | Counts down through negative numbers: -1 to -5 |
| 6 | `range(1, 10, 15)` | The step is bigger than the whole range, so only 1 is printed |
| A | `range(3, 3)` | Start equals stop: nothing to loop over |
| B | `range(1, 5, -1)` | Wrong direction: empty |
| C | `range(0)` | Zero rounds |

**About `end=" "`:** normally `print()` moves to a new line after each call. `print(i, end=" ")` puts a space after the value instead, so the numbers appear on one line. The `\n\n` at the start of the next heading then moves down two lines.

**Learning points:**

- `range(len(...))` is useful when you need the position as well as the item. (Python also has `enumerate()`, which gives both at once. See "Beyond the Book", Extra 3.)
- An empty range never causes an error.
- A step of `0` does. That line is commented out at the end of the cell. Remove the `#` from the `for` line and the `print(i)` line to see the `ValueError`, then put them back.

````python
# --- 1. Looping over a String using Indexes ---
# We use len() to find the total characters, and range() to get the index numbers.
s = "Hello!"
print("String iteration by index:")
for i in range(len(s)):
    print(f"Index {i} contains character: {s[i]}")

# --- 2. Looping over a List ---
my_list = [10, 20, 30, 40]
print("\nList iteration by index:")
for i in range(len(my_list)):
    print(f"Item at index {i} is {my_list[i]}")

# --- 3. Custom Steps (Counting by 2s) ---
print("\nEven numbers from 0 to 8:")
for i in range(0, 10, 2):  # Starts at 0, stops before 10, jumps by 2
    print(i, end=" ")

# --- 4. Reverse Order (Countdown) ---
print("\n\nCountdown from 10 down to 1:")
for i in range(10, 0, -1): # Start 10, Stop 0 (exclusive), Step -1
    print(i, end=" ")

# --- 5. Negative Start and Stop ---
print("\n\nNegative range (-1 down to -5):")
for i in range(-1, -6, -1):
    print(i, end=" ")

# --- 6. Large Steps ---
print("\n\nStep larger than the range:")
for i in range(1, 10, 15):
    print(i) # Only prints 1, then stops because 1 + 15 > 10

# NO OUTPUT:- Empty Ranges. Dont cause errors. They just don't run the loop.

# A. Equal Start and Stop: Nothing to loop over.
print("\nEqual start and stop (3, 3):")
for i in range(3, 3):
    print("This will not print")

# B. Impossible Direction: You can't go "down" from 1 to 5 with a negative step.
print("Impossible direction (1 to 5 with -1):")
for i in range(1, 5, -1):
    print("This will not print")

# C. Range(0): This means "loop zero times."
print("Range(0):")
for i in range(0):
    print("This will not print")

# WHAT CAUSES A "VALUE ERROR"? a STEP of 0.
# Because you cannot "move" by zero, Python stops the script.

# for i in range(1, 10, 0):  <-- THIS WOULD CAUSE AN ERROR
#    print(i)

print("\n--- End of Exercises ---")

````

### <a name="dead-code-in-different-situations"></a>Dead code in different situations

Section 4.15 showed three kinds of dead code. The same ideas turn up in other places too:

| Situation | Example | Why the code is dead |
|---|---|---|
| Inside a function, after `return` | `return total` followed by `print(total)` | `return` leaves the function first |
| After an unconditional `raise` | `raise ValueError("bad")` followed by more lines | The error leaves the block first |
| A branch that can never be `True` | `if False:` or `while False:` | The condition is fixed |
| A branch hidden by an earlier one | `if n % 2 == 0:` then `elif n % 4 == 0:` | Every multiple of 4 is also even, so it is always caught by the first test |
| An `else` after a loop that always breaks | `while True:` with a `break`, followed by `else:` | A `while True` loop can only end with `break`, and `break` always skips the `else` |

Tools called **linters** (such as `pylint` or `flake8`) and some editors, such as VS Code, can point out unreachable code for you.

### <a name="the-iterator-protocol-how-a-for-loop-really-works"></a>The iterator protocol: how a for loop really works

How does a `for` loop "talk" to a list? It uses the **iterator protocol**.

| Term | Meaning | Examples |
|---|---|---|
| **Iterable** | Any object you can loop over | strings, lists, tuples, sets, dictionaries, ranges |
| **Iterator** | An object that hands out the items of an iterable one at a time and remembers where it has got to | what `iter()` returns |
| `iter(obj)` | Makes an iterator from an iterable | `it = iter([10, 20])` |
| `next(it)` | Gets the next item from the iterator. When there are no items left, it raises `StopIteration` | `next(it)` gives `10`, then `20` |

When you write `for x in my_list:`, Python quietly does this:

1. It calls `iter(my_list)` to get an iterator.
2. It calls `next()` on the iterator again and again, putting each item into `x` and running the body.
3. When `next()` raises `StopIteration`, the loop ends. (If there is an `else`, it runs now.)

![What a for loop does behind the scenes](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-04-nb-20-iterator-protocol.png)

The next cell does the same thing by hand.

- The first three `next()` calls give `'a'`, `'b'` and `'c'`.
- The fourth call would raise `StopIteration`. To keep the notebook running, it is written as `next(it, "no more items")`. The second argument is a default value that `next()` returns instead of raising the error.

````python
letters = ['a', 'b', 'c']
it = iter(letters)        # step 1: get an iterator from the list

print(next(it))           # a
print(next(it))           # b
print(next(it))           # c
print(next(it, "no more items"))   # the list is used up

# This is what "for letter in letters:" does for you behind the scenes
````

**Try it:** remove `, "no more items"` from the last `next()` call and run the cell again. You will see the `StopIteration` error that a `for` loop normally catches for you.

### <a name="extra-2-a-range-is-a-lazy-sequence"></a>Extra 2: A range is a lazy sequence

Section 4.16 said that `range()` does not store all its numbers. You can still ask it questions as if it were a list, and it answers without building the list:

| Question | Code |
|---|---|
| How many numbers? | `len(r)` |
| Is a number in it? | `in` |
| What is at a position? | `r[index]` (negative positions work too) |

`range(0, 1_000_000_000, 7)` holds over 142 million numbers, but it takes only a few bytes of memory. (The underscores in `1_000_000_000` just make a long number easier to read.)

````python
r = range(0, 1_000_000_000, 7)
print(len(r))          # how many numbers
print(700 in r)        # True: 700 is a multiple of 7
print(701 in r)        # False
print(r[3])            # 21: the number at position 3
print(r[-1])           # the last number
````
