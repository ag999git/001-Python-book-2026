## <a name="5-1-introduction"></a>5.1 Introduction

Functions go by many names in programming: subroutines, routines, procedures, methods or subprograms.

A **function** is a named block of statements that does a job. You write the block once, give it a name, and then use it again and again just by calling that name. A function can also:

- take **parameters** (inputs), and
- **return** a result (a value or an object).

## <a name="5-2-why-do-we-need-functions"></a>5.2 Why do we need functions?

- **Reuse:** write the code once and use it many times. Python's own modules are examples of this. The `random` module, for instance, is a ready-made set of functions for random numbers.
- **Decomposition:** break a problem into smaller units, each with one job.
- **Readability:** a long, complicated script becomes a set of short, named parts.

##### Notes on Example 2: lazy evaluation

`range()` saves memory with a trick called **lazy evaluation**. Suppose you need the numbers 1 to 10. You could store them all in a list, but the list takes up memory. `range(1, 11)` instead works out each number only when it is needed.

A `range` is a special **lazy sequence**. It is not a list, and it is not a generator either. (Generators are another kind of lazy object that you will meet later in the book.) Chapter 4, "Beyond the Book", showed how a range can answer `len()`, `in` and indexing without building a list.

#### <a name="example-3-common-math-functions"></a>Example 3: Common math functions

**What it does:** uses six `math` functions and the constant `math.pi`.

**How it does it:**

| Example | Call | Result | Note |
|---|---|---|---|
| 1 | `math.sqrt(16)` | `4.0` | Square root. Always a float |
| 2 | `math.factorial(5)` | `120` | 5 x 4 x 3 x 2 x 1 |
| 3 | `math.pow(2, 3)` | `8.0` | Always a float. (`2 ** 3` gives the int `8`) |
| 4 | `math.ceil(4.3)` | `5` | Rounds **up** to the next whole number |
| 5 | `math.floor(4.7)` | `4` | Rounds **down** |
| 6 | `math.isclose(0.1 + 0.2, 0.3)` | `True` | The safe way to compare floats (see Chapter 2) |
| 7 | `math.pi * 5 ** 2` | `78.539...` | Area of a circle of radius 5 |

**Learning point:** `ceil()` and `floor()` also work for negative numbers: `math.ceil(-4.3)` is `-4` and `math.floor(-4.3)` is `-5`. Rounding "up" means towards plus infinity.

````python

import math

# Example 1: Using math.sqrt() to calculate the square root
num = 16
sqrt_num = math.sqrt(num)
print(sqrt_num)  # Output: 4.0

# Example 2: Using math.factorial() to calculate the factorial
fact_num = 5
factorial_result = math.factorial(fact_num)
print(factorial_result)  # Output: 120

# Example 3: Using math.pow() to calculate power
base = 2
exponent = 3
power_result = math.pow(base, exponent)
print(power_result)  # Output: 8.0

# Example 4: Using math.ceil() to round up a number
float_num = 4.3
ceil_num = math.ceil(float_num)
print(ceil_num)  # Output: 5

# Example 5: Using math.floor() to round down a number
float_num = 4.7
floor_num = math.floor(float_num)
print(floor_num)  # Output: 4

# Example 6: Using math.isclose() to compare two floating-point numbers
a = 0.1 + 0.2
b = 0.3
is_close = math.isclose(a, b)
print(is_close)  # Output: True

# Example 7: Using math.pi constant
circle_radius = 5
circle_area = math.pi * (circle_radius ** 2)
print(circle_area)  # Output: 78.53981633974483

````

## <a name="5-7-the-random-module"></a>5.7 The random module

The `random` module makes **pseudo-random** numbers: numbers that look random but are really produced by a fixed formula.

Common uses:

1. games, such as rolling a die or shuffling a pack of cards,
2. simulations,
3. testing, for example making random test data, and
4. machine learning, for example picking random samples.

| Category | Function | What it does | Example | Sample output |
|---|---|---|---|---|
| Random float | `random.random()` | A float from 0.0 up to, but not including, 1.0 | `random.random()` | `0.3748` |
| Random whole number | `random.randint(a, b)` | A whole number from a to b, both included | `random.randint(1, 6)` | a value from 1 to 6 |
| Random whole number | `random.randrange(start, stop, step)` | One number picked from `range(start, stop, step)` | `random.randrange(0, 10, 2)` | 0, 2, 4, 6 or 8 |
| Random float | `random.uniform(a, b)` | A float between a and b | `random.uniform(1.5, 5.5)` | `3.21...` |
| Pick one item | `random.choice(seq)` | One random item from a sequence | `random.choice([2, 4, 6])` | `4` |
| Pick several (repeats allowed) | `random.choices(seq, k=n)` | n items, the same item may come more than once | `random.choices([1, 2, 3], k=4)` | `[3, 1, 3, 2]` |
| Pick several (no repeats) | `random.sample(seq, k)` | k different items | `random.sample([1, 2, 3, 4], 2)` | `[3, 1]` |
| Shuffle | `random.shuffle(lst)` | Mixes up a list **in place**. Returns `None` | `random.shuffle(L)` | `L` becomes `[3, 1, 2]` |
| Repeatable results | `random.seed(x)` | Sets the starting point of the generator | `random.seed(5)` then `random.random()` | always `0.6229...` |
| Random bits | `random.getrandbits(k)` | A whole number made of k random bits | `random.getrandbits(8)` | 0 to 255 |

**Why "pseudo-random"?** Python uses a formula called the **Mersenne Twister**. It starts from a value called the **seed**. The same seed always gives the same sequence of numbers, so the numbers are not truly random: they only **appear** random. A real die is different: you cannot predict the next roll. If you do not set a seed, Python picks one for you (from the operating system's source of randomness, or from the current time), so each run is different.

`random.seed()` is sometimes called a "bookkeeping" function, because it does not produce a random number itself. It only sets up the generator so that the other functions can work. "Beyond the Book", Extra 1 shows how a seed makes results repeatable.

#### <a name="example-4-six-random-functions"></a>Example 4: Six random functions

**What it does:** makes a random float, a random whole number, a random choice, a shuffled list, a random sample and a random float in a range.

**How it does it:**

| Example | Call | What you get |
|---|---|---|
| 1 | `random.random()` | A float in `[0.0, 1.0)` |
| 2 | `random.randint(1, 10)` | A whole number from 1 to 10 |
| 3 | `random.choice(my_list)` | One of East, West, North, South |
| 4 | `random.shuffle(my_numbers)` | The same list, in a new order |
| 5 | `random.sample(range(1, 21), 5)` | 5 different numbers from 1 to 20 |
| 6 | `random.uniform(1.5, 5.5)` | A float between 1.5 and 5.5 |

**About `[0.0, 1.0)`:** a square bracket means the end is included; a round bracket means it is not. So `0.0` can come up, but every number is less than `1.0`.

**Learning points:**

- **Your output will be different from the saved output, and different every time you run the cell.** That is the point of random numbers.
- `shuffle()` changes the list itself and returns `None`. So write `random.shuffle(my_numbers)` and then print `my_numbers`. Writing `print(random.shuffle(my_numbers))` would just print `None`.
- `sample()` never repeats an item; `choices()` can.

````python

import random

# Example 1:- random.random(). # It generates a random number
# in semi-open range [0.0, 1.0). The square bracket means
# "closed interval on the left" and the round parenthesis means
# "open interval on the right". So, 0.0 is possible, but all
# returned numbers # will be strictly less than 1.0.
random_float = random.random()
print(random_float) # Output: random float between 0.0 and 1.0

# Example 2:- random.randint(a, b). # It generates a
# random integer between a and b (both inclusive).
random_integer = random.randint(1, 10)
print(random_integer)  # Output: random integer between 1 and 10

# Example 3:- random.choice(sequence) It selects a random
# element from the given sequence (list, tuple, string, etc.)
my_list = ['East', 'West', 'North', 'South']
random_choice = random.choice(my_list)
print(random_choice)  # Output: A random element from the list

# Example 4:- random.shuffle(list). The list elements are
# shuffled in place changing their order randomly.
my_numbers = [1, 2, 3, 4, 5]
random.shuffle(my_numbers)
print(my_numbers)  # Output: shuffled list of numbers

# Example 5:- random.sample(population, k) returns a list
# of k unique elements chosen from the population sequence or set.
population = range(1, 21)  # Numbers from 1 to 20
sample_size = 5  # Number of elements to sample
random_sample = random.sample(population, sample_size)
print(random_sample) # Output: random sample of 5 numbers from population

# Example 6:- random.uniform(a, b) generates a
# random float number between a and b, including both endpoints a and b.
random_uniform = random.uniform(1.5, 5.5)
print(random_uniform)  # Output: random float between 1.5 and 5.5
````

## <a name="5-13-scope-namespace-and-lifetime-of-variables"></a>5.13 Scope, namespace and lifetime of variables

### <a name="what-is-lifetime"></a>What is lifetime?

- The **lifetime** of a variable is how long it exists in memory: from the moment it is created to the moment it is destroyed.
- When you write `x = 3`, Python (1) creates the object `3` in memory, and (2) ties the name `x` to it.
- The name exists as long as the program can still reach it. When a function ends, its local names are no longer needed.
- Python then frees the memory automatically, using **reference counting** and **garbage collection**. You never delete variables by hand in normal Python.

Scope and lifetime are covered in more depth in a later chapter. For now, you only need global and local variables:

| Property | Global variables | Local variables |
|---|---|---|
| Where defined | Outside every function (usually at the top of the script) | Inside a function body |
| Born | When the line that creates it runs | When the function is called |
| Dies | When the program ends | When the function returns or finishes |
| Scope (visibility) | Wide: can be read anywhere in the script, including inside functions | Narrow: only inside the function where it was created |
| Lifetime | The whole run of the program | Only while the function is running |
| Main use | Constants, or data shared by the whole program | Temporary working values for one task |
| Memory | Stays until the program ends | Freed as soon as the function ends |
| Risk | High: any part of the program can change it, which makes bugs hard to find | Low: changes stay inside the function |

**Why this matters:** local variables keep functions separate from each other, and let Python free memory as soon as a function is done. Too many global variables make a program harder to understand and to fix.

[Back to the Table of Contents](#toc)

#### <a name="example-10-a-global-and-a-local-variable"></a>Example 10: A global and a local variable

**What it does:** defines a global `msg` and a function with a local `name`, calls the function, and then prints `msg`.

**How it does it:**

1. `msg = "Hello"` is at the top level, so it is **global**.
2. Inside `my_function()`, `name = "Alice"` is **local**.
3. The function can see both, so it prints `Hello Alice`.
4. After the call, `print(msg)` works, because `msg` is global.

### <a name="5-13-1-lifetime-compared-with-scope"></a>5.13.1 Lifetime compared with scope

| No. | Property | Scope (the "where") | Lifetime (the "when") |
|---|---|---|---|
| 1 | Definition | The part of the program where a name can be seen and used | The time a variable stays in memory |
| 2 | Question it answers | Can I see this variable from here? | Does this variable still exist? |
| 3 | Decided by | Where the variable is created in the code | When the program enters and leaves a function |
| 4 | Words used | Global scope, local scope, enclosing (non-local) scope | Birth (creation) and death (clean-up) |
| 5 | Picture | A pass: does it let you into this room (function)? | A candle: how long until it burns out? |
| 6 | Typical error | `NameError`: the name is not visible from here | `NameError`: the name no longer exists |

Both problems show up as the same error, `NameError`, which simply means "Python cannot find this name". There are two different reasons:

1. **A scope problem:** the variable exists, but you are looking for it from a place that cannot see it. For example, a local variable cannot be seen from outside its function.
2. **A lifetime problem:** the variable's function has already finished, so the variable is gone.

When Python meets a name, it looks for it in this order, and stops at the first place it finds it:

![Where Python looks for a name, in order](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-05-nb-06-name-lookup.png)

(These four places are often remembered as **LEGB**: Local, Enclosing, Global, Built-in.)

**Note:**

- You never need to free memory by hand.
- Use long-lived global variables with care.
- Loops and `if` blocks do **not** create a new scope in Python, unlike some other languages. A variable created inside a `for` loop still exists after the loop.

### <a name="5-13-2-name-clash-in-local-and-global-scope"></a>5.13.2 Name clash in local and global scope

If the same name is given a value in two different scopes, which one does Python use? The rule is simple:

- **Inside the function**, the local name wins. (There is a way to reach the global one, the `global` keyword; see "Beyond the Book", Extra 2.)
- **Outside the function**, there is no clash, because the local name does not exist there.

[Back to the Table of Contents](#toc)

#### <a name="example-11-a-local-x-hides-the-global-x"></a>Example 11: A local x hides the global x

**What it does:** has a global `x = 'cat'` and a local `x = 'dog'` inside `f3()`. It prints `dog` inside the function and `cat` outside.

**How it does it:**

1. `x = 'cat'` creates a global `x`.
2. Inside `f3()`, `x = 'dog'` creates a **new, local** `x`. It does not change the global one.
3. Inside `f3()`, Python finds the local `x` first, so it prints `dog`. This is called **shadowing**: the local name hides the global one.
4. After `f3()` ends, its local `x` is gone. `print(x)` at the top level finds the global `x`, which is still `cat`.

**Learning point:** assigning to a name inside a function makes it local to that function, even if a global of the same name exists.

````python
# --- GLOBAL SCOPE ---
# Global variable x exists in the "outer" world.
x = 'cat'

def f3():
    # --- LOCAL SCOPE ---
    # Shadowing: This creates a NEW local x.
    # It hides the global x while we are inside this function.
    x = 'dog'

    # Python checks the local "box" first, so it finds 'dog'.
    print('Inside f3() (Local)  ->', x)

# --- MAIN PROGRAM ---
f3() # This triggers the local variable's birth and death.

# Global x was never changed; it was just temporarily hidden.
print('Global scope (Main) ->', x)
````

[Back to the Table of Contents](#toc)

#### <a name="example-12-two-local-scopes-in-one-global-scope"></a>Example 12: Two local scopes in one global scope

**What it does:** has one global `x` and two functions, each with its own local `x`. It prints `dog`, then `rat`, then `cat`.

**How it does it:**

1. The global `x` is `'cat'`.
2. `f1()` creates its own local `x = 'dog'` and prints it. When `f1()` ends, that `x` is gone.
3. `f2()` creates a separate local `x = 'rat'` and prints it. It has no connection to `f1()`'s `x`.
4. The last line prints the global `x`, which was never touched: `cat`.

**Why:** think of the global scope as a big box, and each local scope as a smaller box inside it. Inside a small box, you see the local `x`. Outside the small boxes, you see the global `x`.

![One global scope holding two separate local scopes](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-05-nb-04-two-local-scopes.png)

````python
# --- GLOBAL SCOPE ---
# Global 'x' is the default value for the entire program
x = 'cat'

def f1():
    # --- LOCAL SCOPE 1 ---
    # This 'x' is created specifically for f1()
    x = 'dog'
    print("Inside f1() ->", x)

def f2():
    # --- LOCAL SCOPE 2 ---
    # This 'x' is created specifically for f2()
    # It has no connection to f1's 'x'
    x = 'rat'
    print("Inside f2() ->", x)

# --- MAIN PROGRAM ---
f1()  # Birth and Death of f1's local 'x'
f2()  # Birth and Death of f2's local 'x'

# Global 'x' remains untouched by the functions above
print("Global scope ->", x)
````

### <a name="5-13-3-defining-a-function-inside-another-function-nested-fu"></a>5.13.3 Defining a function inside another function (nested functions)

A function defined inside another function is a **nested function**. Here, `f1()` is the **enclosing** (outer) function, and `f2()` is the **enclosed** (inner) function. Normally, the inner function can only be called from inside the outer one.

Inside `f1()`, the local `x` of `f1()` hides the global `x`. Inside `f2()`, the local `x` of `f2()` hides both the `x` of `f1()` and the global `x`.

[Back to the Table of Contents](#toc)

#### <a name="example-13-three-scopes-one-inside-another"></a>Example 13: Three scopes, one inside another

**What it does:** has a global `x`, an `x` inside `f1()`, and an `x` inside `f2()`, which is nested in `f1()`. Each is printed from its own scope.

**How it does it:**

1. `f1()` is called from the top level.
2. Inside `f1()`, `x = 'dog in f1()'` is created and printed.
3. `f1()` defines `f2()` and then calls it.
4. Inside `f2()`, `x = 'rat in f2()'` is created and printed.
5. Both functions end, their local names are gone, and the global `x` is printed.

| Scope | Value of x | Printed by |
|---|---|---|
| Global | `'cat in global scope'` | the last line |
| Enclosing: `f1()` | `'dog in f1()'` | the `print()` inside `f1()` |
| Local: `f2()` | `'rat in f2()'` | the `print()` inside `f2()` |

**Learning points:**

- `f2` exists only inside `f1`. Try calling `f2()` at the top level: you get a `NameError`.
- Nested functions are useful, especially to build **closures** (Section 5.14), which is how tools such as decorators work. But keep them simple: deep nesting is hard to read.

![Scopes created by a nested function: f2 sits inside f1, which sits inside the global scope](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-05-nb-05-nested-scopes.png)

````python
# Global Scope
x = 'cat in global scope' # This x belongs to the GLOBAL scope

def f1():
    # Local Scope of f1()
    x = 'dog in f1()' # Local x in f1() — shadows global x
    print("x in local scope of f1() ->", x)

    def f2():
        # Enclosed (Nested) Scope inside f1()
        x = 'rat in f2()'  # Local x in f2() shadows f1()'s x
        print("x in local scope of nested f2() ->", x)

    # Call nested function f2()
    f2()

# Function call creates temporary local scopes
f1()

# Local x inside f1() and inside f2() no longer exist here
print("x in global scope ->", x)

# Output:
# x in local scope of f1() -> dog in f1()
# x in local scope of nested f2() -> rat in f2()
# x in global scope -> cat in global scope
````

### <a name="5-13-4-can-a-nested-function-be-called-from-outside"></a>5.13.4 Can a nested function be called from outside?

Yes, if the outer function **returns** the inner function. This hands the inner function to the outside world.

[Back to the Table of Contents](#toc)

#### <a name="example-14-returning-the-inner-function"></a>Example 14: Returning the inner function

**What it does:** `f1()` returns its inner function `f2`, the result is stored in `nested`, and `nested()` then runs `f2`.

**How it does it:**

1. `nested = f1()` calls `f1()`. Inside, `f2` is defined but not called.
2. `return f2` hands back the **function itself**. Notice: no brackets. `return f2()` would call `f2` and return its result (`None`) instead.
3. `f1()` is now finished, but `nested` refers to the function `f2`.
4. `type(nested)` confirms it: `<class 'function'>`.
5. `nested()` calls it, and "Hello from f2!" is printed.

**Why:** in Python, functions are objects, just like numbers and strings. They can be stored in variables, passed around, and returned from other functions.

**Learning point:** the text in triple quotes just below each `def` line is a **docstring**. It describes the function. Try `help(f1)` in a new cell.

![Returning a nested function so it can be called from outside](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-05-nb-07-return-nested.png)

````python
def f1():
    """Outer function that defines and returns another function."""

    def f2():
        """Inner function (Nested)."""
        print("Hello from f2!")

    # We return f2 without parentheses, not f2().
    return f2 # We are returning the function itself, not calling it.

# 1. Call f1(). It returns the function 'f2' and stores it in 'nested'.
# Calling f1() does not execute f2() but returns f2 as a function object.
nested = f1() # 'nested' holds f2 that was defined inside f1()

# 2. At this point, f1() is finished, but 'nested' holds f1's f2.
print(f"Type of 'nested' variable is: {type(nested)}")
# Output: Type of 'nested' variable is: <class 'function'>,
# confirming that 'nested' is indeed a function object.

# 3. We call the inner function from the global scope using the new name.
nested() # Output: Hello from f2!
# Explanation: 'nested' is now a reference to f2() defined inside f1().
````

## <a name="5-14-nested-functions-and-closures-an-advanced-idea"></a>5.14 Nested functions and closures (an advanced idea)

A nested function becomes a **closure** when all three of these are true:

1. the inner function **uses** a variable from the outer function,
2. the outer function **returns** the inner function, and
3. the inner function still **has access** to that variable later, even after the outer function has finished.

Think of a closure as a function carrying a "backpack". The backpack holds the variables from the place where the function was created, even when that place no longer exists.

| | Nested function | Closure |
|---|---|---|
| A function inside a function | Yes | Yes |
| The inner function uses a variable of the outer one | Optional | Required |
| The outer function returns the inner function | No | Yes |
| The outer variables are remembered after the outer function ends | No | Yes |

**Why do we need closures?** Sometimes we want to hand back **both** some data **and** a function that works on that data, and keep the data alive after the outer function ends. Returning just the data, or just a plain function, cannot do this.

[Back to the Table of Contents](#toc)

#### <a name="example-15-a-closure-remembers-a-name"></a>Example 15: A closure remembers a name

**What it does:** makes two greeting functions, one for Alice and one for Bob, and calls them. They print "Hello, Alice!" and "Hello, Bob!".

**How it does it:**

1. `outer_greeter("Alice")` runs. Its parameter `name` is `"Alice"`.
2. It defines `inner_greeter()`, which uses `name`, and returns it.
3. `outer_greeter` has now finished, but `greet_alice` keeps `name = "Alice"` in its backpack.
4. `outer_greeter("Bob")` does the same, making a separate function that remembers `"Bob"`.
5. Calling `greet_alice()` and `greet_bob()` prints each stored name.

**Learning point:** each call to `outer_greeter()` creates a new inner function with its own saved value. The two do not share anything.

````python
def outer_greeter(name):  # 'name' is a parameter of the outer function.
    # This variable 'name' is in the Enclosing scope of inner_greeter
    # It is "captured" by the inner function below

    def inner_greeter():
        # Even after outer_greeter ends, this function
        # keeps 'name' alive so it can use it when called later.
        print(f"Hello, {name}!")

    return inner_greeter # Return the function object

# --- THE CAPTURE ---
# We create two different "instances" of the function
greet_alice = outer_greeter("Alice")
greet_bob = outer_greeter("Bob")

# --- THE EXECUTION ---
# The outer function 'outer_greeter' is long gone,
# but the inner functions still remember their specific data!
greet_alice()  # Output: Hello, Alice!
greet_bob()    # Output: Hello, Bob!
````

### <a name="5-14-1-the-nonlocal-keyword-and-its-use-in-closures"></a>5.14.1 The nonlocal keyword and its use in closures

`nonlocal` is a Python keyword. It is written as one word, with no hyphen. You use it inside a nested function to tell Python:

- "Do not treat this variable as a new local variable."
- "Use the variable of the same name in the nearest outer function."

Normally, if you **assign** to a variable inside a function, Python assumes it is local. `nonlocal` changes that.

| Feature | `nonlocal` | `global` |
|---|---|---|
| Refers to a variable in | the nearest enclosing (outer) function | the global (top-level) scope |
| Typical use | closures and nested functions | shared data for the whole module |
| Can it be used outside a nested function? | No | Yes |

[Back to the Table of Contents](#toc)

### <a name="5-14-2-using-closures-to-keep-state-without-global-variables"></a>5.14.2 Using closures to keep state without global variables

**State** is data that a program remembers over time, or across several function calls. For example, a counter must remember whether it has reached 1, 2 or 3.

You could keep the count in a global variable, but that is usually a bad idea:

- **Unsafe:** any part of the program can change a global by accident.
- **Cluttered:** many globals make the top level of the code messy.
- **Hard to debug:** in a large script, it is hard to find where a global was changed.

A closure is better. It gives a function its own "private memory" that lasts between calls, and that the rest of the program cannot touch.

[Back to the Table of Contents](#toc)

#### <a name="example-16-a-counter-with-private-memory"></a>Example 16: A counter with private memory

**What it does:** makes two separate counters, `counter_a` and `counter_b`, and calls them. Each keeps its own count. It also peeks at the hidden value stored inside `counter_a`.

**How it does it:**

1. `make_counter()` sets `count = 0` and defines `counter()`.
2. Inside `counter()`, `nonlocal count` says: "use the `count` of `make_counter`, not a new local one". Then `count += 1` adds 1 and the new value is returned.
3. `make_counter()` returns `counter`. Each call to `make_counter()` creates a fresh `count`, so `counter_a` and `counter_b` have separate counts.
4. `counter_a.__closure__[0].cell_contents` looks inside the "backpack" of `counter_a` and finds the stored count.

| Call | counter_a's count | counter_b's count | Printed |
|---|---|---|---|
| `counter_a()` | 1 | 0 | Counter A: 1 |
| `counter_a()` | 2 | 0 | Counter A: 2 |
| peek inside `counter_a` | 2 | 0 | The hidden value in counter_a is: 2 |
| `counter_b()` | 2 | 1 | Counter B: 1 |
| `counter_a()` | 3 | 1 | Counter A: 3 |

**Learning points:**

- **Private memory:** each call to `make_counter()` makes a new, private `count`. That is why the two counters do not interfere.
- **No global clutter:** you can make as many independent counters as you like from one function definition.
- **The re-binding rule:** `count += 1` means `count = count + 1`. It **assigns** to `count`, so without `nonlocal`, Python would treat `count` as a new local variable and stop with an `UnboundLocalError`. Try deleting the `nonlocal` line.
- **When `nonlocal` is not needed:** (1) if you only **read** the variable without changing it, as in Example 15; or (2) if the variable holds a mutable object, such as a list, and you change the object's **contents** rather than re-assigning the name. "Beyond the Book", Extra 3 shows case (2). Even then, many programmers prefer `nonlocal`, because it makes the intention clear.
- `__closure__` is a special attribute that every function has. You rarely need it; here it simply proves that the count really is stored inside the function.

![How the closure counter remembers its count between calls](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-05-nb-08-closure-counter.png)

````python
def make_counter():
    # 1. Variable count belongs to the ENCLOSING scope.
    # It acts as the "Private Memory" for the closure.
    count = 0

    def counter():
        # 2. The 'nonlocal' keyword is essential here.
        # It allows the inner function to MODIFY 'count'.
        # Without it, Python would treat 'count' as a new local variable.
        nonlocal count

        count += 1
        return count

    # 3. Return the function object to create the Closure
    return counter

# --- INDEPENDENT STATE ---
# Each call to make_counter() creates a brand new "Private Memory" box.
counter_a = make_counter()
counter_b = make_counter()

print(f"Counter A: {counter_a()}")  # Output: 1
print(f"Counter A: {counter_a()}")  # Output: 2

# Advanced: How to see the "hidden" memory
# Each function object has a __closure__ attribute
cell = counter_a.__closure__[0]
print(f"The hidden value in counter_a is: {cell.cell_contents}")
# Output: The hidden value in counter_a is: 2

# Counter B starts fresh; it cannot see or touch Counter A's memory.
print(f"Counter B: {counter_b()}")  # Output: 1

# Counter A still remembers where it left off.
print(f"Counter A: {counter_a()}")  # Output: 3
````

## <a name="5-15-function-composition"></a>5.15 Function composition

**Function composition** means joining functions so that the output of one becomes the input of the next.

- **The order of working:** in `f(g(x))`, the **inner** function `g(x)` runs first.
- **Data flow:** the result of `g(x)` is passed straight into the outer function `f()`.
- **Final result:** whatever the outermost function returns.

**Why use composition?**

| Benefit | Meaning |
|---|---|
| Modularity | A big task is split into small functions |
| Reuse | The same small functions can be combined in different ways |
| Easier testing | Each small function can be tested on its own |
| Readability | "One function, one job" makes the logic easy to follow |

[Back to the Table of Contents](#toc)

#### <a name="example-17-f-g-x-is-not-the-same-as-g-f-x"></a>Example 17: f(g(x)) is not the same as g(f(x))

**What it does:** defines `f(x)` (add 10) and `g(x)` (multiply by 2), and shows that changing the order changes the answer.

**How it does it:**

| Expression | First step | Second step | Result |
|---|---|---|---|
| `f(g(5))` | `g(5)` = 5 * 2 = 10 | `f(10)` = 10 + 10 = 20 | 20 |
| `g(f(5))` | `f(5)` = 5 + 10 = 15 | `g(15)` = 15 * 2 = 30 | 30 |

**Learning point:** the order of composition matters, just as in maths.

![Working out f(g(5)) and g(f(5))](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-05-nb-09-composition.png)

````python
def f(x):
    """Adds 10 to the input"""
    return x + 10

def g(x):
    """Multiplies the input by 2"""
    return x * 2

# --- Case 1: f(g(x)) ---
# 1. g(5) runs first (5 * 2 = 10)
# 2. f(10) runs second (10 + 10 = 20)
result1 = f(g(5))  # Same as f(g(5)) which is f(10) which is 20

# --- Case 2: g(f(x)) ---
# 1. f(5) runs first (5 + 10 = 15)
# 2. g(15) runs second (15 * 2 = 30)
result2 = g(f(5))  # Same as g(f(5)) which is g(15) which is 30

print(f"f(g(5)) results in: {result1}") # Output: 20
print(f"g(f(5)) results in: {result2}") # Output: 30
````

## <a name="5-16-a-quick-summary"></a>5.16 A quick summary

- A **name** (or "tag") in Python points to an object in memory. When an object is no longer needed, Python frees its memory automatically.
- Names live in a **namespace**. Each namespace has a **scope**: the part of the program where its names can be seen.
- A name gets its scope when it is **first assigned**, usually with `=`. Unlike C++, Python has no way to declare a name without giving it a value.
- A name assigned inside a function has **local** scope. It stops existing when the function ends. The same name can be used again in a different scope without any clash.
- In a local scope, the local name wins over a global one of the same name. Outside the function, only the global name exists.
- Parameters and variables made inside a function cannot be seen from outside. They are local.
- The **lifetime** of a local variable lasts only while its function is running. So an ordinary function does not remember its local values from one call to the next (a closure is the exception, as Example 16 showed).

### <a name="extra-2-the-global-keyword"></a>Extra 2: The global keyword

Example 11 showed that assigning to `x` inside a function creates a new local `x`. If you really want a function to change a global variable, you must say so with the `global` keyword.

- `add_visit()` changes the global `visits`, because of `global visits`.
- Remove the `global visits` line and run the cell again: you get an `UnboundLocalError`, for the same reason as in Example 16.

Use `global` sparingly. For keeping state, a closure (Example 16) is usually the cleaner choice.

````python
visits = 0

def add_visit():
    global visits      # use the global visits, do not make a local one
    visits += 1

add_visit()
add_visit()
print("Visits:", visits)   # Visits: 2
````

### <a name="extra-3-a-closure-counter-without-nonlocal"></a>Extra 3: A closure counter without nonlocal

Section 5.14.2 said that `nonlocal` is not needed if you change the **contents** of a mutable object instead of re-assigning the name. Here `count` is a list holding one number.

`count[0] += 1` changes the item inside the list. The name `count` still points to the same list, so Python does not treat it as a new local variable.

This works, but it is a trick. Many programmers prefer the `nonlocal` version, because it states the intention clearly.

````python
def make_counter():
    count = [0]            # a list, so its contents can be changed

    def counter():
        count[0] += 1      # changes the item; count itself is not re-assigned
        return count[0]

    return counter

c = make_counter()
print(c(), c(), c())       # 1 2 3
````
