## <a name="7-1-introduction"></a>7.1 Introduction

There are two major styles (**paradigms**) of programming:

1. **Procedure-oriented programming (POP)**: the style used in this book so far. A program is mainly a set of functions that take data in, work on it, and give results back.
2. **Object-oriented programming (OOP)**: the program is built from **objects**, each of which holds both some data and the functions (called **methods**) that work on that data.

This chapter introduces OOP in two parts: (1) the ideas and words of OOP, which are much the same in most languages, and (2) how Python puts those ideas into practice.

[Back to the Table of Contents](#toc)

## <a name="7-2-introduction-to-object-oriented-programming"></a>7.2 Introduction to object-oriented programming

As programs grow, keeping data and functions separate becomes hard to manage. OOP groups them together into objects, which makes large programs easier to design, understand, reuse and maintain.

[Back to the Table of Contents](#toc)

### <a name="7-2-1-procedure-oriented-compared-with-object-oriented-progr"></a>7.2.1 Procedure-oriented compared with object-oriented programming

| Aspect | Procedure-oriented (POP) | Object-oriented (OOP) |
|---|---|---|
| Main focus | Functions (procedures) | Objects (data plus methods) |
| Where the data lives | Often global, or passed from function to function | Inside objects (encapsulated) |
| How programs are designed | Top-down | Bottom-up |
| Code reuse | Limited | High, through inheritance and polymorphism |
| Protection of data | Low: data is easy to reach and change | Higher: data hiding and encapsulation |
| Best for | Small, simple programs | Large, complex, real-world applications |

## <a name="7-3-basic-concepts-of-object-oriented-programming"></a>7.3 Basic concepts of object-oriented programming

OOP rests on a few ideas: (1) class and object, (2) abstraction, (3) encapsulation (data hiding), (4) inheritance, and (5) polymorphism.

### <a name="7-3-1-class-object-and-abstraction"></a>7.3.1 Class, object and abstraction

Programs often try to **model** the real world. The real world is full of things (**entities**) that have **characteristics** (what they are) and **behaviour** (what they do). So a model of a real thing needs two parts:

1. its characteristics, which become **attributes** (data), and
2. its behaviour, which becomes **methods** (functions).

A Python object is an "abstraction for data". As you saw in Chapter 1, every Python object has (1) an **identity**, (2) a **type**, and (3) a **value**. The identity is like the object's address in memory. It never changes as long as the object exists.

## <a name="7-4-data-abstraction-and-encapsulation"></a>7.4 Data abstraction and encapsulation

### <a name="abstraction"></a>Abstraction

**Abstraction** means keeping only the details that matter for your purpose and leaving out the rest. It lets you work at a higher level without worrying about how things work inside.

Suppose we are writing software for a vet clinic that keeps records of animal patients. A real pet has many features: species, breed, gender, date of birth, age, weight, height, colour, ear tag number, health, and so on. It eats, drinks, sleeps and makes sounds. For our purpose, we keep only:

| | Kept in our model of a pet |
|---|---|
| Attributes (the "what") | name, species, breed, gender, date of birth, age |
| Behaviour (the "actions") | `make_noise()` (barks, meows ...), `does_tricks()` (fetches a ball, jumps ...) |

![Abstracting the attributes and behaviour of a real pet into a class](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-01-abstraction.png)

**Definition:** abstraction is picking out the essential features of an object and hiding the unimportant details. This reduces complexity and makes the model clearer.

[Back to the Table of Contents](#toc)

### <a name="encapsulation"></a>Encapsulation

In our model, "date of birth" and "age" are linked: if you know the date of birth, the program can work out the age. So the user should only be allowed to set the date of birth, and the program should calculate the age itself. Deciding which data the user may change, and keeping the rest hidden, is **encapsulation**.

**Definition:** encapsulation is packing data (attributes) and the methods that work on it into a single unit (a class), and limiting access to the inner workings.

- **Goal:** protect the data, so outside code cannot change it by accident.
- **Focus:** how an object stores its state and hides its internal logic.

### <a name="7-4-3-the-idea-of-polymorphism"></a>7.4.3 The idea of polymorphism

In the example above we have two classes, `Pet` (the **base** class) and `Fish` (the **derived** class).

## <a name="7-5-operator-overloading"></a>7.5 Operator overloading

**Operator overloading** means that an operator (such as `+`, `-` or `*`) behaves differently depending on the **types** of the values it works on. The operator "looks at" its operands and picks the right action:

| Operands | What `+` does | Example | Result |
|---|---|---|---|
| Two integers | Arithmetic addition | `3 + 4` | `7` |
| Two strings | Concatenation (joining) | `'3' + '4'` | `'34'` |
| An integer and a string | Nothing: it refuses | `3 + '4'` | `TypeError` |

Python does **not** convert one operand to the other type for you. You must convert it yourself, for example `str(3) + '4'` gives `'34'` and `3 + int('4')` gives `7`.

![How the + operator decides what to do](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-04-plus-operator.png)

**How it does it:**

1. `x, y = 3, 4` gives two names their values in one line (multiple assignment). `x + y` is `7`.
2. `x, y = '3', '4'` gives them two strings. Now `x + y` is `'34'`.

## <a name="7-6-function-overloading-in-python"></a>7.6 Function overloading in Python

**Does Python support function overloading?** Not in the usual sense. In Java or C++, you can write several functions with the **same name** but different parameters, and the compiler picks the right one from the arguments you pass. This is called compile-time polymorphism.

Python does not work that way. A function name is just a **name** tied to a function object. If you write `def` twice with the same name, the second `def` simply ties the name to the new function, and the first one is forgotten. (This is the same rule as `x = 1` followed by `x = 2`.)

**The Pythonic way:** instead of writing the same function several times, write one flexible function using:

- **default arguments** (Chapter 6), and
- **a variable number of arguments** with `*args` and `**kwargs` (Chapter 6).

"Beyond the Book", Extra 1 shows one function that handles several kinds of call.

#### <a name="example-2-the-second-definition-replaces-the-first"></a>Example 2: The second definition replaces the first

**What it does:** defines `greet()` twice, then calls it twice. Both calls use the **second** version.

**How it does it:**

1. The first `def greet():` ties the name `greet` to a function with no parameters.
2. The second `def greet(name="Student"):` ties the name `greet` to a new function. The first function is now unreachable: it has become dead code.
3. `greet()` uses the default name, `"Student"`.
4. `greet("Alice")` uses `"Alice"`.

**Learning point:** Python did not choose between two versions. There is only ever one `greet`, the last one defined.

````python
# --- First Definition of the function ---
def greet():
    print("Hello! This is the first version of the function.")

# --- Second Definition of the same function name ---
# This version takes one argument with a default value.
# In Python, this second block completely replaces the first one.
def greet(name="Student"):
    print(f"Hi {name}! This is the second version of the function.")

# --- Testing the function ---
# Even if we call it with no arguments (which the 1st version expected),
# Python only 'knows' the 2nd version exists.
greet()  # Output: Hi Student! This is the second version of the function.
greet("Alice")  # Output: Hi Alice! This is the second version of the function.
````

### <a name="7-7-1-creating-a-simple-class-with-no-init"></a>7.7.1 Creating a simple class with no __init__()

[Back to the Table of Contents](#toc)

#### <a name="example-3-a-simple-pet-class"></a>Example 3: A simple Pet class

**What it does:** creates a class `Pet` with one class attribute and one method, makes an object `my_dog`, and uses both.

**How it does it:**

1. `species = "Canine"` inside the class creates a **class attribute**. It belongs to the class itself and is shared by every object.
2. `def bark(self):` creates a **method**. The first parameter, `self`, stands for the object that calls the method.
3. `Pet.species` reads the attribute straight from the class. No object is needed.
4. `my_dog = Pet()` creates an **instance** (object) of the class.
5. `my_dog.species` finds the attribute. It is not stored on `my_dog` itself, so Python looks in the class and finds `"Canine"`.
6. `my_dog.bark()` runs the method, with `self` set to `my_dog`.

**Learning points:**

- **Class attributes are shared.** There is only one `species` in memory, and every `Pet` object sees it.
- **Every object is the same** so far: without an `__init__()` method, all pets are "Canine". Section 7.8 fixes this.
- **Methods need an object.** `Pet.bark()` gives `TypeError: Pet.bark() missing 1 required positional argument: 'self'`, because no object was given for `self`. Try it in a new cell. (`Pet.bark(my_dog)` works, because then you pass the object yourself.)

![How the main script uses the Pet class and its object](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-06-class-and-instance.png)

When you write `my_dog.species`, Python searches in this order and stops at the first match:

![Where Python looks for obj.name](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-12-attribute-lookup.png)

````python

class Pet:
    # Statement 1: Defining a Class Attribute (shared by all)
    species = "Canine"

    # Statement 2: Defining an Instance Method (a behavior)
    def bark(self):
        print("Woof!")

# --- Using the Class ---

# 1. Accessing data directly from the Class (The Blueprint)
print(Pet.species)    # Output: Canine

# 2. Creating an Instance (The Object)
my_dog = Pet()

# 3. Accessing data and behavior from the Instance
print(my_dog.species) # Output: Canine
my_dog.bark()         # Output: Woof!
````

## <a name="7-8-creating-unique-objects-with-init"></a>7.8 Creating unique objects with __init__()

In Example 3, every pet was the same. Real pets have their own names, breeds and ages. To give each object its own data, a class uses a special method called a **constructor**.

[Back to the Table of Contents](#toc)

### <a name="7-8-1-what-is-a-constructor"></a>7.8.1 What is a constructor?

A **constructor** sets up ("initialises") a new object the moment it is created. It lets whoever creates the object supply that object's own data.

[Back to the Table of Contents](#toc)

### <a name="7-8-2-key-ideas-dunder-and-self"></a>7.8.2 Key ideas: "dunder" and "self"

- **`__init__()`**: in Python the constructor is always named `__init__()`. It has two underscores on each side, so it is called a "double under", or **dunder**, method.
- **`self`**: the most important idea in Python OOP.
    - Outside the class, you use the object's name (such as `p1`) to reach its data.
    - Inside the class, you cannot know what the object will be called, so you use `self` to mean "this particular object".

[Back to the Table of Contents](#toc)

#### <a name="example-4-giving-each-pet-its-own-name-breed-and-age"></a>Example 4: Giving each pet its own name, breed and age

**What it does:** defines `Pet` with an `__init__()` method, creates two different pets, and prints their details two ways.

**How it does it:**

1. `p1 = Pet("Tiger", "Labrador", 3)` creates a new object and calls `__init__()` on it.
2. Inside `__init__()`, `self` is the new object, so `self.name = name` stores `"Tiger"` **on that object**.
3. `p2 = Pet("Kittie", "Siamese Cat", 2)` does the same for a second object, with its own data.
4. `p1.name` reads an attribute directly. `p1.display_info()` calls a method that reads the attributes through `self`.

**Learning points:**

| Point | Meaning |
|---|---|
| The parameter count gap | `__init__(self, name, breed, age)` has **4** parameters, but you pass only **3** arguments. Python passes the object itself as `self`, automatically |
| Local variable compared with attribute | In `self.name = name`, the `name` on the right is a local variable that exists only while `__init__()` runs. `self.name` on the left is an **instance attribute**, stored on the object for as long as the object exists |
| Same names by habit | You could write `self.my_name = name`, but using the same name on both sides is the usual Python style and easier to read |
| No need to call `__init__()` | Python runs it for you the moment you write `Pet(...)` |

![What Python does when you write p1 = Pet("Tiger", "Labrador", 3)](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-07-init-steps.png)

````python
class Pet:
    # The Constructor: Takes 4 parameters (including self)
    def __init__(self, name, breed, age):
        # Assigning the parameters to 'self' (the object)
        self.name = name   # Unique name for this instance
        self.breed = breed # Unique breed for this instance
        self.age = age     # Unique age for this instance

    def display_info(self):
        print(f"Pet Name: {self.name}, Breed: {self.breed}, Age: {self.age}")

# --- Creating unique objects ---
# Note: We pass 3 arguments but __init__ sees 4. Python adds 'self'
# automatically. Each object has its own data (name, breed, age)
# stored in 'self', which is unique to that instance.
p1 = Pet("Tiger", "Labrador", 3)  # The first pet object
p2 = Pet("Kittie", "Siamese Cat", 2)  # The second pet object

# Printing attributes directly
print(f"p1 is {p1.name}, a {p1.breed}.")  # Output: p1 is Tiger, a Labrador
print(f"p2 is {p2.name}, a {p2.breed}.")  # Output: p2 is Kittie, a Siamese Cat

# Calling a method that uses 'self'
p1.display_info()  # Output: Pet Name: Tiger, Breed: Labrador, Age: 3
p2.display_info()  # Output: Pet Name: Kittie, Breed: Siamese Cat, Age: 2
````

##### Notes on Example 4: the rules of self

| Rule | Meaning |
|---|---|
| The first parameter | Every instance method, including `__init__()`, must have `self` as its first parameter. If you leave it out, you get an error when the method is called |
| Inside only | `self` exists only inside the class. Outside, use the object's real name, such as `my_dog` or `p1` |
| Reaching attributes | Inside a method, reach the object's data with `self.`, as in `self.name` |
| Calling other methods | One method calls another with `self.`, as in `self.bark()`, not just `bark()` |
| The automatic drop | When you call a method from outside, you do not pass `self`. You write `my_dog.bark()`, and Python fills in `self` for you |

`self` really is the object. "Beyond the Book", Extra 4 proves it by comparing `id(self)` inside the class with `id(p1)` outside.

## <a name="7-9-constructors-with-default-values"></a>7.9 Constructors with default values

Often most objects share the same values. A kennel where most pets are dogs could make `"Dog"` the default species.

| Feature | Meaning |
|---|---|
| Optional parameters | A parameter with a default value can be left out when the object is created |
| Overwrite rule | If a value is given, it is used; if not, the default is used |
| Order rule | Parameters without defaults must come first; parameters with defaults come after them |
| Keyword arguments | You can give values by name, in any order, for example `Pet("Luna", age=4)` |

[Back to the Table of Contents](#toc)

#### <a name="example-5-a-constructor-with-two-default-values"></a>Example 5: A constructor with two default values

**What it does:** creates three pets, giving one, two and three arguments.

| Call | name | species | age | Output |
|---|---|---|---|---|
| `Pet("Tommy")` | Tommy | Dog (default) | 1 (default) | Tommy is a Dog (Age: 1) |
| `Pet("Luna", "Cat")` | Luna | Cat | 1 (default) | Luna is a Cat (Age: 1) |
| `Pet("Fluffy", "Rabbit", 5)` | Fluffy | Rabbit | 5 | Fluffy is a Rabbit (Age: 5) |

**Learning points:**

- In the second call, the default that is **overridden** is `species`. The default that is still **used** is `age`.
- To give an age but keep the default species, use a keyword argument: `Pet("Rex", age=3)`.

````python
# This code demonstrates the use of default values in a class constructor

class Pet:
    # No default value  for name. But defaults for species and age given.
    def __init__(self, name, species="Dog", age=1):
        self.name = name
        self.species = species
        self.age = age

# 1. Using both Defaults: Only providing the name Tommy.
# species defaults to "Dog", age defaults to 1.
p1 = Pet("Tommy")
print(f"{p1.name} is a {p1.species} (Age: {p1.age})")
# Output: Tommy is a Dog (Age: 1)

# 2. Overriding 1 Default, namely species: Provide name and species.
# species overrides the default "Dog", but age still defaults to 1.
p2 = Pet("Luna", "Cat")  # Creating an instance with name and species. Age takes the default value.
print(f"{p2.name} is a {p2.species} (Age: {p2.age})")
# Output: Luna is a Cat (Age: 1)

# 3. Full Customization: Provide all three arguments. No default used
p3 = Pet("Fluffy", "Rabbit", 5)  # Creating instance with name, species & age
print(f"{p3.name} is a {p3.species} (Age: {p3.age})")
# Output: Fluffy is a Rabbit (Age: 5)
````

## <a name="7-10-class-with-methods-defining-the-behaviour-of-a-class"></a>7.10 Class with methods: defining the behaviour of a class

A class is more than a box for data; its objects can **do** things. The functions defined inside a class are called **methods**.

| Feature | Meaning |
|---|---|
| `self` is needed | Every method takes `self` as its first parameter, so it can reach the object's data |
| Encapsulation | Instead of changing an object's data directly from outside, you call methods to **set** or **get** it |
| Internal attributes | `__init__()` can create attributes that are not passed in as arguments, and give them a starting value |

[Back to the Table of Contents](#toc)

#### <a name="example-6-a-setter-and-a-getter-for-the-mood"></a>Example 6: A setter and a getter for the mood

**What it does:** creates a pet whose mood starts as `"Happy"`, reads the mood, changes it to `"Sleepy"`, and reads it again.

**How it does it:**

1. `__init__()` stores `name` and `species` from its parameters, and also sets `self.mood = "Happy"`, which is **not** a parameter. Every new pet starts happy.
2. `set_mood(self, new_mood)` is a **setter**: it changes the attribute. From outside you pass only the mood: `my_pet.set_mood("Sleepy")`.
3. `get_mood(self)` is a **getter**: it returns the current value. It needs only `self`, because it just reads data already stored on the object.

**Learning points:**

- An object can get its attributes in two ways: (1) from arguments passed to the constructor, or (2) set to a starting value inside `__init__()`.
- The brackets in `my_pet.get_mood()` are needed: they tell Python to **run** the method. Without them, you get the method object itself (as in Chapter 6).
- Methods keep all the changes to an object's state in one place, which makes bugs easier to find in large programs.

![How the getter and setter methods change and read the mood](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-08-getter-setter.png)

````python
class Pet:
    # Constructor with a default value for species and an additional attribute
    # mood which not in the constructor parameters. So when you create a Pet
    # object, you only provide name and optionally species. You cannot set
    # mood at creation, it always starts as "Happy". But you can change it
    # later using the setter method..
    def __init__(self, name, species="Dog"):
        self.name = name  # This attribute is set from the constructor parameter
        self.species = species  # This has default value of "Dog" if not given
        # This attribute is NOT in the constructor parameters.
        # Every pet starts with a default mood of 'Happy'.
        self.mood = "Happy"  # This is an attribute that is initialized in the
        # constructor but not passed as a parameter.

    # Method to update the mood (Setter)
    def set_mood(self, new_mood):
        # We use self.mood to change the attribute for this specific pet
        self.mood = new_mood

    # Method to retrieve the mood (Getter)
    def get_mood(self):
        # Returns the current value of the attribute
        return self.mood

# --- Testing the Methods ---

# 1. Create a pet object
my_pet = Pet("Rocky")  # We only provide the name. Species defaults to "Dog".
# Mood defaults to "Happy".
print(f"{my_pet.name} is a {my_pet.species}.")  # Output: Rocky is a Dog.

# 2. Check the default mood using the getter method
print(f"Default mood: {my_pet.get_mood()}") # Output: Default mood: Happy

# 3. Change the mood using the setter method
my_pet.set_mood("Sleepy")  # This changes mood attribute of my_pet to "Sleepy"

# 4. Check the mood again
print(f"Updated mood: {my_pet.get_mood()}")  # Output: Updated mood: Sleepy
````

## <a name="7-11-managing-the-factory-class-variables-and-static-methods"></a>7.11 Managing the factory: class variables and static methods

So far we have used **instance variables** (such as a pet's name), which belong to one object. Sometimes you need data about the **whole group**, such as a count of how many pets have been created.

Think of a car factory:

| Kind of variable | Belongs to | Car factory example | Pet example |
|---|---|---|---|
| Instance variable | One object | The colour of one car | `self.name` |
| Class variable | The class (shared by all objects) | The number of cars the factory has made | `Pet.count` |

[Back to the Table of Contents](#toc)

#### <a name="example-7-counting-pets-with-a-class-variable-and-a-static-m"></a>Example 7: Counting pets with a class variable and a static method

**What it does:** counts every `Pet` that is created, and prints the total using a static method.

**How it does it:**

1. `count = 0` sits directly inside the class, not inside a method, and has no `self.`. So it is a **class variable**.
2. Each time a pet is created, `__init__()` runs `Pet.count += 1`.
3. `@staticmethod` marks `display_total()` as a **static method**: it takes no `self`, and it is called on the class itself: `Pet.display_total()`.

**Learning points:**

- Inside `__init__()` the code uses `Pet.count`, not `self.count`. Writing `self.count += 1` would create a separate `count` on each object, and the class total would stay 0.
- A static method cannot see any one object's data, such as `self.name`. It is for utility jobs that concern the class as a whole.
- `@staticmethod` is a **decorator** (Chapter 6). Placing it just above the method is the standard, readable way to declare a static method.

| Feature | Instance method | Static method |
|---|---|---|
| First parameter | `self` | none |
| How it is called | `my_pet.bark()` | `Pet.display_total()` |
| Can reach | the object's data (`self.name` and so on) and class variables | only class variables, through the class name |
| Purpose | behaviour of one object | utility jobs for the whole class |

![How the class variable count keeps a total for all pets](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-09-class-variable.png)

````python
class Pet:
    count = 0

    def __init__(self, name, species="Dog"):
        Pet.count += 1 # A shorthand for Pet.count = Pet.count + 1
        self.name = name
        self.species = species
        print(f"Created {self.name}. Total pets now: {Pet.count}")

    # 1. The Modern Decorator tells Python this is a Static Method
    @staticmethod
    def display_total():
        # This belongs to the Class, no 'self' needed
        print(f"--- Global Pet Count: {Pet.count} ---")

# --- Using the Class ---
p1 = Pet("Tommy")  # Create first pet object. Increments count to 1.
p2 = Pet("Luna", "Cat")  # Create second pet object. Increments count to 2.

# 2. Calling the static method. No object needed, we call it on class itself.
Pet.display_total()  # Output: --- Global Pet Count: 2 ---
````

## <a name="7-12-class-decorators"></a>7.12 Class decorators

A **class decorator** is a function that (1) takes a class as its argument, (2) changes it, and (3) returns it. It changes the blueprint itself, at the moment the class is defined, for example by adding attributes or methods.

A function decorator wraps each **call** of a function. A class decorator usually changes the **structure** of a class.

[Back to the Table of Contents](#toc)

### <a name="7-12-1-the-script-decorating-a-class"></a>7.12.1 The script: decorating a class

#### <a name="example-8-adding-a-location-a-status-and-a-method-to-a-class"></a>Example 8: Adding a location, a status and a method to a class

**What it does:** writes a decorator `add_pet_status` and uses it on a `Dog` class. Every `Dog` then has a location, a vaccination flag and a `display_status()` method, even though the `Dog` class itself does not define them.

**How it does it:**

1. `@add_pet_status` above `class Dog:` means: after building the class, run `Dog = add_pet_status(Dog)`.
2. Inside the decorator, `cls` is the `Dog` class. `cls.location = "Delhi"` and `cls.is_vaccinated = True` add two class attributes.
3. `display_status(self)` is an ordinary function defined inside the decorator. `cls.display_status = display_status` attaches it to the class, which turns it into a method.
4. `return cls` hands back the same class, now upgraded.
5. `Dog("Tiger")` makes an object that can use all the new parts.

**Learning points:**

- Every `cls.something = ...` line adds a feature that all future objects of the class will have.
- If you left out `cls.display_status = display_status`, the function would exist only inside the decorator, and objects could not use it.

| Feature | Function decorator | Class decorator |
|---|---|---|
| Works on | one function or method | a whole class |
| When its code runs | the wrapper runs every time the function is called | once, when the class is defined |
| Typical uses | logging, timing, checking permissions | adding attributes, registering classes, validation |

![What the class decorator does to the Dog class](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-10-class-decorator.png)

````python
# --- THE CLASS DECORATOR ---
def add_pet_status(cls):
    # A decorator that 'injects' extra info into a class blueprint

    # 1. Add simple attributes
    cls.location = "Delhi"
    cls.is_vaccinated = True

    # 2. Add a simple method which uses the injected attribute
    def display_status(self):
        if  self.is_vaccinated:
            status = "Protected"
        else:
            status = "Not Protected"

        return f"{self.name} is at {self.location}. Status: {status}"

    # Attach method to the class. Inject it into class blueprint. If you
    # don't do this, the method won't be available to instances of the class
    cls.display_status = display_status

    # 3. Return the class (The Blueprint is now upgraded)
    return cls

# --- APPLYING THE DECORATOR ---
@add_pet_status
class Dog:
    def __init__(self, name):
        self.name = name

# --- TESTING ---
my_dog = Dog("Tiger")

print(f"Location: {my_dog.location}")      # Directly access injected attribute
# Output: Location: Delhi
print(f"Summary:  {my_dog.display_status()}") # Call the injected method
# Output: Summary:  Tiger is at Delhi. Status: Protected
````

[Back to the Table of Contents](#toc)

### <a name="7-12-2-why-functools-is-not-needed-here"></a>7.12.2 Why functools is not needed here

In Chapter 6, the function decorator **replaced** the original function with a new `wrapper` function. Such a wrapper loses the original function's name and docstring, which is why real decorators usually use `functools.wraps` to copy them across.

This class decorator does not replace anything. It (1) takes the `Dog` class, (2) adds new variables and a method to it, and (3) returns the very same class. The class keeps its identity, so there is nothing to copy.

[Back to the Table of Contents](#toc)

### <a name="7-12-3-when-would-a-class-decorator-need-functools"></a>7.12.3 When would a class decorator need functools?

Only if it **wrapped** the methods inside the class, replacing them with new wrapper functions.

## <a name="7-13-data-hiding-name-mangling-and-pseudo-private-variables"></a>7.13 Data hiding, name mangling and pseudo-private variables

**Data hiding** means protecting a class's internal data so that code outside the class does not change it directly. Sometimes a programmer wants some variables to be used only inside the class, and not changed by other programmers.

**Name mangling:** Python supports this idea by quietly renaming any name that starts with **two underscores** inside a class. `__variable` in class `ClassName` is stored as `_ClassName__variable`. This also avoids name clashes when classes inherit from each other.

**Pseudo-private variables:** Python has no truly private variables like Java or C++. A mangled name is only "pseudo-private":

1. it is not meant to be used from outside, but
2. it can still be reached if you know its mangled name, `_ClassName__variable`.

[Back to the Table of Contents](#toc)

#### <a name="example-9-a-normal-variable-and-a-mangled-variable"></a>Example 9: A normal variable and a mangled variable

**What it does:** creates class `Pet` with a normal class variable `name` and a mangled one `__type`, and reads both from an object.

**How it does it:**

1. `name = "Cat"` is stored as it is, so `p.name` works.
2. `__type = "Animal"` is stored as `_Pet__type`, so `p._Pet__type` works.

**Learning point:** try `p.__type` in a new cell. You get `AttributeError: 'Pet' object has no attribute '__type'`, because no attribute has that exact name.

![How name mangling renames __type inside class Pet](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-07-nb-11-name-mangling.png)

````python
# This code demonstrates the concept of name mangling in Python classes.
class Pet:

    name = "Cat"          # normal variable
    __type = "Animal"     # mangled variable

p = Pet()
print("Normal variable:", p.name)  # Output: Normal variable: Cat
print("Mangled variable:", p._Pet__type)  # Output: Mangled variable: Animal
````

##### Notes on Example 9: rules for name mangling

| Rule | Example name | Mangled? | Stored as (inside class Pet) | Remarks |
|---|---|---|---|---|
| Two leading underscores | `__age` | Yes | `_Pet__age` | The standard way to make a pseudo-private variable |
| Two leading underscores and at most one trailing underscore | `__age_` | Yes | `_Pet__age_` | Still mangled |
| One leading underscore | `_age` | No | `_age` | By convention means "for internal use", but nothing is renamed |
| Two underscores on both sides | `__init__` | No | `__init__` | Kept for Python's special ("magic") methods |
| No leading underscore | `age` | No | `age` | A normal, public name |

**Key points:** mangling does not make a variable truly private. It lowers the chance of name clashes, it is mostly useful with inheritance, and the variable can still be reached as `_ClassName__variable`.

## <a name="7-15-implicit-inheritance-from-object"></a>7.15 Implicit inheritance from object

Inheritance from `object` is **implicit**: even if you do not name a parent class, Python makes your class inherit from `object`.

**The word "object" is used in two ways:**

| | Meaning |
|---|---|
| `object` (lowercase, in code) | the built-in root class that every class inherits from |

### <a name="7-15-1-the-str-method"></a>7.15.1 The __str__() method

`__str__()` returns the **text version** of an object. It is used automatically by `print(obj)` and `str(obj)`. If your class does not define it, the one inherited from `object` is used, which gives text like `<__main__.Pet object at 0x7f...>`.

[Back to the Table of Contents](#toc)

#### <a name="example-10-doc-bases-and-str"></a>Example 10: __doc__, __bases__ and __str__()

**What it does:** prints the class docstring, the parent classes, and the default text form of an object.

| Line | Output |
|---|---|
| `Pet.__doc__` | `<<Pet class docstring>>` |
| `Pet.__bases__` | `(<class 'object'>,)`: `Pet` inherits from `object` |
| `print(..., p)` | `<__main__.Pet object at 0x...>`: `print()` called `__str__()` for us |

**Learning points:**

````python
class Pet:
    """<<Pet class docstring>>"""

    def __init__(self, name):
        self.name = name

print("Class docstring ->", Pet.__doc__)
# Output: Class docstring -> <<Pet class docstring>>

print("Base class of Pet ->", Pet.__bases__)
# Output: Base class of Pet -> (<class 'object'>,)

# Create an object
p = Pet("Tom")

print("Object representation using print() ->", p)
# print() internally calls the __str__() method

print("Calling __str__() explicitly ->", p.__str__())
# Output will be something like: <__main__.Pet object at 0x...>
````

### <a name="7-15-3-the-name-attribute-a-module-attribute"></a>7.15.3 The __name__ attribute (a module attribute)

Every module has a `__name__` attribute.

4. This lets a file contain code that runs only when the file is run directly.
5. `__name__` belongs to the module, not to a class. (A **class** also has a `__name__`, but that is the class's own name, such as `"Pet"`.)

### <a name="7-15-4-the-module-attribute"></a>7.15.4 The __module__ attribute

`__module__` gives the name of the module in which a class or function was defined. It helps you find where something came from, which is useful in large projects.

#### <a name="example-12-pet-py-run-directly"></a>Example 12: pet.py, run directly

**What it does:** this is the file `pet.py` from the book. In a notebook, the cell is run directly, so the `if __name__ == "__main__":` part runs and prints the "executed as a script" messages.

**How it does it:**

2. `if __name__ == "__main__":` checks how the file is being used.
3. Run directly: `Pet.__name__` is `"Pet"` and `Pet.__module__` is `"__main__"`.
4. Imported by another file: the `else` part runs instead, and `Pet.__module__` is `"pet"`.

````python
class Pet:
    pass

if __name__ == "__main__":
    print("Pet is being executed as a script.")
    print("Pet.__name__ is->", Pet.__name__)  # Pet.__name__ is-> Pet
    print("Pet.__module__ is->", Pet.__module__)  # Pet.__module__ is-> __main__
else:
    print("Pet is being imported as a module.")
    print("Pet.__name__ is->", Pet.__name__)  # Pet.__name__ is-> Pet
    print("Pet.__module__ is->", Pet.__module__)  # Pet.__module__ is-> pet
````

### <a name="7-15-6-the-del-method-destructor"></a>7.15.6 The __del__() method (destructor)

`__del__()` is called when an object is about to be destroyed.

1. Python calls it automatically, when the object has no references left.
2. After that, the garbage collector frees the memory.
3. `__del__()` is not the same as the `del` statement. `del p` only removes the **name** `p`. The object is destroyed (and `__del__()` called) only when **no** names refer to it any more.

"Beyond the Book", Extra 3 shows this.

### <a name="extra-1-one-flexible-function-instead-of-overloading"></a>Extra 1: One flexible function instead of overloading

Section 7.6 said that Python uses default values and `*args` instead of overloading. This one `area()` function handles a square (one value), a rectangle (two values) and any number of values in a row, with no second definition.

````python
def area(*sides):
    if len(sides) == 1:
        return sides[0] * sides[0]      # a square
    if len(sides) == 2:
        return sides[0] * sides[1]      # a rectangle
    return "Please give one or two sides"

print(area(4))       # 16
print(area(4, 5))    # 20
print(area())        # Please give one or two sides
````

### <a name="extra-2-defining-your-own-str"></a>Extra 2: Defining your own __str__()

Example 10 showed the unhelpful default text of an object. If you define `__str__()` in your class, `print()` uses your version instead. It must **return** a string.

````python
class Pet:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def __str__(self):
        return f"{self.name} the {self.species}"

p = Pet("Tommy", "Dog")
print(p)          # Tommy the Dog
print(str(p))     # the same
````

### <a name="extra-3-del-removes-a-name-del-runs-when-the-object-goes"></a>Extra 3: del removes a name; __del__() runs when the object goes

`del a` removes only the name `a`. The object is still reachable through `b`, so it is not destroyed. When `b` is deleted too, no names are left, and Python calls `__del__()`.

(This is how standard Python, CPython, behaves. It frees an object as soon as nothing refers to it.)

````python
class Pet:
    def __init__(self, name):
        self.name = name

    def __del__(self):
        print(f"{self.name} is being removed from memory")

a = Pet("Tommy")
b = a                  # two names, one object
del a
print("After del a: the object still exists, reached through b")
del b                  # now no names are left, so __del__ runs
````

### <a name="extra-4-self-really-is-the-object"></a>Extra 4: self really is the object

`self` inside a method and the object's name outside the class refer to the **same** object. Their `id()` numbers are equal.

````python
class Pet:
    def show_id(self):
        print("id(self) inside the class :", id(self))

p1 = Pet()
p1.show_id()
print("id(p1) outside the class  :", id(p1))
````
