## <a name="8-1-introduction"></a>8.1 Introduction

So far we have written classes that stand alone. Python also lets you build a **new class from an existing class**. This is called **inheritance**. It is one of the main ideas of object-oriented programming, and most of this chapter is about it.

[Back to the Table of Contents](#toc)

## <a name="8-2-what-is-inheritance"></a>8.2 What is inheritance?

- Inheritance means **creating a new class from an existing class**.
- The new class **reuses** what the existing class already has, and can **extend** it with new things.
- Think of a family tree. You may get your eye colour or your surname from your parents. In the same way, a **child class** (also called a **subclass** or **derived class**) gets attributes and methods from a **parent class** (also called a **superclass** or **base class**).
- **Goal:** reuse code. You write the logic once in the parent and every child can use it.
- **The root of all classes:** every class you write inherits from a built-in class called `object`, even if you do not say so.
- **The `pass` keyword:** if the child should be exactly like the parent, with nothing new, the body of the child class is just `pass`.

[Back to the Table of Contents](#toc)

### <a name="key-characteristics-of-inheritance"></a>Key characteristics of inheritance

| Property | What it means |
|---|---|
| Code reusability | You do not rewrite the same code for similar objects. |
| Transitivity | If B inherits from A, and C inherits from B, then C also has what A has. |
| Overriding | A child can change how a parent method works. For example, a dog speaks differently from a general animal. |
| Method resolution | Python looks for a method in the child first. If it is not there, it looks in the parent. |

### <a name="example-1-type-and-isinstance-with-a-parent-and-a-child-clas"></a>Example 1: `type()` and `isinstance()` with a parent and a child class

**What the code does.** It creates a class `Parent` and a class `Child` that inherits from it. Both classes are empty (`pass`). It makes one object of each: `p1` from `Parent` and `c1` from `Child`. Then it asks two kinds of questions: "What is the exact class of this object?" (`type()`), and "Does this object count as an instance of this class?" (`isinstance()`).

**How it works.** `class Child(Parent):` is the line that creates the inheritance. `type()` always gives the one class the object was made from. `isinstance()` is more generous. It also says `True` when the object's class is a **child** of the class you ask about.

**Why it matters.** A child object can be used anywhere a parent object is expected. `isinstance()` is the normal way to test this. It is better than comparing `type(x) == SomeClass`, because it also accepts subclasses.

**Learning points**

1. `c1` is a `Child`, and because `Child` inherits from `Parent`, `c1` is also a `Parent`.
2. The link is one way only. `p1` is a `Parent` but not a `Child`. A parent does not know about its children.
3. `type()` answers "exactly which class?". `isinstance()` answers "is it this kind of thing?".

The flowchart shows how `isinstance()` reaches its answer.

![How isinstance(obj, SomeClass) decides between True and False](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-01-isinstance.png)

````python
class Parent:
    pass

class Child(Parent):  # Child inherits from Parent.
    # Child is a subclass of Parent and Parent is a superclass of Child
    pass

p1 = Parent()  # Create an instance of Parent
c1 = Child()  # Create an instance of Child

# Let's analyze the types and inheritance relationships
# type() will show us the class of the instance
print("type of p1:", type(p1))  # <class '__main__.Parent'>
print("type of c1:", type(c1))  # <class '__main__.Child'>

# isinstance() will check if an instance belongs to a class or its subclasses
print(f"Is p1 an instance of Parent? {isinstance(p1, Parent)}")  # True
print(f"Is c1 an instance of Child? {isinstance(c1, Child)}")  # True
print(f"Is c1 an instance of Parent? {isinstance(c1, Parent)}")
# True, because Child inherits from Parent
print(f"Is p1 an instance of Child? {isinstance(p1, Child)}")
# False, because Parent does not inherit from Child
````

## <a name="8-3-mro-method-resolution-order"></a>8.3 MRO (Method Resolution Order)

The **MRO** is the **order in which Python searches classes for a method or an attribute**.

When you call `obj.method()`, Python asks, in this order:

1. Is the method in the object's own class?
2. If not, is it in the parent class?
3. If not, is it in the parent's parent, and so on?
4. If it is found nowhere, Python raises an `AttributeError`.

Every class has an attribute `__mro__`. It is a tuple that lists the classes in the exact order Python will search them.

[Back to the Table of Contents](#toc)

### <a name="example-2-showing-the-mro-of-object-animal-and-dog"></a>Example 2: Showing the MRO of `object`, `Animal` and `Dog`

**What the code does.** It prints the MRO of three classes: the built-in `object`, a class `Animal` (which inherits from `object` without saying so), and a class `Dog` that inherits from `Animal`.

**How it works.** `ClassName.__mro__` is read like any other attribute. Python builds it when the class is created.

**Why it matters.** When a program calls the "wrong" method, the MRO tells you why. It is the map Python follows.

**Learning points**

1. The MRO of `object` holds only `object` itself. It has no parent.
2. `Animal` does not name a parent, but its MRO still ends with `object`.
3. The MRO of `Dog` is `Dog`, then `Animal`, then `object`.
4. The MRO is:
    - **linear**: it is one single list,
    - **top-down**: the child comes first, then the parents,
    - **deterministic**: it is always the same order,
    - **ends at `object`**: the last class is always `object`.

![How Python searches the MRO of Dog for a method](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-02-mro-lookup.png)

````python
print(object.__mro__)  # Shows the mro for the base 'object' class
# Output: (<class 'object'>,)
# So object → object (itself)

class Animal:
    def speak(self):
        print("Animal speaks")

print(Animal.__mro__)  # Shows the method resolution order for the Animal class
# Output: (<class '__main__.Animal'>, <class 'object'>)
# So Animal → object

class Dog(Animal):
    pass

print(Dog.__mro__)  # Shows the method resolution order for the Dog class
# Output: (<class '__main__.Dog'>, <class '__main__.Animal'>, <class 'object'>)
# So Dog → Animal → object
````

## <a name="8-4-role-of-the-object-class"></a>8.4 Role of the object class

`object` is the top-most class in Python. Every class inherits from it. You do not import it or create it, because it is built in. It gives every object a basic set of methods, so that any object can be (1) created, (2) initialised, (3) printed, and (4) compared.

[Back to the Table of Contents](#toc)

### <a name="core-methods-inherited-from-object"></a>Core methods inherited from `object`

| Method | Purpose | When it is called | What the default version does | Returns |
|---|---|---|---|---|
| `__new__` | Creates the object | Before the object exists | Allocates memory and returns a new instance | the new object |
| `__init__` | Initialises the object | Just after the object is created | Nothing | `None` (it must not return anything else) |
| `__str__` | Friendly text for users | `print(obj)` or `str(obj)` | `<ClassName object at address>` | `str` |
| `__repr__` | Exact text for developers | `repr(obj)`, or when the interpreter shows the object | Same as the default `__str__` | `str` |
| `__eq__` | Equality test | `obj1 == obj2` | Compares identity, like `is` | `bool` |
| `__hash__` | Hash value | When the object is a set member or a dictionary key | Based on the object's identity | `int` |

Points to remember:

- If `__new__` does not return an instance of the class, `__init__` is not called.
- If you write your own `__eq__`, think about writing `__hash__` too. Two objects that are equal must have the same hash.
- If you write only `__repr__`, `print()` uses it as well, because the default `__str__` falls back to `__repr__`.

## <a name="8-6-inheritance-as-an-is-a-relationship"></a>8.6 Inheritance as an "IS-A" relationship

Inheritance models an **IS-A** relationship. A subclass is a special kind of its superclass. It gets all the data and behaviour of the parent, and then it may keep them, add to them, or change them.

In simple words: if a Dog **is a** Pet, then anywhere your code expects a Pet, a Dog can be used and will work.

A quick test: say the sentence "A ___ is a ___." If it sounds right ("A Dog is a Pet"), inheritance fits. If it sounds wrong ("A Dog is a Collar"), it does not.

## <a name="8-7-what-is-super"></a>8.7 What is super()?

`super()` is used inside a child class to call a method of the parent class.

In simple words, it means: "go to my parent and use its method."

[Back to the Table of Contents](#toc)

### <a name="why-we-need-super"></a>Why we need super()

If a child class writes its own `__init__()`, the parent's `__init__()` is **not** called automatically. Python does not do it for you (unlike Java or C++). So if you do not write `super().__init__(...)`, the attributes that the parent's `__init__()` would have set will simply not exist in the child object.

With `super()` you can:

1. reuse what the parent already does,
2. avoid writing the same code twice, and
3. keep the design clean. Each class looks after its own part.

[Back to the Table of Contents](#toc)

### <a name="syntax"></a>Syntax

```python
super().method_name(arguments)
```

Note that you do **not** pass `self`. `super()` already knows which object it is working for.

[Back to the Table of Contents](#toc)

### <a name="what-happens-inside"></a>What happens inside

When you call `super().method()`, Python:

1. looks at the MRO of the object's class,
2. finds the class that comes **after** the current class in that MRO, and
3. calls `method()` from there.

So `super()` does not strictly mean "call my immediate parent". It means "call the **next** method in the MRO". In single inheritance these are the same thing. In multiple inheritance they can be different.

[Back to the Table of Contents](#toc)

### <a name="example-3-super-in-constructors-and-methods-over-three-level"></a>Example 3: super() in constructors and methods, over three levels

**What the code does.** It builds a chain of three classes: `Pet`, then `Dog(Pet)`, then `Puppy(Dog)`. Each class has its own `__init__()` and its own `walk()`. Each one prints a message and then calls the same method in the class above it with `super()`. Finally it prints the MRO of `Puppy`.

**How it works.**

- `Puppy.__init__` prints its message and calls `super().__init__(name)`. That runs `Dog.__init__`.
- `Dog.__init__` prints its message and calls `super().__init__(name)`. That runs `Pet.__init__`, which stores `self.name`.
- `walk()` works the same way. Each class prints its own line and then passes the call up.

This passing of the call up the chain is called **constructor chaining** (for `__init__`) or **method chaining** (for other methods).

**Why it matters.** Each class only deals with its own part. `Pet` is the only class that sets `name`. If `Pet` changes, `Dog` and `Puppy` still work and get the change for free.

**Learning points**

1. Look at the order of the messages. In `Dog.__init__` and `Puppy.__init__` the `print()` comes before `super().__init__()`, so the child's message appears first. If you swap the two lines, the parent's message appears first.
2. `self.name` is set only in `Pet`, yet `Dog.walk()` and `Puppy.walk()` can use it. It is stored on the same object.
3. The MRO of `Puppy` is `Puppy`, `Dog`, `Pet`, `object`. `super()` follows this list.

![Constructor chaining with super() when Puppy("Tommy") is created](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-03-super-chain.png)

````python
# Demonstration of super() with multi-level inheritance (MRO)

class Pet:
    def __init__(self, name):
        self.name = name  # Instance variable to store name of pet
        print("Pet constructor called")

    def walk(self):  # Parent method
        print(f"{self.name} is walking (Pet method)")

class Dog(Pet):  # Dog inherits from Pet
    def __init__(self, name):  # Overrides the constructor of Pet
        print("Dog constructor called")
        super().__init__(name)  # Calls Pet constructor

    def walk(self):  # Overriding + super() to call parent method
        print(f"{self.name} is walking (Dog method)")
        super().walk()          # Calls next in MRO → Pet

# Create object of Dog
d = Dog("Tommy")  # Constructor chaining: Dog → Pet
# Expected Output: on object creation:
# Dog constructor called
# Pet constructor called
print("Calling walk():")
d.walk()  # Expected Output:
# Tommy is walking (Dog method)
# Tommy is walking (Pet method)

# NEW CLASS (Derived from Dog)
class Puppy(Dog):  # Puppy inherits from Dog
    def __init__(self, name):  # Overrides the constructor of Dog
        print("Puppy constructor called")
        super().__init__(name)  # Calls Dog → then Pet

    def walk(self):  # Overriding + super() to call parent method
        print(f"{self.name} is walking (Puppy method)")
        super().walk()          # Calls Dog → then Pet

# Create object of Puppy
p = Puppy("Tommy")  # Constructor chaining: Puppy → Dog → Pet
# Expected Output: on object creation:
# Puppy constructor called
# Dog constructor called
# Pet constructor called

print("Calling walk():")
p.walk()  # Expected Output:
# Tommy is walking (Puppy method)
# Tommy is walking (Dog method)
# Tommy is walking (Pet method)

print("MRO of Puppy:")  # MRO shows order in which classes are looked up when
# searching for a method. MRO for Puppy is Puppy → Dog → Pet → object.
print(Puppy.__mro__)  # Output:(<class '__main__.Puppy'>,
# <class '__main__.Dog'>, <class '__main__.Pet'>, <class 'object'>)
````

## <a name="8-8-the-four-types-of-is-a"></a>8.8 The four types of "IS-A"

A child class can relate to its parent in four ways.

| No. | Type | What the child does | How it thinks |
|---|---|---|---|
| 1 | **Implicit (pass-through)** | It is an exact copy of the parent. The body is just `pass`. | "I want exactly what the parent has, under a different name." |
| 2 | **Extended (additive)** | It keeps everything and **adds** new methods or attributes. This is the most common type. | "I am a version of the parent, with some extra tools." |
| 3 | **Specific (replacement)** | It **overrides** a parent method with completely different logic. | "I am a type of parent, but I do this task in my own way." |
| 4 | **Cooperative (refinement)** | It overrides a method but also calls the parent version with `super()`, then adds its own work. | "I add my own touch, but I still want the parent to do its normal job." |

Where each type is used:

- **Pass-through** is useful for giving a clear name to a role, even when the data is the same. For example, `class Admin(User): pass` lets your code tell admins apart from other users.
- **Replacement** is used when the parent only has a default or dummy behaviour that is wrong for the child.
- **Cooperative** is the most robust. If the parent's logic changes later (say, a bug fix), the child gets the change automatically.

[Back to the Table of Contents](#toc)

### <a name="how-the-four-types-use-the-parent"></a>How the four types use the parent

| Type | Uses super()? | Parent methods | Child's role |
|---|---|---|---|
| Pass-through | No | Used as they are | None, just `pass` |
| Extension | No, or optional | Used as they are | Adds new methods or attributes |
| Replacement | No | Hidden by the child's version | Replaces the parent logic completely |
| Cooperation | Yes | Called and extended | Adds logic on top of the parent logic |

[Back to the Table of Contents](#toc)

### <a name="example-4-the-four-types-of-is-a-inheritance"></a>Example 4: The four types of IS-A inheritance

**What the code does.** It creates a base class `Pet` with `__init__()`, `speak()` and `eat()`, and then one child class for each type:

| Class | Type | What it changes |
|---|---|---|
| `BasicPet` | Pass-through | Nothing |
| `Retriever` | Additive | Adds `fetch()` |
| `Cat` | Replacement | Its own `speak()` prints "Meow!" instead of the generic sound |
| `Dog` | Cooperative | Its own `__init__()` adds `weight`; its own `speak()` growls, calls `Pet.speak()`, then barks |

It then makes one object of each and calls a method on it.

**How it works.** When `p1.speak()` runs, Python finds no `speak()` in `BasicPet`, so it uses the one in `Pet`. `p3.speak()` finds `speak()` in `Cat` first, so the `Pet` version is never reached. `p4.speak()` finds `speak()` in `Dog`, and that method calls `super().speak()` in the middle.

**Why it matters.** Most real class designs use these four patterns. Knowing their names helps you choose the right one and explain your design to others.

**Learning points**

1. `BasicPet("Rocky")` works even though `BasicPet` has no `__init__()`. The one in `Pet` is used.
2. `Retriever` still has `speak()` and `eat()` from `Pet`, as well as its new `fetch()`.
3. `Cat` never prints the generic sound. The parent method is hidden, not deleted: `Pet.speak(p3)` would still work.
4. `Dog` needs `super().__init__(name)`. Without it, `self.name` would not exist and `speak()` would fail.

![The four types of IS-A inheritance, all derived from Pet](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-04-four-isa-types.png)

````python
class Pet:
    def __init__(self, name):  # Constructor
        self.name = name

    def speak(self):  # Method to be overridden
        print(f"{self.name} makes a generic pet sound.")

    def eat(self):  # Method to demonstrate code reuse
        print(f"{self.name} is eating.")

# --- TYPE 1: Pass-through (Verbatim) ---
# Logic: "I am exactly like a Pet."
class BasicPet(Pet):  # No new attributes/ methods. Inherits everything as-is
    pass

# --- TYPE 2: Additive (Extension) ---
# Logic: "I am a Pet, but I also have a breed and can fetch."
class Retriever(Pet):  # Inherits from Pet and adds new behavior
    def fetch(self):
        print(f"{self.name} is fetching the ball!")

# --- TYPE 3: Replacement (Overriding) ---
# Logic: "I am a Pet, but I don't make generic sounds; I ONLY purr."
class Cat(Pet):  # Inherits from Pet but replaces the speak() method
    def speak(self):  # Overrides the speak() method of Pet
        print(f"{self.name} says: Meow!")

# --- TYPE 4: Refinement (Cooperative) ---
# Logic: "I am a Pet. I'll let the Parent handle the name,
# but I will add 'barking' to the speak behavior."
class Dog(Pet):  # Inherits from Pet and refines the speak() method
    def __init__(self, name, weight):
        super().__init__(name)  # Let Pet handle the name
        self.weight = weight    # Dog handles the weight

    def speak(self):
        print(f"{self.name} lets out a loud growl...")
        super().speak()         # Then do the normal pet sound
        print(f"{self.name} barks!")

# --- Demonstration ---

print("--- Type 1: BasicPet ---")
p1 = BasicPet("Rocky")  # Inherits everything from Pet without changes
p1.speak()

print("--- Type 2: Retriever (Extension) ---")
p2 = Retriever("Buddy")  # Has additional fetch() method
p2.fetch() # New method

print("--- Type 3: Cat (Replacement) ---")
p3 = Cat("Kitty")  # Overrides speak() method
p3.speak() # Parent's generic sound is gone

print("--- Type 4: Dog (Refinement) ---")
p4 = Dog("Leo", 30)  # Adds weight attribute, and refines speak() method
p4.speak() # Combines Parent logic and Child logic
````

### <a name="what-happens-to-init-when-a-derived-object-is-made"></a>What happens to `__init__()` when a derived object is made

There are three possibilities:

1. **The derived class has no `__init__()` of its own.** The parent's `__init__()` is used automatically.
2. **The derived class has its own `__init__()`, but it does not call `super().__init__()`.** Only the child's `__init__()` runs. It completely replaces the parent's, so the parent's attributes are never set.
3. **The derived class has its own `__init__()` and it calls `super().__init__()`.** The object is set up partly by the child's `__init__()` and partly by the parent's.

In Example 4, `BasicPet`, `Retriever` and `Cat` are case 1, and `Dog` is case 3.

## <a name="8-9-multiple-inheritance"></a>8.9 Multiple inheritance

**Multiple inheritance** means a class inherits from more than one base class. The parent classes are listed inside the brackets, separated by commas.

[Back to the Table of Contents](#toc)

### <a name="example-5-the-syntax-of-multiple-inheritance"></a>Example 5: The syntax of multiple inheritance

**What the code does.** The book shows the line `class DerivedClass(Base1, Base2, Base3):`. On its own, that line cannot run, because `Base1`, `Base2` and `Base3` do not exist yet. So this cell first creates three tiny base classes, then uses the book's line, and finally prints the MRO.

**Learning points**

1. The order of the parents in the brackets is the order Python searches them: `Base1` first, then `Base2`, then `Base3`.
2. `DerivedClass` gets `hello()` from `Base1`, `bye()` from `Base2` and `wave()` from `Base3`.

````python
# Three small base classes, so that the book's line can run
class Base1:
    def hello(self):
        print("hello from Base1")

class Base2:
    def bye(self):
        print("bye from Base2")

class Base3:
    def wave(self):
        print("wave from Base3")

class DerivedClass(Base1, Base2, Base3):  # the syntax from the book
    pass

d = DerivedClass()
d.hello()
d.bye()
d.wave()
print(DerivedClass.__mro__)
````

[Back to the Table of Contents](#toc)

### <a name="the-problem-with-multiple-inheritance-and-the-c3-rule"></a>The problem with multiple inheritance, and the C3 rule

What if the same method exists in more than one parent? Python must pick one. It uses the MRO to decide.

Python 3 builds the MRO with an algorithm called **C3 linearization**. Old-style classes in Python 2 used a simple "depth-first, left-to-right" search. That simple search goes wrong for the **diamond problem**, where two parents share the same grandparent:

![The diamond shape: two parents that share one grandparent](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-07-diamond.png)

A pure depth-first search from `Dog1` would visit `Walker`, then `Pet` (the grandparent), and only then `Swimmer`. That is not sensible, because `Swimmer` is closer to `Dog1` than `Pet` is. C3 follows two rules:

1. A child is always checked **before** its parents.
2. Parents are checked **in the order they are listed** in the brackets.

So for `class DerivedClass(Base1, Base2, Base3)`, Python checks the class itself, then `Base1`, `Base2`, `Base3`, and then goes on to their parents. A shared grandparent is checked only after all the classes that inherit from it.

## <a name="8-10-example-script-for-multiple-inheritance"></a>8.10 Example script for multiple inheritance

In this example we see how Python chooses a method when two parents both provide it.

[Back to the Table of Contents](#toc)

### <a name="example-6-walker-swimmer-and-the-order-of-the-parents"></a>Example 6: Walker, Swimmer, and the order of the parents

**What the code does.**

1. **Base class.** `Pet` has a method `info()` that every child inherits.
2. **Two parent classes.** `Walker(Pet)` and `Swimmer(Pet)` each define their own `action()`. `Walker.action()` prints "Walking..." and `Swimmer.action()` prints "Swimming...".
3. **Two child classes with multiple inheritance.** `Dog1(Walker, Swimmer)` and `Dog2(Swimmer, Walker)`. They have the same two parents, but in a different order. Neither child writes its own `action()`.
4. **The test.** It calls `action()` on a `Dog1` object and on a `Dog2` object, and prints both MROs.

**How it works.** For `d1.action()`, Python searches the MRO of `Dog1`: `Dog1`, `Walker`, `Swimmer`, `Pet`, `object`. It finds `action()` in `Walker` first and stops there. For `Dog2`, the MRO starts `Dog2`, `Swimmer`, so the `Swimmer` version runs.

**Why it matters.** With multiple inheritance, **the order of the parent classes changes the behaviour**. Swapping two names in the brackets makes the same call do something different.

**Learning points**

1. The order of the parent classes matters.
2. Python follows the MRO to find a method.
3. The first match in the MRO is the one that runs. The search then stops.
4. Methods of the base class (`info()`) are inherited by all.

![The MRO of Dog1 and Dog2 depends on the order of the parent classes](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-06-multiple-mro.png)

````python
# A. BASE CLASS
class Pet:  # Base class for all pets
    def info(self):  # Common method for all pets
        print("I am a Pet")

# B. FIRST PARENT CLASS
class Walker(Pet):  # Walker inherits from Pet
    def action(self): # Override Pet's action() by implementing for Walker
        print("Walking... (Walker class)")

# C. SECOND PARENT CLASS
class Swimmer(Pet): # Swimmer inherits from Pet
    def action(self): # Override Pet's action() by implementing for Swimmer
        print("Swimming... (Swimmer class)")

# D. CHILD CLASS 1 (Walker first, then Swimmer). This will affect the MRO
class Dog1(Walker, Swimmer):  # Doesnt override action()
    pass  # So use the method from first parent in MRO (Walker)

# E. CHILD CLASS 2 (Swimmer first, then Walker). This will change the MRO
class Dog2(Swimmer, Walker):  #  Doesnt override action()
    pass  # So use the method from first parent in MRO (Swimmer)

# F. OBJECT CREATION
d1 = Dog1()  # Create Dog1 object
d2 = Dog2()  # Create Dog2 object

# G. METHOD CALLS
d1.action() # For Dog1 Walker comes first so Walker's action() will be used
# Output: Walking... (Walker class)

d2.action() # For Dog2 Swimmer comes first so Swimmer's action() will be used
# Output: Swimming... (Swimmer class)

# H. SHOW MRO
print("MRO of Dog1:", Dog1.__mro__)
# MRO for Dog1 is Dog1 → Walker → Swimmer → Pet → object
print("MRO of Dog2:", Dog2.__mro__)
# MRO for Dog2 is Dog2 → Swimmer → Walker → Pet → object
````

## <a name="8-11-composition-the-has-a-relationship"></a>8.11 Composition: the "HAS-A" relationship

**Composition** means that one class holds an object of another class as one of its attributes. The outer object does not **become** that other thing. It **owns** or **uses** it.

- **Rule of thumb:** if you can say "Class A **has a** Class B", use composition.
- **Purpose:** combine different pieces of behaviour without forcing them into a rigid family tree.
- **Examples:** a Dog has a Collar. A Car has an Engine.

[Back to the Table of Contents](#toc)

### <a name="example-7-a-dog-that-has-a-collar"></a>Example 7: A Dog that has a Collar

**What the code does.** It creates a small class `Collar` with one attribute, `color`. The class `Dog` does **not** inherit from `Collar`. Instead, `Dog.__init__()` makes a new `Collar("Red")` and stores it in `self.collar`.

**How it works.** `d.collar` is a full `Collar` object. So `d.collar.color` means: take the dog's collar, then take that collar's colour. Each dot moves one step into the object.

**Why it matters.** "A Dog is a Collar" is nonsense, so inheritance would be the wrong tool. Composition keeps the two classes separate. You can change `Collar` without touching the family tree of `Dog`.

**Learning points**

1. Composition is built with an ordinary assignment inside `__init__()`: `self.collar = Collar("Red")`.
2. `Dog` does not get the methods of `Collar`. You reach them through the attribute: `d.collar.something()`.
3. Every `Dog` gets its own new `Collar` object.

![What happens when d = Dog("Tommy") builds a Collar inside the Dog](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-08-composition.png)

````python

class Collar:  # Component class for composition
    def __init__(self, color):
        self.color = color

class Dog:  # Dog class that has a Collar (composition)
    def __init__(self, name):
        self.name = name
        self.collar = Collar("Red")   # Has-A relationship

d = Dog("Tommy")  # Create Dog object, which creates a Collar object inside it
# Accessing attributes of Dog and its composed Collar
print(d.name, d.collar.color)  # Output: Tommy Red
````

## <a name="8-12-abstract-methods-and-abstract-classes"></a>8.12 Abstract methods and abstract classes

An **abstract method** is a method that is **declared but not implemented**. It works as a placeholder. It tells every child class: "you must write this method yourself".

We need abstract methods when (1) every child class must follow the same rule (for example, "every pet can speak"), but (2) each child does it in its own way (a dog barks, a cat meows). In short, **the parent decides what must exist, and the child decides how it works**.

The simplest way to make an abstract method is to have it raise `NotImplementedError`. This is a built-in exception, made for exactly this purpose.

[Back to the Table of Contents](#toc)

### <a name="example-8-an-abstract-method-made-with-notimplementederror"></a>Example 8: An abstract method made with NotImplementedError

**What the code does.**

1. **Base class `Pet`.** Its `speak()` has no real code. It only raises `NotImplementedError("Subclass must implement speak()")`.
2. **Child class `Dog`.** It writes its own `speak()`. This is called a **concrete** implementation. So `Dog` objects can safely call `speak()`.
3. **Child class `Cat`.** It does **not** write `speak()`. It inherits the version from `Pet`, which raises the error.
4. **Test.** A `Dog` is created and `speak()` works. A `Cat` is also created with no problem. The call `c.speak()` is commented out, because it would stop the program.

**The logic in four steps**

1. The parent defines the method, but it raises an error instead of doing work.
2. The child inherits the method.
3. If the child **overrides** the method with real code, calls succeed.
4. If the child forgets to override it, calling the method crashes with `NotImplementedError`.

**Learning points**

1. **The parent defines the structure.** `Pet` says that `speak()` must exist.
2. **The child must implement it.** `Dog` does, so it works. `Cat` does not, so it fails.
3. **The error happens late, at run time.** Python happily lets you create a `Cat`. The error appears only when `speak()` is actually called.

````python
# A. BASE CLASS
class Pet:  # Base class for all pets
    def speak(self):  # Abstract method
        raise NotImplementedError("Subclass must implement speak()")

# B. CHILD CLASS 1
class Dog(Pet):  # Dog inherits from Pet
    def speak(self):  # Concrete method
        print("Dog barks")

# C. CHILD CLASS 2
class Cat(Pet):  # Cat inherits from Pet
    pass   # Does NOT implement speak()

# D. TESTING
d = Dog()  # Create Dog object
d.speak()   # Works

c = Cat()  # Create Cat object
# c.speak()   # NotImplementedError: Subclass must implement speak()
````

````python
try:
    c.speak()
except NotImplementedError as e:
    print("NotImplementedError:", e)
````

A class is called **abstract** if it contains one or more abstract methods. It is meant to be a pattern for other classes, not something you make objects from.

## <a name="8-13-the-abc-module-the-new-way-of-making-abstract-classes"></a>8.13 The abc module (the "new way" of making abstract classes)

`abc` stands for **Abstract Base Classes**. This module gives Python a proper way to define "blueprints" that other classes must follow.

- **Main goal:** make sure a subclass provides a given interface, that is, a list of required methods.
- **Early failure:** with `NotImplementedError`, the program fails only when the missing method is called. With `abc`, it fails **the moment you try to create the object** (`c = Cat()`).
- **Why use it?**
    1. **Enforcement:** every object in your program really has the methods it claims to have.
    2. **Clarity:** other programmers can see at a glance which methods they must write.

[Back to the Table of Contents](#toc)

### <a name="8-13-1-abstractmethod"></a>8.13.1 @abstractmethod

`@abstractmethod` is a **decorator**. As you saw in Chapter 7, a decorator changes the function written just below it.

| Part | What it does |
|---|---|
| **The flag** | `@abstractmethod` sets a hidden attribute on the function: `__isabstractmethod__ = True`. |
| **The guard** | The class `ABC` acts like a security guard. When you try to create an object, it checks the class and all its parents. If any method still carries the flag and has not been overridden with real code, the object is not created and you get a `TypeError`. |
| **`pass` in the body** | The body of an abstract method is usually just `pass`, because that code never runs. The method is there only to fix its name and parameters. |

The usual pattern:

```python
from abc import ABC, abstractmethod

class SomeBase(ABC):          # 1. inherit from ABC
    @abstractmethod           # 2. mark each required method
    def required_method(self):
        pass
```

[Back to the Table of Contents](#toc)

### <a name="example-9-an-abstract-base-class-with-abstractmethod"></a>Example 9: An abstract base class with @abstractmethod

**What the code does.** It imports `ABC` and `abstractmethod` from the `abc` module. `Pet` inherits from `ABC` and marks `speak()` as abstract. `Dog` writes its own `speak()`, so `Dog()` works. `Cat` does not, so `Cat()` would fail. That line is left commented out.

**How it works.** When `Dog()` is called, the guard finds that `Dog.speak` is a normal method, so the object is made. For `Cat()`, the guard finds that `speak` is still the abstract one from `Pet`, so it refuses.

**Learning points**

1. A class with abstract methods must inherit from `ABC`, otherwise `@abstractmethod` has no effect.
2. You cannot create an object from `Pet` itself either. `Pet()` also raises `TypeError`.
3. The error now comes **early**, when the object is created, not later when a method is called.

![When the error appears: NotImplementedError compared with the abc module](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-09-abstract-timing.png)

````python
from abc import ABC, abstractmethod

class Pet(ABC):  # Abstract Base Class
    @abstractmethod
    def speak(self):
        pass

class Dog(Pet):
    def speak(self):
        print("Dog barks")

class Cat(Pet):  # Cat inherits from Pet
    pass   # ERROR at object creation

d = Dog()
d.speak()  # Dog barks

# c = Cat()   # TypeError Cannot create an object of Cat class
# because it does not implement the abstract method speak() from Pet.
# Advantage → early detection of error at object creation time,
# Unlike NotImplementedError → Late detection at runtime when method is called
````

Raising `NotImplementedError` is an **informal** way to make an abstract method. Using the `abc` module is the **formal** way.

## <a name="8-14-real-subclasses-and-virtual-subclasses"></a>8.14 Real subclasses and virtual subclasses

A class can be related to another class in two ways.

**1. Real subclass (normal inheritance)**

- A class inherits directly from another, with brackets: `class Dog(Pet)`.
- The child knows its parent, but the parent does not keep a list of its children.
- A real subclass shows up in (1) `__mro__`, (2) `issubclass()`, and (3) `isinstance()`.

**2. Virtual subclass (registration)**

- The class does **not** inherit from the parent.
- Instead, it is **registered** with a parent that is an ABC: `Pet.register(RobotDog)`.
- The child gets **no code** from the parent.
- The link is **logical**, not physical. It only affects type checks.
- This works only with abstract base classes.

[Back to the Table of Contents](#toc)

### <a name="real-and-virtual-subclasses-side-by-side"></a>Real and virtual subclasses side by side

| Feature | Real subclass | Virtual subclass |
|---|---|---|
| Inheritance | Yes | No |
| Code reuse | Yes | No |
| Appears in the MRO | Yes | No |
| `issubclass()` | `True` | `True` |
| `isinstance()` | `True` | `True` |
| Needs an ABC | No | Yes |
| Kind of link | Physical | Logical |

**Why use virtual subclasses?** (1) To define a common interface. (2) To let classes that are not related still count as a particular type. This is useful in frameworks, plug-ins and APIs, where you want to accept a class written by someone else without changing its code.

[Back to the Table of Contents](#toc)

### <a name="example-10-a-real-subclass-and-a-virtual-subclass"></a>Example 10: A real subclass and a virtual subclass

**What the code does.**

1. **Abstract base class.** `Pet(ABC)` is empty. It is used only as a common type.
2. **Real subclass.** `Dog(Pet)` inherits from `Pet`.
3. **Virtual subclass.** `RobotDog` has no parent class. It is then registered with `Pet.register(RobotDog)`.
4. **Type checks.** `issubclass()` checks class against class, and `isinstance()` checks object against class. All four checks return `True`.
5. **MRO.** `Dog.__mro__` contains `Pet`. `RobotDog.__mro__` would not, because there is no real inheritance.

**Learning points**

1. After `register()`, Python treats `RobotDog` as a kind of `Pet` for type checks.
2. `RobotDog` gets nothing from `Pet`. If `Pet` had a method, a `RobotDog` object could not use it.
3. `register()` does not check anything. Python trusts you that `RobotDog` really provides what `Pet` promises.

![A real subclass (Dog) and a virtual subclass (RobotDog) of Pet](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-10-virtual-subclass.png)

````python
from abc import ABC

# A. BASE CLASS (ABC)
class Pet(ABC):  # Abstract Base Class
    pass

# B. REAL SUBCLASS
class Dog(Pet):  # Dog inherits from Pet
    def speak(self):
        print("Dog barks")

# C. VIRTUAL SUBCLASS. Does NOT inherit from Pet
class RobotDog:  # But will be registered as virtual subclass
    def speak(self):
        print("Robot dog beeps")

# D. REGISTER RobotDog as virtual subclass of Pet without actual inheritance
Pet.register(RobotDog)
# RobotDog → is subclass of Pet for isinstance() and issubclass() checks
# But doesnt inherit from Pet so doesnt have access to methods/ attributes of Pet

# E. Create Objects
d = Dog()
r = RobotDog()

# F. Real subclass checks
print(issubclass(Dog, Pet))       # True. Dog is a real subclass of Pet.
print(isinstance(d, Pet))         # True. Dog is an instance of Pet.

# G. Virtual subclass checks
print(issubclass(RobotDog, Pet))  # True. RobotDog is a virtual subclass of Pet.
print(isinstance(r, Pet))         # True. RobotDog is an instance of Pet.

# H. MRO Check
print(Dog.__mro__)        # Shows inheritance
# print(RobotDog.__mro__) # Does NOT include Pet
````

## <a name="8-15-registration"></a>8.15 Registration

Registration is not about inheritance. It is about **grouping, classifying and finding** classes.

- Inheritance says **what you are**.
- Registration says **how you are treated**.

[Back to the Table of Contents](#toc)

### <a name="why-register-classes"></a>Why register classes?

1. **To group unrelated classes.** Two classes may share no code but behave in a similar way. For example, `Dog` and `RobotDog` below do not inherit from each other, yet both can be treated as "pets".
2. **Dynamic discovery.** Instead of a long `if ... elif` chain that picks a class by name, the program keeps the classes in a dictionary and looks them up: `registry[name]()`.
3. **Plug-in systems.** Many real applications (for example, Django apps, Flask extensions and payment gateways) do not know all their classes in advance. The classes register themselves.
4. **Loose coupling.** Without registration, the main program must know every class. With registration, classes add themselves and the main program stays clean.

[Back to the Table of Contents](#toc)

### <a name="example-11-grouping-classes-and-finding-them-with-a-registry"></a>Example 11: Grouping classes, and finding them with a registry

**What the code does.** Part A is the book's pair of unrelated classes, `Dog` and `RobotDog`. Parts B and C are not in the book. They turn the book's two short fragments (the `if ... elif` chain and `registry[name]()`) into code that runs, so you can compare them.

**How it works.** `registry` is an ordinary dictionary. Its keys are names and its values are **classes**, not objects. `registry[name]` gives back a class, and the `()` after it creates an object of that class.

**Learning points**

1. A class is an object too. It can be stored in a dictionary like any other value.
2. To add a new kind of pet, you add one entry to the dictionary. The code that uses it does not change.

````python
# A. From the book: two classes with no inheritance link
class Dog:
    def speak(self): print("Dog barks")
class RobotDog:
    def speak(self): print("Robot beeps")

# B. Without a registry: an if ... elif chain
def make_pet_v1(name):
    if name == "dog":
        return Dog()
    elif name == "robot":
        return RobotDog()

# C. With a registry: a dictionary from names to classes
registry = {"dog": Dog, "robot": RobotDog}

def make_pet_v2(name):
    return registry[name]()

make_pet_v1("robot").speak()
make_pet_v2("dog").speak()
````

## <a name="8-16-containers-in-python"></a>8.16 Containers in Python

A **container** is an object that holds several values.

| Type | Description | Examples | Key feature |
|---|---|---|---|
| Sequence | Stores items in order | `list`, `tuple`, `str` | Indexed by whole numbers (0, 1, 2 ...) |
| Mapping | Stores key-value pairs | `dict` | Indexed by keys |

You can make **your own** container class. You do it by writing special methods, also called **magic methods** or **dunder methods** (short for "double underscore").

| Method | Purpose | Called by |
|---|---|---|
| `__getitem__(self, key)` | Read a value | `obj[key]` |
| `__setitem__(self, key, value)` | Store a value | `obj[key] = value` |

If a class defines:

- `__getitem__()`, its objects can be **read** with square brackets, like a list;
- `__setitem__()`, its objects can be **written** with square brackets.

[Back to the Table of Contents](#toc)

### <a name="example-12-a-custom-container-petstore"></a>Example 12: A custom container, PetStore

**What the code does.** `PetStore` keeps pet names in an internal list, `self.pets`, which starts as `[None, None, None]`. Because it defines `__setitem__()` and `__getitem__()`, you can use square brackets on a `PetStore` object, just as you would on a list. Each method prints a message, so you can see when Python calls it.

**How it works.** Python turns the bracket syntax into method calls:

| You write | Python actually runs |
|---|---|
| `store[0] = "Dog"` | `store.__setitem__(0, "Dog")` |
| `store[0]` | `store.__getitem__(0)` |

**Why it matters.** This is how `list`, `dict` and many library types work inside. It lets your own classes feel like built-in types.

**Learning points**

1. You never call `__setitem__()` or `__getitem__()` yourself. Python calls them when it sees the brackets.
2. In the last three lines, "Getting pet at index ..." is printed **before** "Pet at ...". Python must work out `store[0]` (which prints its message) before `print()` can show the result.
3. `store[2]` was never set, so it is still `None`.

![How square brackets on a PetStore object call __setitem__ or __getitem__](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-11-getitem-setitem.png)

````python
# A. CUSTOM CONTAINER
class PetStore:
    def __init__(self, size):
        # Internal storage (list)
        self.pets = [None] * size
        print("PetStore created with size:", size)

    # Called when we do: store[index] = value
    # index is the position in the list, pet_name is the value being assigned
    def __setitem__(self, index, pet_name):
        print(f"Setting pet at index {index} to '{pet_name}'")
        self.pets[index] = pet_name

    # Called when we do: value = store[index]
    # index is the position in the list
    def __getitem__(self, index):
        print(f"Getting pet at index {index}")
        return self.pets[index]

# B. Create a PetStore and test setting and getting items
store = PetStore(3)  # Create a PetStore with 3 slots

# C. Setting values
store[0] = "Dog"  # Calls __setitem__ to set index 0 to "Dog"
store[1] = "Cat"  # Calls __setitem__ to set index 1 to "Cat"
# store[2] is not set, so it remains None

# D. Getting values
print("Pet at 0:", store[0]) # Calls __getitem__ to get index 0
print("Pet at 1:", store[1]) # Calls __getitem__ to get index 1
print("Pet at 2:", store[2]) # Calls __getitem__ to get index 2 (None, not set)
````

## <a name="8-17-new-compared-with-init"></a>8.17 `__new__()` compared with `__init__()`

Python creates an object in two steps:

| Step | Method | Job |
|---|---|---|
| 1 | `__new__(cls)` | **Creates** the object and returns it |
| 2 | `__init__(self)` | **Initialises** the object, that is, fills in its attributes |

The order is: (1) you call `ClassName()`, (2) Python calls `__new__(cls)`, (3) `__new__` returns the new object, (4) Python calls `__init__(self)` on it.

Note the first parameter. `__new__` receives the **class** (`cls`), because the object does not exist yet. `__init__` receives the **object** (`self`).

![The two steps of object creation: __new__() then __init__()](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-12-new-init.png)

[Back to the Table of Contents](#toc)

### <a name="try-it-watching-the-two-steps"></a>Try it: watching the two steps

This cell is not in the book. Both methods print a message so you can see the order. `__new__` passes the real work to `object.__new__` with `super()`.

````python
class Pet:
    def __new__(cls, name):
        print("1. __new__ is called: creating the object")
        obj = super().__new__(cls)   # object.__new__ does the real work
        return obj

    def __init__(self, name):
        print("2. __init__ is called: setting the attributes")
        self.name = name

p = Pet("Tommy")
print("3. Ready:", p.name)
````

## <a name="8-18-metaclasses-in-python"></a>8.18 Metaclasses in Python

A **metaclass** is a class that **creates other classes**:

1. a metaclass creates a class, and
2. a class creates objects.

By default, Python uses `type` as the metaclass. `type` can be used in two ways:

## <a name="8-19-namespaces-in-python"></a>8.19 Namespaces in Python

A **namespace** is a container that maps **names** to **objects**. Think of it as a dictionary: the keys are the names you create, and the values are the objects in memory. Variables are names; objects are values.

Three key ideas:

1. **Unique identity:** every object that exists has its own `id()`.
2. **Mapping:** the namespace links your variable name to that object.
3. **Independence:** different namespaces can use the same name for different objects without any clash. A `name` inside one function has nothing to do with a `name` inside another.

So if three names `var_name1`, `var_name2` and `var_name3` refer to three objects, the namespace behaves like this dictionary:

```python
namespace = {
    "var_name1": object1,
    "var_name2": object2,
    "var_name3": object3
}
```

Most user-defined classes and their objects keep their namespace in an attribute called `__dict__`.

[Back to the Table of Contents](#toc)

### <a name="the-four-namespaces-legb"></a>The four namespaces: LEGB

Python has four namespaces, nested one inside another:

| Letter | Namespace | Where the names come from |
|---|---|---|
| L | **Local** | Inside the current function (innermost) |
| E | **Enclosing** | Inside an outer function, when functions are nested |
| G | **Global** | The top level of the file (module) |
| B | **Built-in** | Names Python provides, such as `print` and `len` (outermost) |

**Visibility rules:** (1) an inner scope can **see** the names of the outer scopes; (2) an outer scope **cannot see** the names of an inner scope.

![The LEGB order Python uses to find a name](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-08-nb-14-legb.png)

[Back to the Table of Contents](#toc)

### <a name="a-summary-of-the-four-namespaces"></a>A summary of the four namespaces

| Question | Answer |
|---|---|
| Where is the variable created? | Local: inside a function. Enclosing: inside an outer function. Global: at file level. Built-in: provided by Python. |
| When is the namespace created? | Local: when the function is called. Enclosing: when the outer function runs. Global: when the program (module) starts. Built-in: when the interpreter starts. |
| How long does it live? | Local: until the function ends. Enclosing: while the outer function runs (longer if a closure keeps it). Global: for the whole program. Built-in: always. |
| Search order | Local, then Enclosing, then Global, then Built-in. The first match wins. |
| How can you see it? | Local: `locals()`. Global: `globals()`. Built-in: the `builtins` module. |
| What can an inner function reach? | Its own locals, the enclosing names, the globals and the built-ins. This is why closures work. |
| Can it be changed directly? | Local: yes. Enclosing: only with `nonlocal`. Global: only with `global`. Built-in: it should not be changed. |
| Shadowing | The same name in an inner scope hides the outer one, but does not change it. |
| Common mistakes | Forgetting `global` or `nonlocal`; reusing a built-in name such as `list` or `sum` for your own variable. |
| Name not found anywhere | `NameError` |

**Learning points**

## <a name="8-20-keywords-global-and-nonlocal"></a>8.20 Keywords: global and nonlocal

These two keywords control **which variable an assignment changes**.

| Keyword | Purpose | Without it |
|---|---|---|
| `global` | Change a **global** variable from inside a function | The assignment creates a new local variable |
| `nonlocal` | Change a variable of the **enclosing** (outer) function from inside a nested function | The assignment creates a new local variable |

`nonlocal` works only inside nested functions.

Note that these keywords are needed only for **assignment**. Just **reading** an outer variable works without them, thanks to LEGB.

### <a name="example-14-the-effect-of-using-and-not-using-global-and-nonl"></a>Example 14: The effect of using and not using global and nonlocal

**What the code does.** It has two parts, and each part changes two variables: one **with** the keyword ("correct") and one **without** it ("wrong").

- **Part 1, `global`.** `test_global()` assigns to `pet_global_wrong` without `global`, and to `pet_global_correct` after `global pet_global_correct`. After the call, the program prints both global variables.
- **Part 2, `nonlocal`.** `outer_function()` has two local variables. The nested `inner_function()` assigns to `pet_enclosing_wrong` without `nonlocal`, and to `pet_enclosing_correct` after `nonlocal pet_enclosing_correct`. Back in the outer function, both are printed.

**How it works.** Inside a function, Python treats a name as **local** if the function assigns to it anywhere. `global` and `nonlocal` switch this off for the names they list.

**Learning points**

1. Inside the functions, both "wrong" and "correct" variables show the new text. The difference only appears **outside**.
2. After the calls, only the "correct" variables have changed. The "wrong" assignments made new local variables that disappeared when the function ended.
3. `global` reaches the top level of the file. `nonlocal` reaches only the nearest enclosing function.

````python
# A. GLOBAL VARIABLES
pet_global_correct = "Global Dog (Original)"  # In global scope.
pet_global_wrong = "Global Cat (Original)"  # In global scope.

# B. FUNCTION: GLOBAL KEYWORD DEMO. Here pet_global_wrong is modified without
# using 'global', so it will not change the global variable. pet_global_correct
# is modified using 'global', so it will change the global variable.
def test_global():
    print("--- Inside test_global() ---")

    # WRONG: No 'global' keyword → creates LOCAL variable
    pet_global_wrong = "Modified without global (Wrong)"
    print("Inside function (wrong global):", pet_global_wrong)
    # Output: Inside function (wrong global): Modified without global (Wrong)

    # CORRECT: Using 'global' → modifies actual global variable
    global pet_global_correct
    pet_global_correct = "Modified with global (Correct)"
    print("Inside function (correct global):", pet_global_correct)
    # Output: Inside function (correct global): Modified with global (Correct)

# C. CALL FUNCTION
test_global()

# D. Check results outside function
print("--- After test_global() ---")
print("Global WRONG variable:", pet_global_wrong)     # Not changed
print("Global CORRECT variable:", pet_global_correct) # Changed

# E. FUNCTION: NONLOCAL KEYWORD DEMO. Here pet_enclosing_wrong is modified without
# using 'nonlocal', so it will not change enclosing variable. pet_enclosing_correct
# is modified using 'nonlocal' so it will change enclosing variable
def outer_function():
    print("--- Inside outer_function() ---")
    pet_enclosing_correct = "Enclosing Dog (Original)"
    pet_enclosing_wrong = "Enclosing Cat (Original)"

    def inner_function():
        print("--- Inside inner_function() ---")

        # WRONG: No 'nonlocal' → creates local variable
        pet_enclosing_wrong = "Modified without nonlocal (Wrong)"
        print("Inner (wrong nonlocal):", pet_enclosing_wrong)

        # CORRECT: Using 'nonlocal' → modifies enclosing variable
        nonlocal pet_enclosing_correct
        pet_enclosing_correct = "Modified with nonlocal (Correct)"
        print("Inner (correct nonlocal):", pet_enclosing_correct)

    # F. Call inner function. This will execute the code inside inner_function
    # and show effects of modifying variables with & without 'nonlocal' keyword
    inner_function()

    # G. Check results in enclosing scope
    print("--- Back in outer_function() ---")
    print("Enclosing WRONG variable:", pet_enclosing_wrong)     # Not changed
    print("Enclosing CORRECT variable:", pet_enclosing_correct) # Changed

# H. Call outer function
outer_function()
````

### <a name="automatic-registration-with-init-subclass"></a>Automatic registration with `__init_subclass__()`

Python calls the special method `__init_subclass__()` of a base class **every time a new subclass is created**. The base class can use it to keep a registry, with nothing to do by hand. This is approach C from section 8.15.

````python
class Pet:
    registry = {}

    def __init_subclass__(cls, **kwargs):
        super().__init_subclass__(**kwargs)
        Pet.registry[cls.__name__.lower()] = cls   # runs once per new subclass

class Dog(Pet):
    def speak(self): return "Woof"

class Cat(Pet):
    def speak(self): return "Meow"

print(Pet.registry)
print(Pet.registry["cat"]().speak())
````

### <a name="a-read-only-attribute-with-property"></a>A read-only attribute with `@property`

`@property` turns a method into something you read like an attribute, with no brackets. It is a clean way to compute a value on demand, or to protect an attribute from being changed.

````python
class Dog:
    def __init__(self, name, birth_year):
        self.name = name
        self._birth_year = birth_year

    @property
    def age(self):                      # read as d.age, not d.age()
        return 2026 - self._birth_year

d = Dog("Tommy", 2021)
print(d.name, "is", d.age)
try:
    d.age = 10                          # no setter was written, so this fails
except AttributeError as e:
    print("AttributeError:", e)
````

### <a name="saving-memory-with-slots"></a>Saving memory with `__slots__`

Normally every object has its own `__dict__`. If a class lists its attributes in `__slots__`, Python stores them in a fixed, smaller space instead, and no `__dict__` is made. You then cannot add new attributes to the object. This is useful when a program creates millions of small objects.

````python
class Point:
    __slots__ = ("x", "y")
    def __init__(self, x, y):
        self.x, self.y = x, y

pt = Point(2, 3)
print(pt.x, pt.y)
print(hasattr(pt, "__dict__"))          # False: there is no __dict__
try:
    pt.z = 5
except AttributeError as e:
    print("AttributeError:", e)
````

### <a name="cooperative-super-in-a-diamond"></a>Cooperative super() in a diamond

In a diamond, `super()` follows the MRO of the **object**, not the parent named in the class. So each class in the diamond runs exactly once, and the shared grandparent runs last.

````python
class Pet:
    def __init__(self):
        print("Pet")

class Walker(Pet):
    def __init__(self):
        print("Walker")
        super().__init__()

class Swimmer(Pet):
    def __init__(self):
        print("Swimmer")
        super().__init__()     # for a Dog1 object this goes to Pet

class Dog1(Walker, Swimmer):
    def __init__(self):
        print("Dog1")
        super().__init__()

Dog1()
print([c.__name__ for c in Dog1.__mro__])
````

Notice that `super()` inside `Walker` calls `Swimmer`, not `Pet`, because `Swimmer` comes next in the MRO of `Dog1`. This is the meaning of "super() calls the next class in the MRO".
