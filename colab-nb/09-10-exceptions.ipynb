## <a name="9-1-errors-and-exceptions"></a>9.1 Errors and exceptions

An **error** is a problem that stops a program from working correctly. Errors come in two broad kinds.

**Syntax errors (parsing errors).** Python cannot understand the code because it breaks the rules of the language. Python finds these while reading the file, **before** anything runs, so not a single line is executed. Common causes: (1) typing mistakes, (2) missing symbols such as `:` or `)`, (3) wrong indentation.

**Exceptions (run-time errors).** The code is valid, but something goes wrong **while it runs**, for example dividing by zero or reading past the end of a list. An exception interrupts the normal flow. If you handle it with `try-except`, the program can carry on.

[Back to the Table of Contents](#toc)

### <a name="syntax-error-compared-with-exception"></a>Syntax error compared with exception

| Feature | Syntax error | Exception |
|---|---|---|
| When it occurs | Before execution, while Python reads the code | During execution |
| Cause | Code that breaks the rules of the language | A problem in the logic or the data at run time |
| Effect on the program | The program does not start at all | The program stops, unless the exception is handled |
| Example | A missing `:` after `if x > 3` | `[1, 2][5]` gives `IndexError` |
| Can it be handled? | No | Yes, with `try-except` |

## <a name="9-2-the-raise-statement"></a>9.2 The raise statement

So far, Python raised exceptions for us. With the `raise` statement **you** can raise an exception on purpose, when your own rule is broken. The syntax is:

```python
raise ExceptionName("message")
```

The message tells the reader what went wrong. You can raise any built-in exception, such as `ValueError`, or one you wrote yourself (see section 9.12).

[Back to the Table of Contents](#toc)

### <a name="example-1-raising-a-valueerror-for-a-negative-age"></a>Example 1: Raising a ValueError for a negative age

**What the code does.** The class `Pet` checks the `age` it is given. If the age is negative, `__init__()` raises a `ValueError` instead of storing the bad value. The last line tries to create a pet with age `-2`.

**How it works.** `raise` stops `__init__()` at once. The lines that set `self.name` and `self.age` never run, and no `Pet` object is made. There is no `try-except` here, so the exception reaches the top level and Python prints a traceback.

**Why it matters.** It is better to refuse bad data early, with a clear message, than to store it and get strange results later.

**Learning points**

1. `raise` is how your own code reports that a rule has been broken.
2. `ValueError` is the right built-in exception when the type is correct (an integer) but the value is not acceptable.
3. The red traceback below the cell is expected. Its last line reads `ValueError: Age cannot be negative`.

![What happens when Pet("Dog", -2) is created](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-09-nb-02-raise-flow.png)

````python
class Pet:
    def __init__(self, name, age):
        if age < 0:
            raise ValueError("Age cannot be negative")
        self.name = name
        self.age = age

# Creating a pet with invalid age
p1 = Pet("Dog", -2)  # Will raise a ValueError because age cannot be negative
````

## <a name="9-3-the-try-except-else-block"></a>9.3 The try-except-else block

Some code is risky: it may fail because of wrong user input, a missing file, a bad list index, and so on. If such a failure is not handled, the program stops abruptly. `try-except` lets you handle the failure gracefully.

The basic structure is:

```python
try:
    # Risky code
except SomeException:
    # Code to handle error
else:
    # Runs only if no exception occurs
```

The flow of execution:

1. Run the `try` block.
2. Did an error occur?
    - **Yes:** Python looks for an `except` block that matches the exception. If one matches, it runs that block.
    - **No:** Python runs the `else` block.
3. The program carries on after the whole `try` statement.

**The role of `else`:** (1) it runs only if **no** exception occurred, and (2) it keeps the normal logic apart from the error-handling logic.

If an exception occurs and **no** `except` block matches it, the exception is not handled. The same exception travels on to the code that called this code. If nothing handles it anywhere, Python prints a traceback and the program stops.

## <a name="9-4-generic-exception-except-without-a-specific-type"></a>9.4 Generic exception (except without a specific type)

A generic `except` block catches exceptions without naming a particular type. There are two forms:

| Form | What it catches |
|---|---|
| `except Exception as e:` | Every normal error (everything derived from `Exception`). `e` holds the exception object. |
| `except:` (a "bare" except) | Absolutely everything, including `KeyboardInterrupt` (pressing Ctrl+C) and `SystemExit`. This is rarely what you want. |

Its main features: (1) it catches any error; (2) it acts as a safety net; (3) it is usually written **after** the specific `except` blocks; (4) it helps prevent a crash.

[Back to the Table of Contents](#toc)

### <a name="advantages-and-disadvantages-of-a-generic-catch-all-except"></a>Advantages and disadvantages of a generic ("catch-all") except

| | Points |
|---|---|
| **Advantages** | Handles errors you did not expect. Prevents a crash, so the program can continue gracefully. Simple to write, with no need to list exception types. Useful as a final safety layer or fallback. |
| **Disadvantages** | Hides the real problem, so it is hard to know which error occurred. Makes debugging and tracking errors difficult. May catch errors you did not intend to catch, and so hide bugs. Makes the code less clear, because the handling is not explicit. |

[Back to the Table of Contents](#toc)

### <a name="good-practice-with-generic-exceptions"></a>Good practice with generic exceptions

- Always write the **specific** `except` blocks first and the generic one last. The generic block is a last safety net, not your main way of handling errors.
- Good places for a generic block: (a) logging systems, (b) programs used directly by people, where a crash would be unfriendly, (c) a final fallback at the top level of a program.
- Do **not** use a generic block when (1) you know which exception to expect, or (2) the system is critical and precise handling matters.

**The `as` keyword.** In `except SomeError as e:`, the name `e` is a variable that refers to the exception object. You can then look inside it, for example `print(e)`. Section 9.13 covers this in detail.

## <a name="9-5-try-except-else-finally-without-a-generic-exception"></a>9.5 try-except-else-finally without a generic exception

A `finally` block runs **always**: after the `try` block if all went well, after the `except` block if an error was handled, and even when an error is **not** handled.

If there is no generic `except` block, only the listed exceptions are handled. When an exception occurs:

1. Python checks whether it matches one of the `except` blocks.
2. If a match is found, that `except` block runs, then `finally`.
3. If no match is found:
    - the `finally` block runs first,
    - then the exception carries on, Python prints its traceback,
    - and the program stops.

Points to note: (1) only the listed exceptions are handled; (2) `finally` always runs; (3) other exceptions stop the program; (4) there is no fallback, because there is no generic `except`.

[Back to the Table of Contents](#toc)

### <a name="example-3-try-except-else-finally-with-one-specific-except"></a>Example 3: try-except-else-finally with one specific except

**What the code does.** `process_numbers()` divides `number` by `divisor`. It handles only `ZeroDivisionError`. It is called three times:

| Case | Call | What happens | Blocks that run |
|---|---|---|---|
| 1 | `process_numbers(10, 2)` | No error; the result is 5.0 | try, else, finally |
| 2 | `process_numbers(10, 0)` | `ZeroDivisionError`, which is handled | try, except, finally |
| 3 | `process_numbers("10", 2)` | `TypeError`, which is **not** handled | try, finally, then a traceback |

**How it works.** In case 3, dividing a string by an integer raises `TypeError`. No `except` block matches it. Python still runs `finally`, prints "Finally block always executes", and then the exception stops the cell with a red traceback.

**Why it matters.** It shows the one job `finally` does that nothing else can: it runs even when the program is about to crash. That is why it is used for clean-up.

**Learning points**

1. Compare the output of the three cases with the table above.
2. In case 3, "Finally block always executes" is printed **before** the traceback.
3. The red `TypeError` traceback at the end is expected.

![try-except-else-finally without a generic except block](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-09-nb-04-finally-no-generic.png)

````python
def process_numbers(number, divisor):
    try:
        print("1. Performing division...")
        result = number / divisor # may raise ZeroDivisionError or TypeError

    except ZeroDivisionError:
        print(f"Error: Cannot divide by zero (divisor was {divisor})")

    else:
        print("No exception occurred")
        print("Result:", result)

    finally:
        print("Finally block always executes")

# --------- Test Cases ---------

print("Case 1: No Error")
process_numbers(10, 2)
# Case 1 Output: try → else → finally

print("Case 2: Handled Exception")
process_numbers(10, 0)  # ZeroDivisionError → handled by specific except block
# Case 2 Output: try → except(ZeroDivisionError) → finally

print("Case 3: Unhandled Exception")
process_numbers("10", 2)   # TypeError. Trying to divide a string by integer.
# Raises a TypeError which is not specifically handled in the except blocks,
# so it will remain unhandled and will show the traceback error message.
# The finally block will still execute.
# Case 3 Output: try → finally (No else, no except) → traceback error message for TypeError
````

## <a name="9-6-try-except-else-finally-with-multiple-except-blocks-and-"></a>9.6 try-except-else-finally with multiple except blocks and a generic except

Real programs meet many kinds of errors: invalid input, division errors, file errors and so on. A full `try` statement gives you (1) graceful handling with `except`, (2) clean normal logic with `else`, and (3) guaranteed clean-up with `finally`.

The typical structure (this is a pattern, not code that runs):

```python
try:
    # Risky/ suspicious code
except ExceptionType1:  # Handle error 1
    # Handle error 1
except ExceptionType2:  # Handle error 2
    # Handle error 2
except:
    # Handle all remaining errors (generic)
else:
    # Runs if no error
finally:
    # Always runs
```

[Back to the Table of Contents](#toc)

### <a name="points-to-note-about-the-blocks"></a>Points to note about the blocks

| Block | Points |
|---|---|
| `try` | Holds the risky code. |
| `except` | Three possibilities: a **single** exception (handles one specific error), **multiple** `except` blocks (handle different errors separately), or a **generic** `except` (handles all errors). |
| `else` | Runs only if no exception occurred. Keeps the normal logic separate. |
| `finally` | Always runs. Used to close files and for other clean-up. |
| Rules of execution | Only one `except` block runs. Once an exception occurs, the rest of the `try` block is skipped. `finally` runs even if an error is not handled. A bare `except:` must be the last `except` block. |

[Back to the Table of Contents](#toc)

### <a name="example-4-adding-a-generic-except-as-a-safety-net"></a>Example 4: Adding a generic except as a safety net

**What the code does.** This is Example 3 with one change: a generic block, `except Exception as e:`, is added after `except ZeroDivisionError:`. The third call now divides `10` by `"a"`.

| Case | Call | Blocks that run |
|---|---|---|
| 1 | `process_numbers(10, 2)` | try, else, finally |
| 2 | `process_numbers(10, 0)` | try, except ZeroDivisionError, finally |
| 3 | `process_numbers(10, "a")` | try, generic except, finally |

**How it works.** In case 3, `TypeError` does not match `ZeroDivisionError`, so Python moves on to `except Exception as e:`. `TypeError` is derived from `Exception`, so this block matches. `e` holds the exception, and printing it shows its message.

**Learning points**

1. There is no traceback now. The program handles every case and carries on.
2. The generic block comes **after** the specific one. If it came first, it would catch the `ZeroDivisionError` too, and the specific block would never run.
3. `finally` can use the function's parameters. Here it prints the divisor in every case.

![try-except-else-finally with a generic except block](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-09-nb-05-finally-generic.png)

````python
def process_numbers(number, divisor):
    try:
        print("1. Performing division...")
        result = number / divisor  # may cause TypeError or ZeroDivisionError

    except ZeroDivisionError:
        print( f"divisor = {divisor} Error. Cannot divide by zero.")

    except Exception as e:  # Generic except for other exceptions like TypeError
        print("Generic Error:", e)

    else:  # Runs if no exception occurs
        print("No exception occurred")
        print("Result:", result)

    finally:  # Always runs, use for cleanup
        print(f"Finally block always executes. (divisor was {divisor})")

# Case 1: No Error
print("Case 1: No Error")
process_numbers(10, 2)  # No error → result = 5.0
# Output: try → else → finally

# Case 2: ZeroDivisionError.
print("Case 2: ZeroDivisionError")
process_numbers(10, 0)  # ZeroDivisionError → because divisor is zero
# Output: except(ZeroDivisionError) → finally

# Case 3: TypeError not specifically handled so Generic except will catch it
print("Case 3: TypeError")
process_numbers(10, "a")  # TypeError → because divisor is not a number
# Output: except(generic) → finally

````

## <a name="9-7-context-managers-the-with-statement"></a>9.7 Context managers (the with statement)

A **context manager** is an object that (1) sets up a resource, for example opens a file, and (2) cleans it up automatically afterwards, for example closes the file. It is used with the `with` statement.

The basic syntax:

```python
with open("file.txt", "r") as f:
    data = f.read()
```

**Why use a context manager?** To read a file you open it, read it and then close it. But if an exception happens after opening and before closing, the `close()` line is never reached and the file stays open. The `with` statement closes the file in every case.

[Back to the Table of Contents](#toc)

### <a name="without-and-with-a-context-manager"></a>Without and with a context manager

| Without a context manager | With a context manager |
|---|---|
| `f = open("file.txt")` | `with open("file.txt") as f:` |
| `data = f.read()` | `    data = f.read()` |
| `f.close()`: if an exception occurs before this line, the file stays open | The file is always closed automatically, even after an error |

### <a name="example-6-what-the-with-statement-does-inside"></a>Example 6: What the with statement does inside

This listing from the book shows, in simplified form, what Python does for you when it runs a `with` statement. It is a **sketch**, not a program: `fileName`, `mode`, `exc_type`, `exc_value` and `traceback` are not defined, so it is shown here as text rather than as a code cell.

```python
manager = open(fileName, mode)
# open() returns file object as a context manager
# The with statement automatically calls manager.__enter__() to get the file
# object and assigns it to f. It also ensures that manager.__exit__() is called
# to close the file, even if an exception occurs.
f = manager.__enter__()

try:
    content = f.read()
except:
    # Exception handling (outside)
    raise
finally:
    manager.__exit__(exc_type, exc_value, traceback)
    # exc_type, exc_value, traceback are the exception details if an exception
    # occurred, otherwise they are None.
```

**How to read it.**

| Step | Line | Meaning |
|---|---|---|
| 1 | `manager = open(fileName, mode)` | `open()` returns an object that can act as a context manager. |
| 2 | `f = manager.__enter__()` | The start of the `with` block. For a file, `__enter__()` returns the file itself; this is the `f` after `as`. |
| 3 | the `try` block | The body of your `with` block. |
| 4 | `except: raise` | Any exception is passed on (re-raised) to the code outside. |
| 5 | `finally: manager.__exit__(...)` | Always runs, so the file is always closed. The three arguments describe the exception, or are all `None` if there was none. |

````python
manager = open("file.txt", "r")
f = manager.__enter__()            # what "with ... as f" does first
try:
    content = f.read()
finally:
    manager.__exit__(None, None, None)   # what "with" does at the end

print(content)
print("Is the file closed now?", f.closed)
````

## <a name="9-8-using-try-except-to-read-a-file"></a>9.8 Using try-except to read a file

File operations often fail. Common reasons: (1) the file does not exist, (2) permission is denied, (3) the path is wrong. Without handling, such an error ends the program abruptly. The usual pattern:

```python
try:
    # File operation
except SomeException:
    # Handle error
else:
    # Runs if no error
```

### <a name="example-7-try-except-else-finally-for-reading-a-file"></a>Example 7: try-except-else-finally for reading a file

**What the code does.** `myReader()` opens a file with `with`, reads it, and handles three kinds of problems:

| except block | Handles |
|---|---|
| `except FileNotFoundError:` | The file does not exist |
| `except PermissionError:` | The file exists but you may not read it |
| `except Exception as e:` | Anything else (generic safety net) |

The `else` block prints the content only when the read succeeded, and `finally` always prints "3. File operation attempted".

It is tested with four cases:

| Case | Call | Expected blocks |
|---|---|---|
| 1 | a file that exists | try, else, finally |
| 2 | a file that does not exist | try, except FileNotFoundError, finally |
| 3 | a file you may not read | try, except PermissionError, finally |
| 4 | an invalid mode, `'non_existent_mode'` | try, generic except (`ValueError`), finally |

**How it works.** `open()` raises a different exception for each problem, and Python picks the first `except` block that matches. An invalid mode raises `ValueError`, which is not listed, so the generic block catches it.

**Why it matters.** Opening files is one of the most common places for errors in real programs. Handling each case separately gives the user a clear message instead of a crash.

**Learning points**

1. Specific blocks first, generic block last.
2. `with` inside `try` closes the file, and `try` handles errors from `open()` itself.
3. The `'r'` mode only reads, so the file cannot be changed by mistake.

````python
# This script demonstrates use of try-except-else-finally blocks for handling
# file operations. It has except blocks for errors like FileNotFoundError &
# PermissionError. It also has a generic except block to catch any other
# unforeseen exceptions. Using open() in 'r' mode is safer for reading
# files as it does not allow modifications, which can help prevent certain
# types of errors. The else block executes if no exceptions occur, and the
# finally block always executes regardless of whether an exception was raised
# or not, making it ideal for cleanup actions.

def myReader(fileName, mode='r'):
    try:
        print(f"1. Trying to open file in '{mode}' mode...")

        with open(fileName, mode) as f:
            content = f.read()

    except FileNotFoundError:
        print("Error: File not found")

    except PermissionError:
        print("Error: Permission denied")

    except Exception as e:
        print("Generic error (not caught by specific except blocks):", e)

    else:
        print("2. File read successfully")
        print("Content:\n", content)

    finally:
        print("3. File operation attempted")

# Test cases
print("Case 1: Valid file")
myReader("existing_file.txt")  # Make sure this file exists for the test
# Case 1 Output: try → else → finally. (No exceptions occur)

print("Case 2: File not found")
myReader("non_existent_file.txt")  # This file does not exist
# Case 2 Output: try → except(FileNotFoundError) → finally

print("Case 3: Permission denied")
myReader("/root/secret_file.txt")  # Assume file exists but not accessible
# Case 3 Output: try → except(PermissionError) → finally

print("Case 4: Invalid mode")
myReader("existing_file.txt", mode='non_existent_mode')  # Invalid mode → ValueError
# Case 4 Output: try → except(generic) → finally
````

### <a name="try-it-catching-two-exceptions-with-one-except"></a>Try it: catching two exceptions with one except

This cell is not in the book. `to_number()` turns a value into an integer. A string like `"abc"` raises `ValueError`, and a list raises `TypeError`. One `except` block with a tuple handles both. `type(e).__name__` shows which one was caught.

````python
def to_number(value):
    try:
        return int(value)
    except (ValueError, TypeError) as e:
        print("Cannot convert", repr(value), "->", type(e).__name__)
        return None

print(to_number("42"))
print(to_number("abc"))   # ValueError
print(to_number([1, 2]))  # TypeError
````

### <a name="9-9-1-do-not-group-a-parent-exception-with-its-child"></a>9.9.1 Do not group a parent exception with its child

Exceptions are classes, arranged in a tree by inheritance. A child exception is a subclass of its parent, so **catching the parent also catches all its children**. Putting a parent and its child in the same tuple is valid, but the child adds nothing.

For example, `ZeroDivisionError` is a subclass of `ArithmeticError`:

```
Exception
 |
 +-- ArithmeticError
 |    +-- ZeroDivisionError
 |
 +-- ValueError
 |
 +-- TypeError
```

So `except (ArithmeticError, ZeroDivisionError):` does the same job as `except ArithmeticError:`.

## <a name="9-11-exception-hierarchy-in-python"></a>9.11 Exception hierarchy in Python

Exceptions are organised as a tree: (1) some exceptions are parent classes, and (2) others are children (subclasses) of them. The tree helps you (1) group related errors, and (2) handle many errors with one `except`.

The root of the tree is `BaseException`. Every exception in Python inherits from it, directly or indirectly.

[Back to the Table of Contents](#toc)

### <a name="direct-subclasses-of-baseexception"></a>Direct subclasses of BaseException

| Class | Purpose |
|---|---|
| `SystemExit` | Raised when the program exits, for example by `sys.exit()` |
| `KeyboardInterrupt` | Raised when the user presses Ctrl+C (in Colab: the Stop button) |
| `GeneratorExit` | Raised when a generator is closed |
| `Exception` | Base class for all application-level errors |

Two points to remember:

- **Derive your own exceptions from `Exception`, not from `BaseException`.** `BaseException` also covers the system signals. Code that catches it may (1) stop the user from ending the program, and (2) interfere with signals such as Ctrl+C.
- **Parent before child is a mistake.** If an `except` for the parent comes first, it catches the child too, and the child's `except` block is never reached.

### <a name="example-8-the-order-of-except-blocks-for-a-parent-and-a-chil"></a>Example 8: The order of except blocks for a parent and a child exception

**What the code does.** It creates two exceptions of its own: `ParentError`, derived from `Exception`, and `ChildError`, derived from `ParentError`. Then it raises each of them in turn, twice:

- **Case 1 (correct order):** `except ChildError` comes before `except ParentError`.
- **Case 2 (wrong order):** `except ParentError` comes first.

**How it works.** `for x in [ParentError, ChildError]:` loops over the two **classes**, and `raise x()` creates and raises an object of each one. For each exception, Python tries the `except` blocks from the top and stops at the first match. A `ChildError` **is a** `ParentError`, so `except ParentError` matches it.

**Why it matters.** Python gives no warning for the wrong order. The program runs, but one handler is silently dead.

**Learning points**

1. Case 1 prints "Caught ParentError" and then "Caught ChildError". Each exception reaches its own block.
2. Case 2 prints "Caught ParentError" twice. The `except ChildError` block never runs.
3. Rule: **always put the most specific (child) exception first.**

````python
# Step 1: Define custom exceptions
class ParentError(Exception):
    pass

class ChildError(ParentError):   # ChildError inherits from ParentError
    pass

# Case 1: Catch Child before Parent (Correct Order)
print("Case 1: Catch ChildError before ParentError")
for x in [ParentError, ChildError]:
    try:
        raise x()   # Raise ParentError or ChildError

    except ChildError:  # This will catch ChildError but not ParentError
        print("Caught ChildError")

    except ParentError: # ChildError already caught so only catches ParentError
        print("Caught ParentError")

# Case 2: Catch Parent before Child (Incorrect Order)
print("Case 2: Catch ParentError before ChildError")

for x in [ParentError, ChildError]:
    try:
        raise x()

    except ParentError:   # This will catch BOTH ParentError and ChildError
        print("Caught ParentError")

    except ChildError: # This NEVER executes ParentError catches both exceptions
        print("Caught ChildError")
````

## <a name="9-12-user-defined-custom-exceptions"></a>9.12 User-defined (custom) exceptions

Sometimes the built-in exceptions do not describe an error in your program well. Python lets you define your own, called **user-defined** or **custom** exceptions.

Three facts to keep in mind:

1. Exceptions are classes.
2. When an exception is raised, it becomes an object.
3. A custom exception is a class derived from `Exception`.

A custom exception must be raised **explicitly**, with `raise`. Creating an object of the class does not raise it. Python raises built-in exceptions such as `ZeroDivisionError` by itself, but it cannot know when your own rule has been broken. So you write:

```python
raise MyException("message")
```

[Back to the Table of Contents](#toc)

### <a name="9-12-1-the-role-of-str-in-exceptions"></a>9.12.1 The role of `__str__()` in exceptions

`__str__()` is a special (dunder) method. It decides how an object is turned into text. It is used by (1) `print(obj)` and (2) `str(obj)`. For an exception, it decides the message the user sees.

[Back to the Table of Contents](#toc)

### <a name="example-9-a-custom-exception-with-its-own-init-and-str"></a>Example 9: A custom exception with its own `__init__()` and `__str__()`

**What the code does.** It brings together two ideas: raising a custom exception, and deciding how it is shown.

1. `InsufficientBalanceError` is derived from `Exception`. Its `__init__()` stores two values, `balance` and `withdraw`. Its `__str__()` builds a clear message from them.
2. `withdraw_money(balance, amount)` raises this exception when `amount` is larger than `balance`, and catches it in the same function.
3. It is tested with a withdrawal that works (500 from 1000) and one that does not (1500 from 1000).

**How it works, step by step**

| No. | Point | Explanation |
|---|---|---|
| 1 | Passing values when raising | `__init__()` takes `balance` and `withdraw`, so the raise statement passes both: `raise InsufficientBalanceError(balance, amount)`. They are stored as attributes of the exception object. |
| 2 | Creating the exception object | `raise` creates an object of `InsufficientBalanceError` (so `__init__()` runs), throws it, and it travels to the nearest matching `except` block. |
| 3 | `except ... as e` | `e` does not create a new object. It is a name for the same object that was raised. |
| 4 | `__str__()` | `print("Error:", e)` calls `e.__str__()` automatically, so the user sees a clear message. |
| 5 | Without `__str__()` | Python would use a default message, which is less helpful. `__str__()` is not required, but it is strongly recommended. |
| 6 | Flow of control | When the exception is raised, the rest of the `try` block is skipped ("Withdrawal successful" is not printed), and control jumps to the matching `except` block. |
| 7 | The object carries data | It can hold several values and a custom message. This is what makes custom exceptions powerful. |
| 8 | It is ordinary OOP | `InsufficientBalanceError` is a class, and `e` is an object (instance) of it. |
| 9 | The type must match | The block `except InsufficientBalanceError:` runs only for that type (or its subclasses). If nothing matches, the program stops with a traceback. |

![Flow of withdraw_money(1000, 1500) with a custom exception](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-09-nb-09-custom-exception.png)

````python
class InsufficientBalanceError(Exception):
    def __init__(self, balance, withdraw):
        self.balance = balance
        self.withdraw = withdraw

    def __str__(self):  # Custom error message when the exception is printed
        return f"Balance {self.balance} is less than withdrawal {self.withdraw}"

# Function to simulate money withdrawal & show custom exception handling
def withdraw_money(balance, amount):
    try:
        if amount > balance:
            raise InsufficientBalanceError(balance, amount)

        print("Withdrawal successful")

    except InsufficientBalanceError as e:  # Handle custom exception
        print("Error:", e)  #Print __str__() of InsufficientBalanceError class

# Test cases
print("Case 1: Sufficient balance")
withdraw_money(1000, 500)  # Balance is sufficient → Withdrawal successful

print("Case 2: Insufficient balance")
withdraw_money(1000, 1500)  # Balance is insufficient
# Error: Balance 1000 is less than withdrawal 1500
````

### <a name="exception-chaining-with-raise-from"></a>Exception chaining with raise ... from

When you catch one exception and raise another, `raise NewError(...) from old_error` keeps the link between them. The traceback then shows both, with the line "The above exception was the direct cause of the following exception". The link is stored in `__cause__`.

````python
class ConfigError(Exception):
    pass

def read_age(text):
    try:
        return int(text)
    except ValueError as e:
        raise ConfigError(f"Bad age in the settings: {text!r}") from e

try:
    read_age("ten")
except ConfigError as err:
    print("Error:", err)
    print("Caused by:", repr(err.__cause__))
````

### <a name="adding-a-note-to-an-exception-python-3-11-and-later"></a>Adding a note to an exception (Python 3.11 and later)
