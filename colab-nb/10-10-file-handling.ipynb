## <a name="10-1-introduction"></a>10.1 Introduction

A **file** is a named place on a disk where data is stored permanently. RAM is **volatile**: it loses its contents when the power goes off. A disk is **non-volatile**. So we save our work in files.

Python gives you three kinds of tools for files: (1) built-in functions such as `open()`, (2) methods of the file object such as `read()` and `write()`, and (3) operating-system utilities in the `os` module.

## <a name="10-2-a-small-note-on-the-function-signature"></a>10.2 A small note on the function "signature"

From this chapter on we use many built-in and library functions. To read their documentation, you need to understand the word **signature**.

A function's signature tells you three things: (1) the **name** of the function, (2) the **inputs** it expects (its parameters), and (3) the **order** in which you must give them.

[Back to the Table of Contents](#toc)

### <a name="10-2-1-the-two-faces-of-a-function"></a>10.2.1 The two faces of a function

When you look up a function such as `open()`, you may see it written in two ways. They are the same function, seen at two levels of detail.

| | The full signature | The essential signature |
|---|---|---|
| What it is | The complete, technical definition from the official documentation. It lists every option the function offers. | A simplified view, as seen in tutorials, textbooks and most real code. It shows only the parameters that matter for the usual task. |
| Example | `open(file, mode='r', buffering=-1, encoding=None, errors=None, newline=None, closefd=True, opener=None)` | `open(filename, mode, encoding)` |
| How to read it | Anything with `=` (or, in older styles, anything inside `[ ]`) is optional. The value after `=` is the default, used when you say nothing. | It leaves out options such as `buffering` or `opener` that most programmers never change. |

[Back to the Table of Contents](#toc)

### <a name="10-2-2-mandatory-and-optional-parameters"></a>10.2.2 Mandatory and optional parameters

| | Mandatory parameters (no default) | Optional parameters (default arguments) |
|---|---|---|
| Meaning | They have no default value. The function cannot work without them. If you leave one out, Python raises `TypeError`. | They have a fallback value in the signature. If you give a value, Python uses it; if you do not, Python uses the default. |
| How to spot them | They come first and have no `=` sign. | They come after the mandatory ones and are written `name=value`. Older manuals may write them in square brackets: `func(required[, optional])`. |
| Example | In `round(number, ndigits=None)`, `number` is mandatory. You cannot round nothing. | In `round(number, ndigits=None)`, `ndigits` is optional. |

[Back to the Table of Contents](#toc)

### <a name="10-2-3-comparison-at-a-glance"></a>10.2.3 Comparison at a glance

| Feature | Mandatory (essential) | Optional (secondary) |
|---|---|---|
| Syntax | `parameter_name` | `parameter_name=default_value` |
| Presence | Must be given in every call | Can be left out |
| Placement | Before the optional ones | After the mandatory ones |
| In documentation | No `=` and no `[ ]` | Marked with `=` or `[...]` |
| Purpose | Says **what** to act on | Fine-tunes **how** to act |

## <a name="10-6-the-file-object-and-good-practice"></a>10.6 The file object and good practice

### <a name="10-6-1-open"></a>10.6.1 open()

`open()` is where every file operation starts. It opens a file and returns a file object. The **mode** decides what you are allowed to do: `'r'` to read, `'w'` to write, `'a'` to append, and so on.

Its essential signature is `open(file, mode='r', encoding=None)`.

Two warnings:

- **Read mode `'r'`:** if the file does not exist, Python stops with `FileNotFoundError`.
- **Write mode `'w'`:** if the file already exists, its content is thrown away **at once**, before you write anything. Be careful with `'w'`.

In practice `open()` is used together with `with`, so that the file is closed automatically.

[Back to the Table of Contents](#toc)

### <a name="summary-of-the-modes"></a>Summary of the modes

| Mode | Purpose | Must the file exist? | Overwrites existing data? | Where writing starts |
|---|---|---|---|---|
| `'r'` | Read only (text) | Yes | No | Not applicable |
| `'rb'` | Read only (binary) | Yes | No | Not applicable |
| `'r+'` | Read and write (text) | Yes | No | Beginning |
| `'rb+'` or `'r+b'` | Read and write (binary) | Yes | No | Beginning |
| `'w'` | Write only (text) | No (creates it) | Yes (clears it) | Beginning |
| `'wb'` | Write only (binary) | No (creates it) | Yes (clears it) | Beginning |
| `'w+'` | Write and read (text) | No (creates it) | Yes (clears it) | Beginning |
| `'wb+'` or `'w+b'` | Write and read (binary) | No (creates it) | Yes (clears it) | Beginning |
| `'a'` | Append only (text) | No (creates it) | No | End of file |
| `'ab'` | Append only (binary) | No (creates it) | No | End of file |
| `'a+'` | Append and read (text) | No (creates it) | No | End (for writing); use `seek()` to read |
| `'ab+'` or `'a+b'` | Append and read (binary) | No (creates it) | No | End (for writing); use `seek()` to read |
| `'x'` | Create a new file (text) | Must NOT exist | No | Beginning |
| `'xb'` | Create a new file (binary) | Must NOT exist | No | Beginning |

An easy way to remember them: the letter says **what** (`r` read, `w` write, `a` append, `x` exclusive create), `+` adds the other direction, and `b` means binary.

`close()` closes the file and frees the resources the operating system gave to it. After that, the file object cannot be used for reading or writing.

Always close files. If you do not, data may be lost or resources may leak. In modern Python you rarely call `close()` yourself, because `with` does it for you.

## <a name="10-7-control-operations"></a>10.7 Control operations

Every open file has a **file pointer** (also called the cursor or position). It marks where the next read or write will happen. Reading moves the pointer forward.

[Back to the Table of Contents](#toc)

### <a name="10-7-1-tell"></a>10.7.1 tell()

`tell()` returns the current position of the pointer, counted from the beginning of the file. It shows how much of the file has been read or written. It is handy when you read in parts, or when you are debugging.

[Back to the Table of Contents](#toc)

### <a name="10-7-2-seek"></a>10.7.2 seek()

`seek(offset)` moves the pointer to a given position, so you can re-read data or skip part of the file. The most common use is `seek(0)`, which goes back to the start.

Use `seek()` with care. A wrong position can make you skip data, or overwrite the wrong part of the file.

The full signature is `seek(offset, whence=0)`. `whence` says where the offset is counted from: `0` the start, `1` the current position, `2` the end. In text mode only `seek(0)`, a value returned by `tell()`, or `seek(0, 2)` (go to the end) are safe.

The online resource has a study of `seek(offset, whence)` and how it behaves in text and binary files.

## <a name="10-8-read-operations"></a>10.8 Read operations

### <a name="10-8-1-read"></a>10.8.1 read()

`read()` with no argument reads the whole file into one string. `read(n)` reads at most `n` characters. It is simple and convenient for small files. For a very large file, reading everything at once can use a lot of memory, so use it only when you know the file is a manageable size.

[Back to the Table of Contents](#toc)

### <a name="10-8-2-readline"></a>10.8.2 readline()

`readline()` (singular) reads **one line** each time it is called. The line includes its newline character `\n`. It is useful when each line is one record, such as one entry in a log. It is often used in a loop. When the end of the file is reached, it returns an empty string `''`.

[Back to the Table of Contents](#toc)

### <a name="10-8-3-readlines"></a>10.8.3 readlines()

`readlines()` (plural) reads **all** the remaining lines and returns them as a **list** of strings, one element per line. It is convenient but can use a lot of memory for big files. For large files, loop over the file directly: `for line in file:`. Use `readlines()` only when you really need all the lines in a list.

[Back to the Table of Contents](#toc)

### <a name="summary-of-the-read-methods"></a>Summary of the read methods

| Method | What it does | Returns |
|---|---|---|
| `read()` | Reads the entire file | One string |
| `read(n)` | Reads up to `n` characters | One string |
| `readline()` | Reads one line | One string (ends with `\n`, or `''` at the end of the file) |
| `readlines()` | Reads all remaining lines | A list of strings |

[Back to the Table of Contents](#toc)

### <a name="example-1-reading-moving-the-pointer-and-closing"></a>Example 1: Reading, moving the pointer, and closing

**What the code does.**

1. It creates `demo.txt` with three lines, so there is something to read.
2. It opens the file for reading and shows where the pointer is, then uses `read(6)`, `seek(0)`, `readline()` and `readlines()` one after another, printing the position as it goes.
3. After the `with` block, the file is closed automatically.
4. Finally it opens the file again **without** `with`, to show `closed` before and after calling `close()` by hand.

**How it works.** Each read starts where the pointer is and moves it forward. `read(6)` takes the six characters `Line 1` and leaves the pointer at 6. `seek(0)` sends it back to the start, so `readline()` reads the first line again, this time with its `\n`. `readlines()` then takes everything that is left.

**Why it matters.** Knowing where the pointer is explains most surprises in file reading, such as "why did my second `read()` return nothing?" (answer: the pointer was already at the end).

**Learning points**

1. A new file object starts at position 0.
2. `readline()` keeps the `\n`. That is why there is an empty line in the output after "Using readline(): Line 1". `print()` adds its own newline as well.
3. `readlines()` returned only lines 2 and 3, because line 1 had already been read.
4. `f.closed` tells you whether a file object is closed.

![How the file pointer moves in Example 1](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-10-nb-02-file-pointer.png)

````python
# Demonstrating file (1) pointer manipulation (2) reading (3) closing
# 1. Create and write some content first (so that reading works)
with open("demo.txt", "w") as f:
    f.write("Line 1\nLine 2\nLine 3\n")  # Write multiple lines to the file

# 2. Now open the file for reading
with open("demo.txt", "r") as f:
    print("Initial pointer position:", f.tell())   # Output: 0

    # A. read() - read first 6 characters
    print("Using read(6):", f.read(6))  # Output: 'Line 1'
    print("Pointer after read:", f.tell())         # Output: 6

    # B. seek() - move pointer back to beginning
    f.seek(0)
    print("Pointer after seek(0):", f.tell())  # Output: 0

    # C. readline() - read one line
    print("Using readline():", f.readline())  # Output: 'Line 1\n'

    # D. readlines() - read remaining lines as list
    print("Using readlines():", f.readlines()) # Output: ['Line 2\n', 'Line 3\n']

# 3. File is automatically closed after 'with'
print("File closed automatically by with statement.")

# 4. Demonstrating close() explicitly (not recommended when using 'with')
f = open("demo.txt", "r")  # open file again
print("File opened again. Closed?", f.closed)  # Output: False

f.close()   # close()
print("After close(), Closed?", f.closed)  # Output: True
````

## <a name="10-9-write-operations"></a>10.9 Write operations

### <a name="10-9-1-write"></a>10.9.1 write()

`write(string)` writes one string to the file and returns the number of characters written. That number can be used to check the write.

`write()` does **not** add a newline. If you want separate lines, put `\n` at the end of each string yourself. And remember: in `'w'` mode the old content of the file is gone.

[Back to the Table of Contents](#toc)

### <a name="10-9-2-writelines"></a>10.9.2 writelines()

`writelines(list_of_strings)` writes every string in a list (or any sequence of strings) to the file. It is handy when you have many lines at once.

`writelines()` does **not** add newlines between the items either. Each string must already end with `\n` if you want it on its own line. Beginners often forget this and find everything on one long line.

| | `write()` | `writelines()` |
|---|---|---|
| Takes | One string | A list (or other sequence) of strings |
| Adds `\n`? | No | No |
| Returns | Number of characters written | `None` |

[Back to the Table of Contents](#toc)

### <a name="example-2-write-and-writelines"></a>Example 2: write() and writelines()

**What the code does.** It opens `write_demo.txt` in `'w'` mode, writes two lines with `write()` and two more with `writelines()`, and lets `with` close the file.

**How it works.** Every string ends in `\n`, so the file ends up with four separate lines. The file is created if it does not exist; if it exists, its old content is erased first.

**Learning points**

1. The cell prints nothing. Writing to a file does not show anything on the screen. The "Try it" cell below reads the file back so that you can see the result.
2. `'w'` replaces the whole file every time you run the cell. You will never see the lines twice.

![What Example 2 writes into write_demo.txt](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-10-nb-03-write-steps.png)

````python
# Open file in write mode. 'w' will create the file if it does not exist
# and overwrite (delete) existing content if it exists
with open("write_demo.txt", "w") as f:

    # A. write() → writes a single string
    # Note: newline (\n) must be added manually
    f.write("Line 1\n")  # Writes 'Line 1' followed by a newline character
    f.write("Line 2\n")  # Writes 'Line 2' followed by a newline character

    # B. writelines() → writes multiple strings from a list
    # Note: it does NOT add newline automatically
    lines = ["Line 3\n", "Line 4\n"]  # List of strings to write
    f.writelines(lines)  # Writes each string from the list named "lines"

# File is automatically closed here due to 'with'
````

## <a name="10-13-methods-for-binary-files"></a>10.13 Methods for binary files

The methods for binary files are the same as for text files. What changes is the **type** of the data: `bytes` instead of `str`.

| No. | Method | In binary mode, compared with text mode |
|---|---|---|
| 1 | `open()` with `b` | Opens the file for bytes instead of text. Used for images, PDFs and audio. A common mistake is forgetting the `b`, which damages the data. |
| 2 | `read()` | Returns a `bytes` object (`b'...'`), all of it or part of it. Used to copy files and to process media. The output is not meant for humans. |
| 3 | `readline()` | Reads bytes up to and including `b'\n'`. Rarely used, and only when the binary file happens to be organised in lines. |
| 4 | `readlines()` | Returns a list of `bytes` objects, split at `b'\n'`. Of little use with binary data; avoid it for large files. |
| 5 | `write()` | Must be given `bytes`, for example `f.write(b"Hello")`. Giving it a `str` raises `TypeError`; encode the string first. |
| 6 | `writelines()` | Writes a list of `bytes` objects. Every element must be a `bytes` object. |
| 7 | `close()` | The same as for text files. Better to use `with`. |
| 8 | `tell()` | Returns the exact byte position. Very reliable in binary mode. |
| 9 | `seek()` | Moves the pointer to a byte offset. Precise; good for random access. |
| 10 | `flush()` | Forces the buffered bytes onto the disk. Useful for live logging. The same as in text mode. |
| 11 | `truncate()` | Cuts the file at a byte position. Anything after it is lost. |

## <a name="10-18-python-i-o-stream-objects"></a>10.18 Python I/O stream objects

| `sys.stderr` (`write()`) | Shows error messages | Program to outside | Error messages, exceptions | Debugging, error reports | Kept separate from normal output |

## <a name="10-20-pickling-and-unpickling"></a>10.20 Pickling and unpickling

**Pickling** is also called **serialisation** or **flattening**. The key ideas:

- Python objects cannot be written to a file as they are. They must first be turned into a sequence of bytes.
- **Pickling** turns a Python object into a byte stream.
- **Unpickling** turns the byte stream back into the original object.
- **Why?** (1) To store complex Python objects (lists, dictionaries, objects of your own classes), (2) to save the state of a program, (3) to pass data between programs.
- **Persistence** means that data survives after the program ends. Data in RAM is lost when the program stops; files and databases make it persistent.

Python has a standard library module called `pickle` for this. To use it, write `import pickle`.

- **To pickle (write):** (1) open the file in `'wb'` mode, (2) call `pickle.dump(obj, file)`.
- **To unpickle (read):** (1) open the file in `'rb'` mode, (2) call `pickle.load(file)`.

| Operation | Function | File mode | Purpose | Returns |
|---|---|---|---|---|
| Pickling | `pickle.dump(obj, file)` | `'wb'` | Save an object | `None` |
| Unpickling | `pickle.load(file)` | `'rb'` | Load an object | The original object |

Important points:

1. Use pickling for Python objects only; other languages cannot read it.
2. Always open pickle files in binary mode.
3. Use `with`.
4. Never open a pickle file in text mode.
5. Never edit a pickle file by hand.
6. **Never unpickle data from a source you do not trust.** A pickle file can be built to run harmful code when it is loaded.

[Back to the Table of Contents](#toc)

### <a name="example-6-pickling-and-unpickling-a-dictionary"></a>Example 6: Pickling and unpickling a dictionary

**What the code does.** It pickles the dictionary `{'a': 1, 'b': 2}` into `data.pkl`, unpickles it into `new_data`, prints both, and finally prints the raw bytes of the file.

**How it works.** `pickle.dump(data, f)` writes the bytes that describe the dictionary. `pickle.load(f)` reads those bytes and builds a **new** dictionary with the same content.

**Learning points**

1. `new_data` equals `data` (`new_data == data` is `True`), but it is a different object (`new_data is data` is `False`).
2. The raw bytes are not meant to be read by people. Only `pickle` understands them.
3. Your raw bytes may differ slightly from the sample in the comment; they depend on the pickle protocol version.

![Pickling a dictionary to a file and unpickling it again](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-10-nb-07-pickle.png)

````python
import pickle

# STEP 1: Create an object (dictionary)
# Object we want to serialize (convert to byte stream) & store in file.
data = {'a': 1, 'b': 2}

# STEP 2: Pickle (write object to file)
# Open a file in binary write mode to store the pickled data
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)   # Convert object → byte stream and store

# STEP 3: Unpickle (read object from file)
# Open the file in binary read mode to read the pickled data
with open("data.pkl", "rb") as f:
    new_data = pickle.load(f)  # Convert byte stream → object

# STEP 4: Verify
print("Original:", data)  # Print the original dictionary
# Output: Original: {'a': 1, 'b': 2}
print("Unpickled:", new_data)  # Print unpickled dictionary (matches original)
# Output: Unpickled: {'a': 1, 'b': 2}

# STEP 5: Reading raw file (not useful)
# Open the file in binary read mode to read the raw bytes
with open("data.pkl", "rb") as f:
    print("Raw bytes:", f.read())  # Not human-readable
    # Output: b'\x80\x04\x95\x0c\x00\x00\... (binary data)
````

## <a name="10-21-json-for-data-storage"></a>10.21 JSON for data storage

- **JSON** stands for JavaScript Object Notation.
- It is a **text-based** format for storing and exchanging data.
- It is easy for people to read, and almost every language and system understands it. Example: `{"name": "Alice", "marks": 90}`.
- Python provides the `json` module. To use it, write `import json`.

JSON is used in (1) web APIs (REST APIs), (2) data exchange over the internet, (3) configuration files, and (4) NoSQL databases such as MongoDB.

**Security:** JSON is plain text, so data sent over a network can be read if it is intercepted. The solution is to send it over an encrypted connection (HTTPS, which uses SSL/TLS).

### <a name="try-it-the-four-json-functions"></a>Try it: the four json functions

This cell is not in the book. It converts a dictionary to a JSON string and back, and then to a `.json` file and back. Notice how Python's `True` becomes JSON's `true`, and that the file is plain text you can open in the Files panel.

````python
import json

pet = {"name": "Tommy", "age": 3, "vaccinated": True, "tricks": ["sit", "roll"]}

# To and from a string
text = json.dumps(pet)
print("JSON string:", text, type(text))
back = json.loads(text)
print("Back to Python:", back, type(back))

# To and from a file
with open("pet.json", "w") as f:
    json.dump(pet, f, indent=2)        # indent makes the file easy to read

with open("pet.json") as f:
    print("Raw file content:")
    print(f.read())

with open("pet.json") as f:
    loaded = json.load(f)
print("Loaded equals original?", loaded == pet)
````

## <a name="10-22-json-compared-with-pickle"></a>10.22 JSON compared with pickle

| Feature | JSON | Pickle |
|---|---|---|
| Format | Text | Binary |
| Readability | Readable by people | Not readable |
| Security | Safe to load | Unsafe: loading can run code |
| Language support | Almost every language | Python only |
| Data types | Basic: dict, list, str, int, float, bool, None | Almost any Python object |
| Used in APIs | Yes | No |
| Speed | Moderate | Faster |
| File size | Larger | Smaller |

Points to remember about JSON: (1) use it for data exchange and in APIs; (2) keys must be strings in double quotes, as in `{"key": value}`; (3) it cannot store complex objects (such as objects of your own classes) directly; (4) do not send sensitive data without encryption; (5) do not use it for binary data.

The online resource has a comparative study of JSON and pickle: serialising with both, and comparing performance, security and portability.

### <a name="reading-a-big-file-one-line-at-a-time"></a>Reading a big file one line at a time

Looping over a file object reads one line at a time, so even a huge file uses very little memory. This is the memory-safe way to count lines.

````python
with open("big.txt", "w") as f:              # make a file with 100,000 lines
    for i in range(100_000):
        f.write(f"line {i}\n")

count = 0
with open("big.txt") as f:
    for line in f:                           # one line in memory at a time
        count += 1
print("Number of lines:", count)

import os
print("File size in bytes:", os.path.getsize("big.txt"))
````

### <a name="paths-the-modern-way-pathlib"></a>Paths the modern way: pathlib

The `pathlib` module treats a path as an object. The `/` operator joins parts of a path, and the object has handy methods for reading and writing.

````python
from pathlib import Path

folder = Path("notes")
folder.mkdir(exist_ok=True)
file = folder / "todo.txt"                   # joins the parts, like os.path.join
file.write_text("Buy dog food\nBook vet visit\n")

print(file, "exists?", file.exists())
print(file.read_text())
print("Name:", file.name, "| Suffix:", file.suffix, "| Parent:", file.parent)
````

### <a name="two-styles-of-error-handling-lbyl-and-eafp"></a>Two styles of error handling: LBYL and EAFP

**LBYL** ("look before you leap") checks first. **EAFP** ("easier to ask forgiveness than permission") just tries and handles the error. Python programmers usually prefer EAFP, because the file could disappear between the check and the `open()`.

````python
import os

name = "missing.txt"

# LBYL: check first
if os.path.exists(name):
    with open(name) as f:
        print(f.read())
else:
    print("LBYL: the file does not exist")

# EAFP: try, and handle the error
try:
    with open(name) as f:
        print(f.read())
except FileNotFoundError:
    print("EAFP: the file does not exist")
````
