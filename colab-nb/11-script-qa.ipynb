### <a name="q1-reverse-an-array-with-slicing"></a>Q1. Reverse an array with slicing

**Question.** Create a 1D NumPy array containing the numbers from 10 to 49. Then, reverse this array using slicing.

**What the answer does.** `np.arange(10, 50)` builds the 40 numbers from 10 to 49 (the stop value 50 is not included). `arr[::-1]` then reads them backwards.

**How it works.** The full slice syntax is `[start : stop : step]`. With start and stop left out, the slice covers the whole array, and a step of -1 means "move one place at a time, from right to left".

**Learning points**

1. `arange(10, 50)` gives 40 numbers, not 41. The stop value is always excluded.
2. `arr[::-1]` is a **view**, not a copy. It shares memory with `arr`.
3. **Try it:** `arr[::-2]` gives every second number, backwards.

![How arr[::-1] reads an array backwards](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-11-qa-01-reverse-slice.png)

```python
import numpy as np

# Create the array using arange
arr = np.arange(10, 50)

# Reverse the array using the [start:stop:step] slicing syntax with step -1
reversed_arr = arr[::-1]

print("Original:", arr)
print("Reversed:", reversed_arr)
```

### <a name="q3-an-identity-matrix-with-5-on-the-diagonal"></a>Q3. An identity matrix with 5 on the diagonal

**Question.** Initialize a `5x5` identity matrix and change all the diagonal elements to 5.

**What the answer does.** `np.eye(5)` gives a 5 x 5 matrix with 1 on the diagonal and 0 everywhere else. Multiplying the whole matrix by 5 turns each 1 into a 5 and leaves the zeros as they are.

**Why it is written this way.** It avoids a loop: one broadcast multiplication does the whole job.

**Learning points**

1. `np.eye()` always gives floats, so the output shows `5.` and `0.`.
2. This trick works only because everything off the diagonal is 0.

```python
import numpy as np

# np.eye creates a square matrix with 1s on the diagonal
identity = np.eye(5)

# Multiply the entire matrix by 5 (Broadcasting a scalar)
# Since only the diagonal has 1s, only they become 5
five_diag = identity * 5

print(five_diag)
```

### <a name="q4-minimum-maximum-and-memory-address"></a>Q4. Minimum, maximum and memory address

**Question.** Generate a `10x10` array with random values. Find the minimum and maximum values and their memory addresses.

**What the answer does.** `np.random.rand(10, 10)` gives 100 random floats between 0 and 1. `data.min()` and `data.max()` return the smallest and largest. `data.ctypes.data` gives the memory address where the array's block of data starts.

**How it works.** A NumPy array keeps all its values in one continuous block of memory. `ctypes.data` is the address of the first byte of that block.

**Learning points**

1. `min()` and `max()` without an axis look at the whole array.
2. The address is a large number, and it is different every time you run the cell.
3. The address printed here is the start of the **array**, not the address of the smallest or largest value. The Notes cell below shows how to work those out.

```python
import numpy as np

# Create random floats in [0, 1)
data = np.random.rand(10, 10)

# Use built-in aggregate functions
min_val = data.min()
max_val = data.max()

print(f"Min: {min_val}, Max: {max_val}")
# The data pointer shows the memory start location
print(f"Memory Address: {data.ctypes.data}")
```

[Back to the Table of Contents](#toc)

### <a name="notes-on-q4"></a>Notes on Q4

To reach the address of a particular element, start at the array's address and add its position times the size of one element. `argmin()` and `argmax()` give the positions.

```python
data = np.random.rand(10, 10)
start = data.ctypes.data
size = data.itemsize                   # bytes per element, 8 for float64

i_min = data.argmin()                  # position in the flattened array
i_max = data.argmax()

print("Min:", data.flat[i_min], "at flat position", i_min)
print("Max:", data.flat[i_max], "at flat position", i_max)
print("Address of the array :", start)
print("Address of the min   :", start + i_min * size)
print("Address of the max   :", start + i_max * size)
print("Row and column of the min:", np.unravel_index(i_min, data.shape))
```

### <a name="q6-a-checkerboard-pattern"></a>Q6. A checkerboard pattern

**Question.** Create an `8x8` matrix and fill it with a checkerboard pattern (0s and 1s) using slicing.

**What the answer does.** It starts with an 8 x 8 array of zeros and then writes 1 into two sets of positions: odd rows with even columns, and even rows with odd columns.

**How it works.** `check[1::2, ::2]` selects rows 1, 3, 5, 7 and columns 0, 2, 4, 6. `check[::2, 1::2]` selects rows 0, 2, 4, 6 and columns 1, 3, 5, 7. Together they cover exactly the black squares of a chessboard.

**Learning points**

1. A slice can be used on both sides of the equals sign.
2. `dtype=int` keeps the output as 0 and 1 rather than 0. and 1.
3. **Try it:** swap the two assignments around, so that the top-left square becomes 1.

![Building a checkerboard with two slice assignments](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-11-qa-02-checkerboard.png)

```python
import numpy as np

check = np.zeros((8, 8), dtype=int)

# Every second row starting from index 1, every second column starting from index 0
check[1::2, ::2] = 1

# Every second row starting from index 0, every second column starting from index 1
check[::2, 1::2] = 1

print(check)
```

### <a name="q8-view-compared-with-copy"></a>Q8. View compared with copy

**Question.** Demonstrate the "View vs Copy" behavior by slicing an array and modifying the slice. Verify the original array.

**What the answer does.** It slices the first three elements and changes one of them, then prints the original array to show that it changed too. It then takes a `.copy()` of the same slice, changes that, and prints the original again to show that this time nothing changed.

**How it works.** A basic slice does not move any data. It only records where to look, so the slice and the original share the same memory. `.copy()` makes a real second copy of the data.

**Learning points**

1. After `slice_view[0] = 99`, the original starts with 99.
2. After `safe_copy[1] = 88`, the original still shows 2 at that position.
3. Rule of thumb: slicing gives a view; fancy indexing (a list of positions) and boolean masking give a copy.

![Slicing gives a view; integer indexing and masking give a copy](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-11-nb-01-view-vs-copy.png)

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5])

# Slicing creates a 'view' (shared memory)
slice_view = original[0:3]
slice_view[0] = 99

print("Modified Original:", original) # Should show 99 at index 0

# To keep original safe, use .copy()
safe_copy = original[0:3].copy()
safe_copy[1] = 88

print("Original after modifying copy:", original) # Should NOT show 88
```

### <a name="q10-normalising-a-matrix-to-the-range-0-to-1"></a>Q10. Normalising a matrix to the range 0 to 1

**Question.** Normalize a random `5x5` matrix so that all values lie between 0 and 1.

**What the answer does.** It applies the min-max formula (x - min) / (max - min) to every element at once.

**How it works.** `rand_arr.min()` and `rand_arr.max()` are single numbers, so they are broadcast across the whole matrix. After the calculation the smallest value becomes exactly 0 and the largest exactly 1.

**Learning points**

1. Normalising puts different measurements on the same scale. It is a common first step in machine learning.
2. The shape does not change; only the values do.
3. If every value were the same, max - min would be 0 and you would get a division-by-zero warning.

```python
import numpy as np

rand_arr = np.random.rand(5, 5) * 100

# Normalization formula: (x - min) / (max - min)
normalized = (rand_arr - rand_arr.min()) / (rand_arr.max() - rand_arr.min())

print(normalized)
```

### <a name="q12-fifty-points-from-0-to-2-pi-and-their-sines"></a>Q12. Fifty points from 0 to 2 pi, and their sines

**Question.** Generate 50 evenly spaced numbers between 0 and 2π and calculate their sine values.

**What the answer does.** `np.linspace(0, 2 * np.pi, 50)` gives exactly 50 points, including both ends. `np.sin(x)` then works on all 50 at once.

**Why `linspace` and not `arange`?** Here the **number of points** matters, not the step size. That is what `linspace` is for.

**Learning points**

1. `np.pi` is NumPy's value of pi.
2. `np.sin` is a universal function (ufunc): it is applied to every element.

### <a name="q16-stacking-two-arrays-vertically-and-horizontally"></a>Q16. Stacking two arrays, vertically and horizontally

**Question.** Stack two 2D arrays vertically and then horizontally.

**What the answer does.** `np.vstack((a, b))` puts `b` under `a`, giving 4 rows and 2 columns. `np.hstack((a, b))` puts `b` beside `a`, giving 2 rows and 4 columns.

**How it works.** Both functions take **one argument**: a tuple of the arrays to join. Vertical stacking needs the same number of columns; horizontal stacking needs the same number of rows.

**Learning points**

1. Think of rows as records (students, readings) and columns as attributes (subjects, sensors).
2. `np.concatenate((a, b), axis=0)` and `axis=1` do the same two jobs.

**What the answer does.** For the matrix [[1, 2], [2, 1]] it returns the eigenvalues 3 and -1, and the matching eigenvectors.

**How it works.** `np.linalg.eig()` returns two things: a 1D array of eigenvalues, and a 2D array whose **columns** are the eigenvectors, each scaled to length 1. So `eigenvectors[:, 0]` belongs to `eigenvalues[0]`.

### <a name="q26-singular-value-decomposition-of-a-3-x-2-matrix"></a>Q26. Singular Value Decomposition of a 3 x 2 matrix

**Question.** Perform a Singular Value Decomposition (SVD) on a `3x2` matrix.

**What the answer does.** It splits the matrix into three parts, U, the singular values S, and V transpose, so that A = U Σ V<sup>T</sup>.

**How it works.** By default `np.linalg.svd()` returns the "full" matrices: U is 3 x 3 here, and Vt is 2 x 2. S always comes back as a **1D array** of singular values, in decreasing order, not as a matrix.

**Learning points**

1. The singular values are 9.53 and 0.51. The first one carries most of the information in the matrix.
2. To rebuild the matrix you need S as a diagonal matrix: `np.diag(S)`.
3. `np.linalg.svd(A, full_matrices=False)` gives the smaller "economy" version, which is what you normally want. The Notes cell shows the rebuild.

```python
import numpy as np

A = np.array([[1, 2], [3, 4], [5, 6]])

# svd returns U, Sigma (as 1D), and VT (Transpose of V)
U, S, VT = np.linalg.svd(A)

print("U:\n", U)
print("Singular Values (Sigma):", S)
print("V Transpose:\n", VT)
```

**The difference.**

| Method | What it returns | Effect of changing the result |
|---|---|---|
| `.flatten()` | Always a new **copy** | The original is untouched |
| `.ravel()` | A **view** when it can, otherwise a copy | Changing it can change the original |

**Learning points**

2. The answer prints only `flat_copy`. The Notes cell below shows the difference in behaviour.

```python
import numpy as np

mat = np.array([[1, 2], [3, 4]])

# .flatten() always returns a copy (safe)
flat_copy = mat.flatten()

# .ravel() returns a view if possible (memory efficient)
flat_view = mat.ravel()

print(flat_copy)
```

### <a name="q33-a-rolling-average-with-np-convolve"></a>Q33. A rolling average with np.convolve

**Question.** Implement a simple "Rolling Average" using `np.convolve`.

**What the answer does.** It slides a window of three values across the data and gives the average of each window.

**How it works.** The weights `[1/3, 1/3, 1/3]` say "add three neighbours and divide by 3". `np.convolve` multiplies the window by the data at every position and adds up the products. `mode='valid'` keeps only the positions where the window fits completely, so 8 values give 6 results.

**Learning points**

1. A rolling (or moving) average smooths out short-term jumps. It is used on share prices, sensor readings and weather data.
2. The three modes are `'full'` (10 values here), `'same'` (8 values, padded with zeros at the ends), and `'valid'` (6 values).
3. Changing the weights changes the filter. Weights that add up to 1 keep the overall level of the data.

![How np.convolve computes a rolling average](https://raw.githubusercontent.com/ag999git/001-Python-book-2026/main/resources/ch-11-qa-03-rolling-average.png)

```python
import numpy as np

data = np.array([1, 2, 3, 4, 5, 6, 7, 8])
window_size = 3

# Create a uniform weight window
weights = np.ones(window_size) / window_size

# valid mode ensures the window fits completely within the data
rolling_avg = np.convolve(data, weights, mode='valid')

print(rolling_avg)
```

**What the answer does.** It creates a file on disk and treats it as a 1000 x 1000 array of `float32` values. It fills it with random numbers and then calls `flush()` to make sure everything is written.

**How it works.** A memory-mapped array behaves like a normal array, but the data stays in the file. The operating system loads only the pieces you touch. So you can work with files far larger than the computer's memory.

**Learning points**

1. `mode='w+'` creates the file (or overwrites it). Use `mode='r'` to read an existing one, and `mode='r+'` to read and change it.

### <a name="q37-unique-values-and-how-often-they-occur"></a>Q37. Unique values and how often they occur

**Question.** Find the unique elements and their counts in an array.

**What the answer does.** `np.unique(data, return_counts=True)` returns two arrays: the distinct values, in order, and how many times each one appears. `dict(zip(values, counts))` pairs them up into a dictionary, which prints neatly.

**Learning points**

1. This is the NumPy version of a frequency table.
2. `np.unique()` always returns the values **sorted**.
3. Other useful options: `return_index=True` (where each value first appears) and `return_inverse=True` (how to rebuild the original array).
4. In NumPy 2, the printed dictionary shows each number as `np.int64(2)`. That is only how NumPy prints its own integer type. For a plain Python dictionary, write `dict(zip(values.tolist(), counts.tolist()))`.

```python
import numpy as np

data = np.array([1, 2, 2, 3, 3, 3, 4])

# return_counts=True gives the frequency of each item
values, counts = np.unique(data, return_counts=True)

print(dict(zip(values, counts)))
```

### <a name="q38-repeating-a-pattern-with-np-tile"></a>Q38. Repeating a pattern with np.tile

**Question.** Tile an array to repeat a pattern.

**What the answer does.** `np.tile(pattern, 3)` lays the three numbers down three times, one after the other.

**How it works.** `tile` repeats the **whole array**, like laying tiles on a floor. `np.repeat` is different: it repeats each element in place, so `np.repeat([1,2,3], 3)` gives `[1 1 1 2 2 2 3 3 3]`.

**Learning points**

1. For a 2D result, give a tuple: `np.tile(pattern, (2, 3))` gives 2 rows, each with the pattern 3 times.
2. `tile` is handy for building test data and repeating patterns such as a checkerboard.

```python
import numpy as np

pattern = np.array([1, 2, 3])

# Repeat the pattern 3 times horizontally
tiled = np.tile(pattern, 3)

print(tiled) # [1, 2, 3, 1, 2, 3, 1, 2, 3]
```

### <a name="q39-comparing-floats-with-np-allclose"></a>Q39. Comparing floats with np.allclose

**Question.** Use `np.allclose` to check if two arrays are mathematically equal within a small tolerance.

**What the answer does.** It shows that `0.1 + 0.2 == 0.3` is `False` in floating-point arithmetic, and that `np.allclose()` correctly says the two are the same for practical purposes.

**How it works.** Numbers such as 0.1 cannot be stored exactly in binary, so a tiny error creeps in: 0.1 + 0.2 is really 0.30000000000000004. `np.allclose()` allows a small tolerance (by default about 1 part in 100 million).

**Learning points**

1. Never compare the results of floating-point calculations with `==`.
2. `np.allclose()` gives one True or False for the whole array; `np.isclose()` gives an element-by-element answer.
3. You can set your own tolerance: `np.allclose(a, b, rtol=1e-3)`.

```python
import numpy as np

a = np.array([0.1 + 0.2])
b = np.array([0.3])

# Floating point precision often makes '==' fail
print("Equality:", a == b) 

# allclose is the best practice for comparing floats
print("Close enough:", np.allclose(a, b))
```

### <a name="q40-a-custom-ufunc-with-np-frompyfunc"></a>Q40. A custom ufunc with np.frompyfunc

**Question.** Create a custom ufunc (Universal Function) using `np.frompyfunc`.

**What the answer does.** It takes an ordinary Python function and wraps it so that it can be applied to a whole array. The 1 and 1 say "one input, one output".

**Learning points**

1. The result has `dtype=object`, so it looks like numbers but is not a normal numeric array. Add `.astype(int)` or use `np.vectorize(my_logic)` if you need a proper numeric result.
2. This is **convenience, not speed**. Python still runs the function once per element. A real ufunc such as `data**2 + 10` is far faster.
3. Use it only when the logic cannot be written with NumPy operations.

```python
import numpy as np

def my_logic(x):
    return x**2 + 10

# Convert standard Python function to a NumPy ufunc
# 1 input, 1 output
vectorized_logic = np.frompyfunc(my_logic, 1, 1)

data = np.array([1, 2, 3])
print(vectorized_logic(data))
```

### <a name="sorting-and-sorting-by-another-column"></a>Sorting, and sorting by another column

`np.sort()` returns the sorted values, and `np.argsort()` returns the **positions** that would sort the array. The positions are what you need when one array must be reordered to match another.

```python
import numpy as np

names = np.array(["Alice", "Bob", "Charlie", "David"])
marks = np.array([85, 92, 78, 88])

order = np.argsort(marks)[::-1]        # positions, highest mark first
print("Rank order:", order)
for place, i in enumerate(order, start=1):
    print(f"{place}. {names[i]:<8} {marks[i]}")

print("Top two marks:", np.sort(marks)[::-1][:2])
```

### <a name="percentiles-and-a-quick-summary-of-a-dataset"></a>Percentiles and a quick summary of a dataset

```python
rng = np.random.default_rng(seed=0)
data = rng.normal(loc=60, scale=12, size=500)     # 500 exam marks

print(f"Count  : {data.size}")
print(f"Mean   : {data.mean():.2f}")
print(f"Std    : {data.std(ddof=1):.2f}")
print(f"Min    : {data.min():.2f}")
print(f"25%    : {np.percentile(data, 25):.2f}")
print(f"Median : {np.median(data):.2f}")
print(f"75%    : {np.percentile(data, 75):.2f}")
print(f"Max    : {data.max():.2f}")
print("How many above 75:", np.count_nonzero(data > 75))
```

### <a name="counting-with-a-boolean-array"></a>Counting with a boolean array

A True counts as 1 and a False as 0, so `sum()` and `mean()` on a mask answer "how many?" and "what fraction?".

```python
marks = np.array([85, 32, 78, 92, 25, 60])

passed = marks >= 40
print("Mask        :", passed)
print("How many passed:", passed.sum())
print("Pass rate      :", f"{passed.mean() * 100:.1f}%")
print("Any failures?  :", (~passed).any())
print("All above 20?  :", (marks > 20).all())
```
