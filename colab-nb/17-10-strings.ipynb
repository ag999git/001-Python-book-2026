## <a name="17-4-string-indexing-and-immutability"></a>17.4 String indexing and immutability
Strings in Python are an ordered collection of characters. "Ordered" means that it is a sequence of characters.
In Python, the individual characters are accessed by their index, which starts from 0 and ends at one less than the length of the string. Thus, if the length of a string is n, then its index will be from 0 to n-1. (This is similar to C++). However, in Python, we can also access the individual characters of a string using negative indices as well.
The last character in the string has a negative index of -1, and the first character (For a string of n characters) has an index of –n.
The indexing applies to string variables as well as string literals.
The other important characteristic of a string sequence is its immutability, which means you cannot change any character appearing at a particular index (position) of the string.
This is clear from the following:


```python
# String immutability and indexing
myString = "Hello World!"
print(myString[0])  #'H'. Indexing starts at 0
print(myString[-1]) #'!'. Negative indexing starts from the end
```

## <a name="17-5-escape-sequence"></a>17.5 Escape sequence
In Python, inside a string literal (which begins with a single, double, or triple quote), a backslash, i.e., \ has the magical power of an "escape character". This means that a backslash isn't interpreted as a backslash, but as a signal that the next character is to be given a special interpretation.


```python
# Escape Sequences Demonstration Script
print("I want a Backslash: \\")  # Backslash character
print('It\'s a cold day!')  # Single quote inside string
print("He said, \"Good Morning\"")  # Double quotes inside string
print("Hello\nWorld!")  # New line between Hello and World
print("Hello\tWorld!")  # Tab space between Hello and World
print("First Line\rSecond Line")  # Carriage return
print("Column1\bColumn2")  # Backspace character
print("Unicode Character: \u03A9")  # Greek capital letter Omega (Ω)
```

## <a name="17-7-traversing-strings-iterating-through-characters"></a>17.7 Traversing Strings: Iterating Through Characters
Traversing a string means visiting each character in it, one by one, from the first index to the last. Since a string is a sequence with every character sitting at a known index, traversal lets you print characters individually, count them, search for specific ones, or apply logic to each one in turn — for example, counting vowels or checking for uppercase letters. In Python, this is typically done using either a for loop or a while loop.
Method 1: Direct Character Iteration (The for Loop)
The most common and clean way to traverse a string is by using a for loop. This pattern treats the string directly as an iterable sequence. Python automatically extracts each character, assignment-by-assignment, without requiring you to manually manage index numbers.


```python
msg = "Python"
# Iterating directly over the characters
for char in msg:
    print(char, end="-")
# Output: P-y-t-h-o-n-
```

### <a name="method-2-index-based-iteration-the-while-loop"></a>Method 2: Index-Based Iteration (The while Loop)
Alternatively, you can traverse a string by combining its index sequence with a loop counter variable. This approach is highly useful when your application needs to know the exact position index of a character while processing it, or when you need to skip characters dynamically.


```python
msg = "Python"
index = 0
# Iterating using index boundaries
while index < len(msg):
    print(f"Index {index}: {msg[index]}")
    index += 1
# Output: Index 0: P Index 1: y Index 2: t Index 3: h Index 4: o Index 5: n
```

### <a name="string-slicing"></a>String slicing

Let us look at how Python extracts sub-segments from our standard string s1 = "Hello World!" using different slicing syntax patterns.

```python
s1 = "Hello World!"

# Category 1: Standard Forward Slicing (Step = 1)
print(s1[0:5])   # Output: 'Hello'. Extracts indices 0 to 4
print(s1[:5])    # Output: 'Hello'. Omitted start defaults to 0
print(s1[6:])    # Output: 'World!'. Omitted end goes to the absolute end
print(s1[:])     # Output: 'Hello World!'. Creates complete copy of the string

# Category 2: Extended Slicing (Skipping and Reversing with Steps)
print(s1[::2])   # Output: 'HloWrd'. Takes every 2nd character: 0, 2, 4...
print(s1[::-1])  # Output: '!dlroW olleH'. Reverses the entire string

# Category 3: Negative Index Slicing
print(s1[-6:-1]) # Output: 'World'. Extracts from index -6 up to -2
print(s1[-7:])   # Output: 'World!'. From 7th last character to the very end

# Category 4: Reverse Slicing with Explicit Bounds
print(s1[11:5:-1])  # Output: '!dlroW'. Steps backward from index 11 down to 6
```

### <a name="1-the-ord-function-character-to-integer"></a>1. The ord() Function (Character to Integer)
The ord() function takes a single-character string as an argument and returns its corresponding integer Unicode code point.


```python
# Finding the code points of English letters
print(ord('A'))    # Output: 65
print(ord('B'))    # Output: 66
print(ord('a'))    # Output: 97

# Finding the code point of a space and a symbol
print(ord(' '))    # Output: 32
print(ord('$'))    # Output: 36

# Finding the code point of an emoji
print(ord('🐍'))   # Output: 128013
```

### <a name="why-this-matters-for-string-comparisons"></a>Why This Matters for String Comparisons
Understanding code points explains why string comparison results can sometimes surprise beginners. Let us look at two common rules:
1.	Uppercase letters are "smaller" than lowercase letters: Because ord('A') is 65 and ord('a') is 97, evaluating 'A' < 'a' results in True.
2.	Alphabetical comparison is strictly index-by-index: Python compares strings starting from index 0. If characters match, it moves to index 1, and so on, until it finds a difference.


```python
# Comparing case
print("Is 'Apple' < 'apple'?", 'Apple' < 'apple')  # Output: True (as 65 < 97)

# Step-by-step comparison
# 'B' (66) is greater than 'A' (65), so Python stops immediately
print("Is 'Blue' greater than 'Apple'?", 'Blue' > 'Apple')   # Output: True
```

### <a name="a-word-of-caution-find-versus-index"></a>A word of caution: .find() versus .index()

A Word of Caution: .find() vs .index(): Python actually gives you two ways to search for a substring: .find() and .index(). Both search left to right, and both return the index of the first match. The difference is entirely in how they handle failure.
•	.find() returns -1 if the substring isn't found.
•	.index() raises a ValueError and crashes your program if the substring isn't found.


```python
text = "abracadabra"
print(text.find("xyz"))   # Output: -1. Returns quietly
print(text.index("xyz"))  # Raises: ValueError: substring not found
```

### <a name="7-checking-beginnings-and-endings-startswith-and-endswith"></a>7. Checking Beginnings and Endings: .startswith() and .endswith()
Often you don't need to search for a substring anywhere in a string—you only care whether it appears at a specific end. Checking a file's extension, verifying a URL's protocol, or confirming that a command begins with a particular keyword are all "edge" checks rather than "anywhere" checks. Python provides .startswith() and .endswith() for exactly this, and both return a Boolean.


```python
# Demonstrating .startswith() and .endswith()

filename = "report_2026.pdf"
url = "https://www.python.org"

print(filename.startswith("report"))  # Output: True
print(filename.endswith(".pdf"))      # Output: True
print(filename.endswith(".docx"))     # Output: False

print(url.startswith("https"))        # Output: True

# A common real-world use: routing based on file type
if filename.endswith(".pdf"):  # True if the filename ends with ".pdf"
    print("Opening as a PDF document.")
elif filename.endswith(".docx"):  # True if the filename ends with ".docx"
    print("Opening as a Word document.")
else:                             # If the filename doesn't match known types
    print("Unsupported file type.")
```

### <a name="checking-what-a-string-contains-isalpha-isdigit-and-more"></a>Checking what a string contains: .isalpha(), .isdigit() and more

•	`.isalpha()` — Returns True if the string contains only letters (a–z, A–Z) and has at least one character.

•	`.isdigit()` — Returns True if the string contains only digits (0–9) and has at least one character.

•	`.isalnum()` — Returns True if the string contains only letters and/or digits (no spaces, no punctuation) and has at least one character.

•	`.isspace()` — Returns True if the string contains only whitespace characters (spaces, tabs, newlines) and has at least one character.

•	`.isupper()` — Returns True if all the cased characters in the string are uppercase (a string with no cased characters at all returns False).

•	`.islower()` — Returns True if all the cased characters in the string are lowercase.

Example script showing implementation of these string methods:


```python
print("abcd".isalpha())   # Output: True. Only letters
print("1234".isdigit())   # Output: True. Only digits
print("abc123".isalnum())   # Output: True. Letters and digits, no space
print("   ".isspace())   # Output: True. Only whitespace
print("HELLO".isupper())   # Output: True. All cased characters are uppercase
print("hello".islower())   # Output: True. All cased characters are lowercase

# A common real-world use: validating input before conversion
age_input = input("Enter your age: ")

if age_input.isdigit():
    age = int(age_input)
    print(f"You are {age} years old.")
else:
    print("Please enter a valid number.")
```

## <a name="17-13-checking-for-palindromes"></a>17.13 Checking for Palindromes
Algorithmic Execution Steps (Using Negative Indexing)
-  Step 1: Calculate the absolute length of the input string (n).
-  Step 2: Compute the comparison limit endpoint: "comp"=n//2.

-  Step 3: Initialize a loop tracking counter variable i=0.
-  Step 4: While i<"comp" , systematically extract and compare the forward position character at index i with its mirror negative index relative position at -(i + 1).
-  Step 5: Evaluate the match condition:
	If a pair mismatch is detected, terminate immediately and return False.
	If the pair matches, increment the loop counter (i=i+1) and check the next inner layer.
-  Step 6: If the loop exits naturally without hitting a mismatch, the string is verified as symmetrical. Return True.
Python Implementation Script

This implementation mirrors the exact tracking variables, conditional diamond checks, and counter increments defined in our algorithmic flowchart.
The script (Using negative index) to check for a palindrome is given below:


```python
def is_palindrome(in_string):
    print(f"\nEvaluating Input String: '{in_string}'")
    n = len(in_string)  # Length of the input string
    comp = n // 2  # Number of comparisons needed (half the length of string)

    i = 0  # Initialize the loop counter for comparisons
    while i < comp:  # Loop until we reach the midpoint of the string
        forward_char = in_string[i]  # Character from start of string
        backward_char = in_string[-(i + 1)]  # Character from end of string
        if forward_char != backward_char:  # Mismatch found, not a palindrome
            print(f"In iteration {i}: '{forward_char}' != '{backward_char}'")
            return False  # Return False immediately if a mismatch is found

        print(f" -> Iteration {i}: '{forward_char}' matches '{backward_char}'")

        i = i + 1  # Increment the loop counter for the next comparison

    # 8. Executed if the loop finishes cleanly (No path taken)
    # Flowchart Block: Result = True
    return True

# EXECUTION AND VERIFICATION TEST CASES

# Case 1: Valid Palindrome (Odd length)
print("Is racecar a palindrome?:", is_palindrome("racecar"))

# Case 2: Invalid Palindrome (Mismatched central character)
print("Is racexar a palindrome?:", is_palindrome("racexar"))
```

## <a name="17-14-method-2-slicing-and-direct-equality-comparison"></a>17.14 Method 2: Slicing and Direct Equality Comparison.


An alternative, highly "Pythonic" strategy to check for a palindrome is to generate a reversed copy of the entire string in memory and compare it directly to the original using the equality operator (==).

Rather than walking through index pairs using a loop, we leverage extended slicing with a negative step ([::-1]) to reverse the sequence in a single operational step.
Algorithmic Execution Steps (Using String Reversal)

-  Step 1: Accept the input string (s).
-  Step 2: Generate a fully reversed copy of the string using the expression rev = s[::-1].
-  Step 3: Evaluate the conditional equality check (s == rev):

    -  If the strings are identical (Yes), categorize the input as a Palindrome.
    -    If the strings do not match (No), categorize it as Not a Palindrome.


Python Implementation Script (Matching the Reversal Flowchart)
This script perfectly mirrors the linear logic and direct equality branch of your second flowchart.


```python
# SCRIPT: PALINDROME VALIDATION ALGORITHM (STRING REVERSAL METHOD)
# This script checks for a palindrome by creating a mirrored copy of the input
# string using extended slicing and performing a direct equality comparison.

def check_palindrome_reversal(s):
    print(f"\nEvaluating Input String (s): '{s}'")


    # Creates a reversed copy of the string using a stride/step of -1
    rev = s[::-1]
    print(f" -> Generated Reversed String (rev): '{rev}'")

    if s == rev:
        # Flowchart Block: Palindrome
        print(" -> Decision: Strings match! Result = True")
        return "Palindrome"
    else:
        # Flowchart Block: Not Palindrome
        print(" -> Decision: Strings mismatch! Result = False")
        return "Not Palindrome"

# EXECUTION AND VERIFICATION TEST CASES

# Case 1: Valid Palindrome Match
result_1 = check_palindrome_reversal("level")
print(f"Is 'level' a palindrome? {result_1}")

# Case 2: Mismatched Case Sincerity (Python is case-sensitive!)
result_2 = check_palindrome_reversal("madams")
print(f"Is 'madams' a palindrome? {result_2}")
```
