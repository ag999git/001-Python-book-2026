18.1.1 Singleton: The Lone Instance
The Singleton pattern ensures that a class has only one instance and provides a global point of access to it. This is invaluable when managing shared resources, such as a single connection to a database or a configuration manager that must remain consistent across your entire application.



````python
class Singleton:
    _instance = None  # Class variable to store the single instance
    def __new__(cls):
        if cls._instance is None:
            # Create the instance only if it does not exist
            cls._instance = super().__new__(cls)
        return cls._instance

s1 = Singleton()  # Create the first instance of Singleton
s2 = Singleton() # Create the second instance of Singleton
# s1 and s2 should point to the same instance of Singleton
# 'is' checks if both variables point to the exact same memory object
print(f"Are s1 & s2 same? {s1 is s2}")  # Output: Are s1 & s2 same? True
````

18.1.2 Factory Method: Choosing the Blueprint
The Factory Method provides an interface for creating objects but allows subclasses or methods to decide which class to instantiate. Instead of calling MyClass() directly, you call a method that returns the appropriate object based on the input parameters provided.


````python
class Dog:
    def greet(self):
        return "Woof!"  # Dog's greeting
class Cat:
    def greet(self):
        return "Meow!"  # Cat's greeting

# get_pet()→ factory function that returns pet instance based on the pet type
def get_pet(pet_type):
    if pet_type == "dog": # Return a Dog instance if the pet type is "dog"
        return Dog()
    return Cat()  # Return a Cat instance for any other pet type

# Create instances of Dog and Cat using the get_pet function
pet_dog = get_pet("dog")  # Create a dog instance
print(f"dog→ {pet_dog.greet()}")  # Output: dog→ Woof!

pet_cat = get_pet("cat")  # Create a cat instance
print(f"cat→ {pet_cat.greet()}")  # Output: cat→ Meow!
````

18.1.3 Abstract Factory Pattern—Creating Related Objects Together

Problem Without Abstract Factory: Suppose we create objects directly:
```python
button = WindowsButton()
checkbox = MacCheckbox()
```
This creates a problem.
The program now contains objects from different families:
•	Windows button
•	Mac checkbox
They may not look or behave consistently.
We need a way to ensure that all objects belong to the same family.
Abstract Factory Solution: We create a factory for each family.



````python
# ----- Objects/ Products Created by Factories -----

# Step 1: Create WindowsButton product class
class WindowsButton:
    def display(self):
        print("Displaying Windows Button")

# Step 2: Create WindowsCheckbox product class
class WindowsCheckbox:
    def display(self):
        print("Displaying Windows Checkbox")

# Step 3: Create MacButton product class
class MacButton:
    def display(self):
        print("Displaying Mac Button")

# Step 4: Create MacCheckbox product class
class MacCheckbox:
    def display(self):
        print("Displaying Mac Checkbox")

# ----- Abstract Factories -----

# Step 5: Create WindowsFactory class that creates Windows objects/products
class WindowsFactory:

    def create_button(self):
        return WindowsButton()

    def create_checkbox(self):
        return WindowsCheckbox()

# Step 6: Create MacFactory class that creates Mac objects/products
class MacFactory:

    def create_button(self):
        return MacButton()

    def create_checkbox(self):
        return MacCheckbox()

# ----- Client Code -----

# Step 7: Create client function that uses factory to create objects/products
def create_screen(factory):

    # Step 8: Create button using the received factory
    button = factory.create_button()

    # Step 9: Create checkbox using the received factory
    checkbox = factory.create_checkbox()

    # Step 10: Display the created button
    button.display()

    # Step 11: Display the created checkbox
    checkbox.display()

# Step 12: Create WindowsFactory object
windows_factory = WindowsFactory()

# Step 13: Create MacFactory object
mac_factory = MacFactory()

# Step 14: Use WindowsFactory to create Windows family products
create_screen(windows_factory)

# Step 15: Use MacFactory to create Mac family products
create_screen(mac_factory)
````

18.1.4 Builder: Constructing Piece-by-Piece
The Builder pattern is used when an object requires many steps to construct. Instead of a massive constructor with twenty parameters, you use a "builder" object. You call methods like .set_color(), .set_engine(), and .add_features(), and finally call .build() to receive your fully formed object.


````python
class Robot:
    def __init__(self):
        self.parts = []  # Initialize an empty list to hold the robot's parts
    def add_arm(self):
        self.parts.append("Arm")  # Add an "Arm" to the robot's parts list
        return self  # Return self→ allows chaining: .add_arm().add_arm()

# Create a Robot instance and add two arms to it
bot = Robot().add_arm().add_arm()  # Adds two arms to the robot
print(f"Robot parts: {bot.parts}")  # ['Arm', 'Arm']

# Can add another arm to the Robot instance
bot.add_arm()  # Adds another arm to the robot
print(f"After adding another arm: {bot.parts}")  # ['Arm', 'Arm', 'Arm']
````

18.1.5 Prototype: Cloning Efficiency
The Prototype pattern creates new objects by copying (cloning) an existing instance. This is highly efficient when the cost of initializing a new object from scratch is expensive (e.g., loading large datasets or querying a database).


````python
class Prototype:  # Prototype class→ shows Prototype design pattern
    def __init__(self):
        self.data = [1, 2, 3] # Complex data
    def clone(self):  # Clone method→ creates a copy of the object
        # Create a new instance and copy internal data manually
        new_obj = Prototype()  # Create a new instance of Prototype
        new_obj.data = list(self.data)  # Copy the data
        return new_obj  # Return the new object

original_obj = Prototype()  # Create an instance of Prototype
copy_obj = original_obj.clone()  # Clone the original object
# 'original_obj.data == copy_obj.data' checks if their data is the same
print(f"Data match: {original_obj.data == copy_obj.data}")
# 'original_obj is not copy_obj' verifies they are different objects in memory
print(f"Are they different objects? {original_obj is not copy_obj}")

````

18.2.1 Decorator
The Decorator pattern allows you to "wrap" an object to add new behaviors dynamically. Instead of changing the original class, you create a decorator that adds functionality before or after the original method runs. It's like adding a leather case to a smartphone—the phone still works, but now it has new features.
The following script shows the use of decorators in two different forms: (1) Form 1: The manual wrapper, (2) Form 2: Using the @ decorator.


````python
# --- THE DECORATOR LOGIC ---
class FrameDecorator:
    """This class adds a decorative frame around a greeting."""
    def __init__(self, target):
        self.target = target

    # Form 1: Explicit method call. Adds a frame around the greeting.
    def render(self):
        return f"*** Welcome, {self.target()}! ***"

    # Form 2: Magic method for @ syntax. Also adds a frame around the greeting.
    def __call__(self):
        return f"*** Welcome, {self.target()}! ***"

# --- FORM 1: MANUAL PATTERN (Object Wrapping) ---
def get_user_name():
    return "Alice"

# Manually wrapping the function
manual_wrapper = FrameDecorator(get_user_name)

# --- FORM 2: @ SYNTAX (Automatic Wrapping) ---
@FrameDecorator
def get_user_name_syntax():
    return "Bob"

# --- EXECUTION ---
print("Form 1 (Manual):", manual_wrapper.render())
print("Form 2 (Syntax):", get_user_name_syntax())

# Output:
# Form 1 (Manual): *** Welcome, Alice! ***
# Form 2 (Syntax): *** Welcome, Bob! ***
````

18.2.2 Adapter
The Adapter pattern acts as a bridge between two incompatible interfaces. If you have a legacy system that returns data in XML format, but your new application only understands JSON, an Adapter sits in the middle, converts the format, and allows the systems to talk to each other.
The following script shows how the Adapter pattern allows two incompatible classes to work together by creating a "translator" class.

````python
# The incompatible legacy class (expects a specific method name)
class OldPrinter:
    def print_legacy_format(self, text):
        return f"Old format: {text}"

# The Adapter: Bridges the gap
class PrinterAdapter:
    def __init__(self, old_printer):  # Store the old printer instance
        self.printer = old_printer

    # New method name that the client expects. Refers to old method internally.
    def print_new_format(self, text):  # Returns result of old printer's method
        print(f"Adapter: Converts to old printer's method for text: '{text}'")
        return self.printer.print_legacy_format(text)

# Usage
old_machine = OldPrinter()  # Create an instance of the old printer
# Use adapter to call new method name, which internally calls the old method
adapter = PrinterAdapter(old_machine)
# print_new_format()→ calls print_legacy_format() of OldPrinter
# Text "Hello World" internally passed to print_legacy_format()
output = adapter.print_new_format("Hello World")
print(output)  # Output: "Old format: Hello World"
````

18.2.3 Facade
The Facade pattern provides a simple, unified interface to a complex set of subsystems. Rather than a client having to interact with five different classes to get a result, the Facade provides a single "one-stop-shop" method that handles the complexity behind the scenes.
The following script implements the Facade pattern by providing a simple "front door" to a complex set of internal classes.


````python
# Internal complex subsystems
class Lights:
    def turn_on(self): # Turn on the lights
        return "Lights are on."

class Heater:
    def turn_on(self): # Turn on the heater
        return "Heater is on."

# The Facade: Provides a simple interface to the complex subsystems
class SmartHomeFacade:
    def start_morning(self):  # Start the morning routine
        # The client only calls this one method
        # returns a simple string indicating the actions taken
        return f"{Lights().turn_on()} {Heater().turn_on()}"

# Usage
home = SmartHomeFacade()  # Create a facade instance
print(home.start_morning())  # Output: "Lights are on. Heater is on."
````

18.2.4 Composite
The Composite pattern allows you to treat individual objects and groups of objects the same way. Think of a file system: a folder can contain files or other folders. By treating them uniformly, you can calculate the size of a folder by simply asking every item inside it (whether it's a file or another folder) for its size.
The following script shows how the Composite pattern allows you to treat a single object and a group of objects the same way by using a tree structure.


````python
class File:  # class File represents a file with a given size
    def __init__(self, size):
        self.size = size  # size of the file

    def get_size(self):  # This method returns the size of the file
        return self.size

class Folder: # class Folder can contain files and other folders
    def __init__(self):
        # We start with an empty list; we don't know the count yet
        self.contents = []

    # First add all the items to the folder
    def add(self, item):
        # This adds any object that has a 'get_size' method
        self.contents.append(item)  # Add the item to the folder's contents

    # Now we need to calculate the total size of the folder
    def get_size(self):  # This method returns the total size of the folder
        total = 0  # Initialize total size to 0
        # The folder iterates over whatever is inside
        for item in self.contents:
            total = total + item.get_size()
        return total  # Return the total size of all items in the folder

# Usage
my_folder = Folder()  # Creating a new folder
# Now add some files to folder. Use add() to add files of different sizes
my_folder.add(File(10)) # Adding a file of size 10
my_folder.add(File(50)) # Adding a file of size 50
my_folder.add(File(5))  # Adding a file of size 5

# Now get the total size of the folder. Use get_size()
total_size = my_folder.get_size()  # Calculate the total size of the folder
print(f"Total size: {total_size}")  # Total size: 65
````

18.2.5 18.3.5 Proxy
The Proxy pattern acts as a "placeholder" for another object. It controls access to the original object, allowing you to perform actions like lazy-loading (waiting until the object is actually needed), caching (saving results), or security checks before the real object is ever touched.


````python
from abc import ABC, abstractmethod

# Abstract Base Class (ABC)
# Defines the common interface that every image class must follow.
class Image(ABC):
    # Every subclass must provide its own implementation of display()
    @abstractmethod
    def display(self):
        pass

# Internal class that performs the actual work
# The leading underscore (_) indicates that this class is intended
# for internal use and should normally not be used directly.
class _RealImage(Image):
    # Load the image from disk
    def __init__(self, filename):
        self.filename = filename
        print(f"Loading '{self.filename}' from disk...")

    # Display the image
    def display(self):
        print("display() method of _RealImage called.")
        print(f"Displaying '{self.filename}'")

# Proxy class→ Controls access to the real image object.
class ImageProxy(Image):
    # Store the filename without loading the image
    def __init__(self, filename):
        self.filename = filename
        # Initially, the real image is not loaded, so we set it to None
        self.real_image = None

    # Display the image
    def display(self):
        print("display() method of ImageProxy called.")

        # Has the real image already been created?
        if self.real_image is None: # None means real image not created yet
            print("Proxy: Image not loaded yet.")
            # Lazy Loading: Load real image only when it is actually needed.
            self.real_image = _RealImage(self.filename)

        else:  # Means real image already created→ skip the loading step
            print("Proxy: Image already loaded. Skipping loading step.")

        # Forward the request to the real object
        self.real_image.display()

# Client Code→ client interacts with the proxy instead of the real object.
photo = ImageProxy("holiday.jpg")

# First request
print("First request:")
photo.display()

# Second request
print("Second request:")
photo.display()
````

18.4.1 Observer: The Pub-Sub Model
The Observer pattern defines a one-to-many dependency. When one object (the Subject) changes state, all its dependents (Observers) are notified and updated automatically.
In the Observer Pattern, there are two main participants:
Subject/Publisher  ─────────►  Observer/Subscribers
The Subject is the object whose state changes. It is also called Publisher because it "publishes" the information, that is, it sends the information to all who are registered with it.
The Observers are interested in those changes. The observers are called Subscribers because they have subscribed to receive updates.


````python
from abc import ABC, abstractmethod

# Subject (Publisher)
# Maintains a list of observers and notifies them of changes.
class WeatherStation:
    def __init__(self):
        # Initially, no observers are registered.
        self._observers = []

    # Register a new observer.
    def attach(self, observer):
        self._observers.append(observer)

    # Remove an observer.
    def detach(self, observer):
        self._observers.remove(observer)

    # Notify every registered observer.
    def notify(self, temperature):
        print("\nWeatherStation: Temperature changed.\n")
        for observer in self._observers:
            observer.update(temperature)

# Abstract Base Class (ABC)
class Observer(ABC):
    @abstractmethod
    def update(self, temperature):
        pass

# Observer 1. Must implement the update() method.
class PhoneDisplay(Observer):
    def update(self, temperature):
        print(f"Phone Display: Temperature = {temperature}°C")

# Observer 2. Must implement the update() method.
class LaptopDisplay(Observer):
    def update(self, temperature):
        print(f"Laptop Display: Temperature = {temperature}°C")

# Observer 3. Must implement the update() method.
class WeatherLogger(Observer):
    def update(self, temperature):
        print(f"From WeatherLogger: Recorded {temperature}°C")

# Client Code
station = WeatherStation()  # Create the subject.

# Create three observers.
phone = PhoneDisplay()
laptop = LaptopDisplay()
logger = WeatherLogger()

# Register all observers.
station.attach(phone)
station.attach(laptop)
station.attach(logger)

# Notify all observers.
station.notify(32)
````


18.4.2 Strategy: Swapping Algorithms
The Strategy pattern allows you to define a family of algorithms, encapsulate each one, and make them interchangeable. This lets the algorithm vary independently from the clients that use it.


````python
from abc import ABC, abstractmethod

# Abstract Base Class (ABC) Every payment strategy must implement pay() method
class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

# Concrete Strategy 1
class CreditCardPayment(PaymentStrategy):
    # Pay using a credit card
    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card.")

# Concrete Strategy 2
class BitcoinPayment(PaymentStrategy):
    # Pay using Bitcoin
    def pay(self, amount):
        print(f"Paid ₹{amount} using Bitcoin.")

# Concrete Strategy 3
class UPIPayment(PaymentStrategy):
    # Pay using UPI
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI.")

# Context. Uses whichever payment strategy is currently selected.
# strategy is passed to the ShoppingCart at runtime, allowing the client to
# change the payment method without modifying the ShoppingCart class.
class ShoppingCart:
    # Store the initial payment strategy
    def __init__(self, strategy):
        self.strategy = strategy

    # Change the payment strategy at runtime
    def set_strategy(self, strategy):
        self.strategy = strategy

    # Complete the payment
    def checkout(self, amount):
        print("ShoppingCart: Processing payment...")
        # Delegate the payment to the selected strategy
        self.strategy.pay(amount)

# Client Code

# Create a shopping cart that initially uses a credit card.
cart = ShoppingCart(CreditCardPayment())
print("First Checkout")
cart.checkout(1000)

# Switch to another payment method.
cart.set_strategy(BitcoinPayment())
print("Second Checkout")
cart.checkout(1000)

# Switch again without changing the ShoppingCart class.
cart.set_strategy(UPIPayment())
print("Third Checkout")
cart.checkout(1000)
````

18.4.3 Command Method
Command: Turns a request into a stand-alone object. This allows you to parameterize objects with different requests, queue operations, and support undoable operations.


````python
from abc import ABC, abstractmethod

# Abstract Base Class (ABC). Every command must implement execute().
class Command(ABC):
    @abstractmethod
    def execute(self):
        pass

# Receiver. The object that actually performs the work.
class Bulb:
    # Turn the bulb on
    def turn_on(self):
        print("bulb is ON.")

    # Turn the bulb off
    def turn_off(self):
        print("bulb is OFF.")

# Concrete Command (TurnOnCommand gets receiver (bulb) as a parameter)
class TurnOnCommand(Command):
    # Store the receiver (bulb)
    def __init__(self, bulb):
        self.bulb = bulb

    # Execute the request
    def execute(self):
        self.bulb.turn_on()  # Call the receiver's (bulb) turn_on() method

# Concrete Command (TurnOffCommand gets receiver (bulb) as a parameter)
class TurnOffCommand(Command):
    # Store the receiver (bulb)
    def __init__(self, bulb):
        self.bulb = bulb

    # Execute the request
    def execute(self):
        self.bulb.turn_off()  # Call the receiver's (bulb) turn_off() method

# Invoker
# Executes whichever command it receives.
class RemoteControl:
    def press_button(self, command):
        command.execute()

# Client Code

# Create the receiver.
my_bulb = Bulb()

# Create the commands.
turn_on = TurnOnCommand(my_bulb)
turn_off = TurnOffCommand(my_bulb)

# Create the invoker.
remote = RemoteControl()

# Execute different commands.
remote.press_button(turn_on)
remote.press_button(turn_off)
````

18.4.4 Iterator:
Provides a way to access the elements of an aggregate object sequentially without exposing its underlying representation.


````python
# Iterator class. Returns one element at a time.
class NumberIterator:

    # Step 1: Store the collection and initialise the position.
    def __init__(self, numbers):
        self.numbers = numbers  # Store the collection of numbers
        self.index = 0  # Initialize the position to first element (index 0)

    # Step 2: Return the next element.
    def __next__(self):
        # Stop when every element has been visited.
        if self.index >= len(self.numbers):  # Check if index is out of bounds
            raise StopIteration    # Raise StopIteration→ iteration complete
        # Retrieve the current element.
        value = self.numbers[self.index]  # Get the current element
        # Move to the next position.
        self.index += 1  # Increment the index to point to the next element
        return value

# Collection class
class NumberCollection:

    # Step 3: Store the collection.
    def __init__(self):
        self.data = [10, 20, 30, 40]  # Initialize collection with some numbers

    # Step 4: Return an iterator object.
    def __iter__(self):
        # Return an instance of NumberIterator initialized with collection data
        return NumberIterator(self.data)

# Client Code

# Step 5: Create the collection.
numbers = NumberCollection()

# Step 6: Iterate through the collection.
for value in numbers:
    print(value)
````

18.4.5 Template Method:
Defines the program skeleton of an algorithm in a method, called the template method, which defers some steps to subclasses.


````python
from abc import ABC, abstractmethod

# Abstract Base Class (ABC)
class ReportGenerator(ABC):

    # Step 1: Template Method. Defines the overall algorithm.
    def generate_report(self):
        self.open_report()
        self.write_content()
        self.save_report()

    # Step 2: Common step. Opens the report.
    def open_report(self):
        print("Opening report (common step)...")

    # Step 3: Common step. Saves the report.
    def save_report(self):
        print("Saving report (common step)...")

    # Step 4: Step that subclasses must implement (abstract method)
    @abstractmethod
    def write_content(self):
        pass

# 1st Concrete subclass. Overrides the write_content() method
# to provide specific behavior for sales reports.
class SalesReport(ReportGenerator):

    # Step 5: Write sales-specific content
    def write_content(self):
        print("Writing sales data (Specific to SalesReport)...")

# 2nd Concrete subclass. Overrides the write_content() method
# to provide specific behavior for student reports.
class StudentReport(ReportGenerator):

    # Step 6: Write student-specific content
    def write_content(self):
        print("Writing student marks (Specific to StudentReport)...")

# Client Code

# Step 7: Create a SalesReport object.
sales_report = SalesReport()  # Create an instance of SalesReport

# Step 8: Execute the template method. call→ generate_report()→ which calls
# open_report(), write_content(), and save_report(). The write_content() method
# will be the one defined in SalesReport, so→ print "Writing sales data...".
sales_report.generate_report()
# Step 9: Create a StudentReport object.
student_report = StudentReport()  # Create an instance of StudentReport

# Step 10: Execute the same template method. call→ generate_report()→ which calls
# open_report(), write_content(), and save_report(). The write_content() method
# will be the one defined in StudentReport, so→ print "Writing student marks...".
student_report.generate_report()
````

18.5.3 Mixins and Multiple Inheritance
A mixin is a small class that provides a specific feature. It is not usually designed to exist independently. Instead, it adds reusable behavior to another class.
Example:
A class may need: (1) logging ability, (2) serialization ability (Saving to disk), (3) validation ability.
Instead of creating a huge parent class, we can create small mixins.
Example script


````python
class LoggingMixin:
    # Mixin class that provides logging functionality
    def log(self, message):
        print("LOG:", message)

# class "User" has its own responsibility: managing user-related operations
# Logging is added as an extra capability using mixin
class User(LoggingMixin):  # Inherits from LoggingMixin to gain logging ability

    def __init__(self, name):
        self.name = name

    def register(self): # Use log method of LoggingMixin→ log user registration
        print(f"Registering user: {self.name}")
        self.log("User registration completed")

    def greet(self):  # Own responsibility of "User": greeting the user
        print(f"Hello {self.name}")

user = User("Alex")
user.register()
user.greet()
````

18.5.5 Duck Typing as an Alternative to Formal Interfaces
Many languages require explicit interfaces. Python often follows: "If it behaves like the required object, use it." This idea is called duck typing. The famous explanation: If something walks like a duck and talks like a duck, treat it as a duck.


````python
class Dog:
    def speak(self):  # Dog's speak method returns "Bark"
        return "Bark"

class Cat:
    def speak(self):  # Cat's speak method returns "Meow"
        return "Meow"

def make_sound(obj):  # make_sound() takes an object→ calls its speak() method
    print(type(obj).__name__, "says", obj.speak())

make_sound(Dog())  # Output: Dog says Bark
make_sound(Cat())  # Output: Cat says Meow
# Both objects work because they provide: speak()
````
