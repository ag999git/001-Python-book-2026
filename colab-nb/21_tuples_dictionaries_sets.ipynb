**Simulating Changes via Reassignment**

You cannot change a tuple object, but you can build a completely new tuple out of old parts and reassign it to your variable.


````python
t1 = (1, 2, 3)
print(id(t1))  #
t2 = t1 + (4, 5)  # Concatenation creates a brand new tuple object
print(t2)         # Output: (1, 2, 3, 4, 5)
print(id(t2)) # id of t2 is different from id of t1
print("Is t1 equal to t2?", t1 == t2)  # Output: False
````

**Extended Unpacking (Python 3+):**

Uses the `*` operator to capture remaining values into a flexible list structure.

````python
a, *rest = (1, 2, 3, 4)
print(a)      # Output: 1
print(rest)   # Output: [2, 3, 4] (Notice this unpacks into a mutable list)
print(type(rest))  # Output: <class 'list'>
````

**21.4 Creating tuples from iterables**

An iterable is an object that can return its elements one at a time. You can create tuples from iterables like:-
1. string

2. list

3. another tuple

4. dictionary (from keys and from values)

5. set


````python
# Creating tuples from different iterables

# 1. From a string
string_iterable = "hello"
tuple_from_string = tuple(string_iterable)
print("Tuple from string:", tuple_from_string)
# Output: ('h', 'e', 'l', 'l', 'o')

# 2. From a list
list_iterable = [1, 2, 3, 4, 5]
tuple_from_list = tuple(list_iterable)  # Converts list to tuple
print("Tuple from list:", tuple_from_list)
# Output: (1, 2, 3, 4, 5)

# 3. From another tuple
another_tuple_iterable = (10, 20, 30)
tuple_from_another_tuple = tuple(another_tuple_iterable)
print("Tuple from another tuple:", tuple_from_another_tuple)
# Output: (10, 20, 30)

# 4. From a dictionary (keys)
dict_iterable = {'a': 1, 'b': 2, 'c': 3}
tuple_from_dict = tuple(dict_iterable)
print("Tuple from dictionary keys:", tuple_from_dict)
# Output: ('a', 'b', 'c')

# 5. From a dictionary (values)
tuple_from_dict_values = tuple(dict_iterable.values())
print("Tuple from dictionary values:", tuple_from_dict_values)
# Output: (1, 2, 3)

# 6. From a set
set_iterable = {100, 200, 300}
tuple_from_set = tuple(set_iterable)
print("Tuple from set:", tuple_from_set)
# Output: (100, 200, 300) (order may vary)
````

**Script showing use of various operations on tuples**

The following script shows the use of various operations on tuples. Explanations wherever needed are provided in hash (#) comments.


````python
# Sample tuple
sample_tuple = (1, 2, 3, 4, 5)

# 1. Length of the tuple
length = len(sample_tuple)
print("Length of tuple:", length)  # Output: 5

# 2. Accessing elements by index
first_element = sample_tuple[0]
print("First element:", first_element)  # Output: 1
last_element = sample_tuple[-1]  # last element using -ve index
print("Last element:", last_element)  # Output: 5

# 3. Slicing the tuple
sliced_tuple = sample_tuple[1:4]  # Slices from index 1 to 3
print("Sliced tuple:", sliced_tuple)  # Output: (2, 3, 4)

# 4. Concatenation
tuple1 = (1, 2)
tuple2 = (3, 4)
concatenated_tuple = tuple1 + tuple2  # Concatenates two tuples
print("Concatenated", concatenated_tuple) # Output: (1, 2, 3, 4)

# 5. Repetition
repeated_tuple = tuple1 * 3 # Repeats the tuple three times
print("Repeated", repeated_tuple)  # Output: (1, 2, 1, 2, 1, 2)

# 6. Membership testing
is_member = 3 in sample_tuple # Checks if 3 is in the tuple
print("Is 3 in tuple?:", is_member)  # Output: True

# 7. Iteration
print("Iterating through tuple:")
for item in sample_tuple:  # Go over each element in tuple
    print(item)
# Output: 1 2 3 4 5 (each on new line)

# 8. Counting occurrences
count_of_2 = sample_tuple.count(2) # Counts occurrences of 2
print("Count of 2 in tuple:", count_of_2)  # Output: 1

# 9. Finding index of an element
index_of_4 = sample_tuple.index(4) # index of first 4
print("Index of 4 in tuple:", index_of_4)  # Output: 3

# 10. Finding maximum and minimum values
max_value = max(sample_tuple)
min_value = min(sample_tuple)
print("Max value:", max_value)  # Output: 5
print("Min value:", min_value)  # Output: 1

# 11. Summing elements
sum_of_elements = sum(sample_tuple)
print("Sum of elements:", sum_of_elements)  # Output: 15

# 12. Sorting elements
unsorted_tuple = (3, 1, 4, 2)
sorted_list = sorted(unsorted_tuple)
print("Sorted elements:", sorted_list)  # Output: [1, 2, 3, 4]

# 13. Unpacking tuples
a, b, c, d, e = sample_tuple # Unpacks values into variables
print("Unpacked values:", a, b, c, d, e)  # Output: 1 2 3 4 5

# 14. Extended unpacking
first, *middle, last = sample_tuple # Unpack to first middle last
# *middle collects all middle elements into a list
print("First:", first)  # Output: 1
print("Middle:", middle)  # Output: [2, 3, 4]. It is a list
print("Last:", last)  # Output: 5

# 15. Accessing nested tuples
nested_tuple = ((1, 2), (3, 4)) # 3 is in 2nd tuple 1st position
element_2_1 = nested_tuple[1][0] # Gets 1st element of 2nd tuple
print("Element at (1,0):", element_2_1)  # Output: 3

# 16. Equality check
tuple_a = (1, 2, 3)
tuple_b = (1, 2, 3)
are_equal = tuple_a == tuple_b
print("Are tuples equal?:", are_equal)  # Output: True

# 17. Identity check
are_identical = tuple_a is tuple_b
print("Tuples identical?:", are_identical) # Output: False

# 18. Reassignment
tuple_a = tuple_a + (4,) # Reassigns tuple_a to a new tuple
print("Reassigned tuple_a:", tuple_a)  # Output: (1, 2, 3, 4)
````

**21.5.2 Swapping tuples**


Traditional Way of Swapping, is to use temporary variables. This method works in all programming languages, as shown below:-


````python
n1 = 100
n2 = 200
temp = n1
n1 = n2
n2 = temp
print("n1 ->", n1, "n2 ->", n2)  # n1 -> 200 n2 -> 100
# Pythonic Way of Swapping (Using Tuples)
# Python allows swapping without a temporary variable.
t1, t2 = (1, 'one'), (2, 'two')
t1, t2 = t2, t1  # Swapping tuples
print("After swapping:")
print("t1:", t1)  # Output: (2, 'two')
print("t2:", t2)  # Output: (1, 'one')

````


**3. Collection Mechanics: Hashability and Dictionary Keys**

"In Python, a dictionary key must be immutable and hashable — you'll see the formal definition of hashability shortly, when dictionaries are introduced.

For now, the practical rule is: lists are mutable, so they're unhashable and can't be used as dictionary keys or set elements.

Tuples, however, are structurally immutable, making them excellent candidates for composite dictionary keys (like a `(latitude, longitude)` grid lookup).


The Hashability Condition: A tuple is only hashable if every single element inside it is also hashable. If a tuple contains a mutable object (like a list), the tuple itself becomes unhashable.


````python
# Valid: Tuple containing integers (hashable)
valid_key = (1, 2)
my_dict = {valid_key: "Coordinates A"}

# Invalid: Tuple containing a list (unhashable!)
invalid_key = (1, [2, 3])
# my_dict = {invalid_key: "Coordinates B"} -> Raises TypeError: unhashable type: 'list'
````

**Sorting, Reversing, and Simulating Depletions**

Because tuples are immutable, they completely lack in-place modification methods such as `.sort()` and `.reverse()`.

Similarly, you cannot pop or delete an element from a tuple. If you need to "remove" data, you must use slicing or comprehensions to build an entirely new tuple object that excludes the unwanted elements.


````python
# Simulating removal via slicing
original_tuple = ('a', 'b', 'c', 'd')
# We want to "remove" index 2 ('c')
new_tuple = original_tuple[:2] + original_tuple[3:]
print(new_tuple)  # Output: ('a', 'b', 'd')
````

**Dictionary comprehension examples**

````python
# 1: Mapping strings to their lengths
words = ["ant", "bat", "cat", "dog"]
word_lengths = {word: len(word) for word in words}
print(word_lengths)  # Output: {'ant': 3, 'bat': 3, 'cat': 3, 'dog': 3}

# 2: Mapping numbers to their cubes
numbers = [1, 2, 3, 4]
cube_dict = {num: num ** 3 for num in numbers}
print(cube_dict)  # Output: {1: 1, 2: 8, 3: 27, 4: 64}

# 3: Mapping letters to their ASCII values
letters = "ABCDE"
ascii_values = {ch: ord(ch) for ch in letters}
print(ascii_values)  # Output: {'A': 65, 'B': 66, 'C': 67, 'D': 68, 'E': 69}

# 4: Mapping even numbers to their squares. Using a condition in comprehension
even_squares = {x: x ** 2 for x in range(1, 10) if x % 2 == 0}
print(even_squares)  # Output: {2: 4, 4: 16, 6: 36, 8: 64}

# 5: Reversing a dictionary. Swapping keys and values in a dictionary
grades = {"Alice": 90, "Bob": 85, "Charlie": 95}
reverse = {value: key for key, value in grades.items()}
print(reverse)  # Output: {90: 'Alice', 85: 'Bob', 95: 'Charlie'}
````

The following script shows common errors/ pitfalls in dictionary comprehension.

**21.13 Traversing a Dictionary**

Traversing (also called iterating) a dictionary means visiting each element in the dictionary one at a time.

During traversal, you can read, process, or display the contents of the dictionary.

Since every dictionary entry consists of a key and its corresponding value, Python provides three common ways to iterate over a dictionary depending on what information you need.


````python
# Sample dictionary
my_dict = {"key1": "value-1", "key2": "value-2", "key3": "value-3"}

# Method 1: Traversing Keys (Default Behaviour)
for key in my_dict:
    print(f"Key: {key} | Value: {my_dict[key]}")

# Output→ Key: key1 | Value: value-1, Key: key2 | Value: value-2, Key: key3 | Value: value-3
# Explanation: When a dictionary is used directly in a for loop, Python
# automatically iterates over its keys.
# To obtain the corresponding value, use my_dict[key]

# Method 2: Traversing Values Only
for value in my_dict.values():
    print(value)
# Output (Shown on a single line): value-1; value-2; value-3
# Explanation: Use .values() when only values without the keys are needed.

# Method 3: Traversing Key-Value Pairs
for key, value in my_dict.items():
    print(f"{key} -> {value}")
# Output (On single line): key1 -> value-1, key2 -> value-2, key3 -> value-3
# Explanation: # The .items() method returns each dictionary entry as a
# (key, value) tuple. # Python automatically unpacks each tuple into the
# variables key and value. This is the recommended and most Pythonic way to
# traverse dictionaries when both the key and value are needed.
````

**Modifying a Dictionary During Traversal**

Do not add or remove keys while iterating over a dictionary.
```python
my_dict = {"A": 1, "B": 2, "C": 3}
for key in my_dict:
    if key == "B":
        del my_dict[key]
# Output
# RuntimeError: dictionary changed size during iteration.
```
Python raises this error because the dictionary structure changes while it is being traversed.

**Safe Technique**

```python
# If keys need to be deleted, iterate over a copy of the keys.
my_dict = {"A": 1, "B": 2, "C": 3}

for key in list(my_dict.keys()):
    if my_dict[key] == 2:
        del my_dict[key]

print(my_dict) # Output: {'A': 1, 'C': 3}
```
Using list(`my_dict.keys()`) creates a snapshot of the keys before the loop begins, allowing the original dictionary to be modified safely.


**21.14 Dictionary View Objects (Python 3)**

What are Dictionary View Objects?

In Python 3, the dictionary methods: (1) `dict.keys()` (2) `dict.values()` and (3) `dict.items()` do not return lists.

Instead, they return special objects called dictionary view objects.

A view object acts like a live window into the dictionary.

Unlike a list, it does not store a separate copy of the data. Instead, it always reflects the current contents of the dictionary.


````python
# Practical Demonstration
# Sample dictionary
d1 = {"k1": "v1", "k2": "v2", "k3": "v3"}

# 1. Create a keys view object
keys_view = d1.keys()
print(keys_view)  # Output: dict_keys(['k1', 'k2', 'k3'])
print(type(keys_view))  # Output: <class 'dict_keys'>

# 2. Convert the view into a list
keys_list = list(keys_view)
print(keys_list)  # Output: ['k1', 'k2', 'k3']

# 3. Modify the dictionary. Add a new key-value pair
d1["k4"] = "v4"
print("\nAfter adding a new key...")
# The view updates automatically
print(keys_view)  # Output: dict_keys(['k1', 'k2', 'k3', 'k4'])

# 4. The list remains unchanged
print(keys_list)  # Output: ['k1', 'k2', 'k3']
````

````python
# Practical Example
student = {"name": "Alice", "age": 20, "city": "Paris"} # Sample dictionary
print(len(student))  # Output: 3
empty_dict = dict()  # Creates an empty dictionary
print(empty_dict)  # Output: {}
print(type(student))  # Output: <class 'dict'>
print(str(student))  # Output: "{'name': 'Alice', 'age': 20, 'city': 'Paris'}"
print("age" in student)  # Output: True
print("marks" in student)  # Output: False

# student2 is a new dictionary.
student2 = {"city": "Paris", "age": 20, "name": "Alice"}
# '==' checks for equality of content
print(student == student2)  # Output: True
# while 'is' checks for identity (whether they are the same object in memory)
print(student is student2)  # Output: False
````

**21.16 Common Dictionary Methods**


Dictionary methods can be grouped according to their purpose.


````python
# Practical Demonstration
student = {"name": "Alice", "age": 20, "city": "Paris"} # Sample dictionary

# Safe access
print("age→", student.get("age"))  # Output: 20
print("marks→", student.get("marks", "Not Available"))  # Output: Not Available

# Keys
print("keys→", student.keys())  # Output: dict_keys(['name', 'age', 'city'])

# Values
print("values→", student.values())# Output: dict_values(['Alice', 20, 'Paris'])

# Items
print("items→", student.items())
# Output: dict_items([('name', 'Alice'), ('age', 20), ('city', 'Paris')])

# Traversal
for key, value in student.items():
    print(key, ":", value)
# Output: name : Alice, age : 20, city : Paris (each on new line)

# Update
student.update({"age": 21, "country": "India"})
print("updated→", student)
# Output: {'name': 'Alice', 'age': 21, 'city': 'Paris', 'country': 'India'}

# Copy
student_copy = student.copy()
print("copy→", student_copy)

# Pop
removed = student.pop("city")
print("removed→", removed)
print("after pop→", student)

# setdefault
student.setdefault("grade", "A")
print("after setdefault→", student)

# Clear
temp = student.copy()
temp.clear()
print("after clear→", temp)

# fromkeys
subjects = dict.fromkeys(["Math", "Science", "English"], 0)
print("fromkeys→", subjects)
````

## SETS

**The Classic `{}` Trap: Using `{}` to Create an Empty Set**

Many beginners assume that `{}` creates an empty set because non-empty sets are written using curly braces. However, in Python, {} creates an empty dictionary, not an empty set.

To create an empty set, always use the `set()` constructor.


````python
# Incorrect: Creates an empty dictionary
empty_literal = {}  # Creates an empty dictionary not a set

# Correct: Creates an empty set
empty_set = set()

print(type(empty_literal))  # Output: <class 'dict'>
print(type(empty_set))  # Output: <class 'set'>
````

**Note: `len()` and `type()` are general Python built-in functions.**


They are commonly used with sets but are not exclusive to them.

Examples:


````python
# 1. Creating a set using literals
colors = {"red", "green", "blue"}
print(colors)  # Output: {'red', 'green', 'blue'}

# 2. Creating an empty set
numbers = set()
print(numbers)  # Output: set()

# 3. Creating a set from a list
values = [10, 20, 20, 30, 30]
s = set(values)
print(s)  # Output: {10, 20, 30}

# 4. Using len() and type() to inspect sets
s = {10, 20, 30}
print(len(s))  # Output: 3
print(type(s)) # Output: <class 'set'>

````

**Script 1: Basic Set Comprehensions**

````python
# Step 1: Create a list with duplicate values
numbers = [1, 2, 3, 3, 4, 2, 5, 5, 5, 6, 4]
print("Original List:→", numbers)
# Output: Original List:→ [1, 2, 3, 3, 4, 2, 5, 5, 5, 6, 4]

# Step 2: Create a set using a comprehension
unique_numbers = {n for n in numbers}
print("Unique Values:→", unique_numbers)
# Output: Unique Values:→ {1, 2, 3, 4, 5, 6} (order may vary)

# Step 3: Create a set of squares
squares = {n * n for n in numbers}
print("Squares:→", squares)
# Output: Squares:→ {1, 4, 9, 16, 25, 36} (order may vary)

# Step 4: Create a set of odd squares
odd_squares = {n * n for n in numbers if n % 2 != 0}
print("Odd Squares:→", odd_squares)
# Output: Odd Squares:→ {1, 9, 25} (order may vary)
````

**Script 2: Nested Set Comprehension**

````python
# Step 1: A nested list
matrix = [
    [1, 2, 3, 4, 5, "cat"],
    [2, 3, 4, 5, 6, "dog"],
    [4, 5, 6, 7, 8, "cat"],
]

# Step 2: Collect all unique values (Set Comprehension)
unique_values_comprehension = {value
                 for row in matrix
                 for value in row}

print("Unique Values (Comprehension):→", unique_values_comprehension)

# Step 3: Using 2 nested loops to achieve the same result
unique_values_loop = set()
for row in matrix:          # Outer loop
    for value in row:       # Inner loop
        unique_values_loop.add(value)

print("Unique Values (Loop):→", unique_values_loop)
````

**21.24 Common mistakes**

Mistake 1 – Indexing a Set
```python
languages = {"Python", "Java", "C++"}

try:
    print(languages[0])
except TypeError as e:
    print(e)

# Output: 'set' object is not subscriptable
```
**Why did this happen?**

Lists, tuples and strings support indexing because they maintain the position of every element. A set is unordered. Since its elements do not have fixed positions, indexing is not supported.

Mistake 2 – Adding Mutable Objects
```python
try:
    s = {1, 2, [3, 4]}
except TypeError as e:
    print(e)

# Output: unhashable type: 'list'
```

**Why did this happen?**

Lists are mutable, meaning they can change after creation. Set elements must be hashable, so lists cannot be stored inside a set.
```python
# Correct Approach
s = {1, 2, (3, 4)}
print(s)  # Output: {1, 2, (3, 4)}
```
**A tuple is immutable, so it can be stored in a set.**

Rule of Thumb: You can store numbers, strings, booleans, and tuples inside a set (since they are immutable), but never lists, dictionaries, or other standard sets.

**Mistake 3- Assuming a set is ordered**

```python
colors = {"Red", "Green", "Blue"}
print(colors) # Output: {'Green', 'Blue', 'Red'}
```
You might wonder "Why didn't Python print them in the order I entered them?"
Explanation: set is unordered. The display order is not guaranteed and should not be relied upon. If the order changes, it does not mean that any data has been lost.




**Mistake 5: The "Nested Tuple" Trap**


A tuple is immutable, so beginners often assume all tuples can be put into a set.

However, if a tuple contains even a single mutable element (like a list), the entire tuple becomes unhashable.


````python
# Step 1: Create a tuple containing a mutable list
mutable_inside_tuple = (1, 2, [3, 4])

# Step 2: Attempt to add it to a set
try:
    bad_set = {10, mutable_inside_tuple}
except TypeError as e:
    # Hashability is recursive. List inside tuple makes it unhashable
    print("Error Caught:", e)
# Output: # Error Caught: unhashable type: 'list'
````

**Mistake 6: Attempting Set-in-Set Nesting (Use frozenset Instead)**

Because sets are mutable, you cannot create a set of sets directly.

Attempting to do so raises `TypeError: unhashable type: 'set'`.

To nest sets, you must convert the inner set into an immutable frozenset.


````python
# Step 1: Attempt direct nesting of mutable sets (Will fail)
try:
    invalid_nested_set = {{1, 2}, {3, 4}}
except TypeError as e:
    # Explanation: Sets are mutable and therefore unhashable
    print("Error Caught:", e)
# Output: Error Caught: unhashable type: 'set'

# Step 2: Correct approach using frozenset
valid_nested_set = {frozenset({1, 2}), frozenset({3, 4})}
print("Correct: using frozenset:", valid_nested_set)
# Output: Correct: using frozenset: {frozenset({1, 2}), frozenset({3, 4})}
````

**Membership Testing and Traversing a Set**

One of the most common operations on a set is checking whether an element is present.

Python provides the in and not in operators for this purpose. You can also visit every element in a set using a for loop.

Since sets are unordered, the elements may be displayed in a different order each time the program runs.

The following script demonstrates membership testing using in and not in, followed by traversing all the elements of a set using a for loop.


````python
# Create a set of programming languages
languages = {"Python", "Java", "C++", "JavaScript"}

# Membership testing
print("Is Python in the set?   ", "Python" in languages)  # Output: True
print("Is Ruby in the set?     ", "Ruby" in languages)  # Output: False
print("Is Ruby not in the set? ", "Ruby" not in languages)  # Output: True

# Traversing the set
print("\nLanguages in the set:")

for language in languages:
    print(language)
# Output: Python, Java, C++, JavaScript (order may vary)
````

**Timing Demonstration: Comparing Lookup Performance: List vs. Set**


One of the biggest advantages of a set is its fast membership testing. While a list usually searches its elements one by one, a set uses hashing to locate an element much more efficiently.

The following script compares the time taken to search for the same element in a large list and an equivalent set.

The exact timings may vary from one computer to another, but the set is usually much faster for large collections.

On average, membership testing in a set takes `O(1)` time, whereas a list typically requires `O(n)` time because it searches sequentially.


````python
import time

# Create a large list and an equivalent set
numbers_list = list(range(1000000))
numbers_set = set(numbers_list)

# Search for the last element
target = 999999

# Measure lookup time in the list
start = time.perf_counter()
target in numbers_list
list_time = time.perf_counter() - start

# Measure lookup time in the set
start = time.perf_counter()
target in numbers_set
set_time = time.perf_counter() - start

print("Lookup Time Comparison")
print(f"List : {list_time:.6f} seconds")  # Larger time due to linear search
print(f"Set  : {set_time:.6f} seconds") # Smaller time due to hash-based lookup
````

**Common Mistake 2: Modifying a Set While Iterating**


Do not add or remove elements from a set while iterating over it. Changing the size of the set during iteration raises a `RuntimeError`.


```python

            numbers.remove(num)    # Changing the set while looping

for num in numbers.copy():

        numbers.remove(num)

```

**21.31 Worked Practical Example:**

Students Enrolled in Elective Subjects
Suppose a class offers two elective subjects: Python Programming and Data Science. Some students have enrolled in only one subject, while others have enrolled in both.
Using Python sets, answer the following questions:

-  1.	Which students have enrolled in at least one elective? (`Union`)

-  2.	Which students have enrolled in both electives? (`Intersection`)

-  3.	Which students have enrolled only in Python Programming? (`Difference`)

-  4.	Which students have enrolled in exactly one elective? (`Symmetric Difference`)

-  5.	Is every member of the Project Team enrolled in Python Programming? (`Subset/Superset`)

-  6.	Do the Python Programming students and the Photography Club have any students in common? (`Disjoint Sets`)

Python Script


````python
# A. Students enrolled in two elective subjects
python_students = {"Alice", "Bob", "Charlie", "David"}
data_science_students = {"Charlie", "David", "Eve", "Frank"}
print("Python Programming:→", python_students)
# Output: {'Alice', 'Bob', 'Charlie', 'David'}
print("Data Science:→", data_science_students)
# Output: {'Charlie', 'David', 'Eve', 'Frank'}

# 1. Union: Students enrolled in at least one elective
all_students = python_students | data_science_students
print("Students in at least one elective→", all_students)
# Output: {'Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'}

# 2. Intersection: Students enrolled in both electives
both_subjects = python_students & data_science_students
print("Students enrolled in both electives→", both_subjects)
# Output: {'Charlie', 'David'}

# 3. Difference: Students enrolled only in Python Programming
python_only = python_students - data_science_students
print("Students enrolled only in Python Programming→", python_only)
# Output: {'Alice', 'Bob'}

# 4. Symmetric Difference: Students enrolled in exactly one elective
one_subject_only = python_students ^ data_science_students
print("Students enrolled in exactly one elective→",one_subject_only)
# Output: {'Alice', 'Bob', 'Eve', 'Frank'}

# B. Relationship operations
# Every member of the project team studies Python
# project_team is sub set of python_students
project_team = {"Charlie", "David"}

# 5. Is every project team member enrolled in Python?
print("Project team is a subset of Python students:→",
      project_team.issubset(python_students))
# Output: True

# 6. Does Python contain all project team members?
print("Python students form a superset of project team:→",
      python_students.issuperset(project_team))
# Output: True

# Photography Club has completely different students
photography_club = {"Grace", "Heidi"}

# 7. Do the two groups have any common students?
print("Python students and Photography Club are disjoint:→",
      python_students.isdisjoint(photography_club))
# Output: True
````

**21.32 Common Mistakes**

**Mistake 1: Using Set Operators with a List**

Set operators (`|, &, -, ^`) work only with sets.

If one operand is a list or another iterable, Python raises a `TypeError`.
```python
numbers = {1, 2, 3}
values = [3, 4, 5]

try:
    # Error: values is a list, not a set
    print(numbers & values)
except TypeError as e:
    print("Error:", e)
# Output: Error: unsupported operand type(s) for &: 'set' and 'list'

# Correct Approach 1: Use the method
print(numbers.intersection(values))
# Output: {3}

# Correct Approach 2: Convert the list to a set
print(numbers & set(values))
# Output: {3}
```
Explanation
•	Operators such as `|, &, -`, and `^` require both operands to be sets.
•	Methods such as `intersection()`, `union()`, and `difference()` can accept other iterables, including lists and tuples.

**Common Mistake 2: Confusing `A - B` with `B - A`**

The difference operation depends on the order of the sets.
```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

print("A - B =", set_a - set_b)  # Output: A - B = {1, 2}
print("B - A =", set_b - set_a)  # Output: B - A = {4, 5}
```
Explanation
-  A - B returns elements that are in A but not in B.
-  B - A returns elements that are in B but not in A.
-  Since the order changes, the results are different.
-  Remember: Set difference is not commutative.
`A - B ≠ B - A`


21.33 frozenset
What is a frozenset? A frozenset is an immutable version of a set.

Once it is created, you cannot add, remove, or modify its elements.
Since a frozenset cannot change, it is hashable. This allows it to be used as a dictionary key or as an element of another set.

**Remember:**
-  set `→` Mutable.
-  frozenset `→` Immutable


````python
# Step 1: Create two subject combinations
group1 = frozenset({"Python", "Mathematics"})
group2 = frozenset({"Physics", "Chemistry"})

# Step 2: Use frozensets as dictionary keys
# group1 and group2→ immutable & hashable→ can be used as dictionary keys
classrooms = {group1: "Room 101", group2: "Room 202"}
print("Classroom:", classrooms[group1])
# Output: Classroom: Room 101

# Step 3: Store frozensets inside another set
subject_groups = {group1, group2}
print("Number of subject groups:", len(subject_groups))  # Output: 2

# Step 4: Attempt to modify a frozenset
try:
    group1.add("English")
except AttributeError as e:
    print("Error:", e)
# Output: Error: 'frozenset' object has no attribute 'add'
````
