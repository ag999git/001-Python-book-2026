Linear search

````python
import random
import time
import matplotlib.pyplot as plt

# Linear search function. # It checks every item one by one.
# If the target found→ True. Else→ False
def linear_search(data, target):
    for item in data:
        if item == target:
            return True
    return False

sizes = [1000, 10000, 100000, 500000, 1000000] # Different values of n.
target = -1 # Keep target = -1 to ensure it is not found→ worst case scenario
times = [] # Stores search times for different list sizes

# Test linear search for different list sizes
for n in sizes:
    numbers = [] # Create an empty list
    for i in range(n): # Generate n random numbers
        value = random.randint(1, 100000)
        numbers.append(value)

    start = time.perf_counter() # Start measuring time
    linear_search(numbers, target) # Call linear_search() to search for target
    end = time.perf_counter() # Stop measuring time
    elapsed = end - start  # Calculate total time taken
    times.append(elapsed)  # Store the time for plotting later
    print("Items:", n, "| Time:", elapsed, "seconds")

# Plot: x-axis  → number of items (n), y-axis  → search time
# If algorithm is O(n)→ time should increase roughly in proportion to n.
plt.plot(sizes, times, marker="o")
plt.xlabel("Number of items (n)")
plt.ylabel("Search time (seconds)")
plt.title("Linear Search: n versus Time")
plt.grid(True)
plt.show()
````

Binary search
"Binary Search"=O(log⁡n)

````python
import random
import time
import matplotlib.pyplot as plt

# Binary search function. The list must already be sorted.
# It checks middle item→ removes half of remaining list after each check.
def binary_search(data, target):
    left = 0  # Initialize left pointer
    right = len(data) - 1  # Initialize right pointer
    while left <= right:
        mid = (left + right) // 2  # Find middle position
        if data[mid] == target:   # Target found
            return True

        elif target > data[mid]: # Target is on the right side
            left = mid + 1

        else:  # Target is on the left side
            right = mid - 1

    return False  # Target not found

sizes = [1000, 10000, 100000, 500000, 1000000] # Different values of n
times = []  # Store execution times for each list size 1000, 10000, ....

print("Binary Search Timing")
for n in sizes:
    numbers = []  # Create an empty list
    for i in range(n): # Generate n random numbers
        value = random.randint(1, 1000000)
        numbers.append(value)

    numbers.sort()   # Binary search requires sorted data
    target = -1  # Target = -1 to ensure it is not found→ worst case scenario
    start = time.perf_counter()     # Start timer
    binary_search(numbers, target)  # Perform binary search
    end = time.perf_counter()     # Stop timer
    elapsed = end - start  # Calculate total time taken
    times.append(elapsed)  # Store the time for plotting later
    print("Items:", n, "| Time:", elapsed, "seconds")

plt.plot(sizes, times, marker="o") # Plot input size versus search time
plt.xlabel("Number of items (n)")
plt.ylabel("Search time (seconds)")
plt.title("Binary Search: n versus Time")
plt.grid(True)
plt.show()
````

Bubble sort

````python
def bubble_sort(numbers):
    n = len(numbers)
    # Each pass places the one largest
    # remaining element at the end
    for pass_no in range(n - 1):
        # Compare neighboring items. Last pass_no items are already sorted
        # So→ (n - pass_no - 1) because we don't check the last pass_no items
        # Example if pass_no = 0, we check n - 1 items,
        # if pass_no = 1, we check n - 2 items, etc.
        for i in range(n - pass_no - 1):
            if numbers[i] > numbers[i + 1]: # Wrong order? Swap them
                numbers[i], numbers[i + 1] = (numbers[i + 1], numbers[i])

        print("After pass", pass_no + 1, ":",numbers)

numbers = [6,5,4,3,2,1]  # Worst case example. The list is in reverse order.
print("Original:→", numbers)
bubble_sort(numbers)  # Sort the list in place
print("Sorted:→", numbers)
````

The following script implements insertion sort ():

````python
def insertion_sort(numbers):
    # Start from the second item as first item is already considered sorted.
    for index in range(1, len(numbers)):
        key = numbers[index]  # Item that we want to insert
        position = index - 1  # Compare with items on the left

        # Move bigger items one position to the right
        while position >= 0 and numbers[position] > key:
            numbers[position + 1] = numbers[position]  # Move item to the right
            position = position - 1  # Move to the next item on the left

        # Insert key at its correct position
        numbers[position + 1] = key
        print("After inserting", key, ":", numbers)

numbers = [5, 4, 2, 3, 10, 9, 8, 1]  # Test the function
print("Original:", numbers)  # [5, 4, 2, 3, 10, 9, 8, 1]
insertion_sort(numbers)  # Sort the list in place
print("Sorted:", numbers)  # [1, 2, 3, 4, 5, 8, 9, 10]
````

Selection sort

````python
def selection_sort(numbers):
    n = len(numbers)  # Number of items in the list
    print("Original:", numbers)  # Print original list

    for position in range(n - 1): # Each pass fixes one position
        min_index = position # Assume current position has the smallest item
        for index in range(position + 1, n): # Search remaining unsorted part

            if numbers[index] < numbers[min_index]: # if True, found smaller item
                min_index = index  # Update index of smallest item

        # Put smallest item at correct position
        if min_index != position: # Swap smallest with item at current position
            temp = numbers[position] # Store current item in temporary variable
            numbers[position] = numbers[min_index]
            numbers[min_index] = temp

        print("After pass", position + 1, ":", numbers)

    return numbers

# Test selection sort
numbers = [5, 4, 2, 3, 7, 9, 10]
selection_sort(numbers)
print("Sorted:", numbers)  # Sorted: [2, 3, 4, 5, 7, 9, 10]
````

Part 2: Optimized Bubble Sort

````python
def bubble_sort_optimised(numbers):
    n = len(numbers)
    for pass_no in range(n - 1):
        is_swapped = False          # Assume the list is already sorted

        for i in range(n - pass_no - 1):  # Compare neighbouring items
            if numbers[i] > numbers[i + 1]: # Wrong order? Swap them
                numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]
                is_swapped = True   # A swap happened — list was not yet sorted

        print("After pass", pass_no + 1, ":", numbers)

        if not is_swapped:          # No swaps this pass — list is fully sorted
            print("Early exit after pass", pass_no + 1)
            break                   # Stop immediately — no more passes needed

# Test 1: Already sorted list — should exit after just 1 pass
numbers = [1, 2, 3, 4, 5, 6]
print("Original:", numbers)
bubble_sort_optimised(numbers)
print("Sorted:", numbers)

# Test 2: Reverse sorted list — worst case, all passes needed
numbers = [6, 5, 4, 3, 2, 1]
print("\nOriginal:", numbers)
bubble_sort_optimised(numbers)
print("Sorted:", numbers)
````

What is the Standard Library?
Python's creators wrote thousands of lines of highly optimized code in the C language to handle common programming tasks.
````python
# Without the Standard Library — counting manually takes multiple lines
fruits = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']

count = {}
for fruit in fruits:
    if fruit in count:
        count[fruit] += 1
    else:
        count[fruit] = 1
print(count)
# Output: {'apple': 3, 'banana': 2, 'orange': 1}

# With the Standard Library — Counter does the same in one line
from collections import Counter

result = Counter(fruits)
print(result)
# Output: Counter({'apple': 3, 'banana': 2, 'orange': 1})
````

17.7.3 defaultdict

````python
# But defaultdict automatically creates a default value when a key is missing. Example:
from collections import defaultdict
students = defaultdict(list)
students["Math"].append("John")
students["Science"].append("Mary")
print(students)
# Output: defaultdict(<class 'list'>, {'Math': ['John'], 'Science': ['Mary']})
````

Real-World Example: Grouping Students by Grade
One of the most common uses of defaultdict in real programs is grouping—taking a flat list of items and organizing them into categories. Without defaultdict, grouping requires several extra lines just to handle the first item in each new category. With defaultdict, the grouping logic collapses into a single line per item.
The example below takes a flat list of (student, grade) pairs and groups the students under their respective grades:


````python
from collections import defaultdict

# A flat list of (student name, grade) pairs
results = [
    ("Alice",   "A"),
    ("Bob",     "B"),
    ("Charlie", "A"),
    ("Diana",   "C"),
    ("Eve",     "B"),
    ("Frank",   "A"),
]

# --- Without defaultdict ---
# Must check whether the key exists before appending
grades_plain = {}
for student, grade in results:
    if grade not in grades_plain:       # Does this grade key exist yet?
        grades_plain[grade] = []        # No — create an empty list first
    grades_plain[grade].append(student) # Now it is safe to append

print("Without defaultdict:", grades_plain)
# Output: {'A': ['Alice','Charlie','Frank'], 'B': ['Bob','Eve'], 'C': ['Diana']}

# --- With defaultdict ---
# Missing key is handled automatically — no if-check needed
grades_default = defaultdict(list)
for student, grade in results:
    grades_default[grade].append(student)   # One line — no guard needed

print("With defaultdict:", grades_default)
# Output: defaultdict(<class 'list'>, {'A': ['Alice', 'Charlie', 'Frank'],
#                                      'B': ['Bob', 'Eve'],
#                                      'C': ['Diana']})
````

•	OrderedDict → provides additional ordering-related features
The question is why use OrderedDict? It is because it provides extra ordering operations.
A normal dict in modern Python maintains insertion order, but it does not provide special methods for directly manipulating that order. If we want to change the order, we have to manually remove and reinsert items.
OrderedDict provides built-in methods for order manipulation, such as move_to_end() and popitem(last=False). These operations can be simulated using a normal dictionary, but OrderedDict provides them directly as part of its API.
Example: Using a normal dict and OrderedDict to move an item to the end. Then using OrderedDict.move_to_end(target_key, last = False) to move the item at target_key to the beginning.


````python
data = {"a": 1, "b": 2, "c": 3}
value = data.pop("a")  # Remove item with key "a" & store in variable 'value'.
data["a"] = value  # Insert item back at the end of the dictionary.
print(data)  # Output: {'b': 2, 'c': 3, 'a': 1}
# We had to:
# 1.    remove the item
# 2.    store its value
# 3.    insert it again

# Using OrderedDict
from collections import OrderedDict
ordered_dict = OrderedDict({"a": 1, "b": 2, "c": 3})
# The parameter last=True (default) moves the item to the end,
# while last=False moves it to the beginning.
ordered_dict.move_to_end("a")
print(ordered_dict)  # Output: OrderedDict([('b', 2), ('c', 3), ('a', 1)])
ordered_dict.move_to_end("b", last=False)  # Move 'b' to the beginning
print(ordered_dict)  # Output: OrderedDict([('b', 2), ('a', 1), ('c', 3)])
````

Important Rules for ChainMap: Writing always affects only the first dictionary:
This is the most important rule to remember. When you add a new key or update an existing one through a ChainMap, the change goes exclusively into the first dictionary in the chain. The other dictionaries are never touched. In the example above, that means any writes go into user_settings, leaving default_settings completely unchanged—which is usually exactly what you want, but it can cause confusion if you forget it.


````python
from collections import ChainMap

default_settings = {"theme": "light", "font": "Arial", "volume": 5}
user_settings = {"theme": "dark"}

app_settings = ChainMap(user_settings, default_settings)

# Writing through ChainMap only affects the FIRST dict (user_settings)
app_settings["volume"] = 10

print(user_settings)
# Output: {'theme': 'dark', 'volume': 10}  ← changed
print(default_settings)
# Output: {'theme': 'light', 'font': 'Arial', 'volume': 5}  ← untouched
````

Simplified Signature of heappop()
heapq.heappop(heap)
heap → The heap list from which the smallest item has to be removed.
Items are removed in priority order, not insertion order. The heapq module helps us create priority queues efficiently. Example:


````python
import heapq
tasks = []  # Create an empty list to represent the priority queue
heapq.heappush(tasks, (1, "Emergency"))  # Task with 1→ (highest priority)
heapq.heappush(tasks, (3, "Normal"))  # Task with 3→ (lowest priority)
heapq.heappush(tasks, (2, "Important"))
print(heapq.heappop(tasks))  # Output: (1, 'Emergency')
````

Python does not have a built-in max-heap, but there is a standard, widely used workaround: store all values as their negatives. Since the min-heap always gives you the smallest value, storing -90 instead of 90 means the min-heap will surface -90 first—and when you negate it again on the way out, you get 90. The largest original value effectively becomes the smallest stored value.

````python
import heapq

# Min-heap (default behaviour)
# Smallest value is always retrieved first

scores_min = []
heapq.heappush(scores_min, 50)
heapq.heappush(scores_min, 90)
heapq.heappush(scores_min, 30)
heapq.heappush(scores_min, 75)

print("Min-heap — smallest first:")
print(heapq.heappop(scores_min))    # Output: 30
print(heapq.heappop(scores_min))    # Output: 50

# Max-heap (negation workaround)
# Store values as negatives → min-heap surfaces the most negative first
# Negate again on the way out to recover the original value

scores_max = []
heapq.heappush(scores_max, -50)   # Store 50 as -50
heapq.heappush(scores_max, -90)   # Store 90 as -90 most negative → get first
heapq.heappush(scores_max, -30)   # Store 30 as -30
heapq.heappush(scores_max, -75)   # Store 75 as -75

print("\nMax-heap — largest first:")
print(-heapq.heappop(scores_max))   # Negate on the way out → Output: 90
print(-heapq.heappop(scores_max))   # Output: 75
````

17.10.1 Queue—FIFO Queue
Queue (FIFO): A standard Queue follows the First-In, First-Out (FIFO) rule. It acts just like the deque from the collections module, but it is designed to be safe for multiple threads. You use put(item) to add an item to the back of the line, and get() to remove an item from the front. Example:


````python
import queue
my_queue = queue.Queue() # Create a FIFO queue
my_queue.put("Task 1") # Add an item to the queue
my_queue.put("Task 2") # Add another item to the queue
print(my_queue.get()) # Output: Task 1. Added first→ Removed first (FIFO)
print(my_queue.get()) # Output: Task 2. Added second→ Removed second (FIFO)
````

17.10.2 LifoQueue (LIFO - Stack)
A LifoQueue follows the Last-In, First-Out (LIFO) rule. Think of a stack of plates. When you add a plate, you put it on top. When you need a plate, you take the one from the top. The last plate you put on is the first one you take off. Just like the standard Queue, you use put() to add to the top and get() to remove from the top. Example:


````python
import queue
plate_stack = queue.LifoQueue() # Create a stack using LifoQueue
plate_stack.put("Red Plate") # Add a plate to the stack
plate_stack.put("Blue Plate")
print(plate_stack.get()) # Output: Blue Plate. Last in, first out (LIFO)
print(plate_stack.get()) # Output: Red Plate. First in, last out (FILO)
````

17.11 A Simple Thread-Safety Demo
The three examples above showed queue, LifoQueue, and PriorityQueue being used by a single thread—one piece of code running in sequence. That is useful for understanding the data structures, but it does not show the real reason the queue module exists: safe communication between multiple threads running at the same time.
A thread is a unit of work that Python can run concurrently with other threads inside the same program. Think of two workers sharing one whiteboard. If both try to write at exactly the same moment, the result is garbled. Thread-safety means the data structure manages this automatically—only one thread can modify it at a time, even if both try simultaneously.
The example below creates two worker threads. Each thread picks up tasks from a shared queue.Queue() and process them. Because queue.Queue() is thread-safe; both workers can read from the same queue simultaneously without corrupting each other's data:


````python
import queue
import threading

# Shared queue — both worker threads read from this
task_queue = queue.Queue()

# Worker function — each thread runs this
def worker(worker_id):
    while True:
        task = task_queue.get()         # Get next task safely
        if task is None:                # None is the stop signal
            break
        print(f"Worker {worker_id} processing: {task}")
        task_queue.task_done()          # Signal that this task is complete

# Create and start two worker threads
thread1 = threading.Thread(target=worker, args=(1,))
thread2 = threading.Thread(target=worker, args=(2,))
thread1.start()
thread2.start()

# Add tasks to the shared queue
for task in ["Task A", "Task B", "Task C", "Task D", "Task E"]:
    task_queue.put(task)

# Send stop signal — one None per worker thread
task_queue.put(None)    # Stops thread 1
task_queue.put(None)    # Stops thread 2

# Wait for both threads to finish
thread1.join()
thread2.join()

print("All tasks complete.")
````

17.14 dataclasses — Structured Data Classes
Many programs need classes that mainly store data.
Examples: (1) student records (2) product information (3) configuration settings (4) API responses
A normal class requires writing a lot of repeated code. The dataclasses module helps reduce this work.
The Boilerplate Problem: Suppose we create a simple Student class:


````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __repr__(self):
        return f"Student({self.name}, {self.age})"

print(Student("Alice", 20))  # Output: Student(Alice, 20)
````

What is @dataclass? @dataclass is a decorator that automatically creates common methods for a class. It can generate: (1) __init__() (2) __repr__() (3) __eq__() (comparison) and much more. Example:

````python
from dataclasses import dataclass
@dataclass
class Student:  # @dataclass decorator→ generates methods like __init__()
    name: str  # type annotations for fields
    age: int
    marks: float

s1 = Student("Anita", 20, 85.5)  # We did not have to write an __init__() method
print(s1)  # Output: Student(name='Anita', age=20, marks=85.5)
````

Default Values: Dataclasses can have default values. Example:

````python
from dataclasses import dataclass
@dataclass
class Student:
    name: str
    age: int = 18  # Default value for age is set to 18

s1 = Student("Anita")
print(s1)  # Output: Student(name='Anita', age=18)
````

Creating Default Values Safely: For simple values, we can write:
```python
@dataclass
class Student:
    name: str
    age: int = 18
```

But mutable objects such as lists need special handling because objects may accidentally share the same list.
```python
subjects = []  # This is unsafe:
```
The solution: Use field() as shown below:


````python
from dataclasses import dataclass, field
@dataclass
class Student:
    name: str
    subjects: list = field(default_factory=list)

s1 = Student("Anita")
s2 = Student("Bob")
s1.subjects.append("Python")  # This affects s1's subjects list, not s2's.

print(s1.subjects)  # Output: ['Python']
print(s2.subjects)  # Output: []. Each instance→ own separate list of subjects
````

frozen=True—Immutable Dataclasses
Sometimes, once you create an object, you don't want anyone to change its data. For example, if you pull a specific record from a database, you probably shouldn't be able to accidentally change the ID number.
You can make a dataclass "frozen" (completely read-only) by passing frozen=True to the decorator. This is similar to a namedtuple (which you might see in older Python code), but dataclasses are better because they allow you to set default values and use type hints.


````python
from dataclasses import dataclass
@dataclass(frozen=True)
class Student:
    name: str
    age: int

s1 = Student("Anita", 20)
s1.age = 25 # Output: FrozenInstanceError
# frozen=True makes the object behave like an immutable record.
````

The following script compares a standard recursive Fibonacci function against a cached one using time.perf_counter() to see the speed difference:

````python
from functools import lru_cache
import time

# 1. Standard slow recursive function
def fib(n):
    if n < 2: return n
    return fib(n - 1) + fib(n - 2)

# 2. Optimized version with caching
@lru_cache(maxsize=None)
def fib_cached(n):
    if n < 2: return n
    return fib_cached(n - 1) + fib_cached(n - 2)

# Test the uncached version
t0 = time.perf_counter()
fib(32)
print(f"Without cache: {time.perf_counter() - t0:.5f} seconds")

# Test the cached version
t1 = time.perf_counter()
fib_cached(32)
print(f"With cache:    {time.perf_counter() - t1:.5f} seconds")

# Inspect the cache — shows how many calls were stored and retrieved
print("Cache info:", fib_cached.cache_info())
````

17.15.2 functools.partial()—Fixing Function Arguments
Sometimes you have a function that takes several arguments, but in a particular part of your program, one of those arguments is always the same. Instead of passing that repeated argument every single time, partial() lets you create a new function with that argument permanently fixed.


````python
from functools import partial

# Basic example
def power(x, n):
    return x ** n

square = partial(power, n=2)    # n is permanently fixed to 2
cube   = partial(power, n=3)    # n is permanently fixed to 3

print(square(5))    # Output: 25
print(cube(3))      # Output: 27

# Real-world example: tax calculator
def calculate_tax(amount, rate):
    return amount * rate

# In India, GST on electronics is fixed at 18%
# Create a specialised function with rate locked in
electronics_tax = partial(calculate_tax, rate=0.18)

# Now call it with just the amount — rate is already fixed
print("Electronics Tax on 10000:", electronics_tax(10000))   # Output: 1800.0
print("Electronics Tax on 25000:", electronics_tax(25000))   # Output: 4500.0
print("Electronics Tax on 5000:", electronics_tax(5000))    # Output:  900.0
````

17.15.3 functools.partial()—Fixing Function Arguments
Sometimes we want a new function with some arguments already filled. Example:


````python
from functools import partial
def power(x, n):  # A function that raises x to the power of n
    return x ** n

square = partial(power, n=2)  # n is fixed to 2, so square(x) will compute x^2
print(square(5))  # Output: 25
````

17.15.4 functools.reduce() — Reducing to One Value
reduce() repeatedly applies a function to combine items. reduce() can use any function that accepts two arguments and returns a result. The simplified signature is:
```python
reduce(function, iterable)
```
It repeatedly applies:
```python
function(previous_result, next_item)
```
Example:


````python
from functools import reduce
def multiply(a, b):
    return a * b

result = reduce(multiply, [2, 3, 4, 5]) # Calculates 2 * 3 * 4 * 5 = 120
print(result)  # Output: 120
````

17.15.5 itertools.chain()—Combining Iterables
chain() joins multiple iterables into one sequence.
Simplified Signature of itertools.chain()
```python
itertools.chain(*iterables)
```

Arguments → `*iterables` means: it accepts any number of iterables. Each iterable can be:  (1) list, (2) tuple, (3) string, (4) range, (5) other iterable objects.
Note: chain() does not return a list as output. itertools.chain() returns a chain object, which is an iterator. If you want a list output, you need to cast it to list. Example:


````python
from itertools import chain
# Chaining 2 lists, a tuple, a string, & a range object into a single iterable
numbers = chain([1, 2],[3, 4], (5, 6), "ABC", range(7, 10))
print(list(numbers)) # Output: [1, 2, 3, 4, 5, 6, 'A', 'B', 'C', 7, 8, 9]
````

17.15.6 itertools.groupby()—Grouping Data
groupby() groups consecutive items having the same key. Usually, the data should be sorted first. Simplified Signature
```python
itertools.groupby(iterable, key=None)
```

Arguments: (1) iterable → The data to group (list, tuple, string, etc.) (2) key → A function that decides the group category


````python
from itertools import groupby
data = ["A", "A", "B", "B", "C", "C", "C"]
for key, group in groupby(data):
    print(key, list(group))
# Output:
# A ['A', 'A']
# B ['B', 'B']
# C ['C', 'C', 'C']
````

17.15.7 itertools.product()—Cartesian Product
product() gives all possible combinations. Example:


````python
from itertools import product
colors = ["Red", "Blue"]
sizes = ["Small", "Medium"]
print(list(product(colors, sizes)))
# Output: [('Red', 'Small'), ('Red', 'Medium'), ('Blue', 'Small'), ('Blue', 'Medium')]
````

17.15.8 itertools.combinations()—Choosing Without Repetition
combinations() gives all possible ways to choose a fixed number of items from a collection, where order does not matter, and each item is used only once.
Simplified signature:
```python
itertools.combinations(iterable, r)
```

r is the number of items to choose each time.


````python
from itertools import combinations

# Choose 2 players from a group of 4 to form pairs
players = ["Alice", "Bob", "Charlie", "Diana"]
pairs = list(combinations(players, 2))
print(pairs)
# Output:
# [('Alice', 'Bob'), ('Alice', 'Charlie'), ('Alice', 'Diana'),
#  ('Bob', 'Charlie'), ('Bob', 'Diana'), ('Charlie', 'Diana')]

# Choose 2 numbers from [1, 2, 3, 4]
print(list(combinations([1, 2, 3, 4], 2)))
# Output: [(1, 2), (1, 3), (1, 4), (2, 3), (2, 4), (3, 4)]
````

17.15.9 itertools.permutations()—All Arrangements
permutations() gives all possible orders. Example:


````python
from itertools import permutations
my_colors = ['red', 'green', 'blue']
# Create a list of all permutations of the colors
all_permutations = list(permutations(my_colors))
print(all_permutations)  # Total 3! = 6 permutations
# Output: [('red', 'green', 'blue'), ('red', 'blue', 'green'), ('green', 'red', 'blue'), ('green', 'blue', 'red'), ('blue', 'red', 'green'), ('blue', 'green', 'red')]
````

17.16 Data Serialization: pickle and json
So far in this chapter, all data exists only while your program is running. The moment it stops, everything is lost. Serialization is the process of saving Python objects to a file so they can be restored later—or sent to another program.
Python provides two main modules for this: pickle for Python-only storage and json for universal, human-readable storage.
pickle—Saving Python Objects
pickle converts any Python object—lists, dictionaries, dataclasses, custom objects—directly into bytes and saves them to a file. The reverse process (loading back) is called unpickling.


````python
import pickle

# Saving (pickling)
student = {"name": "Anita", "age": 20, "marks": [85, 90, 78]}

with open("student.pkl", "wb") as f:   # "wb" = write bytes
    pickle.dump(student, f)             # Save object to file

print("Saved.")

# Loading (unpickling)
with open("student.pkl", "rb") as f:   # "rb" = read bytes
    loaded = pickle.load(f)             # Restore object from file

print(loaded)
# Output: {'name': 'Anita', 'age': 20, 'marks': [85, 90, 78]}
````

json—Universal, Human-Readable Storage
json saves data as plain, readable text in a format understood by virtually every programming language. Open a .json file in any text editor, and you can read it directly—unlike a .pkl file, which is binary and unreadable.


````python
import json

# Saving to JSON in a file name "student.json"
student = {"name": "Anita", "age": 20, "marks": [85, 90, 78]}

with open("student.json", "w") as f:   # "w" = write text
    json.dump(student, f, indent=4)     # indent=4 makes the file readable

print("Saved.")

# Loading from JSON from the file "student.json"
with open("student.json", "r") as f:   # "r" = read text
    loaded = json.load(f)

print(loaded)
# Output: {'name': 'Anita', 'age': 20, 'marks': [85, 90, 78]}
````
